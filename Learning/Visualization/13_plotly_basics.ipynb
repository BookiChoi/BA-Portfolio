# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** Plotly is a fundamentally different kind of visualization library from everything in Sections 2-12: instead of producing a static image, every chart is an interactive HTML/JavaScript element. `plotly.express` (usually imported as `px`) is the high-level interface — like Seaborn, you hand it a DataFrame and column names and it builds the whole chart. The lower-level `plotly.graph_objects` interface is covered in Section 14.

**KR:** Plotly는 섹션 2~12의 모든 것과 근본적으로 다른 종류의 시각화 라이브러리입니다: 정적 이미지 대신 모든 차트가 인터랙티브 HTML/JavaScript 요소입니다. plotly.express(보통 px로 임포트)는 고수준 인터페이스입니다 -- Seaborn처럼 DataFrame과 컬럼명만 넘기면 전체 차트를 만들어줍니다. 저수준 인터페이스인 plotly.graph_objects는 섹션 14에서 다룹니다.

### Matplotlib/Seaborn vs Plotly (비교)

| Aspect / 항목 | Matplotlib/Seaborn (Sections 2-12) | Plotly (Sections 13-14) |
|---|---|---|
| Output / 출력 형태 | Static image / 정적 이미지 | Interactive web element (HTML/JS) / 인터랙티브 웹 요소 |
| Hovering / 마우스 오버 | Not possible, needs manual annotation / 불가능, 직접 주석 필요 | Automatic tooltip with exact values / 자동으로 정확한 값이 툴팁 표시 |
| Zoom / 확대·축소 | Not possible / 불가능 | Drag to zoom, double-click to reset / 드래그해 확대, 더블클릭으로 원상복구 |
| Legend / 범례 | Always shows everything / 항상 전체 표시 | Click an entry to hide/show that series / 범례 클릭으로 시리즈 숨기기·표시 |
| Save / 저장 | Image file only (.png, .svg) / 이미지 파일만 | Interactive HTML or static image / 인터랙티브 HTML 또는 정적 이미지 |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** A stakeholder reading a static chart can only see what YOU chose to annotate. A Plotly chart lets them explore on their own — hover a random point to check its exact value, zoom into a busy cluster, or click a legend entry to isolate one series — all without asking you to re-make the chart. This is especially valuable for anything shared as a link rather than presented live.

**KR:** 정적 차트를 보는 이해관계자는 당신이 주석을 단 부분만 볼 수 있습니다. Plotly 차트는 그들이 직접 탐색할 수 있게 합니다 -- 아무 점에나 호버해서 정확한 값을 확인하거나, 복잡한 구간을 확대하거나, 범례를 클릭해 시리즈 하나만 남겨볼 수 있습니다 -- 차트를 다시 만들어 달라고 요청할 필요 없이요. 발표로 보여주기보다 링크로 공유하는 자료라면 특히 유용합니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Dashboards embedded in an internal website, a Slack/email link someone can explore on their own, or any deliverable where the audience will look at it without you in the room to explain each point.

**KR:** 사내 웹사이트에 삽입할 대시보드, 슬랙·이메일로 공유해 직접 살펴볼 수 있는 링크, 또는 각 지점을 설명해줄 당신이 그 자리에 없는 상태로 보게 될 산출물이라면 어디든 해당됩니다.

# 📝 Syntax (문법)

## Basic Syntax (기본 문법)

````python
import plotly.express as px
import pandas as pd
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [820, 890, 950, 1020, 980, 1150]
df = pd.DataFrame({'month': months, 'revenue': revenue})

fig = px.line(df, x='month', y='revenue', title='Monthly Revenue', markers=True)
# Plotly has NO global rcParams -- font must be set per-figure
# Plotly에는 전역 rcParams가 없으므로 폰트는 Figure마다 설정해야 함
fig.update_layout(font=dict(family='AppleGothic'))
fig.show()
````

## Common Variations (자주 쓰는 변형)

````python
# color= splits data into multiple colored series automatically
# color=를 지정하면 데이터가 여러 색상 시리즈로 자동 분리됨
months2 = ['Jan', 'Feb', 'Mar', 'Apr']
online = [820, 890, 950, 1020]
offline = [650, 630, 600, 590]
df2 = pd.DataFrame({
    'month': months2 * 2,
    'revenue': online + offline,
    'channel': ['Online'] * 4 + ['Offline'] * 4
})

fig = px.line(df2, x='month', y='revenue', color='channel', markers=True,
              color_discrete_map={'Online': '#2E86AB', 'Offline': '#F4A261'})
fig.update_layout(font=dict(family='AppleGothic'), hovermode='x unified')
fig.show()

# px.bar with color= needs barmode='group' explicitly (default is 'relative' -- see Mistake 3 below)
# color=가 있는 px.bar는 barmode='group'을 명시해야 함 (기본값은 'relative' -- 아래 실수 3 참고)
fig2 = px.bar(df2, x='month', y='revenue', color='channel', barmode='group')
fig2.update_layout(font=dict(family='AppleGothic'))
fig2.show()
````

# 🧪 Small Examples (예제)

## Example 1 — px.line: Monthly Revenue Trend (월별 매출 추이)

````python
# Business question: how did monthly revenue trend, and can a viewer check the
# exact number for any month without you annotating every point?
# 비즈니스 질문: 월별 매출은 어떻게 변했으며, 매 점에 주석을 달지 않아도
# 보는 사람이 아무 달이나 정확한 숫자를 확인할 수 있는가?

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [820, 890, 950, 1020, 980, 1150]
df = pd.DataFrame({'month': months, 'revenue': revenue})

fig = px.line(df, x='month', y='revenue', title='Monthly Revenue Trend', markers=True)
fig.update_layout(font=dict(family='AppleGothic'))
fig.show()

print("Hover over any point in the rendered chart to see its exact value in a tooltip.")
print("렌더링된 차트에서 아무 점에나 마우스를 올리면 정확한 값이 툴팁으로 표시됩니다.")
````

## Example 2 — px.bar: Revenue by Region x Channel (지역×채널별 매출)

````python
# Business question: how does revenue compare across regions AND channels at once?
# 비즈니스 질문: 지역과 채널을 동시에 비교하면 매출이 어떻게 다른가?

regions = ['Seoul', 'Busan', 'Daegu', 'Incheon'] * 2
channel = ['Online'] * 4 + ['Offline'] * 4
sales = [520, 240, 195, 290, 220, 140, 120, 160]
df = pd.DataFrame({'region': regions, 'channel': channel, 'sales': sales})

# IMPORTANT: when color= is set, Plotly's default barmode is 'relative' (stacked-looking),
# NOT 'group' -- pass barmode='group' explicitly for side-by-side bars.
fig = px.bar(df, x='region', y='sales', color='channel', barmode='group',
             title='Revenue by Region x Channel',
             category_orders={'region': ['Seoul', 'Incheon', 'Busan', 'Daegu']},
             color_discrete_map={'Online': '#2E86AB', 'Offline': '#F4A261'},
             text='sales')
fig.update_layout(font=dict(family='AppleGothic'), legend_title_text='Channel')
fig.update_traces(texttemplate='%{text}', textposition='outside')
fig.show()

total_by_region = df.groupby('region')['sales'].sum().sort_values(ascending=False)
print("Total sales by region (지역별 총 매출):")
print(total_by_region)
````

## Example 3 — px.scatter: Branch Employees vs Revenue (지점 직원수-매출 관계)

````python
# Business question: which branch is both large (employees) and highly
# profitable (bubble size), and which region does it belong to?
# 비즈니스 질문: 어느 지점이 규모(직원 수)도 크고 수익성(버블 크기)도 높으며,
# 어느 지역에 속하는가?

branches = ['Gangnam', 'Hongdae', 'Sinchon', 'Jamsil', 'Suwon', 'Busan', 'Daegu']
employees = [45, 28, 22, 38, 30, 25, 20]
revenue = [850, 420, 380, 720, 480, 390, 310]
profit = [180, 65, 55, 140, 90, 60, 45]
region_type = ['Metro'] * 5 + ['Regional'] * 2
df = pd.DataFrame({'branch': branches, 'employees': employees, 'revenue': revenue,
                    'profit': profit, 'region_type': region_type})

fig = px.scatter(df, x='employees', y='revenue', size='profit', color='region_type',
                  hover_name='branch',
                  hover_data={'profit': ':,.0f', 'region_type': False},
                  title='Branch Employees vs Revenue (bubble size = profit)',
                  labels={'employees': 'Employees', 'revenue': 'Revenue ($K)'},
                  color_discrete_map={'Metro': '#2E86AB', 'Regional': '#F4A261'},
                  size_max=40)
fig.update_layout(font=dict(family='AppleGothic'), legend_title_text='Region Type')
fig.show()

df['efficiency'] = df['revenue'] / df['employees']
best = df.loc[df['efficiency'].idxmax()]
print(f"Highest revenue-per-employee: {best['branch']} ({best['efficiency']:.1f} $K/person)")
print(f"직원 1인당 매출 최고: {best['branch']} ({best['efficiency']:.1f} $K/인)")
````

## Example 4 — px.histogram: Order Amount Distribution by Customer Type (고객 유형별 주문 금액 분포)

````python
# Business question: how much do new customers' order amounts overlap with
# existing customers', and where's the dividing line?
# 비즈니스 질문: 신규 고객의 주문 금액은 기존 고객과 얼마나 겹치며, 그 경계는 어디인가?

np.random.seed(42)
new_customers = np.random.normal(350, 100, 200)
existing_customers = np.random.normal(550, 180, 200)
df = pd.DataFrame({
    'order_amount': np.concatenate([new_customers, existing_customers]),
    'customer_type': ['New'] * 200 + ['Existing'] * 200
})

fig = px.histogram(df, x='order_amount', color='customer_type', barmode='overlay',
                    opacity=0.6, nbins=25,
                    title='Order Amount Distribution by Customer Type',
                    labels={'order_amount': 'Order Amount ($)'},
                    color_discrete_map={'New': '#2E86AB', 'Existing': '#E63946'})
fig.update_layout(font=dict(family='AppleGothic'), legend_title_text='Customer Type')

new_mean = df[df['customer_type'] == 'New']['order_amount'].mean()
existing_mean = df[df['customer_type'] == 'Existing']['order_amount'].mean()
fig.add_vline(x=new_mean, line_dash='dash', line_color='#2E86AB')
fig.add_vline(x=existing_mean, line_dash='dash', line_color='#E63946')
fig.show()

print(f"New avg: {new_mean:.0f} / Existing avg: {existing_mean:.0f}")
print(f"신규 평균: {new_mean:.0f} / 기존 평균: {existing_mean:.0f}")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Forgetting `fig.show()` (fig.show() 호출 빼먹기)
A `px.line()`/`px.bar()`/etc. call only BUILDS the figure object; nothing renders until you call `.show()` on it, exactly like Matplotlib's `plt.show()`.  
px.line()/px.bar() 등의 호출은 Figure 객체를 만들 뿐이며, plt.show()와 마찬가지로 .show()를 호출해야 렌더링됩니다.

✅ **Fix / 해결:** End every chart-building cell with `fig.show()`.

### Mistake 2 — Expecting a Global Font Setting Like `plt.rcParams` (plt.rcParams 같은 전역 폰트 설정을 기대함)
Unlike Matplotlib's `plt.rcParams`, which applies to every chart drawn afterward, Plotly has no equivalent global setting — Korean labels break unless you call `fig.update_layout(font=dict(family='AppleGothic'))` on every single figure.  
이후 그리는 모든 차트에 적용되는 Matplotlib의 plt.rcParams와 달리, Plotly에는 동등한 전역 설정이 없습니다 -- 매 Figure마다 fig.update_layout(font=dict(family='AppleGothic'))를 호출하지 않으면 한글 라벨이 깨집니다.

✅ **Fix / 해결:** Wrap font setup in a small helper function you call after building every figure, if you're making several.

### Mistake 3 — `px.bar(..., color=...)` Defaults to `barmode='relative'`, Not `'group'` (barmode 기본값이 group이 아니라 relative)
When every x-value maps to only one color, `'relative'` looks identical to grouped bars, hiding the issue — but as soon as one x-value has multiple colors (e.g. one region with two channels), the bars stack instead of sitting side by side.  
하나의 x값에 색이 하나만 대응하면 relative가 그룹 막대와 똑같아 보여 문제가 드러나지 않습니다 -- 하지만 하나의 x값에 여러 색이 대응하는 순간(예: 한 지역에 두 채널), 막대가 나란히 서는 대신 쌓여버립니다.

✅ **Fix / 해결:** Always pass `barmode='group'` explicitly whenever you use `color=` on `px.bar`.

### Mistake 4 — Comparing Multiple Lines with the Default Hover Mode (기본 hover 모드로 여러 선 비교하기)
The default `hovermode` only shows the value of the single series closest to your cursor — comparing two lines month by month means moving your mouse back and forth repeatedly.  
기본 hovermode는 커서에 가장 가까운 시리즈 하나의 값만 보여줍니다 -- 두 선을 월별로 비교하려면 마우스를 계속 왔다 갔다 해야 합니다.

✅ **Fix / 해결:** Set `fig.update_layout(hovermode='x unified')` to show every series' value at a given x in one tooltip.

### Mistake 5 — Negative Values in a `size=` Column for `px.scatter` (px.scatter의 size= 컬럼에 음수 값)
Bubble size must be non-negative; a column containing negative numbers (e.g. profit/loss) raises a `ValueError` when passed directly to `size=`.  
버블 크기는 반드시 0 이상이어야 합니다 -- 음수가 포함된 컬럼(예: 손익)을 size=에 그대로 넘기면 ValueError가 발생합니다.

✅ **Fix / 해결:** Transform the column first (e.g. `.abs()` or a separate magnitude column) before using it as `size=`.

### Mistake 6 — Using `trendline='ols'` Without Installing statsmodels (statsmodels 미설치 상태로 trendline='ols' 사용)
`px.scatter(..., trendline='ols')` depends on the separate `statsmodels` package — without it installed, the call raises an import error at runtime.  
px.scatter(..., trendline='ols')는 별도 패키지인 statsmodels에 의존합니다 -- 설치되어 있지 않으면 실행 시 import 에러가 발생합니다.

✅ **Fix / 해결:** Run `pip install statsmodels` first, or skip `trendline=` and use a Section 11 `regplot` when you don't need interactivity for that specific view.

# 💡 Tips (팁)
Useful tips or shortcuts / 유용한 팁과 단축법

- `update_layout(font=dict(family=...))` has to be called per-figure — wrap it in a small helper if you build many charts.<br>update_layout(font=dict(family=...))는 Figure마다 호출해야 합니다 -- 여러 차트를 만든다면 작은 헬퍼 함수로 감싸두세요.
- `hovermode='x unified'` is the single best setting for comparing multiple lines/series at a glance.<br>hovermode='x unified'는 여러 선·시리즈를 한눈에 비교할 때 가장 유용한 설정입니다.
- `color_discrete_map` fixes category-to-color assignment, so colors don't shuffle around if the data's row order changes.<br>color_discrete_map은 범주-색상 매핑을 고정해, 데이터의 행 순서가 바뀌어도 색이 뒤섞이지 않게 합니다.
- `fig.write_html('name.html')` keeps a chart fully interactive when shared as a file; `fig.write_image()` needs `pip install kaleido` and produces a static image instead.<br>fig.write_html('name.html')은 파일로 공유해도 차트가 완전히 인터랙티브하게 유지됩니다; fig.write_image()는 pip install kaleido가 필요하며 정적 이미지를 만듭니다.
- Clicking a legend entry in the rendered chart toggles that series on/off — genuinely useful during a live walkthrough ("let's look at just Online for a second").<br>렌더링된 차트에서 범례 항목을 클릭하면 그 시리즈를 껐다 켤 수 있습니다 -- 실시간 설명 중("잠깐 온라인만 봅시다") 실제로 유용합니다.

# 💼 Business Example (비즈니스 예시)
*How would a Business Analyst use this?*

**Scenario / 시나리오:**

**EN:** Your sales director is dialing into a call from the airport and can't view a static PDF slide clearly on their phone. They ask for a link they can open in any browser, hover over any month to see the exact number, and toggle "Online" or "Offline" off if they want to focus on one channel — all without you re-sending anything.

**KR:** 영업 이사님이 공항에서 통화로 참여하는데, 정적인 PDF 슬라이드를 휴대폰에서는 제대로 볼 수 없습니다. 어떤 브라우저에서든 열 수 있고, 아무 달이나 마우스를 올리면 정확한 숫자를 볼 수 있으며, 한 채널에 집중하고 싶으면 "온라인"이나 "오프라인"을 꺼버릴 수 있는 링크를 요청합니다 -- 다시 보낼 필요 없이요.

**To-do / 할 일:**
- [x] Build a channel comparison line chart with a unified hover tooltip showing both series at once<br>두 시리즈를 한 번에 보여주는 통합 호버 툴팁이 있는 채널 비교 선 그래프 만들기
- [x] Add a horizontal target line so performance vs goal is visible at a glance<br>목표 대비 성과가 한눈에 보이도록 수평 목표선 추가하기
- [x] Save it as an interactive HTML file the director can open and explore without you<br>이사님이 직접 열어 탐색할 수 있는 인터랙티브 HTML 파일로 저장하기

````python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug']
online = [820, 890, 950, 1020, 980, 1150, 1230, 1300]
offline = [650, 630, 600, 590, 570, 550, 530, 510]

df = pd.DataFrame({
    'month': months * 2,
    'revenue': online + offline,
    'channel': ['Online'] * 8 + ['Offline'] * 8
})

fig = px.line(df, x='month', y='revenue', color='channel', markers=True,
              title='Channel Revenue Trend',
              labels={'revenue': 'Revenue ($K)', 'month': ''},
              color_discrete_map={'Online': '#2E86AB', 'Offline': '#F4A261'})
fig.update_layout(font=dict(family='AppleGothic'), hovermode='x unified')
fig.add_hline(y=1000, line_dash='dash', line_color='gray',
              annotation_text='Monthly Target', annotation_position='bottom right')
fig.update_traces(line=dict(width=3), marker=dict(size=9))
fig.show()

fig.write_html('channel_revenue_interactive.html')
print("Saved: channel_revenue_interactive.html -- open in any browser, hover any month, click legend to toggle a channel")
print("저장됨: channel_revenue_interactive.html -- 아무 브라우저에서나 열어 아무 달이나 호버, 범례 클릭으로 채널 토글 가능")

online_growth = (online[-1] - online[0]) / online[0] * 100
offline_growth = (offline[-1] - offline[0]) / offline[0] * 100
print(f"Online growth: {online_growth:+.1f}% / Offline growth: {offline_growth:+.1f}%")
print(f"온라인 성장률: {online_growth:+.1f}% / 오프라인 성장률: {offline_growth:+.1f}%")
````

**EN:** `plotly.express` (`px`) is a high-level interface, like Seaborn, that turns a DataFrame plus column names into a fully interactive chart with hover tooltips, zoom, and clickable legends — the fundamental difference from everything in Sections 2-12 is the OUTPUT (interactive HTML/JS vs. a static image), not the input style. `px.line`, `px.bar`, `px.scatter`, and `px.histogram` cover the same four use cases as Sections 3-5's Matplotlib charts, but each comes with new gotchas: no global font setting, `px.bar`'s `barmode='relative'` default when `color=` is set, and the need for `hovermode='x unified'` to compare multiple series at once. `fig.show()` is required just like `plt.show()`, and `fig.write_html()` is how you hand off a chart that stays interactive after you close Python.

**KR:** plotly.express(px)는 Seaborn과 마찬가지로 DataFrame과 컬럼명을 완전히 인터랙티브한 차트(호버 툴팁, 확대, 클릭 가능한 범례)로 바꿔주는 고수준 인터페이스입니다 -- 섹션 2~12의 모든 것과의 근본적인 차이는 입력 방식이 아니라 출력 결과(정적 이미지 대신 인터랙티브 HTML/JS)입니다. px.line, px.bar, px.scatter, px.histogram은 섹션 3~5의 Matplotlib 차트와 같은 네 가지 용도를 다루지만, 각각 새로운 함정을 동반합니다: 전역 폰트 설정 없음, color=를 지정했을 때 px.bar의 barmode='relative' 기본값, 그리고 여러 시리즈를 한 번에 비교하려면 hovermode='x unified'가 필요하다는 점입니다. plt.show()와 마찬가지로 fig.show()가 필수이며, fig.write_html()은 Python을 닫은 뒤에도 인터랙티브 상태를 유지하는 차트를 넘겨주는 방법입니다.
