# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** `plotly.graph_objects` (usually imported as `go`) is the low-level interface behind everything `plotly.express` builds for you automatically. Instead of handing over a DataFrame and column names, you construct each chart element ("trace") yourself with `go.Scatter`, `go.Bar`, `go.Pie`, etc., and assemble them into a `go.Figure` one `add_trace()` call at a time. In fact, `px.line()` secretly returns a `go.Figure` under the hood — `px` is really just a shortcut that builds `go` objects for you.

**KR:** plotly.graph_objects(보통 go로 임포트)는 plotly.express가 자동으로 만들어주던 모든 것의 뒤에 있는 저수준 인터페이스입니다. DataFrame과 컬럼명을 넘기는 대신, go.Scatter, go.Bar, go.Pie 등으로 각 차트 요소("trace")를 직접 만들고 add_trace() 호출로 하나씩 go.Figure에 조립합니다. 사실 px.line()도 내부적으로는 go.Figure를 반환합니다 -- px는 사실상 go 객체를 대신 만들어주는 단축 도구일 뿐입니다.

### px vs go: when to use which (언제 무엇을 쓰는가)

| Situation / 상황 | Recommended / 추천 |
|---|---|
| One DataFrame, a standard chart type, quickly / 하나의 DataFrame으로 빠르게 표준 차트 | `px` (Section 13) |
| Combine different chart types (bar + line) in one Figure / 서로 다른 차트 종류를 한 Figure에 결합 | `go` |
| Each subplot needs completely different data/axis ranges / 서브플롯마다 완전히 다른 데이터·축 범위 | `go` + `make_subplots` |
| Viewer switches data via a dropdown/button/slider / 드롭다운·버튼·슬라이더로 데이터 전환 | `go`'s `updatemenus` / `sliders` |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** `px` is fast but opinionated — the moment you need two different chart types sharing one x-axis, a dashboard with a pie chart squeezed next to a bar chart, or a chart that lets someone pick "show revenue" vs "show visitor count" with one click, `px` doesn't have a parameter for that. `go` trades one-liner convenience for full control: every trace, axis, and interactive control is something you explicitly build, which is exactly what's needed for genuinely custom dashboards.

**KR:** px는 빠르지만 정해진 방식대로만 동작합니다 -- 하나의 x축을 공유하는 두 가지 다른 차트 종류, 파이 차트와 막대 차트가 나란히 있는 대시보드, 또는 클릭 한 번으로 "매출 보기"와 "방문자 수 보기"를 전환하는 차트가 필요한 순간, px에는 그럴 파라미터가 없습니다. go는 한 줄의 편리함 대신 완전한 제어권을 줍니다: 모든 trace, 축, 인터랙티브 컨트롤을 명시적으로 직접 만들며, 이는 진짜 커스텀 대시보드에 정확히 필요한 방식입니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Anytime the deliverable is a genuine dashboard rather than a single chart: a combo chart showing revenue (bars) against growth rate (a line on a second axis), a one-page KPI summary combining a trend, a composition breakdown, and a ranking, or a single shareable file where the recipient picks which metric to look at instead of you sending three separate charts.

**KR:** 산출물이 단일 차트가 아니라 진짜 대시보드일 때 언제나 해당됩니다: 매출(막대)과 성장률(두 번째 축의 선)을 함께 보여주는 콤보 차트, 추세·구성·순위를 결합한 한 페이지짜리 KPI 요약, 또는 세 개의 차트를 따로 보내는 대신 받는 사람이 직접 지표를 선택할 수 있는 파일 하나.

# 📝 Syntax (문법)

## Basic Syntax (기본 문법)

````python
import plotly.graph_objects as go

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [820, 890, 950, 1020, 980, 1150]

fig = go.Figure()
fig.add_trace(go.Scatter(x=months, y=revenue, mode='lines+markers', name='Revenue',
                          line=dict(color='#2E86AB', width=3)))
fig.update_layout(title='Monthly Revenue', font=dict(family='AppleGothic'))
fig.show()

# NOTE: px.line(df, x='month', y='revenue') does exactly this internally -- this is the
# same result, written by hand with go.Figure() + go.Scatter() + add_trace().
# px.line(df, x='month', y='revenue')이 내부적으로 하는 일을 go.Figure() + go.Scatter()
# + add_trace()로 직접 풀어 쓴 것과 동일한 결과입니다.
````

## Common Variations (자주 쓰는 변형)

````python
# Combine a bar trace and a line trace on TWO different y-axes in one Figure
# 막대 trace와 선 trace를 서로 다른 두 개의 y축에 하나의 Figure로 결합
months2 = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue2 = [820, 890, 950, 1020, 980, 1150]
growth = [None] + [(revenue2[i] - revenue2[i - 1]) / revenue2[i - 1] * 100 for i in range(1, len(revenue2))]

fig = go.Figure()
fig.add_trace(go.Bar(x=months2, y=revenue2, name='Revenue', marker_color='#A8D5E8', yaxis='y'))
fig.add_trace(go.Scatter(x=months2, y=growth, mode='lines+markers', name='MoM Growth (%)',
                          line=dict(color='#E63946', width=3), yaxis='y2'))
fig.update_layout(
    title='Revenue & Month-over-Month Growth',
    font=dict(family='AppleGothic'),
    yaxis=dict(title='Revenue ($K)'),
    yaxis2=dict(title='Growth (%)', overlaying='y', side='right'),
)
fig.show()
````

# 🧪 Small Examples (예제)

## Example 1 — go.Figure: Revenue & Growth Rate Combo Chart (매출 & 성장률 콤보 차트)

````python
# Business question: is revenue growing, and is that growth accelerating or slowing?
# 비즈니스 질문: 매출이 성장하고 있는가, 그 성장은 가속화되는가 둔화되는가?

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug']
revenue = [820, 890, 950, 1020, 980, 1150, 1230, 1300]
growth = [None] + [(revenue[i] - revenue[i - 1]) / revenue[i - 1] * 100 for i in range(1, len(revenue))]

fig = go.Figure()
fig.add_trace(go.Bar(x=months, y=revenue, name='Revenue', marker_color='#A8D5E8', yaxis='y',
                      text=revenue, textposition='outside'))
fig.add_trace(go.Scatter(x=months, y=growth, mode='lines+markers', name='MoM Growth (%)',
                          line=dict(color='#E63946', width=3), yaxis='y2'))
fig.add_hline(y=0, yref='y2', line_dash='dot', line_color='gray')

best_month_idx = growth.index(max(g for g in growth if g is not None))
fig.add_annotation(x=months[best_month_idx], y=growth[best_month_idx], yref='y2',
                    text='Best Growth', showarrow=True, arrowhead=2, ay=-40)

fig.update_layout(
    title='Monthly Revenue & Month-over-Month Growth',
    font=dict(family='AppleGothic'),
    yaxis=dict(title='Revenue ($K)'),
    yaxis2=dict(title='Growth (%)', overlaying='y', side='right'),
    legend=dict(x=0.01, y=1.15, orientation='h'),
)
fig.show()

valid_growth = [g for g in growth if g is not None]
avg_growth = sum(valid_growth) / len(valid_growth)
print(f"Average MoM growth: {avg_growth:.1f}%")
print(f"평균 전월 대비 성장률: {avg_growth:.1f}%")
print(f"Best growth month: {months[best_month_idx]} ({growth[best_month_idx]:.1f}%)")
print(f"최고 성장월: {months[best_month_idx]} ({growth[best_month_idx]:.1f}%)")
````

## Example 2 — make_subplots: Half-Year Sales Dashboard (상반기 매출 대시보드)

````python
# Business question: can leadership see the trend, the channel mix, and the
# product ranking on one page, each independently explorable?
# 비즈니스 질문: 경영진이 추세, 채널 구성, 제품 순위를 한 페이지에서
# 각각 독립적으로 탐색할 수 있는가?

from plotly.subplots import make_subplots

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [820, 890, 950, 1020, 980, 1150]
channels = ['Online', 'Offline', 'Partner']
channel_sales = [520, 380, 250]
products = ['Product A', 'Product B', 'Product C', 'Product D']
product_sales = [320, 280, 210, 150]

fig = make_subplots(
    rows=2, cols=2,
    specs=[[{'type': 'xy', 'colspan': 2}, None],
           [{'type': 'domain'}, {'type': 'xy'}]],
    subplot_titles=('Monthly Revenue Trend', 'Channel Mix', 'Product Ranking'),
    row_heights=[0.55, 0.45]
)
fig.add_trace(go.Scatter(x=months, y=revenue, mode='lines+markers', name='Revenue',
                          line=dict(color='#2E86AB', width=3)), row=1, col=1)
fig.add_trace(go.Pie(labels=channels, values=channel_sales,
                      marker_colors=['#2E86AB', '#E63946', '#F4A261'], hole=0.4), row=2, col=1)
fig.add_trace(go.Bar(x=product_sales[::-1], y=products[::-1], orientation='h',
                      marker_color='#2E86AB', text=product_sales[::-1], textposition='outside'),
              row=2, col=2)

fig.update_layout(title='Half-Year Sales Dashboard', font=dict(family='AppleGothic'),
                   showlegend=False, height=700)
fig.update_xaxes(title_text='Revenue ($K)', row=2, col=2)
fig.show()

total_revenue = sum(revenue)
top_product_share = product_sales[0] / sum(product_sales) * 100
print(f"6-month total revenue: ${total_revenue}K")
print(f"6개월 누적 매출: ${total_revenue}K")
print(f"Top product ({products[0]}) share of product sales: {top_product_share:.1f}%")
print(f"1위 제품({products[0]}) 매출 비중: {top_product_share:.1f}%")
````

## Example 3 — Interactive Dropdown: Switch Between KPIs (드롭다운으로 KPI 전환하기)

````python
# Business question: can ONE file let a viewer pick between revenue, visitors,
# and conversion rate themselves, instead of you sending three separate charts?
# 비즈니스 질문: 세 개의 차트를 따로 보내는 대신, 파일 하나로 매출·방문자수·
# 전환율을 보는 사람이 직접 선택하게 할 수 있는가?

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [820, 890, 950, 1020, 980, 1150]
visitors = [12000, 13500, 12800, 15200, 16800, 18200]
conversion = [1.8, 1.9, 2.1, 2.0, 2.4, 2.9]

fig = go.Figure()
fig.add_trace(go.Bar(x=months, y=revenue, name='Revenue', visible=True, marker_color='#2E86AB'))
fig.add_trace(go.Bar(x=months, y=visitors, name='Visitors', visible=False, marker_color='#E63946'))
fig.add_trace(go.Bar(x=months, y=conversion, name='Conversion Rate', visible=False, marker_color='#F4A261'))

fig.update_layout(
    title='Selected Metric: Revenue',
    font=dict(family='AppleGothic'),
    updatemenus=[dict(
        type='dropdown',
        buttons=[
            dict(label='Revenue ($K)', method='update',
                 args=[{'visible': [True, False, False]}, {'title': 'Selected Metric: Revenue'}]),
            dict(label='Visitors', method='update',
                 args=[{'visible': [False, True, False]}, {'title': 'Selected Metric: Visitors'}]),
            dict(label='Conversion Rate (%)', method='update',
                 args=[{'visible': [False, False, True]}, {'title': 'Selected Metric: Conversion Rate'}]),
        ],
        x=1.0, y=1.2, xanchor='right'
    )]
)
fig.show()

metrics = {'Revenue': revenue, 'Visitors': visitors, 'Conversion Rate': conversion}
fastest = max(metrics, key=lambda k: (metrics[k][-1] - metrics[k][0]) / metrics[k][0])
print(f"Fastest-growing metric: {fastest}")
print(f"가장 빠르게 성장한 지표: {fastest}")
print("Selecting a different option in the dropdown switches the chart instantly --")
print("all 3 bar sets are pre-drawn, only their `visible` flag changes.")
print("드롭다운에서 다른 옵션을 선택하면 차트가 즉시 전환됩니다 -- 3개의 막대 세트가")
print("모두 미리 그려져 있고, visible 속성만 바뀝니다.")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Not Specifying `mode=` on `go.Scatter` (go.Scatter에 mode= 미지정)
If you don't set `mode=`, Plotly picks it FOR you based on how many data points there are — 20 or fewer gets `'lines+markers'`, more than 20 gets `'lines'` only. The exact same code can render differently depending only on data size.  
mode=를 지정하지 않으면 Plotly가 데이터 점 개수에 따라 자동으로 선택합니다 -- 20개 이하면 'lines+markers', 초과하면 'lines'만. 즉 같은 코드가 데이터 크기에 따라 다르게 렌더링될 수 있습니다.

✅ **Fix / 해결:** Always set `mode='lines+markers'` (or whatever you intend) explicitly.

### Mistake 2 — Forgetting `yref='y2'` on a Dual-Axis Chart (이중 축 차트에서 yref='y2' 빠뜨리기)
On a dual-axis chart, `fig.add_hline()` and `fig.add_annotation()` default to the LEFT (primary) axis unless told otherwise — a reference line meant for the right-axis series ends up positioned incorrectly.  
이중 축 차트에서 fig.add_hline()과 fig.add_annotation()은 별도로 지정하지 않으면 기본적으로 왼쪽(1차) 축을 기준으로 합니다 -- 오른쪽 축 시리즈를 위한 기준선이 엉뚱한 위치에 그려집니다.

✅ **Fix / 해결:** Pass `yref='y2'` explicitly when the line/annotation belongs to the secondary axis.

### Mistake 3 — Adding a Pie Chart to a Subplot Without `type='domain'` (type='domain' 없이 서브플롯에 파이 차트 추가)
`make_subplots`'s default subplot type is `'xy'` (has x/y axes); a pie chart has no axes at all, so dropping one into a plain `'xy'` cell raises a `ValueError`.  
make_subplots의 기본 서브플롯 타입은 'xy'(x·y축이 있음)입니다 -- 파이 차트는 축이 아예 없으므로 일반 'xy' 칸에 넣으면 ValueError가 발생합니다.

✅ **Fix / 해결:** Set that cell's `specs` entry to `{'type': 'domain'}` before adding a `go.Pie` trace to it.

### Mistake 4 — `visible` List Length Not Matching the Number of Traces (visible 리스트 길이가 trace 개수와 불일치)
A dropdown button's `args=[{'visible': [...]}]` list needs exactly one entry per trace in the whole figure. Getting the count wrong doesn't raise a Python error — the mismatch only shows up as broken behavior once you actually click the button in a browser.  
드롭다운 버튼의 args=[{'visible': [...]}] 리스트는 Figure 전체의 trace 개수와 정확히 같은 길이여야 합니다. 개수가 틀려도 Python 단계에서는 에러가 나지 않고, 실제로 브라우저에서 버튼을 눌렀을 때만 잘못된 동작으로 드러납니다.

✅ **Fix / 해결:** Keep the `visible` list's length in sync with `len(fig.data)` every time you add or remove a trace.

### Mistake 5 — Starting With More Than One Trace `visible=True` (visible=True인 trace가 2개 이상으로 시작)
If two dropdown-controlled traces both start as `visible=True`, the chart opens showing both stacked/overlapping series at once instead of a clean single view.  
드롭다운으로 제어되는 두 trace가 모두 visible=True로 시작하면, 차트가 열렸을 때 하나가 아니라 두 시리즈가 겹쳐서 표시됩니다.

✅ **Fix / 해결:** Exactly one trace should start `visible=True`; every other trace in the same dropdown group starts `visible=False`.

### Mistake 6 — Mixing `px` and `go` Incorrectly (px와 go를 잘못 섞어 쓰기)
Trying `fig.add_trace(px.bar(...))` raises an error — `add_trace()` expects a `go` trace object (`go.Bar(...)`, `go.Scatter(...)`), not a whole Figure that `px` already built.  
fig.add_trace(px.bar(...))처럼 시도하면 에러가 발생합니다 -- add_trace()는 px가 이미 만든 전체 Figure가 아니라 go trace 객체(go.Bar(...), go.Scatter(...))를 기대합니다.

✅ **Fix / 해결:** Build the trace directly with `go.Bar(...)`/`go.Scatter(...)` when adding it to a `go.Figure` you're assembling by hand.

# 💡 Tips (팁)
Useful tips or shortcuts / 유용한 팁과 단축법

- `yaxis2=dict(overlaying='y', side='right')` is the manual way to add a second y-axis; `make_subplots(specs=[[{'secondary_y': True}]])` does the same thing with cleaner, auto-aligned syntax inside a subplot.<br>yaxis2=dict(overlaying='y', side='right')는 두 번째 y축을 수동으로 추가하는 방법입니다; 서브플롯 안에서는 make_subplots(specs=[[{'secondary_y': True}]])가 더 깔끔하고 자동 정렬되는 문법으로 같은 일을 합니다.
- `colspan` in `make_subplots`'s `specs` works just like Section 9's Matplotlib `GridSpec` slicing (`gs[0, :]`) — one subplot spanning the full width.<br>make_subplots의 specs에 있는 colspan은 섹션 9의 Matplotlib GridSpec 슬라이싱(gs[0, :])과 똑같이 동작합니다 -- 한 서브플롯이 전체 너비를 차지합니다.
- `hole=0.4` on `go.Pie` is the one-line way to turn a pie chart into a donut chart.<br>go.Pie에 hole=0.4를 주면 파이 차트가 한 줄로 도넛 차트가 됩니다.
- Order matters when you `add_trace()` multiple times into the same subplot cell — later traces are drawn on top of earlier ones.<br>같은 서브플롯 칸에 add_trace()를 여러 번 호출할 때는 순서가 중요합니다 -- 나중에 추가한 trace가 먼저 추가한 것 위에 그려집니다.
- A dropdown's `args` list can update layout properties (like `title`) at the same time as `visible`, so the chart's title always matches what's currently shown.<br>드롭다운의 args 리스트는 visible과 동시에 title 같은 레이아웃 속성도 업데이트할 수 있어, 차트 제목이 항상 현재 보여지는 내용과 일치하게 유지됩니다.

# 💼 Business Example (비즈니스 예시)
*How would a Business Analyst use this?*

**Scenario / 시나리오:**

**EN:** Your team lead wants a single file that works as three different reports depending on who opens it: the finance team wants to see revenue, the marketing team wants visitors, and the growth team wants conversion rate — but you don't want to build and maintain three separate charts.

**KR:** 팀장님이 여는 사람에 따라 세 가지 다른 보고서 역할을 하는 파일 하나를 원합니다: 재무팀은 매출을, 마케팅팀은 방문자 수를, 성장팀은 전환율을 보고 싶어합니다 -- 하지만 세 개의 차트를 따로 만들고 유지보수하고 싶지는 않습니다.

**To-do / 할 일:**
- [x] Build one figure with three metrics, each as its own trace<br>세 지표를 각각 별도의 trace로 담은 Figure 하나 만들기
- [x] Add a dropdown so each team can select their own metric<br>각 팀이 자신의 지표를 선택할 수 있도록 드롭다운 추가하기
- [x] Make sure only one metric is visible at a time when the file first opens<br>파일을 처음 열었을 때 한 번에 하나의 지표만 보이도록 만들기
- [x] Save it as one interactive HTML file for all three teams<br>세 팀 모두를 위한 인터랙티브 HTML 파일 하나로 저장하기

````python
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
revenue = [2400, 2650, 2820, 3100]
visitors = [45000, 48000, 51000, 58000]
conversion = [2.1, 2.3, 2.4, 2.6]

fig = go.Figure()
fig.add_trace(go.Bar(x=quarters, y=revenue, name='Revenue ($K)', visible=True, marker_color='#2E86AB'))
fig.add_trace(go.Bar(x=quarters, y=visitors, name='Visitors', visible=False, marker_color='#4C956C'))
fig.add_trace(go.Bar(x=quarters, y=conversion, name='Conversion Rate (%)', visible=False, marker_color='#F4A261'))

fig.update_layout(
    title='Quarterly KPI Dashboard -- Revenue',
    font=dict(family='AppleGothic'),
    updatemenus=[dict(
        type='dropdown',
        buttons=[
            dict(label='Revenue ($K)', method='update',
                 args=[{'visible': [True, False, False]}, {'title': 'Quarterly KPI Dashboard -- Revenue'}]),
            dict(label='Visitors', method='update',
                 args=[{'visible': [False, True, False]}, {'title': 'Quarterly KPI Dashboard -- Visitors'}]),
            dict(label='Conversion Rate (%)', method='update',
                 args=[{'visible': [False, False, True]}, {'title': 'Quarterly KPI Dashboard -- Conversion Rate'}]),
        ],
        x=1.0, y=1.15, xanchor='right'
    )]
)
fig.show()

fig.write_html('quarterly_kpi_dashboard.html')
print("Saved: quarterly_kpi_dashboard.html -- one file, three teams, each picks their own metric")
print("저장됨: quarterly_kpi_dashboard.html -- 파일 하나로 세 팀이 각자 지표를 선택")

q_growth = {'Revenue': (revenue[-1] - revenue[0]) / revenue[0] * 100,
            'Visitors': (visitors[-1] - visitors[0]) / visitors[0] * 100,
            'Conversion Rate': (conversion[-1] - conversion[0]) / conversion[0] * 100}
for name, g in q_growth.items():
    print(f"{name} full-year growth: {g:+.1f}%")
    print(f"{name} 연간 성장률: {g:+.1f}%")
````

# 📝 Summary (요약)
*Write today's concept in 3~5 sentences.*

**EN:** `plotly.graph_objects` (`go`) is the low-level counterpart to Section 13's `plotly.express`: instead of a DataFrame and column names, you build every trace by hand with `go.Scatter`, `go.Bar`, `go.Pie`, etc., and assemble them into a `go.Figure` with `add_trace()`. This unlocks three things `px` can't do alone: combining different chart types (like a bar-and-line combo with a second y-axis), laying out a true multi-panel dashboard with `make_subplots` — including mixed types like a line, a donut, and a bar in one Figure — and adding interactive controls like a dropdown (`updatemenus`) that lets a viewer switch which pre-drawn trace is visible, all inside one shareable file. The recurring gotchas are all about explicitness: `go` never guesses what you mean, so `mode=`, `yref=`, subplot `type=`, and the `visible` list all need to be stated by hand.

**KR:** plotly.graph_objects(go)는 섹션 13의 plotly.express에 대응하는 저수준 도구입니다: DataFrame과 컬럼명 대신, go.Scatter, go.Bar, go.Pie 등으로 모든 trace를 직접 만들고 add_trace()로 go.Figure에 조립합니다. 이를 통해 px 혼자서는 할 수 없는 세 가지가 가능해집니다: 서로 다른 차트 종류 결합(두 번째 y축이 있는 막대+선 콤보), make_subplots로 진짜 다중 패널 대시보드 구성(선·도넛·막대를 한 Figure에 섞는 것 포함), 그리고 미리 그려둔 trace 중 무엇을 보여줄지 전환하는 드롭다운(updatemenus) 같은 인터랙티브 컨트롤을 파일 하나 안에 추가하는 것입니다. 반복되는 함정은 모두 명시성에 관한 것입니다: go는 절대 의도를 추측하지 않으므로, mode=, yref=, 서브플롯 type=, visible 리스트를 모두 직접 명시해야 합니다.

**Q1 (EN).** What does `isinstance(px.line(df, x='a', y='b'), go.Figure)` return, and what does that tell you about the relationship between `px` and `go`?  
**Q1 (KR).** `isinstance(px.line(df, x='a', y='b'), go.Figure)`는 무엇을 반환하며, 이것이 px와 go의 관계에 대해 무엇을 말해주나요?

It returns True, because px.line() is already a go.Figure.  
px는 go 객체를 대신 만들어 주는 단축 도구입니다.

**Q2 (EN).** Why does a `go.Scatter` trace with `yaxis='y2'` need a matching `yaxis2=dict(overlaying='y', side='right')` in `update_layout`?  
**Q2 (KR).** `yaxis='y2'`인 go.Scatter trace가 `update_layout`에 짝이 맞는 `yaxis2=dict(overlaying='y', side='right')`를 필요로 하는 이유는 무엇인가요?

yaxis='y2' only names the axis; the layout has to create it.  
overlaying='y'는 같은 그림 위에 겹치고, side='right'는 눈금을 오른쪽에 둡니다.

**Q3 (EN).** You add a `go.Pie` trace to a `make_subplots` cell and get a `ValueError`. What's the most likely cause, and how do you fix it?  
**Q3 (KR).** make_subplots의 한 칸에 go.Pie trace를 추가했더니 ValueError가 발생했습니다. 가장 가능성 높은 원인은 무엇이며, 어떻게 고치나요?

The cell is still the default 'xy' type, and a pie has no axes.  
그 칸의 specs를 {'type': 'domain'}으로 바꿉니다.

**Q4 (EN).** You build a dropdown with 3 buttons for a figure that has 3 traces, but clicking a button shows the wrong chart or no chart at all. What's the most likely bug in your `visible` lists?  
**Q4 (KR).** trace가 3개인 Figure에 버튼 3개짜리 드롭다운을 만들었는데, 버튼을 누르면 잘못된 차트가 나오거나 아무것도 안 나옵니다. visible 리스트에서 가장 가능성 높은 버그는 무엇인가요?

Each visible list must have one boolean per trace, in add order.  
trace가 3개면 [True, False, False]처럼 길이 3이어야 하고, True 위치가 버튼과 맞아야 합니다.

**Q5 (EN).** Why does `fig.add_trace(px.bar(df, x='a', y='b'))` raise an error, and what should you pass to `add_trace()` instead?  
**Q5 (KR).** `fig.add_trace(px.bar(df, x='a', y='b'))`가 에러를 던지는 이유는 무엇이며, add_trace()에는 대신 무엇을 넘겨야 하나요?

px.bar() returns a whole Figure, and add_trace() wants one trace.  
go.Bar(...)를 넘깁니다.
