# 🧠 Concept · 개념

## What is it? · 무엇인가?

Data visualization is the process of turning raw numbers and text into charts, graphs, and maps so people can spot patterns, trends, outliers, and relationships at a glance. In DA (Data Analyst) and BA (Business Analyst) work, visualization isn't decoration — it's the core communication tool that turns analysis results into decisions.

데이터 시각화(Data Visualization)란 숫자와 텍스트로 이루어진 원시 데이터를 그래프, 차트, 지도 등 시각적 형태로 변환하여 사람이 패턴·경향·이상치·관계를 직관적으로 파악할 수 있게 만드는 과정입니다. DA(Data Analyst)·BA(Business Analyst) 직무에서 시각화는 "예쁘게 꾸미는 작업"이 아니라 분석 결과를 의사결정으로 연결하는 핵심 커뮤니케이션 도구입니다.

## Why do we use it? · 왜 사용하는가?

**1. People process images far faster than tables.** Scanning 100 rows of a sales table to find "which month spiked" takes real effort. Seeing the same data as a line chart makes the spike obvious in under a second — the processing speed is simply different.

**1) 사람은 표보다 그림을 압도적으로 빠르게 처리합니다.** 100개의 행으로 이루어진 매출 표를 눈으로 훑어서 "어느 달에 매출이 급증했는지" 찾는 것과, 같은 데이터를 꺾은선 그래프로 보는 것은 인지 속도 자체가 다릅니다.

**2. Anscombe's Quartet.** Statistician Francis Anscombe's famous 1973 example: four datasets share nearly identical summary statistics (mean, variance, correlation), yet look completely different once plotted — a clean line, a curve, a line distorted by one outlier, and a near-vertical cluster. It's the most-cited example in visualization education for one reason: summary statistics alone can hide the real shape of the data, and only a chart reveals it.

**2) 앤스컴의 4중주 (Anscombe's Quartet).** 통계학자 프랜시스 앤스컴이 1973년에 제시한 유명한 사례입니다. 서로 다른 4개의 데이터셋이 평균·분산·상관계수 등 요약 통계량은 거의 동일하지만, 실제로 그래프로 그려보면 완전히 다른 모양(깔끔한 직선 관계 / 곡선 관계 / 이상치 하나 때문에 왜곡된 직선 / 수직선에 가까운 분포)을 보입니다. "요약 통계량만 보고 데이터를 판단하면 안 되며, 반드시 시각화로 확인해야 한다"는 것을 보여주는, 시각화 교육에서 가장 자주 인용되는 사례입니다.

**3. The interview perspective.** Being able to explain *why* you picked a particular chart for a particular situation is judged just as heavily as being able to write the code for it.

**3) 면접 관점.** "왜 이 상황에 이 차트를 선택했는가"를 논리적으로 설명하는 능력은, 코드를 짤 줄 아는 것만큼 중요하게 평가됩니다.

## When is it used in Business Analytics? · 언제 비즈니스 분석에서 사용되는가?

Visualization plays three distinct roles in DA/BA work, and each one has a different audience:

시각화는 DA/BA 실무에서 3가지 역할을 하며, 역할마다 대상이 다릅니다:

| Role / 역할 | Purpose / 목적 | Audience / 대상 |
|---|---|---|
| Exploration (EDA) / 탐색 | The analyst quickly gets a feel for the data / 분석가 본인이 데이터를 빠르게 파악 | Myself / 나 자신 |
| Analysis / 분석 | Test hypotheses, find patterns and relationships / 가설 검증, 패턴·관계 발견 | Myself + fellow analysts / 나 자신 + 동료 분석가 |
| Communication / 커뮤니케이션 | Deliver insight and drive a decision / 인사이트를 전달하고 의사결정을 이끌어냄 | Leadership, non-technical teams / 경영진, 현업 부서(비개발자) |

Notice how the *same* underlying data can need three completely different chart treatments depending on which of these three rows applies — quick and messy for exploration, precise for analysis, polished and simplified for communication. This idea reappears in the EDA → Analysis → Reporting workflow later in this section.

같은 데이터라도 이 3가지 역할 중 어디에 해당하느냐에 따라 완전히 다른 방식의 차트가 필요하다는 점에 주목하세요 — 탐색 단계는 빠르고 거칠게, 분석 단계는 정밀하게, 커뮤니케이션 단계는 다듬어지고 단순하게. 이 개념은 이 섹션 뒤쪽의 EDA → 분석 → 보고 워크플로우에서 다시 등장합니다.

# 📝 Syntax · 문법

## Basic Syntax · 기본 문법

There's no charting code in this section — it's the conceptual foundation the rest of the course builds on. Instead, "syntax" here means encoding the chart-selection logic itself as plain Python, so you have a reusable reference you can call any time you're unsure which chart to use.

이 섹션에는 실제 차트를 그리는 코드가 없습니다 — 이후 모든 섹션의 개념적 토대이기 때문입니다. 대신 여기서 "문법"은 차트 선택 로직 자체를 파이썬으로 표현하는 것을 의미하며, 어떤 차트를 써야 할지 헷갈릴 때마다 참고할 수 있는 재사용 가능한 자료가 됩니다.

````python
# Basic Syntax: represent the purpose -> chart type guide as a dictionary
# 기본 문법: 목적 -> 차트 유형 가이드를 딕셔너리로 표현하기

chart_guide = {
    "Comparison": {
        "question_en": "How do values differ across categories?",
        "question_kr": "항목별로 크기가 어떻게 다른가?",
        "chart": "Bar, Column",
        "section": 4,
    },
    "Trend": {
        "question_en": "How does a value change over time?",
        "question_kr": "시간에 따라 어떻게 변했는가?",
        "chart": "Line, Area",
        "section": 3,
    },
    "Distribution": {
        "question_en": "How is a single numeric variable spread out?",
        "question_kr": "데이터가 어떻게 퍼져 있는가?",
        "chart": "Histogram, Box, Violin",
        "section": 5,
    },
    "Relationship": {
        "question_en": "Is there a correlation between two or more variables?",
        "question_kr": "두 변수 사이에 상관관계가 있는가?",
        "chart": "Scatter, Bubble",
        "section": 6,
    },
    "Composition": {
        "question_en": "What share does each part contribute to the whole?",
        "question_kr": "전체에서 각 부분의 비중은?",
        "chart": "Pie, Donut, Treemap, Stacked Bar",
        "section": 7,
    },
}

for purpose, info in chart_guide.items():
    print(f"{purpose:12s} -> {info['chart']:28s} (Section {info['section']})")
````

## Common Variations · 주요 변형

The same dictionary can power a small lookup helper. In practice, you always ask "what's my core question?" first, then let the answer tell you which chart family to use — never the other way around.

같은 딕셔너리로 간단한 조회 함수도 만들 수 있습니다. 실무에서는 항상 "내 핵심 질문이 무엇인가?"를 먼저 묻고, 그 답이 어떤 차트 계열을 써야 할지 알려주게 해야 합니다 — 반대 순서가 아닙니다.

````python
# Common Variations: a lookup helper built on top of chart_guide
# 주요 변형: chart_guide를 활용한 조회 함수

def recommend_chart(purpose: str) -> str:
    """Return the recommended chart type(s) for a given analytic purpose.
    주어진 분석 목적에 맞는 추천 차트 유형을 반환합니다."""
    info = chart_guide.get(purpose)
    if info is None:
        return "Unknown purpose - check spelling (Comparison / Trend / Distribution / Relationship / Composition)"
    return info["chart"]

print("Trend       ->", recommend_chart("Trend"))
print("Composition ->", recommend_chart("Composition"))
print("Unknown     ->", recommend_chart("Something Else"))
````

# 🧪 Small Examples · 예제

## Example 1: Same Statistics, Different Shape · 예제 1: 같은 통계량, 다른 모양

The real Anscombe's Quartet uses four 11-point datasets. Below is a small-scale version of the same idea: two tiny datasets that can be made to share (almost) the same mean, yet tell very different stories once you actually look at how the numbers are spread out.

실제 앤스컴의 4중주는 11개의 점으로 이루어진 4개 데이터셋을 사용합니다. 아래는 같은 아이디어를 축소한 버전입니다: 평균은 (거의) 같게 만들 수 있지만, 숫자가 실제로 어떻게 퍼져 있는지 보면 전혀 다른 이야기를 들려주는 두 개의 작은 데이터셋입니다.

````python
# Example 1: two datasets that can share a mean but have very different shapes
# 예제 1: 평균은 같게 만들 수 있지만 모양은 전혀 다른 두 데이터셋

dataset_steady = [48, 50, 49, 51, 50, 49, 51, 50]      # steady around 50
dataset_spike  = [10, 12, 11, 13, 190, 9, 11, 10]        # one huge outlier

mean_steady = sum(dataset_steady) / len(dataset_steady)
mean_spike  = sum(dataset_spike) / len(dataset_spike)

print(f"Steady dataset mean / 평균: {mean_steady:.1f}")
print(f"Spike  dataset mean / 평균: {mean_spike:.1f}")
print()
print("The means are nowhere near equal here on purpose - try lowering that 190")
print("in dataset_spike until both means land close together. Even then, plotting")
print("both datasets side by side would show two completely different stories.")
print()
print("일부러 두 평균이 전혀 다르게 설정되어 있습니다 - dataset_spike의 190을")
print("낮춰서 두 평균이 비슷해지도록 만들어 보세요. 그렇게 만들어도, 두 데이터셋을")
print("나란히 그래프로 그려보면 완전히 다른 이야기가 보일 것입니다.")
````

## Example 2: The Purpose-Based Chart Selection Table · 예제 2: 목적 기반 차트 선택표

Before picking a chart, always answer the "core question" column first. The most common real-world mistake is reaching for a Pie chart when the real question is about a trend, or a Line chart when the real question is about composition.

차트를 고르기 전에는 항상 "핵심 질문" 열에 먼저 답해야 합니다. 실무에서 가장 흔한 실수는 실제로는 추세를 묻는 질문인데 Pie 차트를 쓰거나, 실제로는 비율을 묻는 질문인데 Line 차트를 쓰는 것입니다.

````python
# Example 2: walking through business questions using chart_guide
# 예제 2: chart_guide를 활용해 비즈니스 질문 살펴보기

business_questions = [
    "Which of our 4 stores sold the most last month?",
    "Has our website traffic grown over the past 6 months?",
    "What percentage of revenue comes from each sales channel?",
]

mapped_purpose = ["Comparison", "Trend", "Composition"]

for question, purpose in zip(business_questions, mapped_purpose):
    print(f"Q: {question}")
    print(f"   -> Purpose: {purpose} | Chart: {recommend_chart(purpose)}")
    print()
````

## Example 3: Matplotlib vs Seaborn vs Plotly vs Tableau Public · 예제 3: Matplotlib vs Seaborn vs Plotly vs Tableau Public

One-line summary: Matplotlib is the "engine," Seaborn is the "fast EDA skin" (built on Matplotlib), Plotly is "interactive charts for the web," and Tableau is the "no-code dashboard you share with non-developers."

한 줄 요약: Matplotlib은 "엔진", Seaborn은 "빠른 EDA용 스킨"(Matplotlib 기반), Plotly는 "웹에 올릴 인터랙티브 차트", Tableau는 "비개발자와 공유하는 실무 대시보드"입니다.

````python
# Example 3: comparing the four tools as structured data
# 예제 3: 4개 도구를 구조화된 데이터로 비교하기

library_comparison = {
    "Matplotlib": {
        "type": "Low-level plotting engine",
        "best_for": "Fine-grained control, static report images",
        "downside": "Verbose syntax, plain default style",
    },
    "Seaborn": {
        "type": "Statistical viz library (built on Matplotlib)",
        "best_for": "Fast EDA, distributions/relationships with good defaults",
        "downside": "Deep customization still needs Matplotlib syntax",
    },
    "Plotly": {
        "type": "Interactive web visualization library",
        "best_for": "Dashboards, hover/zoom, instant HTML sharing",
        "downside": "Learning curve, can slow down on very large data",
    },
    "Tableau Public": {
        "type": "No-code BI tool",
        "best_for": "Drag-and-drop dashboards for non-developers",
        "downside": "Public version publishes the whole workbook openly",
    },
}

for tool, info in library_comparison.items():
    print(f"[{tool}]")
    print(f"  Type      : {info['type']}")
    print(f"  Best for  : {info['best_for']}")
    print(f"  Downside  : {info['downside']}")
    print()
````

## Example 4: The EDA → Analysis → Reporting Workflow · 예제 4: EDA → 분석 → 보고 워크플로우

Even within a single project, professionals switch tools by stage: Pandas `.plot()`/Seaborn for fast EDA (speed over polish), Matplotlib/Plotly for deep analysis (precise control, interactive exploration), and Matplotlib/Tableau for the final report (polish over speed, clear storytelling). This guide itself follows that same order: Matplotlib deep-dive → Seaborn → Plotly → Tableau.

실무에서는 하나의 프로젝트 안에서도 단계에 따라 다른 도구를 사용합니다: 빠른 EDA에는 Pandas `.plot()`/Seaborn(속도 > 미관), 심층 분석에는 Matplotlib/Plotly(정밀한 제어, 인터랙티브 탐색), 최종 보고에는 Matplotlib/Tableau(미관 > 속도, 명확한 스토리텔링). 이 가이드도 같은 순서(Matplotlib 정밀 제어 → Seaborn → Plotly → Tableau)를 따릅니다.

````python
# Example 4: modeling the 3-stage workflow as a list of stages
# 예제 4: 3단계 워크플로우를 리스트로 모델링하기

workflow = [
    {
        "stage": "1. EDA (Exploratory Data Analysis)",
        "goal_en": "Skim the data quickly and get a feel for it",
        "goal_kr": "데이터를 빠르게 훑어보고 감을 잡는다",
        "tools": ["Pandas .plot()", "Seaborn"],
        "priority": "Speed > polish",
    },
    {
        "stage": "2. Analysis",
        "goal_en": "Test a specific hypothesis or dig into a relationship",
        "goal_kr": "특정 가설을 검증하거나 관계를 깊이 파고든다",
        "tools": ["Matplotlib (fine control)", "Plotly (interactive)"],
        "priority": "Precision, filtering, zooming",
    },
    {
        "stage": "3. Reporting",
        "goal_en": "Deliver the insight to a decision-maker",
        "goal_kr": "의사결정권자에게 인사이트를 전달한다",
        "tools": ["Matplotlib (fixed image)", "Tableau (shared dashboard)"],
        "priority": "Polish > speed",
    },
]

for step in workflow:
    print(step["stage"])
    print(f"  Goal (EN): {step['goal_en']}")
    print(f"  Goal (KR): {step['goal_kr']}")
    print(f"  Tools    : {', '.join(step['tools'])}")
    print()
````

# ⚠️ Common Mistakes · 자주 하는 실수

### Mistake 1: Picking a chart before deciding the question · 실수 1: 질문을 정하기 전에 차트부터 고르기

Jumping straight to "I'll make a bar chart" without first stating the core question (comparison? trend? composition?) is the single most common beginner mistake — it leads to using a Line chart for composition data, or a Pie chart for a trend.

핵심 질문(비교인가? 추세인가? 비율인가?)을 먼저 정하지 않고 "막대 차트를 그려야지"라고 곧장 결정해버리는 것이 초보자가 가장 흔히 저지르는 실수입니다 — 비율 데이터에 Line 차트를 쓰거나, 추세 데이터에 Pie 차트를 쓰는 결과로 이어집니다.

✅ **Fix:** Always write down the core question first, then look it up in the purpose table from Example 2.  
✅ **해결:** 항상 핵심 질문을 먼저 글로 적고, Example 2의 목적 표에서 찾아보세요.

### Mistake 2: Judging data only from summary statistics · 실수 2: 요약 통계량만으로 데이터를 판단하기

Trusting a mean, median, or correlation number without ever plotting the data is exactly the trap Anscombe's Quartet warns against — very different datasets can share nearly identical statistics.

데이터를 한 번도 시각화해보지 않고 평균·중앙값·상관계수 같은 숫자만 믿는 것은 앤스컴의 4중주가 경고하는 바로 그 함정입니다 — 완전히 다른 데이터셋이 거의 동일한 통계량을 가질 수 있습니다.

✅ **Fix:** Treat a quick plot as a mandatory step in every analysis, not an optional nice-to-have.  
✅ **해결:** 빠른 시각화를 "있으면 좋은 것"이 아니라 모든 분석의 필수 단계로 취급하세요.

### Mistake 3: Using the same polish level for exploration and for the boardroom · 실수 3: 탐색용과 보고용에 같은 완성도를 쓰기

Sending a raw, unlabeled EDA scatter plot straight to leadership (or over-polishing every throwaway exploration chart) wastes time in both directions — match the chart's polish level to its audience using the 3-role table.

라벨도 없는 EDA용 산점도를 그대로 경영진에게 보내거나, 반대로 버릴 탐색용 차트 하나하나를 지나치게 다듬는 것은 양쪽 다 시간 낭비입니다 — 3가지 역할 표에 맞춰 차트의 완성도를 대상에 맞추세요.
