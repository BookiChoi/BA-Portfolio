# 🧠 Concept · 개념

## What is it? · 무엇인가?

A Line Chart plots data points and connects them in sequence, showing how a value flows and changes along an ordered axis — most often time.

Line Chart(꺾은선 그래프)는 데이터 포인트를 점으로 찍고 순서대로 선으로 연결하여, 값의 변화 흐름을 보여주는 차트입니다.

## Why do we use it? · 왜 사용하는가?

Use it whenever you want to see change over time: rising, falling, staying flat, a sudden shift at one point, or comparing the trend of several things at once — a Line Chart is optimized for exactly these questions.

시간 흐름에 따른 변화(추세)를 보고 싶을 때 사용합니다. "상승/하락/정체", "특정 시점의 급변", "여러 대상의 추세 비교"를 파악하는 데 최적화되어 있습니다.

## When is it used in Business Analytics? · 언제 비즈니스 분석에서 사용되는가?

Common DA/BA situations: monthly/daily revenue trends, website visitor changes, KPI trend monitoring, before/after comparison of a marketing campaign, tracking inventory levels over time.

DA/BA 실무 상황: 월별·일별 매출 추이, 웹사이트 방문자 수 변화, KPI(핵심성과지표) 트렌드 모니터링, 마케팅 캠페인 전후 효과 비교, 재고 수준 변화 추적 등.

The x-axis must be ordered, sequential data (date, month, quarter, time). Unordered categorical data (product name, region name) doesn't fit a Line Chart — that's Section 4's Bar Chart instead:

x축은 순서가 있는 연속형 데이터(날짜, 월, 분기, 시간 등)여야 합니다. 순서가 없는 범주형 데이터(제품명, 지역명 등)에는 부적합하며, 이런 경우는 섹션 4의 Bar Chart를 사용해야 합니다:

| | Line Chart | Bar Chart (Section 4) |
|---|---|---|
| Core question / 핵심 질문 | How does it change over time? / 시간에 따라 어떻게 변했는가? | How do values differ across categories? / 항목별로 크기가 어떻게 다른가? |
| X-axis requirement / X축 요구사항 | Ordered, sequential / 순서가 있는 연속형 | No order needed / 순서 필요 없음 |
| Example / 예시 | Monthly revenue trend / 월별 매출 추이 | Store-by-store sales comparison / 지점별 매출 비교 |

# 📝 Syntax · 문법

## Basic Syntax · 기본 문법

The core pattern: a list of ordered x-values, a list of y-values, `ax.plot()`, then label everything.

핵심 패턴: 순서가 있는 x값 리스트, y값 리스트, `ax.plot()`, 그리고 라벨을 다는 것입니다.

````python
# Basic Syntax: a single line chart
# 기본 문법: 단일 라인 차트

import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'AppleGothic'   # macOS: 'AppleGothic', Windows: 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

months = ['1월', '2월', '3월', '4월', '5월', '6월']
revenue = [850, 920, 880, 1050, 980, 1150]

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(months, revenue, marker='o', color='#2E86AB', linewidth=2)
ax.set_title('월별 매출 추이')
ax.set_xlabel('월')
ax.set_ylabel('매출 (백만원)')
ax.grid(True, alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
````

## Common Variations · 주요 변형

Real work usually compares more than one series on the same chart. Give each `ax.plot()` call its own `label=`, and use these style options to keep multiple lines readable:

실무에서는 하나의 선보다 여러 대상을 겹쳐서 비교하는 경우가 훨씬 많습니다. 각 `ax.plot()` 호출마다 `label=`을 지정하고, 아래 스타일 옵션으로 여러 선이 겹쳐도 알아보기 쉽게 만드세요:

| Option / 옵션 | Example values / 값 예시 | Meaning / 의미 |
|---|---|---|
| linestyle | `'-'` `'--'` `':'` `'-.'` | solid / dashed / dotted / dash-dot / 실선 / 파선 / 점선 / 파점선 |
| marker | `'o'` `'s'` `'^'` `'D'` | circle / square / triangle / diamond / 원 / 사각형 / 삼각형 / 다이아몬드 |
| linewidth | a number, e.g. `2` | line thickness / 선 굵기 |
| alpha | `0` to `1` | transparency, useful when lines overlap / 투명도 (선이 겹칠 때 유용) |

💡 A practical tip: instead of using many bright colors, make the one line you want to highlight bold and colorful, and fade the rest to thin gray — it makes "what to look at" far clearer in a report.

💡 실무 팁: 여러 색을 쓰기보다, 강조하고 싶은 선 1개만 진한 색+굵게, 나머지는 회색+얇게 처리하면 보고서에서 "무엇을 보라는 건지" 훨씬 명확해집니다.

````python
# Common Variations: comparing two lines with different styles
# 주요 변형: 서로 다른 스타일로 두 선 비교하기

import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False

months = ['1월', '2월', '3월', '4월', '5월', '6월']
product_a = [850, 920, 880, 1050, 980, 1150]
product_b = [600, 650, 700, 680, 750, 800]

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(months, product_a, marker='o', color='#2E86AB', linewidth=2, label='제품 A')
ax.plot(months, product_b, marker='s', color='#E63946', linewidth=2, linestyle='--', label='제품 B')
ax.set_title('제품별 매출 추이 비교')
ax.set_xlabel('월')
ax.set_ylabel('매출 (백만원)')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

# 🧪 Small Examples · 예제

## Example 1: Highlight One Series, Gray Out the Rest · 예제 1: 하나만 강조하고 나머지는 회색 처리

The "highlight + gray" pattern from the tip above, applied to three channels at once. The viewer's eye goes straight to Online without needing a caption to explain it.

위 팁에서 소개한 "강조+회색" 패턴을 채널 3개에 적용한 예시입니다. 캡션 설명 없이도 시선이 바로 온라인 채널로 향합니다.

````python
# Example 1: highlight one line, fade the others to gray
# 예제 1: 한 선만 강조하고 나머지는 회색으로 처리하기

import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False

months = ['1월', '2월', '3월', '4월', '5월', '6월']
online = [120, 135, 128, 150, 142, 168]
offline = [95, 90, 88, 85, 80, 78]
partner = [40, 42, 38, 45, 41, 44]

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(months, online, color='#2E86AB', linewidth=2.5, marker='o', label='온라인 (강조)')
ax.plot(months, offline, color='gray', linewidth=1, alpha=0.6, label='오프라인')
ax.plot(months, partner, color='gray', linewidth=1, alpha=0.6, label='제휴몰')
ax.set_title('채널별 매출 추이 - 온라인 강조')
ax.set_xlabel('월')
ax.set_ylabel('매출 (백만원)')
ax.legend()
plt.tight_layout()
plt.show()
````

## Example 2: A Reference Line with axhline() · 예제 2: axhline()으로 기준선 추가하기

`ax.axhline(y=...)` draws a horizontal reference line across the whole chart — perfect for showing a target or benchmark alongside the actual trend.

`ax.axhline(y=...)`는 차트 전체를 가로지르는 수평 기준선을 그립니다 — 실제 추세 옆에 목표치나 벤치마크를 함께 보여주기에 좋습니다.

````python
# Example 2: plotting actual revenue against a fixed target line
# 예제 2: 실제 매출과 고정 목표선을 함께 그리기

import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False

months = ['1월', '2월', '3월', '4월', '5월', '6월']
revenue = [82, 89, 95, 101, 97, 112]
target = 100

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(months, revenue, marker='o', color='#2E86AB', linewidth=2)
ax.axhline(y=target, color='gray', linestyle='--', linewidth=1.5)
ax.text(0, target + 1.5, f'목표: {target}', fontsize=9, color='gray')
ax.set_title('월별 매출 vs 목표선')
ax.set_ylabel('매출 (백만원)')
plt.tight_layout()
plt.show()
````

## Example 3: Shading a Time Range with axvspan() · 예제 3: axvspan()으로 구간 강조하기

`ax.axvspan(start, end)` shades a vertical band across a range of x-values — ideal for marking "this is when the campaign ran" directly on the trend.

`ax.axvspan(시작, 끝)`은 x값 구간에 세로 음영을 칠합니다 — 추세 위에 "이 구간에 캠페인이 진행됐다"를 직접 표시하기에 좋습니다.

````python
# Example 3: shading the weeks a campaign was active
# 예제 3: 캠페인이 진행된 주차 구간에 음영 표시하기

import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False

weeks = ['1주', '2주', '3주', '4주', '5주', '6주']
visitors = [1200, 1250, 1300, 1600, 1750, 1820]

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(weeks, visitors, marker='o', color='#2E86AB', linewidth=2)
ax.axvspan(3, 5, color='#2E86AB', alpha=0.08)
ax.annotate('캠페인 기간', xy=(4, 1700), xytext=(4, 2000), ha='center',
            arrowprops=dict(facecolor='black', arrowstyle='->', lw=1.3), fontsize=9)
ax.set_title('주간 방문자 추이 - 캠페인 구간 강조')
ax.set_ylabel('방문자 수 (명)')
plt.tight_layout()
plt.show()
````

# ⚠️ Common Mistakes · 자주 하는 실수

### Mistake 1: x-axis categories aren't sorted in time order · 실수 1: x축 카테고리가 시간 순서와 다르게 정렬됨

If the list you pass as x-values isn't already in chronological order, the line zigzags back and forth instead of showing a clean trend.

x값으로 넘기는 리스트가 이미 시간순으로 정렬되어 있지 않으면, 선이 깔끔한 추세 대신 뒤죽박죽 지그재그로 그려집니다.

✅ **Fix:** Sort the underlying list chronologically before it ever reaches `ax.plot()`.  
✅ **해결:** `ax.plot()`에 넘기기 전에 리스트 자체를 시간순으로 미리 정렬하세요.

### Mistake 2: Forgetting label= on every line · 실수 2: 여러 선을 그릴 때 label= 누락

Calling `ax.legend()` does nothing useful if none of your `ax.plot()` calls included a `label=` argument — the legend box shows up empty.

`ax.plot()` 호출 중 하나라도 `label=` 인자를 빠뜨리면, `ax.legend()`를 호출해도 범례가 비어 있게 됩니다.

✅ **Fix:** Add `label=` to every single `ax.plot()` call before you draw multiple lines, not just some of them.  
✅ **해결:** 여러 선을 그릴 때는 일부가 아니라 각 `ax.plot()` 호출마다 반드시 `label=`을 지정하세요.

### Mistake 3: Missing values (None / NaN) break the line · 실수 3: 데이터에 결측치(None / NaN)가 있음

A gap in the middle of your data makes the line visibly break at that point — this is actually correct, expected Matplotlib behavior, not a bug.

데이터 중간에 빈 값이 있으면 그 지점에서 선이 눈에 띄게 끊어집니다 — 이는 버그가 아니라 Matplotlib의 의도된 정상 동작입니다.

✅ **Fix:** Recognize this as intentional; if you don't want the gap, use `.dropna()` or interpolate the missing values first.  
✅ **해결:** 의도된 동작임을 인지하고, 끊김을 원하지 않으면 `.dropna()` 또는 보간(interpolate) 처리를 먼저 하세요.

### Mistake 4: Overlaying 5 or more lines · 실수 4: 선을 5개 이상 겹쳐 그림

Once you cross roughly 5 overlapping lines, the colors blur together and readability drops sharply.

대략 5개를 넘어 선을 겹쳐 그리면 색이 뒤섞여 가독성이 크게 떨어집니다.

✅ **Fix:** Use the "highlight one, gray the rest" technique from Example 1 so only the important line stands out.  
✅ **해결:** Example 1의 "하나만 강조, 나머지는 회색" 기법을 사용해 중요한 선 하나만 눈에 띄게 하세요.

# 💡 Tips · 팁
Useful tips or shortcuts · 유용한 팁과 지름길

- Always sort your x-axis list chronologically before plotting — a Line Chart assumes the data is already in order.  
그리기 전에 항상 x축 리스트를 시간순으로 정렬하세요 — Line Chart는 데이터가 이미 순서대로 있다고 가정합니다.
- Give every line a `label=` the moment you draw it, so `ax.legend()` is never left empty later.  
선을 그리는 그 즉시 `label=`을 지정하세요, 나중에 `ax.legend()`가 비어있지 않도록.
- When comparing many series, bold + color the one line you want the viewer to notice, and fade the rest to thin gray.  
여러 계열을 비교할 때는, 시선을 끌고 싶은 선 하나만 굵고 진하게, 나머지는 얇은 회색으로 처리하세요.
- Reach for `axhline()`/`axvspan()` whenever the viewer needs a fixed reference point — a target, or a before/after boundary.  
보는 사람에게 고정된 기준점(목표치, 전후 경계)이 필요할 때는 `axhline()`/`axvspan()`을 활용하세요.

캠페인 전/후 평균과 성장률을 계산해 숫자 근거로 출력하기

````python
# Business Example: weekly visitor trend by channel, with campaign impact
# 비즈니스 예제: 채널별 주간 방문자 추이 및 캠페인 효과

import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False

# self-contained dataset: 3 marketing channels over 8 weeks
# 자체 포함형 데이터셋: 3개 마케팅 채널의 8주간 데이터
weeks = ['1주', '2주', '3주', '4주', '5주', '6주', '7주', '8주']
organic = [1200, 1350, 1280, 1420, 1600, 1550, 1700, 1850]
paid_ads = [800, 950, 1100, 1050, 1200, 1400, 1350, 1500]
social = [400, 420, 380, 500, 650, 700, 680, 750]
target = 1500   # weekly visitor target / 주간 목표 방문자 수

# before(1-4)/after(5-8) average - numeric proof of campaign impact
# 캠페인 전(1~4주) / 후(5~8주) 평균 - 캠페인 효과의 숫자 근거
organic_before = sum(organic[:4]) / 4
organic_after = sum(organic[4:]) / 4
growth_rate = (organic_after - organic_before) / organic_before * 100

fig, ax = plt.subplots(figsize=(11, 6))

ax.plot(weeks, organic, marker='o', color='#2E86AB', linewidth=2.5, label='자연유입 (Organic)')
ax.plot(weeks, paid_ads, marker='s', color='#E63946', linewidth=2.5, linestyle='--', label='유료광고 (Paid Ads)')
ax.plot(weeks, social, marker='^', color='#F4A261', linewidth=2.5, linestyle=':', label='소셜미디어 (Social)')

# target line
ax.axhline(y=target, color='gray', linestyle='-.', linewidth=1.5, alpha=0.8)
ax.text(0, target + 40, f'주간 목표: {target}명', fontsize=9, color='gray')

# shade the post-campaign weeks and annotate the launch point
ax.axvspan(4, 7, color='#2E86AB', alpha=0.06)
ax.annotate('신규 캠페인 시작', xy=('5주', 1600), xytext=('4주', 2050),
            arrowprops=dict(facecolor='black', arrowstyle='->', lw=1.5),
            fontsize=10, fontweight='bold')

ax.set_title('채널별 주간 방문자 추이 (최근 8주)', fontsize=14, fontweight='bold')
ax.set_xlabel('주차')
ax.set_ylabel('방문자 수 (명)')
ax.legend(loc='upper left')
ax.grid(True, alpha=0.3, linestyle='--')
ax.set_ylim(0, 2300)

plt.tight_layout()
plt.show()

print(f"Before campaign avg / 캠페인 전 평균: {organic_before:.0f}")
print(f"After campaign avg / 캠페인 후 평균: {organic_after:.0f}")
print(f"Growth rate / 성장률: {growth_rate:+.1f}%")
````

A Line Chart plots ordered data points and connects them, making it the right tool whenever the core question is "how did this change over time" — not "how do these categories compare," which is Bar Chart's job instead. Multiple series can be compared on one chart by giving each `ax.plot()` call a distinct `label=`, `linestyle`, and `marker`, and the "highlight one, gray the rest" pattern keeps busy charts readable. `axhline()` and `axvspan()` add fixed reference points — a target value or a highlighted time range — that turn a plain trend line into a chart that argues a specific point. Finally, remember that an unsorted x-axis, a missing `label=`, or more than about 5 overlapping lines are the most common ways a Line Chart goes wrong.

Line Chart는 순서가 있는 데이터 포인트를 선으로 연결하며, 핵심 질문이 "이게 시간에 따라 어떻게 변했는가"일 때 적합한 도구입니다 — "이 카테고리들이 어떻게 다른가"는 대신 Bar Chart의 역할입니다. 각 `ax.plot()` 호출에 서로 다른 `label=`, `linestyle`, `marker`를 지정해 하나의 차트에서 여러 계열을 비교할 수 있으며, "하나만 강조, 나머지는 회색" 패턴은 복잡한 차트를 계속 읽기 쉽게 유지해줍니다. `axhline()`과 `axvspan()`은 목표치나 강조 구간 같은 고정 기준점을 추가해, 단순한 추세선을 특정 주장을 뒷받침하는 차트로 바꿔줍니다. 마지막으로, 정렬되지 않은 x축, 누락된 `label=`, 5개 이상 겹친 선이 Line Chart가 가장 흔히 잘못되는 지점임을 기억하세요.
