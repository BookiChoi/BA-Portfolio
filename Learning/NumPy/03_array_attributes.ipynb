# 🧠 Concept

## What is it?
*(Explain it in your own words.)*

NumPy arrays carry metadata about themselves — four read-only attributes that tell you their shape, dimensionality, element type, and total size, all without touching the actual data.

NumPy 배열은 자기 자신에 대한 메타데이터를 함께 가지고 있습니다 — 실제 데이터를 건드리지 않고도 형태, 차원, 원소 타입, 전체 크기를 알려주는 4개의 읽기 전용 속성입니다.

### The four attributes at a glance / 한눈에 보는 4가지 속성

| Attribute / 속성 | Returns / 반환값 | Example / 예시 |
|---|---|---|
| `.shape` | Tuple of dimension sizes (각 차원의 크기 튜플) | `(3, 4)` |
| `.ndim` | Number of dimensions (차원의 개수) | `2` |
| `.dtype` | Data type of elements (원소의 데이터 타입) | `float64` |
| `.size` | Total element count (전체 원소 개수) | `12` |

## Why do we use it?
*(When is it useful?)*

Because most array bugs — a shape mismatch, a silent string-instead-of-number column, an unexpected extra dimension — show up immediately once you check these four attributes, often before you've even run the operation that would have failed.

대부분의 배열 관련 버그 — shape 불일치, 조용히 문자열로 남아있는 숫자 컬럼, 예상 밖의 차원 — 는 이 4가지 속성만 확인해도 연산을 실행하기도 전에 바로 드러나기 때문입니다.

## When is it used in Business Analytics?
*(Real-world use case)*

- Confirming a DataFrame → ndarray conversion produced the shape you expected.  
DataFrame → ndarray 변환 결과가 예상한 shape인지 확인할 때.
- Validating two arrays match in shape before an operation between them.  
두 배열 사이 연산을 하기 전 shape이 맞는지 검증할 때.
- Catching numbers that were silently read as strings (`dtype=object`) from a CSV.  
CSV에서 숫자가 조용히 문자열(`dtype=object`)로 읽힌 경우를 잡아낼 때.
- Using `.size` as the denominator when calculating a missing-value ratio.  
결측치 비율을 계산할 때 `.size`를 분모로 사용할 때.

# 📝 Syntax

## Basic Syntax

````python
import numpy as np

arr = np.array([[1200, 1350, 1100], [1400, 1600, 1500]])

print(arr.shape)   # (2, 3) -> 2 rows, 3 columns
print(arr.ndim)    # 2 -> two dimensions
print(arr.dtype)   # int64 -> element type
print(arr.size)    # 6 -> total elements (2 x 3)
````

## Common Variations

````python
import numpy as np

arr = np.array([[1200, 1350, 1100], [1400, 1600, 1500]])

# unpack .shape directly
rows, cols = arr.shape
print(f"rows: {rows}, cols: {cols}")

# check ndim before trusting an operation
if arr.ndim != 2:
    raise ValueError(f"Expected 2D, got {arr.ndim}D")
print("ndim check passed")

# dtype changes after astype() — the original array is unaffected until reassigned
float_arr = arr.astype(float)
print(arr.dtype, "->", float_arr.dtype)

# .size vs len() are NOT the same for 2D+ arrays
print(len(arr))    # 2 -> only the first dimension
print(arr.size)    # 6 -> every element
````

# 🧪 Small Examples

## Example 1: `.shape` — Validating a DataFrame → ndarray Conversion
### 3-1. .shape — 배열의 형태

````python
import numpy as np
import pandas as pd

# Business: convert a DataFrame to ndarray and verify the shape
df = pd.DataFrame({
    'sales': [1200, 1350, 1100, 1400],
    'visitors': [3200, 3500, 2800, 3700],
    'conversion_rate': [0.038, 0.039, 0.039, 0.038]
})

data = df.to_numpy()
print("Data shape:", data.shape)   # (4, 3) -> 4 rows, 3 columns
print(f"{data.shape[0]} records, {data.shape[1]} features")
````

## Example 2: `.ndim` — Guarding a Function Against the Wrong Shape
### 3-2. .ndim — 차원 수

````python
import numpy as np

def safe_process(arr):
    """Only processes 2D arrays — validates first."""
    if arr.ndim != 2:
        raise ValueError(f"Expected a 2D array. Got: {arr.ndim}D")
    return arr.mean(axis=0)

data_2d = np.array([[1200, 3200], [1300, 3500], [1100, 3800]])
print(data_2d)
print("Column-wise mean:", safe_process(data_2d))
````

## Example 3: `.dtype` — Fixing Numbers That Were Read as Strings
### 3-3. .dtype — 데이터 타입

````python
import numpy as np
import pandas as pd

# Business: a CSV column that looks numeric but was read as text
df = pd.DataFrame({'revenue': ['1200000', '1300000', '1100000']})

sales = df['revenue'].to_numpy()
print("dtype before:", sales.dtype)   # object -> it's text!

sales_num = sales.astype(float)
print("dtype after:", sales_num.dtype)
print("Average revenue:", np.mean(sales_num))
````

## Example 4: `.size` — Calculating a Missing-Value Ratio
### 3-4. .size — 전체 원소 수

````python
import numpy as np

# Business: what fraction of this year's monthly data is missing?
monthly_data = np.array([1200, np.nan, 1100, 1400, np.nan, 1500,
                          1300, 1250, np.nan, 1700, 1800, 2000])

nan_count = np.sum(np.isnan(monthly_data))
total = monthly_data.size
nan_ratio = nan_count / total * 100

print(f"Total data points: {total}")
print(f"Missing values: {nan_count}")
print(f"Missing ratio: {nan_ratio:.1f}%")
````

# ⚠️ Common Mistakes

**Mistake 1 — Calling these as methods instead of attributes.**
`.shape`, `.ndim`, `.dtype`, and `.size` all have NO parentheses — writing `arr.shape()` raises a `TypeError` because you're trying to "call" a tuple.  
`.shape`, `.ndim`, `.dtype`, `.size`는 전부 괄호가 없습니다 — `arr.shape()`처럼 쓰면 튜플을 "호출"하려는 셈이라 `TypeError`가 발생합니다.  
✅ **Fix:** Access them like plain attributes — `arr.shape`, not `arr.shape()`.  
✅ **해결법:** 일반 속성처럼 접근하세요 — `arr.shape()`가 아니라 `arr.shape`입니다.

**Mistake 2 — Misreading a 1D array's shape.**
A 1D array's `.shape` is `(5,)` — a one-element tuple with a trailing comma, not just `5`. And `(5,)` is a different shape from `(5, 1)` — a 1D array of 5 elements is not the same as a 5×1 column vector.  
1D 배열의 `.shape`은 `(5,)`입니다 — 그냥 `5`가 아니라 쉼표가 붙은 1개짜리 튜플입니다. 그리고 `(5,)`와 `(5, 1)`은 다른 shape입니다 — 5개짜리 1D 배열은 5×1 열벡터와 같지 않습니다.  
✅ **Fix:** If a function unexpectedly complains about dimensions, print `.shape` and check for a missing or extra `1` in the tuple.  
✅ **해결법:** 함수가 예상 밖으로 차원 관련 에러를 낸다면 `.shape`을 출력해서 튜플에 `1`이 빠졌거나 더 있는지 확인하세요.

**Mistake 3 — Assuming `len(arr)` equals `arr.size`.**
For a 2D+ array, `len(arr)` returns only the size of the FIRST dimension (the row count), while `arr.size` returns the total number of elements. `np.zeros((3, 4))` has `len() == 3` but `.size == 12`.  
2D 이상의 배열에서 `len(arr)`은 첫 번째 차원(행 수)의 크기만 반환하지만, `arr.size`는 전체 원소 개수를 반환합니다. `np.zeros((3, 4))`는 `len()`이 `3`이지만 `.size`는 `12`입니다.  
✅ **Fix:** Use `.size` whenever you actually want the total element count, and `.shape[0]` when you specifically want the row count.  
✅ **해결법:** 전체 원소 개수가 필요하면 `.size`를, 행 수가 필요하면 `.shape[0]`을 사용하세요.

**Mistake 4 — Trusting `np.sum()` on a `dtype=object` array.**
If an array's dtype is `object` (often from mixed or string data), `np.sum()` may raise an error or silently do string concatenation instead of numeric addition.  
배열의 dtype이 `object`(주로 혼합 데이터나 문자열에서 발생)라면, `np.sum()`이 에러를 내거나 숫자 덧셈 대신 조용히 문자열 연결을 해버릴 수 있습니다.  
✅ **Fix:** Check `.dtype` before summing; convert with `.astype(float)` or `.astype(int)` first if needed.  
✅ **해결법:** 합계를 구하기 전에 `.dtype`을 확인하고, 필요하면 먼저 `.astype(float)`이나 `.astype(int)`로 변환하세요.

**Mistake 5 — Assigning a decimal into an int-dtype array.**
If an array's dtype is already `int`, assigning `1.9` into one of its slots truncates to `1` silently — no warning, no error.  
배열의 dtype이 이미 `int`라면, 그 자리에 `1.9`를 대입해도 경고나 에러 없이 조용히 `1`로 잘려서 들어갑니다.  
✅ **Fix:** If you need to preserve decimals, make sure the array's dtype is `float` before assigning into it.  
✅ **해결법:** 소수점을 보존해야 한다면, 값을 대입하기 전에 배열의 dtype이 `float`인지 먼저 확인하세요.

**Mistake 6 — Defaulting to `float32` for financial figures.**
`float32` uses half the memory of `float64`, but it also has less precision — for money or other figures where exactness matters, that precision loss can silently accumulate.  
`float32`는 `float64`보다 메모리를 절반만 사용하지만, 정밀도도 낮습니다 — 금액처럼 정확도가 중요한 값에서는 이 정밀도 손실이 조용히 누적될 수 있습니다.  
✅ **Fix:** Use `float64` (NumPy's default) for financial data; reach for `float32` only when memory is the actual bottleneck.  
✅ **해결법:** 금융 데이터에는 (NumPy의 기본값인) `float64`를 사용하고, 메모리가 실제로 병목일 때만 `float32`를 고려하세요.

# 💼 Business Example
*How would a Business Analyst use this?*

**Scenario / 시나리오:**
You just received a new CSV of customer transactions. Before running any analysis, you need to run a quick data-quality check: confirm the shape matches what you expected, make sure numeric columns weren't accidentally read as text, and calculate what fraction of the data is missing.
방금 새로운 고객 거래 CSV를 받았습니다. 분석을 시작하기 전, 빠른 데이터 품질 점검이 필요합니다: shape이 예상과 일치하는지 확인하고, 숫자 컬럼이 실수로 텍스트로 읽히지 않았는지 확인하고, 결측치 비율을 계산합니다.

**To-do / 할 일:**
- [x] Convert the DataFrame to a NumPy array and check `.shape`.  
DataFrame을 NumPy 배열로 변환하고 `.shape`을 확인합니다.
- [x] Check `.dtype` on the revenue column — is it actually numeric?  
매출 컬럼의 `.dtype`을 확인합니다 — 실제로 숫자 타입인가요?
- [x] Convert to float with `.astype()` if needed.  
필요하다면 `.astype()`으로 float로 변환합니다.
- [x] Use `.size` to calculate the missing-value ratio.  
`.size`를 사용해 결측치 비율을 계산합니다.

````python
import numpy as np
import pandas as pd

# A fresh CSV load, before any cleaning
df = pd.DataFrame({
    'revenue': ['1200000', '1350000', '1100000', '1400000', '1250000'],
    'visitors': [3200, 3500, 2900, 3700, 3100]
})

data = df.to_numpy()
print("Shape:", data.shape)
print("ndim:", data.ndim)

# Check whether "revenue" came in as numbers or as text
revenue = df['revenue'].to_numpy()
print("\nrevenue dtype before:", revenue.dtype)   # object -> it's actually text!

revenue_num = revenue.astype(float)
print("revenue dtype after:", revenue_num.dtype)
print("Average revenue:", np.mean(revenue_num))

# Simulate a data-quality check: how much of a related KPI array is missing?
kpi_array = np.array([1200000, np.nan, 1100000, np.nan, 1250000])
missing_ratio = np.sum(np.isnan(kpi_array)) / kpi_array.size * 100
print(f"\nMissing ratio in this KPI array: {missing_ratio:.1f}%")
````

*Write today's concept in 3~5 sentences.*

Every NumPy array carries four attributes that describe it without touching the data itself: `.shape` (a tuple of dimension sizes), `.ndim` (how many dimensions), `.dtype` (the element type), and `.size` (total element count). # ❓ Review Questions

**Q1.** What does `.shape` return for a 1D array of 5 elements, and why is it written that way instead of just `5`?  
5개 원소를 가진 1D 배열의 `.shape`은 무엇을 반환하며, 왜 그냥 `5`가 아니라 그렇게 표기되나요?
The shape is `(5,)`. The comma means it is a one-element tuple, which represents the size of each dimension.  
.shape은 `(5,)`를 반환합니다. ,가 있는 이유는 이것이 원소 하나를 가진 tuple이라는 뜻이며, 각 차원의 크기를 표현하기 때문입니다.

**Q2.** What's the difference between `.ndim` and `.shape`?  
`.ndim`과 `.shape`의 차이는 무엇인가요?
.ndim returns the number of dimensions, while .shape returns the size of each dimension.  
.ndim은 차원의 개수를 반환하고, .shape은 각 차원의 크기를 반환합니다.

**Q3.** Why might a numeric-looking column from a CSV have `dtype=object`, and how do you fix it?
