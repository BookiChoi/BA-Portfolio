## What is it?

This section (Part 1 of 2) turns five foundational statistics into interview-ready, reusable functions: mean, median, variance/standard deviation, growth rate, and change rate. The goal, stated directly in the original material, is to answer "can you calculate this metric?" with both working code and a business interpretation — not just the number.

이번 섹션(2부 중 1부)은 5가지 기초 통계량을 면접 대비가 되는 재사용 가능한 함수로 만듭니다: 평균, 중앙값, 분산/표준편차, 성장률, 증감률. 원문에 명시된 목표는 "이 지표 계산할 수 있어요?"라는 질문에 코드와 비즈니스 해석을 함께 답하는 것입니다 — 숫자 하나만이 아니라.

### Five KPIs at a glance / 5가지 KPI 한눈에 보기

| KPI | Formula (brief) / 공식(약식) | What it tells you / 알려주는 것 |
|---|---|---|
| Mean / 평균 | Σx / n (or weighted) | Central tendency, sensitive to outliers (중심 경향, 이상치에 민감) |
| Median / 중앙값 | middle value (중간값) | Central tendency, robust to outliers (중심 경향, 이상치에 강함) |
| Std / Var / 표준편차·분산 | spread from mean (평균으로부터의 퍼짐) | Volatility, stability via CV (변동성, CV로 본 안정성) |
| Growth Rate / 성장률 | (this - last) / last × 100, or CAGR | Time-series trend: MoM/YoY/CAGR (시계열 트렌드) |
| Change Rate / 증감률 | (current - previous) / \|previous\| × 100 | Any two-value before/after comparison (임의의 두 값 전후 비교) |

## Why do we use it?
*(When is it useful?)*

Because these five numbers are the vocabulary of every KPI dashboard and every BA/DA interview — and knowing not just the formula but WHEN mean beats median, what counts as a "stable" CV, and the difference between MoM and YoY is what separates a working analyst from someone who memorized `np.mean()`.

이 5가지 숫자는 모든 KPI 대시보드와 모든 BA/DA 면접의 공용어이기 때문이며, 공식뿐 아니라 언제 평균이 중앙값보다 나은지, 어느 정도의 CV가 "안정적"인지, MoM과 YoY의 차이가 무엇인지를 아는 것이 `np.mean()`만 외운 사람과 실전에서 일하는 애널리스트를 구분합니다.

## When is it used in Business Analytics?
*(Real-world use case)*

- Reporting "the average/typical customer spends X" — and knowing which of mean or median to trust.  
  "평균/일반적인 고객은 X를 지출합니다"를 리포트할 때 — 평균과 중앙값 중 무엇을 믿어야 할지 아는 것.
- Comparing channel or team volatility (via CV) before allocating budget or trusting a forecast.  
  예산을 배분하거나 예측을 신뢰하기 전, 채널·팀별 변동성(CV)을 비교할 때.
- Tracking MoM/YoY growth trends and computing a fair CAGR across products with different timelines.  
  MoM/YoY 성장 추세를 추적하고, 기간이 다른 상품들 간 공정한 CAGR을 계산할 때.
- Building a KPI report that shows both absolute and percentage change vs. last period.  
  지난 기간 대비 절대값과 퍼센트 변화를 모두 보여주는 KPI 리포트를 만들 때.

# 📝 Syntax

## Basic Syntax

````python
import numpy as np

data = np.array([1200, 1350, 1100, 1400, 1600, 1500])

print(np.mean(data))                          # simple average
print(np.median(data))                        # outlier-robust center
print(np.std(data), np.var(data))             # spread

prev, curr = data[:-1], data[1:]
print(prev, curr)
growth = (curr - prev) / prev * 100           # period-over-period % change
print(np.round(growth, 1))
````

## Common Variations

````python
import numpy as np

values = np.array([3.8, 4.2, 5.1, 2.9])
weights = np.array([5000, 3000, 800, 2200])

# weighted average -- weights by importance/traffic/volume
print(np.average(values, weights=weights))

# sample std (ddof=1) vs population std (ddof=0)
sample = np.array([12, 15, 11, 14, 13])
print(np.std(sample, ddof=1), np.std(sample, ddof=0))

# CAGR -- fair growth comparison across different time spans
start, end, years = 1000, 1543, 2
cagr = (end / start) ** (1 / years) - 1
print(f"{cagr*100:.2f}%")

# safe division for change-rate style KPIs
previous = np.array([1350, 0, 800])
current = np.array([1600, 500, 950])
pct_change = np.where(previous != 0, (current - previous) / np.abs(previous) * 100, np.nan)
print(pct_change)
````

# 🧪 Small Examples

## Example 1: Mean — Simple, Weighted & Moving Average
### 11-1. 평균 (Mean)

$$\text{Mean} = \frac{x_1 + x_2 + \cdots + x_n}{n} = \frac{\sum x_i}{n}$$

````python
import numpy as np

def calc_mean(arr, axis=None, label=""):
    """Reusable mean function. axis: None(overall), 0(per-column), 1(per-row)."""
    result = np.mean(arr, axis=axis)
    if label:
        tag = f"{result:.2f}" if np.ndim(result) == 0 else f"{result}"
        print(f"[{label}] mean: {tag}")
    return result

# Simple mean
monthly_revenue = np.array([1200, 1350, 1100, 1400, 1600, 1500,
                             1300, 1250, 1450, 1700, 1800, 2000])
avg = calc_mean(monthly_revenue, label="Monthly revenue")

# Weighted mean -- when some values matter more (e.g. higher-traffic channels)
channel_cvr = np.array([3.8, 4.2, 5.1, 2.9])
channel_traffic = np.array([5000, 3000, 800, 2200])
simple_avg = np.mean(channel_cvr)
weighted_avg = np.average(channel_cvr, weights=channel_traffic)
print(f"\nSimple average CVR: {simple_avg:.2f}%")
print(f"Traffic-weighted CVR: {weighted_avg:.2f}%")

# 3-month moving average -- smooths noise to reveal trend
def moving_average(arr, window=3):
    result = np.full(len(arr), np.nan)
    for i in range(window - 1, len(arr)):
        result[i] = np.mean(arr[i - window + 1:i + 1])
    return result

ma3 = moving_average(monthly_revenue)
print("\n3-month moving average (last 3 months):", np.round(ma3[-3:], 0))
````

## Example 2: Median — Outlier-Resistant Center + Quartiles
### 11-2. 중앙값 (Median)

$$\text{Median} = \begin{cases} x_{(n+1)/2} & n \text{ odd} \\ \dfrac{x_{n/2} + x_{n/2+1}}{2} & n \text{ even} \end{cases}$$

````python
import numpy as np

def calc_median(arr, label="", compare_mean=True):
    """Median + a quick skew read by comparing it to the mean."""
    median = np.median(arr)
    if label:
        print(f"[{label}]")
        print(f"  median: {median:,.0f}")
        if compare_mean:
            mean = np.mean(arr)
            skew = ("right-skewed (high-value outliers)" if mean > median * 1.1
                    else "left-skewed (low-value outliers)" if mean < median * 0.9
                    else "roughly symmetric")
            print(f"  mean:   {mean:,.0f}")
            print(f"  shape:  {skew}")
    return median

# Business: customer purchase amounts, with a VIP tail
np.random.seed(7)
purchases = np.concatenate([
    np.random.normal(80000, 20000, 950),
    np.random.normal(800000, 150000, 50)
])
purchases = np.abs(purchases).astype(int)

calc_median(purchases, label="Customer purchase amount")

q1, q2, q3 = np.percentile(purchases, [25, 50, 75])
print(f"\nQ1: {q1:,.0f} | Median: {q2:,.0f} | Q3: {q3:,.0f} | IQR: {q3-q1:,.0f}")
````

## Example 3: Std Dev & Variance — Channel Stability via CV
### 11-3. 분산 & 표준편차 (Variance & Standard Deviation)

$$\text{Variance} = \frac{\sum (x_i - \mu)^2}{n} \ (\text{ddof=0}) \quad \text{or} \quad \frac{\sum (x_i - \mu)^2}{n-1} \ (\text{ddof=1})$$
$$\text{Standard Deviation} = \sqrt{\text{Variance}}$$

````python
import numpy as np

def calc_volatility(arr, label="", ddof=1):
    """Mean, std, variance, and CV -- with a stability read."""
    mean, std, var = np.mean(arr), np.std(arr, ddof=ddof), np.var(arr, ddof=ddof)
    cv = std / mean * 100
    if label:
        stability = ("very stable" if cv < 10 else "stable" if cv < 20
                     else "moderate" if cv < 30 else "unstable")
        print(f"[{label}] mean={mean:.2f}  std={std:.2f}  CV={cv:.1f}%  -> {stability}")
    return {"mean": mean, "std": std, "var": var, "cv": cv}

# Business: ROAS volatility across 3 channels
np.random.seed(3)
channels = {
    "Search": np.array([3.2, 3.5, 3.1, 3.8, 3.3, 3.6, 3.0, 3.7, 3.2, 3.4, 3.5, 3.1]),
    "SNS": np.array([1.8, 5.2, 2.1, 6.3, 1.9, 5.8, 2.0, 6.1, 1.7, 5.5, 2.3, 5.9]),
    "Email": np.array([4.1, 4.2, 4.0, 4.3, 4.1, 4.2, 4.0, 4.3, 4.1, 4.2, 4.1, 4.0]),
}
for ch, data in channels.items():
    calc_volatility(data, label=ch)
````

## Example 4: Growth Rate — MoM, YoY & CAGR
### 11-4. 성장률 (Growth Rate)

$$\text{MoM} = \frac{\text{this month} - \text{last month}}{\text{last month}} \times 100 \qquad \text{CAGR} = \left(\frac{\text{end}}{\text{start}}\right)^{1/n} - 1$$

````python
import numpy as np

def calc_growth_rate(arr, period=1, label=""):
    """period=1 -> MoM, period=12 -> YoY."""
    arr = np.array(arr, dtype=float)
    growth = np.full(len(arr), np.nan)
    valid = arr[:-period] != 0            # guard the DENOMINATOR (the earlier period)
    growth[period:][valid] = (
        (arr[period:][valid] - arr[:-period][valid]) / arr[:-period][valid] * 100
    )
    if label:
        g = growth[~np.isnan(growth)]
        print(f"[{label}] avg {np.mean(g):+.2f}%  max {np.max(g):+.2f}%  "
              f"min {np.min(g):+.2f}%")
    return growth

def calc_cagr(start_val, end_val, periods, label=""):
    """Compound Annual Growth Rate."""
    cagr = (end_val / start_val) ** (1 / periods) - 1
    if label:
        print(f"[{label}] CAGR: {cagr*100:.2f}%")
    return cagr

# Business: 24 months of revenue -- MoM, YoY, and CAGR
np.random.seed(42)
monthly_24 = (1000 + np.linspace(0, 500, 24) + np.random.normal(0, 80, 24)).astype(int)

mom = calc_growth_rate(monthly_24, period=1, label="MoM growth")
yoy = calc_growth_rate(monthly_24, period=12, label="YoY growth")
calc_cagr(monthly_24[0], monthly_24[-1], periods=2, label="2-year CAGR")
````

## Example 5: Change Rate — A Monthly KPI Report Card
### 11-5. 증감률 (Change Rate)

$$\Delta = \text{current} - \text{previous} \qquad \Delta\% = \frac{\text{current} - \text{previous}}{|\text{previous}|} \times 100$$

````python
import numpy as np

def calc_change(current, previous, label="", unit=""):
    """Absolute + percent change, safe against a zero/negative previous value."""
    current = np.array(current, dtype=float)
    previous = np.array(previous, dtype=float)
    change_abs = current - previous
    change_pct = np.where(previous != 0, change_abs / np.abs(previous) * 100, np.nan)
    if label and np.ndim(current) == 0:
        arrow = "up" if change_abs > 0 else ("down" if change_abs < 0 else "flat")
        print(f"[{label}] {previous:,.1f}{unit} -> {current:,.1f}{unit}  "
              f"({arrow} {change_abs:+,.1f}{unit}, {change_pct:+.2f}%)")
    return {"abs": change_abs, "pct": change_pct}

# Business: monthly KPI report -- multiple metrics vs last month
kpis = {
    "Revenue": {"prev": 1350, "curr": 1600},
    "Visitors": {"prev": 3500, "curr": 3200},
    "Conversion rate": {"prev": 3.86, "curr": 4.12},
    "Ad spend": {"prev": 800, "curr": 950},
}
print("=== Monthly KPI change report ===")
for kpi, vals in kpis.items():
    calc_change(vals["curr"], vals["prev"], label=kpi)
````

# ⚠️ Common Mistakes

**Mistake 1 — Trusting a team/department average without checking for outlier distortion.**
One huge deal or one VIP customer can single-handedly drag a small team's average far above what "typical" performance actually looks like.  
한 건의 대형 계약이나 VIP 고객 한 명이 작은 팀의 평균을 실제 "일반적인" 성과보다 훨씬 위로 끌어올릴 수 있습니다.  
✅ **Fix:** Always compute the median alongside the mean — a large gap is your outlier warning sign.  
✅ **해결법:** 항상 평균과 함께 중앙값도 계산하세요 — 둘의 차이가 크면 이상치 경고 신호입니다.

**Mistake 2 — Using a plain average when groups have very different sizes or weights.**
Averaging 4 channels' conversion rates equally ignores that one channel might carry 10x the traffic of another — the "typical" rate should reflect that imbalance.  
4개 채널의 전환율을 동일하게 평균 내면, 한 채널이 다른 채널보다 트래픽이 10배 많을 수 있다는 점을 무시하게 됩니다 — "일반적인" 전환율은 이 불균형을 반영해야 합니다.  
✅ **Fix:** Use `np.average(arr, weights=...)` whenever the underlying groups differ in size or importance.  
✅ **해결법:** 기저 그룹의 크기나 중요도가 다르다면 항상 `np.average(arr, weights=...)`를 사용하세요.

**Mistake 3 — Interpreting variance directly instead of standard deviation.**
Variance is in SQUARED units — a variance of "400" for a metric measured in won doesn't mean anything intuitive on its own.  
분산은 제곱 단위입니다 — 원(KRW) 단위 지표에서 분산이 "400"이라는 것 자체는 직관적인 의미가 없습니다.  
✅ **Fix:** Always report and interpret the standard deviation (same units as the original data), and reserve variance for intermediate calculations.  
✅ **해결법:** 항상 표준편차(원본과 같은 단위)를 보고·해석하고, 분산은 중간 계산용으로만 사용하세요.

**Mistake 4 — Mixing up `ddof=0` (population) and `ddof=1` (sample).**
If your data IS the full population, `ddof=0` is correct; if it's a sample representing a larger population (the more common case, and most textbooks' default), you want `ddof=1`.  
데이터가 모집단 전체라면 `ddof=0`이 맞고, 더 큰 모집단을 대표하는 표본이라면(더 흔한 경우이며 대부분 교과서의 기본값) `ddof=1`을 써야 합니다.  
✅ **Fix:** Default to `ddof=1` unless you're certain your array IS the entire population.  
✅ **해결법:** 배열이 모집단 전체라는 확신이 없다면 기본적으로 `ddof=1`을 사용하세요.

**Mistake 5 — Computing growth rate without guarding against a zero previous value.**
If last month's value was 0, `(this - last) / last` divides by zero — `calc_growth_rate` needs the `valid = arr[:-period] != 0` check for exactly this reason. Note it must mask `arr[:-period]` (the earlier period, which is the DENOMINATOR), not `arr[period:]` — masking the current period lets a zero denominator slip through and produces `inf` instead of `NaN`.  
지난 달 값이 0이었다면 `(이번 - 지난) / 지난`은 0으로 나누게 됩니다 — `calc_growth_rate`에 `valid = arr[:-period] != 0` 확인이 필요한 이유가 바로 이것입니다. 반드시 `arr[:-period]`(과거 기간 = 분모)를 검사해야 하며, `arr[period:]`(현재 기간)를 검사하면 0인 분모가 통과되어 `NaN` 대신 `inf`가 나옵니다.  
✅ **Fix:** Always check the denominator before computing any growth or change rate.  
✅ **해결법:** 성장률이나 증감률을 계산하기 전에는 항상 분모를 확인하세요.

**Mistake 6 — Reading MoM noise as a trend on seasonal data.**
If a business has a strong seasonal pattern (e.g. a December spike), MoM growth will swing wildly every year at the same point — that's not a real trend break.  
비즈니스에 강한 계절성 패턴이 있다면(예: 12월 급증), MoM 성장률은 매년 같은 시점에 크게 요동칩니다 — 이는 실제 추세 변화가 아닙니다.  
✅ **Fix:** For seasonal businesses, compare YoY (same month, one year prior) instead of MoM.  
✅ **해결법:** 계절성이 있는 비즈니스에서는 MoM 대신 YoY(작년 동월 대비)로 비교하세요.

**Mistake 7 — Reporting a change rate when the previous value is 0 or negative.**
`calc_change()` correctly returns `NaN` for these cases — trying to force a percentage anyway produces a number with no real meaning.  
`calc_change()`는 이런 경우 올바르게 `NaN`을 반환합니다 — 그런데도 퍼센트를 억지로 만들면 실제 의미가 없는 숫자가 나옵니다.  
✅ **Fix:** When the previous value is 0 or negative, report the absolute change alone and explain why a percentage isn't meaningful here.  
✅ **해결법:** 이전 값이 0이거나 음수라면 절대값 변화만 보고하고, 왜 퍼센트가 여기서 의미가 없는지 설명하세요.

**Mistake 8 — Getting excited about a huge % change from a tiny base ("Base Effect").**
Going from 2 sales to 6 sales is technically "+200%," but the absolute business impact is still just 4 extra sales — very different from a real business tripling in size.  
판매 2건에서 6건이 되는 것은 기술적으로 "+200%"이지만, 실제 비즈니스 임팩트는 여전히 4건 추가일 뿐입니다 — 실제로 사업이 3배로 커진 것과는 매우 다릅니다.  
✅ **Fix:** Always report the absolute number alongside a large percentage change, especially when the base value is small.  
✅ **해결법:** 특히 기준값(base)이 작을 때는, 큰 퍼센트 변화와 함께 항상 절대값도 함께 보고하세요.

# 💡 Tips
Useful tips or shortcuts

- When reporting "the average," always compute the median alongside it as a free sanity check — if they're close, the mean is trustworthy; if they're far apart, say so.
  "평균"을 리포트할 때는 항상 중앙값도 함께 계산해서 무료 검증 수단으로 쓰세요 — 둘이 비슷하면 평균을 신뢰할 수 있고, 많이 다르면 그 사실을 함께 언급하세요.
- CV (`std/mean × 100`) is the way to compare volatility ACROSS metrics with different units or scales — raw std alone can't do that.
  CV(`std/mean × 100`)는 단위나 스케일이 다른 지표들 간 변동성을 비교하는 방법입니다 — 원시 std만으로는 비교할 수 없습니다.
- CAGR is the fair way to compare growth across periods of different LENGTH — a 3-year 50% total growth and a 1-year 50% total growth are not the same story.
  CAGR은 기간 길이가 다른 성장을 공정하게 비교하는 방법입니다 — 3년간 총 50% 성장과 1년간 총 50% 성장은 같은 이야기가 아닙니다.

# 💼 Business Example
*How would a Business Analyst use this?*

**Scenario / 시나리오:**
In a BA interview, you're given one array of 18 months of revenue and asked, live, to report: the average, whether it's outlier-distorted, the volatility (with an interpretation), and this month's growth rate — all with working code.
BA 면접에서 18개월치 매출 배열 하나를 받고, 그 자리에서 평균, 이상치로 왜곡되었는지 여부, 변동성(해석 포함), 이번 달 성장률을 코드와 함께 보고해달라는 요청을 받은 상황입니다.

**To-do / 할 일:**
- [x] Compute mean and median together to check for distortion.  
      평균과 중앙값을 함께 계산해서 왜곡 여부를 확인합니다.
- [x] Compute CV and classify stability.  
      CV를 계산하고 안정성을 분류합니다.
- [x] Compute this month's growth rate vs last month.  
      이번 달의 전월 대비 성장률을 계산합니다.

````python
import numpy as np

np.random.seed(11)
revenue_18mo = (1400 + np.linspace(0, 300, 18) + np.random.normal(0, 90, 18)).astype(int)

mean, median = np.mean(revenue_18mo), np.median(revenue_18mo)
print(f"Mean: {mean:,.0f}  Median: {median:,.0f}")
print("Distorted by outliers:", "yes" if abs(mean - median) > median * 0.1 else "no")

std = np.std(revenue_18mo, ddof=1)
cv = std / mean * 100
stability = "stable" if cv < 20 else "unstable"
print(f"\nStd: {std:,.0f}  CV: {cv:.1f}%  -> {stability}")

mom = (revenue_18mo[-1] - revenue_18mo[-2]) / revenue_18mo[-2] * 100
print(f"\nLatest MoM growth: {mom:+.1f}%")
````

# ❓ Review Questions

**Q1.** If a dataset's mean is much higher than its median, what does that tell you about the distribution?  
데이터셋의 평균이 중앙값보다 훨씬 높다면, 이는 분포에 대해 무엇을 말해주나요?

A much higher mean than median means a right tail — large outliers are pulling the mean up.  
평균이 중앙값보다 훨씬 크면, 큰 이상치가 평균을 위로 끌어올리고 있다는 뜻입니다.

**Q2.** Why is CV (coefficient of variation) more useful than raw standard deviation when comparing volatility across different channels?  
서로 다른 채널 간 변동성을 비교할 때 CV(변동계수)가 원시 표준편차보다 왜 더 유용한가요?

CV is std / mean × 100, so you can compare volatility even when channels have different scales.  
std는 원 단위라 규모가 다른 채널끼리 비교가 안 되고, CV는 평균 대비 %라서 비교가 됩니다.

**Q3.** What's the difference between MoM growth rate and CAGR, and when would you use each?  
MoM 성장률과 CAGR의 차이는 무엇이며, 각각 언제 사용하나요?

MoM is month-vs-last-month; CAGR is the fair annual rate over several years.  
MoM은 바로 전월 대비(단기·잡음 많음), CAGR은 여러 해를 “매년 이만큼씩 복리 성장”으로 환산한 값입니다.

**Q4.** Why does `calc_growth_rate()` need a `valid = arr[:-period] != 0` check before dividing, and why must the mask use `arr[:-period]` rather than `arr[period:]`?  
`calc_growth_rate()`는 왜 나눗셈 전에 `valid = arr[:-period] != 0` 확인이 필요하며, 왜 마스크가 `arr[period:]`가 아니라 `arr[:-period]`여야 하나요?

You must guard the denominator (the earlier period arr[:-period]), not the current period.  
성장률은 (이번 − 지난) / 지난이라 분모가 지난 값입니다. arr[period:](이번)을 검사하면 지난달이 0인데도 나눗셈이 새어 inf가 납니다. arr[:-period] != 0이어야 지난달이 0인 칸을 nan으로 막습니다.

**Q5.** What is the "Base Effect," and why can it make a percentage change misleading?  
"Base Effect"란 무엇이며, 왜 퍼센트 변화를 오해하게 만들 수 있나요?

a tiny starting number makes % look huge even if the absolute change is small.  
기준값(분모)이 작으면 퍼센트가 쉽게 커집니다.
