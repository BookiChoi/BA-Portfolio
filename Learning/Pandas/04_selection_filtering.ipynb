# 🧠 Concept

## What is it?
*(Explain it in your own words.)*

**English**
Selection & filtering is how you narrow a full DataFrame down to just the rows and columns you actually need — "only Engineering employees," "just the name and salary columns," "the top 3 by rating." pandas gives you several tools for this: bracket selection (`df["col"]`), label/position-based access (`.loc[]` / `.iloc[]`), condition-based row filtering (Boolean indexing, `isin()`, `between()`, `query()`), and ranking (`sort_values()`, `nlargest()`).

**한글**
선택 & 필터링은 전체 DataFrame을 실제로 필요한 행과 열만으로 좁히는 방법입니다 — "개발팀 직원만", "name과 salary 열만", "평점 상위 3명". pandas는 이를 위한 여러 도구를 제공합니다: 대괄호 선택(`df["col"]`), 라벨/위치 기반 접근(`.loc[]` / `.iloc[]`), 조건 기반 행 필터링(Boolean indexing, `isin()`, `between()`, `query()`), 그리고 순위 매기기(`sort_values()`, `nlargest()`).

## Why do we use it?
*(When is it useful?)*

**English**
Real datasets rarely need to be analyzed in full — you group sales by *this* region, chart *this* quarter, report the top *5* products. Filtering first keeps every downstream calculation faster and the output focused on exactly what the question asked. It's also how you act on what EDA found: isolate just the rows with a missing value, or just the duplicated ones, so you can decide what to do with them.

**한글**
실제 데이터셋은 전체를 다 분석해야 하는 경우가 드뭅니다 — *이* 지역별로 매출을 그룹화하고, *이* 분기의 차트를 그리고, 상위 *5*개 제품을 보고합니다. 먼저 필터링을 하면 이후의 모든 계산이 더 빨라지고, 출력이 질문이 요구한 것에 정확히 집중됩니다. 이는 또한 EDA가 발견한 것에 대응하는 방법이기도 합니다 — 결측치가 있는 행만, 또는 중복된 행만 따로 떼어내서 어떻게 처리할지 결정할 수 있습니다.

## When is it used in Business Analytics?
*(Real-world use case)*

**English**
This is the single most-used skill in day-to-day analysis — nearly every request from a stakeholder is a filter question in disguise ("how many customers in Seoul spent over ₩100,000?", "which reps are below quota?"). Fluency here is the difference between writing one clean line and fighting with the data for ten minutes.

**한글**
이는 일상적인 분석에서 가장 많이 쓰이는 단일 스킬입니다 — 이해관계자의 요청은 거의 항상 필터링 질문을 변장하고 있습니다("서울 고객 중 10만원 이상 쓴 사람이 몇 명이야?", "어떤 담당자가 목표 미달이야?"). 이 부분에 능숙한 것과 그렇지 않은 것의 차이는, 깔끔한 한 줄을 쓰는 것과 10분간 데이터와 씨름하는 것의 차이입니다.

### Quick Comparison: JS/TS vs pandas / 빠른 비교

| Concept / 개념 | JavaScript / TypeScript | pandas |
|---|---|---|
| Get one column / 열 하나 가져오기 | `arr.map(o => o.col)` | `df["col"]` → Series |
| Get several columns / 여러 열 가져오기 | `arr.map(o => ({col1:o.col1, col2:o.col2}))` | `df[["col1","col2"]]` → DataFrame |
| Access by position / 위치로 접근 | `arr[2]` | `df.iloc[2]` |
| Access by label / 라벨로 접근 | not built-in / 기본 지원 없음 | `df.loc["label"]` |
| Filter rows / 행 필터링 | `arr.filter(o => o.salary > 100)` | `df[df["salary"] > 100]` or `df.query("salary > 100")` |
| Multiple conditions / 여러 조건 | `arr.filter(o => o.a && o.b)` | `df[(cond1) & (cond2)]` — note `&`, not `&&` |
| Membership in a list / 리스트 포함 여부 | `arr.filter(o => list.includes(o.x))` | `df[df["x"].isin(list)]` |
| Sort + take top N / 정렬 후 상위 N개 | `[...arr].sort(...).slice(0, 3)` | `df.nlargest(3, "col")` |

# 📝 Syntax

## Basic Syntax

````python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Minsu", "Younghee", "Junho"],
    "dept": ["Sales", "Marketing", "Engineering"],
    "salary": [4200000, 3800000, 5100000]
})

# Single brackets -> Series / 대괄호 1개 -> Series
print(employees["dept"])
print()

# Boolean indexing -- keep only rows where the condition is True
# Boolean indexing -- 조건이 True인 행만 유지
print(employees[employees["salary"] > 4000000])
````

## Common Variations

````python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Minsu", "Younghee", "Junho"],
    "dept": ["Sales", "Marketing", "Engineering"],
    "salary": [4200000, 3800000, 5100000]
})

# Double brackets -> DataFrame, even for a single column / 대괄호 2개 -> 단일 열이라도 DataFrame
print(employees[["dept"]])
print()

# .loc -- label-based row + column access / .loc -- 라벨 기반 행 + 열 접근
print("loc[1, 'salary']:", employees.loc[1, "salary"])
print()

# query() -- the same filter, written like a SQL WHERE clause / query() -- 같은 필터를 SQL WHERE절처럼
print(employees.query("salary > 4000000"))
````

# 🧪 Small Examples

## Example 1 — Column Selection: Series vs DataFrame
*(Covers source section 4-1)*

**English:** `df["col"]` with single brackets returns a **Series**. `df[["col1", "col2"]]` with double brackets — a list inside the brackets — returns a **DataFrame**, even if the list has only one column name in it. This distinction decides which methods you can chain next.  
**한글:** 대괄호 1개인 `df["col"]`은 **Series**를 반환합니다. 대괄호 안에 리스트가 들어간 `df[["col1", "col2"]]`(대괄호 2개)은, 리스트 안에 열 이름이 하나뿐이어도 **DataFrame**을 반환합니다. 이 차이가 다음에 어떤 메서드를 체이닝할 수 있는지를 결정합니다.

````python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Minsu", "Younghee", "Junho", "Seoyeon"],
    "dept": ["Sales", "Marketing", "Engineering", "Sales"],
    "salary": [4200000, 3800000, 5100000, 3500000],
    "years": [3, 5, 7, 2]
})

# Single brackets -> Series / 대괄호 1개 -> Series
dept_series = employees["dept"]
print(type(dept_series))
print(dept_series)
print()

# Double brackets (a list inside) -> DataFrame, even with just one column
# 대괄호 2개(안에 리스트) -> 열이 하나뿐이어도 DataFrame
dept_df = employees[["dept"]]
print(type(dept_df))
print(dept_df)
print()

# The real reason this matters -- selecting several columns at once
# 이게 중요한 진짜 이유 -- 여러 열을 한 번에 선택할 때
print(employees[["name", "dept", "salary"]])
````

## Example 2 — loc / iloc: Label vs Position
*(Covers source section 4-2)*

**English:** `.loc[]` selects by **label**; `.iloc[]` selects by **position** (an integer, like a JS array index). With the default `RangeIndex` they happen to agree, but two behaviors set them apart for good: slicing (`.loc[1:3]` **includes** row 3; `.iloc[1:3]` **excludes** it, just like JS's `arr.slice(1,3)`), and custom indexes (once the index is a string or a date, `.iloc[]` never understands it — only `.loc[]` does).  
**한글:** `.loc[]`는 **라벨**로 선택하고, `.iloc[]`는 **위치**(정수, JS 배열 인덱스와 같음)로 선택합니다. 기본 `RangeIndex`에서는 둘이 우연히 일치하지만, 두 가지 동작이 이 둘을 확실히 갈라놓습니다: 슬라이싱(`.loc[1:3]`은 3행을 **포함**하고, `.iloc[1:3]`은 JS의 `arr.slice(1,3)`처럼 3행을 **제외**), 그리고 커스텀 인덱스(인덱스가 문자열이나 날짜가 되면, `.iloc[]`는 그것을 절대 이해하지 못하고 `.loc[]`만 이해함).

````python
import pandas as pd

employees = pd.DataFrame({
    "emp_id": ["E001", "E002", "E003", "E004", "E005", "E006", "E007"],
    "name": ["Minsu", "Younghee", "Junho", "Seoyeon", "Daehyun", "Soyoung", "Jihoon"],
    "dept": ["Sales", "Marketing", "Engineering", "Sales", "Marketing", "Engineering", "Sales"],
    "salary": [4200000, 3800000, 5100000, 3500000, 4500000, 5800000, 4000000]
})

# 1) Single row -- with the default RangeIndex, loc and iloc agree
#    단일 행 -- 기본 RangeIndex에서는 loc와 iloc가 일치함
print("loc[2] == iloc[2]:", employees.loc[2].equals(employees.iloc[2]))
print()

# 2) Single value -- a 3-column frame so position 1 really IS "salary"
#    단일 값 -- 열이 3개뿐인 표라 위치 1이 정확히 "salary"
small = pd.DataFrame({
    "name": ["Minsu", "Younghee", "Junho"],
    "salary": [4200000, 3800000, 5100000],
    "rating": [4.2, 3.8, 4.7]
})
print("loc[2, 'salary']:", small.loc[2, "salary"])
print("iloc[2, 1]      :", small.iloc[2, 1])
print()

# 3) Slicing -- loc INCLUDES the end label, iloc EXCLUDES the end position
#    슬라이싱 -- loc는 끝 라벨을 포함하고, iloc는 끝 위치를 제외함
print("loc[1:3]  ->", len(employees.loc[1:3]), "rows (1,2,3 -- end included)")
print("iloc[1:3] ->", len(employees.iloc[1:3]), "rows (1,2 -- end excluded, like JS's arr.slice(1,3))")
print()

# 4) Custom index -- this is where loc and iloc really diverge
#    커스텀 인덱스 -- loc와 iloc가 진짜로 갈라지는 지점
emp_idx = employees.set_index("emp_id")
print("loc['E002']:", emp_idx.loc["E002", "name"])   # by label -- works / 라벨로 접근 -- 동작함
print("iloc[1]    :", emp_idx.iloc[1]["name"])        # by position -- still works (same row, by coincidence) / 위치로 접근 -- 동작함(우연히 같은 행)

try:
    emp_idx.iloc["E002"]   # iloc NEVER accepts a label, no matter what the index looks like
except TypeError as e:
    print("iloc['E002'] raises:", e)
````

## Example 3 — Boolean Indexing: Single & Compound Conditions
*(Covers source section 4-3)*

**English:** A condition like `employees["salary"] > 4500000` produces a Series of `True`/`False` values, and wrapping it in `df[...]` keeps only the `True` rows. To combine conditions, pandas requires `&` (AND), `|` (OR), and `~` (NOT) — plain Python `and` / `or` / `not` don't work on a whole Series — and every individual condition needs its own parentheses.  
**한글:** `employees["salary"] > 4500000` 같은 조건은 `True`/`False` 값으로 된 Series를 만들고, 이를 `df[...]`로 감싸면 `True`인 행만 남습니다. 조건을 결합하려면 pandas는 `&`(AND), `|`(OR), `~`(NOT)를 요구합니다 — 일반 파이썬의 `and` / `or` / `not`은 Series 전체에 대해 동작하지 않습니다 — 그리고 각 개별 조건은 반드시 괄호로 감싸야 합니다.

````python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Minsu", "Younghee", "Junho", "Seoyeon", "Daehyun", "Soyoung", "Jihoon"],
    "dept": ["Sales", "Marketing", "Engineering", "Sales", "Marketing", "Engineering", "Sales"],
    "salary": [4200000, 3800000, 5100000, 3500000, 4500000, 5800000, 4000000],
    "years": [3, 5, 7, 2, 6, 9, 4]
})

# Single condition / 단일 조건
print("salary > 4,500,000:")
print(employees[employees["salary"] > 4500000])
print()

# AND (&) -- Engineering AND at least 7 years / AND (&) -- 개발팀이면서 근속 7년 이상
print("Engineering with 7+ years:")
print(employees[(employees["dept"] == "Engineering") & (employees["years"] >= 7)])
print()

# OR (|) -- Sales OR salary over 5,000,000 / OR (|) -- 영업팀이거나 연봉 5백만 초과
print("Sales OR salary > 5,000,000:")
print(employees[(employees["dept"] == "Sales") | (employees["salary"] > 5000000)])
print()

# NOT (~) -- everyone except Marketing / NOT (~) -- 마케팅팀이 아닌 직원
print("not Marketing:")
print(employees[~(employees["dept"] == "Marketing")])
````

## Example 4 — isin / between: Multi-Value & Range Filters
*(Covers source section 4-4)*

**English:** `.isin([...])` keeps rows whose value matches *any* item in a list — a shortcut for stacking several `|` conditions. `.between(lo, hi)` keeps rows within a range, **including both ends** — the pandas equivalent of JS's `x >= lo && x <= hi`, in far fewer characters.  
**한글:** `.isin([...])`은 값이 리스트 안의 항목 *중 하나*와 일치하는 행을 남깁니다 — 여러 `|` 조건을 쌓는 것의 단축형입니다. `.between(lo, hi)`는 **양쪽 끝을 모두 포함**하는 범위 안의 행을 남깁니다 — JS의 `x >= lo && x <= hi`에 해당하는 pandas 버전을, 훨씬 적은 글자 수로 표현합니다.

````python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Minsu", "Younghee", "Junho", "Seoyeon", "Daehyun", "Soyoung", "Jihoon"],
    "dept": ["Sales", "Marketing", "Engineering", "Sales", "Marketing", "Engineering", "Sales"],
    "salary": [4200000, 3800000, 5100000, 3500000, 4500000, 5800000, 4000000]
})

# isin -- matches ANY value in the list / isin -- 리스트 중 하나라도 일치
print("dept is Sales or Engineering:")
print(employees[employees["dept"].isin(["Sales", "Engineering"])])
print()

# ~isin -- exclude values in the list / ~isin -- 리스트에 있는 값 제외
print("everyone except Marketing:")
print(employees[~employees["dept"].isin(["Marketing"])])
print()

# between -- inclusive range filter, both ends included / between -- 양쪽 끝을 포함하는 범위 필터
print("salary between 4,000,000 and 5,000,000 (inclusive):")
print(employees[employees["salary"].between(4000000, 5000000)])
````

## Example 5 — query(): Filtering That Reads Like SQL
*(Covers source section 4-5)*

**English:** `query()` produces exactly the same result as Boolean indexing, just written as a string that reads closer to a SQL `WHERE` clause. A Python variable can be referenced inside the string with `@variable_name`.  
**한글:** `query()`는 Boolean indexing과 정확히 동일한 결과를 만들지만, SQL의 `WHERE`절에 더 가깝게 읽히는 문자열로 작성합니다. 파이썬 변수는 문자열 안에서 `@변수명`으로 참조할 수 있습니다.

### `query()` vs Boolean Indexing / 언제 무엇을 쓸까

| | Boolean Indexing | `query()` |
|---|---|---|
| Complex conditions / 조건이 복잡할 때 | gets long and parenthesis-heavy / 길어지고 괄호가 많아짐 | reads closer to natural language / 자연어에 가깝게 읽힘 |
| Column names with spaces / 열 이름에 공백 | works directly / 바로 사용 가능 | needs backticks: `` `col name` `` / 백틱으로 감싸야 함 |
| Referencing a Python variable / 외부 변수 참조 | use the variable directly / 변수명 그대로 사용 | needs `@variable` / `@변수명` 필요 |
| Chaining with `.loc` etc. / `.loc` 등과 체이닝 | very flexible / 결합이 매우 유연 | clean inside a method chain / 메서드 체이닝에 깔끔 |

````python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Minsu", "Younghee", "Junho", "Seoyeon", "Daehyun", "Soyoung", "Jihoon"],
    "dept": ["Sales", "Marketing", "Engineering", "Sales", "Marketing", "Engineering", "Sales"],
    "salary": [4200000, 3800000, 5100000, 3500000, 4500000, 5800000, 4000000],
    "years": [3, 5, 7, 2, 6, 9, 4]
})

# Same filter, two ways -- confirm they produce identical results
# 같은 필터를 두 가지 방식으로 -- 결과가 동일한지 확인
a = employees[(employees["dept"] == "Engineering") & (employees["years"] >= 7)]
b = employees.query("dept == 'Engineering' and years >= 7")
print("identical results:", a.equals(b))
print(b)
print()

# @variable -- reference a Python variable inside the query string
# @변수 -- query 문자열 안에서 파이썬 변수 참조
threshold = 4500000
print(employees.query("salary > @threshold")[["name", "dept", "salary"]])
print()

# isin-style filtering also works inside query() / isin 스타일 필터도 query() 안에서 가능
print(employees.query("dept in ['Sales', 'Engineering']")[["name", "dept"]])
````

## Example 6 — sort_values / nlargest / nsmallest
*(Covers source section 4-6)*

**English:** `.sort_values("col", ascending=False)` sorts by one column. Passing a list to both `sort_values` and `ascending` sorts by multiple columns, each with its own direction. `.nlargest(n, "col")` / `.nsmallest(n, "col")` are shortcuts for "sort, then take the top/bottom N" in one call.  
**한글:** `.sort_values("col", ascending=False)`는 열 하나를 기준으로 정렬합니다. `sort_values`와 `ascending` 둘 다에 리스트를 전달하면 여러 열을 기준으로, 각각 다른 방향으로 정렬합니다. `.nlargest(n, "col")` / `.nsmallest(n, "col")`는 "정렬한 뒤 상위/하위 N개 가져오기"를 한 번에 하는 단축형입니다.

````python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Minsu", "Younghee", "Junho", "Seoyeon", "Daehyun", "Soyoung", "Jihoon"],
    "dept": ["Sales", "Marketing", "Engineering", "Sales", "Marketing", "Engineering", "Sales"],
    "salary": [4200000, 3800000, 5100000, 3500000, 4500000, 5800000, 4000000],
    "rating": [4.2, 3.8, 4.7, 3.5, 4.5, 4.9, 4.1]
})

# Single-column sort, descending / 단일 열 기준 내림차순 정렬
print("sorted by salary (descending):")
print(employees.sort_values("salary", ascending=False))
print()

# Multi-column sort -- dept ascending, salary descending within each dept
# 다중 열 정렬 -- dept는 오름차순, 같은 dept 안에서는 salary 내림차순
print("sorted by dept (asc), then salary (desc):")
print(employees.sort_values(["dept", "salary"], ascending=[True, False]))
print()

# nlargest / nsmallest -- shortcut for sort_values().head(n) / sort_values().head(n)의 단축형
print("top 3 by salary:")
print(employees.nlargest(3, "salary"))
print()
print("bottom 3 by salary:")
print(employees.nsmallest(3, "salary"))
````

## Example 7 — Common Combo Patterns
*(Covers source section 4-7)*

**English:** Pattern A chains filter → column selection → sort into one readable pipeline — the most common shape a real request takes. Pattern B combines `isin()` with `nlargest()` to answer "top N *within* a specific group" in two lines.  
**한글:** 패턴 A는 필터 → 열 선택 → 정렬을 하나의 읽기 쉬운 파이프라인으로 체이닝합니다 — 실제 요청이 가장 흔하게 취하는 형태입니다. 패턴 B는 `isin()`과 `nlargest()`를 결합해서 "특정 그룹 *안에서* 상위 N개"를 두 줄로 답합니다.

````python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Minsu", "Younghee", "Junho", "Seoyeon", "Daehyun", "Soyoung", "Jihoon"],
    "dept": ["Sales", "Marketing", "Engineering", "Sales", "Marketing", "Engineering", "Sales"],
    "salary": [4200000, 3800000, 5100000, 3500000, 4500000, 5800000, 4000000],
    "rating": [4.2, 3.8, 4.7, 3.5, 4.5, 4.9, 4.1]
})

# Pattern A: filter -> select columns -> sort, all in one chain
# 패턴 A: 필터 -> 열 선택 -> 정렬을 한 번에 체이닝
result_a = (
    employees[employees["salary"] >= 4000000]
    [["name", "dept", "salary", "rating"]]
    .sort_values("rating", ascending=False)
)
print("Pattern A -- salary >= 4M, sorted by rating:")
print(result_a)
print()

# Pattern B: isin filter + nlargest -- top N within a specific group of departments
# 패턴 B: isin 필터 + nlargest -- 특정 부서 그룹 안에서 상위 N명
result_b = (
    employees[employees["dept"].isin(["Sales", "Marketing"])]
    .nlargest(3, "salary")[["name", "dept", "salary"]]
)
print("Pattern B -- top 3 salaries in Sales/Marketing:")
print(result_b)
````

# ⚠️ Common Mistakes

### Mistake 1 — Using Python's `and` / `or` / `not` instead of `&` / `|` / `~`
**English:** `df[cond1 and cond2]` feels natural coming from JS's `&&` / `||` / `!`, but it raises `ValueError: The truth value of a Series is ambiguous` — Python's `and`/`or`/`not` expect a single `True`/`False`, not a whole column of them.  
**한글:** JS의 `&&` / `||` / `!`에 익숙하면 `df[cond1 and cond2]`가 자연스럽게 느껴지지만, `ValueError: The truth value of a Series is ambiguous`가 발생합니다 — 파이썬의 `and`/`or`/`not`은 단일 `True`/`False` 하나를 기대하지, 열 전체를 기대하지 않기 때문입니다.

**✅ Fix / 해결법:**  
Always use `&` (AND), `|` (OR), and `~` (NOT) for Series-level Boolean logic — these operate element-by-element across the whole column.  
Series 수준의 Boolean 논리에는 항상 `&`(AND), `|`(OR), `~`(NOT)을 사용하세요 — 이들은 열 전체에 걸쳐 요소 단위로 동작합니다.

### Mistake 2 — Forgetting parentheses around each condition
**English:** `df[df["salary"] > 4000000 & df["dept"] == "Sales"]` looks reasonable but raises a `TypeError`, because `&` binds *tighter* than `>` and `==` in Python — it tries to combine `4000000` and `df["dept"]` before the comparisons even happen.  
**한글:** `df[df["salary"] > 4000000 & df["dept"] == "Sales"]`는 그럴듯해 보이지만 `TypeError`를 일으킵니다 — 파이썬에서 `&`는 `>`나 `==`보다 *더 강하게* 결합하기 때문에, 비교가 일어나기도 전에 `4000000`과 `df["dept"]`를 먼저 결합하려 하기 때문입니다.

**✅ Fix / 해결법:**  
Wrap every individual condition in its own parentheses: `df[(df["salary"] > 4000000) & (df["dept"] == "Sales")]`.  
모든 개별 조건을 각각 괄호로 감싸세요: `df[(df["salary"] > 4000000) & (df["dept"] == "Sales")]`.

### Mistake 3 — Assuming `.loc[1:3]` and `.iloc[1:3]` return the same rows
**English:** `.loc[1:3]` returns **3** rows (label `3` is included), while `.iloc[1:3]` returns only **2** (position `3` is excluded, exactly like JS's `arr.slice(1,3)`). Mixing these up is a classic off-by-one bug when switching between the two.  
**한글:** `.loc[1:3]`은 **3**개 행을 반환하고(라벨 `3`이 포함됨), `.iloc[1:3]`은 **2**개만 반환합니다(위치 `3`이 제외됨, JS의 `arr.slice(1,3)`과 정확히 같음). 이 둘을 헷갈리는 것은 서로 바꿔 쓸 때 생기는 전형적인 off-by-one 버그입니다.

**✅ Fix / 해결법:**  
Remember the rule as a pair: **loc includes, iloc excludes**. When in doubt, print `len(result)` and check it matches what you expected.  
"loc는 포함, iloc는 제외"라는 짝으로 규칙을 기억하세요. 헷갈리면 `len(result)`를 출력해서 예상과 맞는지 확인하세요.

# 💡 Tips
Useful tips or shortcuts / 유용한 팁과 단축법

- `df["col"]` returns a Series; `df[["col"]]` returns a DataFrame — reach for the double-bracket form whenever you need to chain more DataFrame methods afterward.  
`df["col"]`은 Series를, `df[["col"]]`은 DataFrame을 반환합니다 — 이후 DataFrame 메서드를 더 체이닝해야 한다면 대괄호 2개 형태를 사용하세요.
- Once the index is not the default `RangeIndex` (strings, dates), always reach for `.loc[]` — `.iloc[]` never understands a label, only a position.  
인덱스가 기본 `RangeIndex`가 아니게 되면(문자열, 날짜) 항상 `.loc[]`을 사용하세요 — `.iloc[]`은 라벨을 절대 이해하지 못하고 오직 위치만 이해합니다.
- `query()` shines once a filter needs 3+ conditions or a variable reference (`@var`) — for one or two simple conditions, Boolean indexing is usually just as readable.  
`query()`는 조건이 3개 이상이거나 변수 참조(`@변수`)가 필요할 때 빛을 발합니다 — 조건이 한두 개면 Boolean indexing도 보통 그만큼 읽기 쉽습니다.
- `nlargest(n, "col")` is both shorter and clearer in intent than `sort_values("col", ascending=False).head(n)` — reach for it whenever the goal is specifically "top N."  
`nlargest(n, "col")`은 `sort_values("col", ascending=False).head(n)`보다 짧고 의도도 명확합니다 — 목표가 정확히 "상위 N개"일 때 사용하세요.

**한글:** 영업 이사가 요청합니다: "영업 또는 마케팅팀에서 평점이 가장 높은 상위 3명을 보여줘, 단 경력 4년 이상인 사람만." 이 요청을 필터 → 순위 파이프라인으로 그대로 옮겨보세요.

**To Do / 할 일**
- [x] Filter to Sales or Marketing departments / 영업 또는 마케팅 부서로 필터링하기
- [x] Filter to 4+ years of experience / 경력 4년 이상으로 필터링하기
- [x] Combine both conditions correctly with parentheses / 두 조건을 괄호로 올바르게 결합하기

# 📝 Summary
*Write today's concept in 3~5 sentences.*

**English**
Selection and filtering narrow a DataFrame down to exactly what a question asks for. `df["col"]` returns a Series while `df[["col"]]` returns a DataFrame — a distinction that determines which methods you can chain next. `.loc[]` selects by label and includes the end of a slice, while `.iloc[]` selects by position and excludes it, like JS's `.slice()`; the difference matters most once a custom index (like an ID) is set, since `.iloc[]` never accepts a label. Boolean indexing (`&` / `|` / `~`, always parenthesized), `isin()`, and `between()` filter rows by condition, and `query()` offers a more SQL-like way to write the same filters. `sort_values()` orders rows by one or more columns, and `nlargest()` / `nsmallest()` are shortcuts for "sort, then take the top/bottom N."

**한글**
선택과 필터링은 DataFrame을 질문이 요구하는 정확히 그만큼으로 좁히는 작업입니다. `df["col"]`은 Series를, `df[["col"]]`은 DataFrame을 반환하는데 — 이 차이가 다음에 어떤 메서드를 체이닝할 수 있는지를 결정합니다. `.loc[]`는 라벨로 선택하고 슬라이스의 끝을 포함하는 반면, `.iloc[]`는 위치로 선택하고 끝을 제외합니다(JS의 `.slice()`처럼) — 이 차이는 ID 같은 커스텀 인덱스가 설정된 뒤에 가장 중요해지는데, `.iloc[]`는 라벨을 절대 받지 않기 때문입니다. Boolean indexing(`&` / `|` / `~`, 항상 괄호로 감싸기), `isin()`, `between()`은 조건으로 행을 필터링하고, `query()`는 같은 필터를 SQL에 더 가깝게 쓰는 방법을 제공합니다. `sort_values()`는 하나 이상의 열로 행을 정렬하고, `nlargest()` / `nsmallest()`는 "정렬한 뒤 상위/하위 N개를 가져오기"의 단축형입니다.
