# 🧠 Concept

## What is it?
*(Explain it in your own words.)*

**English**
A pandas **Series** is a one-dimensional labeled array — every value is paired with an index (a label), similar to one column in a spreadsheet. A **DataFrame** is a two-dimensional table made of several Series that share the same index, placed side by side — think of it as the whole spreadsheet, not just a single column.

Coming from JS/TS, the biggest mental shift is the **index**. A plain JS array only has positions (`arr[0]`, `arr[1]`, ...). A pandas Series has both a *position* and a *label*, and you get to decide what that label means — an ID, a date, a customer name, anything.

**한글**
pandas **Series**는 인덱스(라벨)와 값이 한 쌍을 이루는 1차원 배열입니다. 스프레드시트의 열(column) 하나와 비슷합니다. **DataFrame**은 같은 인덱스를 공유하는 여러 개의 Series가 옆으로 나란히 붙어 있는 2차원 테이블입니다 — 열 하나가 아니라 스프레드시트 전체라고 생각하면 됩니다.

JS/TS 출신이라면 가장 큰 개념 전환은 **인덱스**입니다. 일반 JS 배열은 위치(`arr[0]`, `arr[1]`, ...)만 가지지만, pandas Series는 위치와 라벨을 둘 다 가지며, 그 라벨이 무엇을 의미할지(ID, 날짜, 고객명 등)는 자유롭게 정할 수 있습니다.

## Why do we use it?
*(When is it useful?)*

**English**
Series and DataFrames let you apply an operation to an entire column at once — a *vectorized* operation — instead of writing a manual loop. `scores * 1.1` updates every value in one line, the way you might wish `array.map(x => x * 1.1)` worked, except faster and with less code. The index also means you can pull out a row by a meaningful label (`"E002"`, `"2024-01-15"`) instead of memorizing a numeric position.

**한글**
Series와 DataFrame을 쓰면 반복문을 직접 작성하지 않고도 열 전체에 한 번에 연산을 적용할 수 있습니다 — 이를 *벡터 연산*이라고 부릅니다. `scores * 1.1`처럼 한 줄이면 모든 값이 갱신됩니다. `array.map(x => x * 1.1)`이 이렇게 동작하길 바랐던 것과 비슷하지만, 더 빠르고 코드도 짧습니다. 인덱스가 있으면 숫자 위치를 외울 필요 없이 의미 있는 라벨(`"E002"`, `"2024-01-15"`)로 바로 행을 꺼낼 수 있습니다.

## When is it used in Business Analytics?
*(Real-world use case)*

**English**
Every analysis starts here. A CSV of orders, an Excel export of employees, a JSON response from an internal API — all of it lands in a DataFrame first. Before you can group sales by region or compute month-over-month growth, you need to know the table's shape (`shape`), what each column actually contains (`dtypes`), and how to point directly at one row (`set_index()` + `.loc[]`). Every later chapter — filtering, cleaning, GroupBy, merging — builds directly on top of what's in this notebook.

**한글**
모든 분석은 여기서 시작합니다. 주문 CSV, 직원 명단 엑셀, 사내 API의 JSON 응답 — 전부 먼저 DataFrame으로 들어옵니다. 지역별 매출을 그룹화하거나 전월 대비 성장률을 계산하기 전에, 테이블의 크기(`shape`)와 각 열이 실제로 어떤 데이터를 담고 있는지(`dtypes`), 특정 행을 바로 가리키는 방법(`set_index()` + `.loc[]`)을 알아야 합니다. 이후 챕터들 — 필터링, 정제, GroupBy, 병합 — 은 전부 이 노트북의 내용 위에 세워집니다.

### Quick Comparison: JS/TS vs pandas / 빠른 비교

| Concept / 개념 | JavaScript / TypeScript | pandas |
|---|---|---|
| 1D collection / 1차원 자료구조 | `Array` — position only / 위치만 있음 | `Series` — position **+** label / 위치 **+** 라벨 |
| 2D collection / 2차원 자료구조 | `Array<Object>` (list of objects) | `DataFrame` — columns share one index / 열들이 인덱스를 공유 |
| Access by position / 위치로 접근 | `arr[1]` | `.iloc[1]` |
| Access by label / 라벨로 접근 | not built-in — needs a `Map` / 기본 지원 없음 — `Map` 필요 | `.loc["label"]` |
| Bulk math on all items / 전체 값에 연산 | `arr.map(x => x * 1.1)` | `series * 1.1` |
| Size info / 크기 정보 | `arr.length` — one number / 숫자 하나 | `df.shape` → `(rows, cols)` tuple |

# 📝 Syntax

## Basic Syntax

````python
import pandas as pd

# Create a Series from a list — index is auto-assigned (0, 1, 2, ...)
# 리스트로 Series 생성 — 인덱스가 자동으로 부여됨 (0, 1, 2, ...)
prices = pd.Series([29000, 45000, 12000, 67000])
print(prices)
print()

# Create a DataFrame from a dict — each key becomes a column
# dict로 DataFrame 생성 — 각 key가 하나의 열(column)이 됨
products = pd.DataFrame({
    "name": ["Laptop", "Mouse", "Keyboard"],
    "price": [1200000, 25000, 45000],
    "stock": [15, 230, 89]
})
print(products)
````

## Common Variations

````python
import pandas as pd

# Series from a dict — the keys become the index
# dict로 Series 생성 — key가 인덱스가 됨
scores = pd.Series({"Korean": 85, "Math": 92, "English": 78})
print(scores)
print()

# Series with a custom index — attach meaningful labels to plain list data
# 커스텀 인덱스가 있는 Series — 리스트 데이터에 의미 있는 라벨 붙이기
ratings = pd.Series([4.5, 3.8, 4.9], index=["Chicken Place", "Cafe", "Snack Bar"], name="Rating")
print(ratings)
print()

# DataFrame from a list of dicts — the exact shape of a typical JSON API response
# list of dicts로 DataFrame 생성 — 일반적인 JSON API 응답과 동일한 형태
orders = pd.DataFrame([
    {"id": 1, "item": "Tumbler", "qty": 2},
    {"id": 2, "item": "Notebook", "qty": 5},
    {"id": 3, "item": "Pen", "qty": 10}
])
print(orders)
print()

# DataFrame from a 2D list + explicit column names
# 2D 리스트 + 열 이름(columns)으로 DataFrame 생성
data = [
    ["Seoul", 9700000],
    ["Busan", 3400000],
    ["Incheon", 3000000]
]
cities = pd.DataFrame(data, columns=["City", "Population"])
print(cities)
````

## Example 1 — Building and Inspecting a Series
*(Covers source section 1-1: Series basics)*

**English:** Build a Series three different ways, then pull out its core attributes and try both label-based and position-based access.
**한글:** 세 가지 방식으로 Series를 만들고, 핵심 속성을 꺼낸 뒤 라벨 접근과 위치 접근을 둘 다 시도해봅니다.

````python
import pandas as pd

# 1) From a list — index is auto-assigned / 리스트로 생성 — 인덱스 자동 부여
prices = pd.Series([29000, 45000, 12000, 67000])

# 2) With a custom index / 커스텀 인덱스 지정
ratings = pd.Series([4.5, 3.8, 4.9], index=["Chicken Place", "Cafe", "Snack Bar"], name="Rating")

# 3) Core attributes — like a richer version of JS's array.length
#    핵심 속성 — JS의 array.length보다 더 많은 정보를 담고 있음
print("values:", ratings.values)   # underlying numpy array / 값만 numpy 배열로
print("index :", ratings.index)    # the labels / 인덱스(라벨) 목록
print("dtype :", ratings.dtype)    # data type / 데이터 타입
print("name  :", ratings.name)     # Series name / Series 이름
print("len   :", len(ratings))     # length / 길이
print()

# Access by label vs by position / 라벨 접근 vs 위치 접근
print("by label   :", ratings["Cafe"])   # label access — pandas-only feature / 라벨 접근 — pandas만의 기능
print("by position:", ratings.iloc[1])   # position access — same idea as JS's arr[1] / 위치 접근 — JS의 arr[1]과 동일한 개념
print()

# A common combo: build + sort in one flow / 자주 쓰는 조합: 생성 + 정렬
scores = pd.Series({"Alice": 88, "Bob": 95, "Charlie": 72}, name="Score")
print(scores.sort_values(ascending=False))
````

## Example 2 — Three Ways to Build a DataFrame
*(Covers source section 1-2: DataFrame creation methods)*

**English:** In real work, a DataFrame almost always comes from one of three shapes: a dict of columns, a list of row-dicts (API responses), or a 2D list with separate column names.  
**한글:** 실무에서 DataFrame은 거의 항상 세 가지 형태 중 하나에서 만들어집니다: 열(column) 단위의 dict, 행(row) 단위 dict의 리스트(API 응답), 또는 2D 리스트 + 별도의 열 이름.

````python
import pandas as pd

# ① dict → keys become columns / dict → key가 열이 됨
products = pd.DataFrame({
    "name": ["Laptop", "Mouse", "Keyboard"],
    "price": [1200000, 25000, 45000],
    "stock": [15, 230, 89]
})
print("① dict:")
print(products)
print()

# ② list of dicts → exactly what fetch() gives you from a JSON API
#    list of dicts → fetch()로 받은 JSON API 응답과 동일한 형태
api_response = [
    {"product": "A", "sales": 100},
    {"product": "B", "sales": 200},
    {"product": "C", "sales": 150}
]
df = pd.DataFrame(api_response)
print("② list of dicts:")
print(df)
print()

# ③ 2D list + columns → row-oriented raw data / 2D 리스트 + columns → 행 단위 원시 데이터
data = [
    ["Seoul", 9700000],
    ["Busan", 3400000],
    ["Incheon", 3000000]
]
cities = pd.DataFrame(data, columns=["City", "Population"])
print("③ 2D list + columns:")
print(cities)
print()

# Common combo: check dtypes right after creation / 자주 쓰는 조합: 생성 직후 dtypes 확인
print("dtypes of df (from ②):")
print(df.dtypes)
````

## Example 3 — Core Attributes: shape / columns / index / dtypes / values
*(Covers source section 1-3: key DataFrame attributes)*

**English:** These five attributes are the first thing to check on any new DataFrame — they answer "how big is it, what's in it, and what type is each column?" in one glance.  
**한글:** 이 다섯 가지 속성은 새로운 DataFrame을 받으면 가장 먼저 확인해야 할 것들입니다 — "크기가 얼마나 되는지, 무엇이 들어있는지, 각 열의 타입이 무엇인지"를 한눈에 알려줍니다.

````python
import pandas as pd

inventory = pd.DataFrame({
    "item": ["A", "B", "C", "D"],
    "qty": [10, 25, 0, 7],
    "price": [1000, 2500, 1200, 800]
})

print("shape  :", inventory.shape)     # (rows, cols) tuple / (행, 열) 튜플
print("columns:", inventory.columns)   # column name list / 열 이름 목록
print("index  :", inventory.index)     # row index / 행 인덱스
print("dtypes :")
print(inventory.dtypes)                # dtype of every column / 각 열의 타입
print()

# .values converts everything to a single numpy array
# .values는 모든 것을 하나의 numpy 배열로 변환함
print("values (mixed types become 'object'):")
print(inventory.values)                # mixed str+int → whole array becomes 'object' / 문자+숫자 혼합 → 전체가 object 타입
print()

# Common combo: use shape to check for an empty DataFrame
# 자주 쓰는 조합: shape로 빈 DataFrame 체크
empty_df = pd.DataFrame({"a": [], "b": []})
if empty_df.shape[0] == 0:
    print("No data / 데이터 없음")
else:
    print(f"{empty_df.shape[0]} rows exist / {empty_df.shape[0]}행 존재")
````

## Example 4 — Managing the Index: set_index() / reset_index()
*(Covers source section 1-4: set_index / reset_index)*

**English:** The default index (0, 1, 2, ...) is just a row number with no meaning. Promoting a meaningful column — an ID, a currency code — to the index turns lookups into one clean line with `.loc[]`.  
**한글:** 기본 인덱스(0, 1, 2, ...)는 의미 없는 일련번호일 뿐입니다. ID, 통화 코드처럼 의미 있는 열을 인덱스로 승격시키면 `.loc[]` 한 줄로 깔끔하게 조회할 수 있습니다.

````python
import pandas as pd

employees = pd.DataFrame({
    "emp_id": ["E001", "E002", "E003"],
    "name": ["Minsu", "Jieun", "Seojun"],
    "dept": ["Sales", "Marketing", "Engineering"]
})
print("before set_index:")
print(employees)
print()

# set_index — promote emp_id from a column to the index
# set_index — emp_id를 열에서 인덱스로 승격
employees_idx = employees.set_index("emp_id")
print("after set_index('emp_id'):")
print(employees_idx)
print()

# Now emp_id is a label, not a column — look up a row directly
# 이제 emp_id는 열이 아니라 라벨 — 바로 행을 조회 가능
print("lookup E002 directly:")
print(employees_idx.loc["E002"])
print()

# reset_index — go back to a plain column + default integer index
# reset_index — 다시 일반 열 + 기본 정수 인덱스로 복원
back_to_normal = employees_idx.reset_index()
print("after reset_index():")
print(back_to_normal)
print()

# Common combo: set_index() as a one-line lookup table
# 자주 쓰는 조합: set_index()로 조회 테이블 만들기
exchange_rate = pd.DataFrame({
    "currency": ["USD", "EUR", "JPY"],
    "rate": [1350.5, 1470.2, 9.15]
}).set_index("currency")
print("USD rate:", exchange_rate.loc["USD", "rate"])
````

## Example 5 — Why dtypes Matter
*(Covers source section 1-5: dtype types and why they matter)*

**English:** JS treats every number as a single `number` type. pandas splits data into several dtypes instead, for memory efficiency and speed. This split causes two very common surprises, shown in the code below.  
**한글:** JS는 모든 숫자를 하나의 `number` 타입으로 처리합니다. pandas는 메모리 효율과 연산 속도를 위해 데이터를 여러 dtype으로 나눕니다. 이 구분 때문에 아주 흔한 문제 두 가지가 아래 코드에서 나타납니다.

| dtype | Meaning / 의미 | Example / 예시 |
|---|---|---|
| `int64` | whole numbers / 정수 | `10`, `-5` |
| `float64` | decimals, can hold `NaN` / 소수 (NaN 가능) | `3.14`, `NaN` |
| `str` *(pandas ≥ 3.0)* | pure text / 순수 문자열 | `"Seoul"` |
| `object` | mixed types in one column, or text on pandas < 3.0 / 한 열에 여러 타입 혼합, 또는 pandas 3.0 미만의 문자열 | `10` & `"78"` mixed |
| `bool` | True / False | `True` |
| `datetime64[ns]` | dates and times / 날짜시간 | `2024-01-01` |
| `category` | fixed set of labels, memory-efficient / 정해진 라벨 집합, 메모리 절약 | `"Low"/"Mid"/"High"` |

> 💡 **Version note / 버전 참고:** This notebook was tested on pandas 3.0+, where pure-text columns get their own `str` dtype. If your `.dtype` output shows `object` instead of `str` for a plain text column, you're likely on pandas < 3.0 — the underlying *behavior* (concatenation instead of addition) is identical either way.  
> 이 노트북은 pandas 3.0 이상에서 테스트되었으며, 순수 텍스트 열은 전용 `str` dtype을 가집니다. 순수 텍스트 열의 `.dtype` 출력이 `str`이 아니라 `object`로 나온다면 pandas 3.0 미만을 사용 중일 가능성이 높습니다 — 어느 쪽이든 실제 *동작*(덧셈이 아니라 이어붙이기)은 동일합니다.

````python
import pandas as pd

# Problem 1: one NaN silently turns an int column into a float column
# 문제 1: NaN 하나가 정수 열을 조용히 float 열로 바꿔버림
mixed = pd.DataFrame({
    "qty_clean": [10, 20, 30],
    "qty_with_na": [10, None, 30]
})
print(mixed.dtypes)
print(mixed)
print("-> 10 became 10.0 because a column with any NaN must be float64")
print("-> NaN이 하나라도 있으면 그 열 전체가 float64가 되므로 10이 10.0으로 표시됨")
print()

# Problem 2: numeric-looking strings don't add, they concatenate
# 문제 2: 숫자처럼 보이는 문자열은 더해지지 않고 이어붙여짐
prices = pd.Series(["1000", "2000", "3000"])
print("dtype:", prices.dtype)          # "str" on pandas 3.0+, "object" on older pandas -- either way, it's text, not numbers
                                        # pandas 3.0 이상은 "str", 이전 버전은 "object" -- 어느 쪽이든 숫자가 아니라 텍스트임
print("sum():", prices.sum())          # "100020003000" — string concatenation! / 문자열 이어붙이기!
print()

# Fix — astype() converts the type explicitly / 해결 — astype()으로 타입을 명시적으로 변환
prices_num = prices.astype(int)
print("after astype(int) - dtype:", prices_num.dtype)
print("after astype(int) - sum() :", prices_num.sum())
print()

# Common combo: cleaning a numeric-looking column straight out of a CSV
# 자주 쓰는 조합: CSV에서 막 읽은 숫자처럼 보이는 열 정리
raw = pd.DataFrame({
    "name": ["A", "B", "C"],
    "amount": ["1500", "2300", "980"]   # often arrives as text from a CSV / CSV에서 흔히 문자열로 들어옴
})
raw["amount"] = raw["amount"].astype(int)
print("cleaned sum:", raw["amount"].sum())
````

## Example 6 — Common Combo Pattern: API Response → Clean → Index → Lookup
*(Covers source section 1-6: frequently-used comprehensive pattern)*

**English:** The most common real-world flow: receive JSON, fix the numeric column's type, and set a lookup index — all in a few lines.  
**한글:** 실무에서 가장 흔한 흐름: JSON을 받아서, 숫자 열의 타입을 고치고, 조회용 인덱스를 설정하는 것 — 몇 줄이면 끝납니다.

````python
import pandas as pd

api_data = [
    {"order_id": "ORD001", "customer": "Chulsoo", "amount": "45000"},
    {"order_id": "ORD002", "customer": "Younghee", "amount": "32000"},
    {"order_id": "ORD003", "customer": "Minjun", "amount": "78000"}
]

df = pd.DataFrame(api_data)
df["amount"] = df["amount"].astype(int)   # string -> number / 문자열 -> 숫자
df = df.set_index("order_id")             # index for direct lookup / 조회용 인덱스
print(df)
print()
print("ORD002 amount:", df.loc["ORD002", "amount"])
print()

# Pattern B: verify structure immediately after receiving data
# 패턴 B: 받은 데이터 구조를 즉시 검증
check = pd.DataFrame({
    "id": [1, 2, 3],
    "score": [85.5, 90.0, 78.5]
})
print(f"rows/cols: {check.shape}")
print(check.dtypes)
````

# ⚠️ Common Mistakes

### Mistake 1 — Assuming numbers with a missing value stay integers
**English:** In JS, `[10, null, 30]` doesn't change the "type" of the array. In pandas, the moment a numeric column contains even one `NaN`, the *entire column* is upgraded to `float64` — so `10` silently displays as `10.0`.  
**한글:** JS에서는 `[10, null, 30]`이 배열의 "타입"을 바꾸지 않습니다. pandas에서는 숫자 열에 `NaN`이 단 하나만 있어도 *그 열 전체*가 `float64`로 승격되어, `10`이 조용히 `10.0`으로 표시됩니다.
**✅ Fix / 해결법:**   
This is expected behavior, not a bug — `NaN` only exists as a float value in pandas. If you need true integers with missing values, use pandas' nullable `Int64` dtype (`df["col"].astype("Int64")`) instead of plain `int64`.  
이는 버그가 아니라 정상 동작입니다 — `NaN`은 pandas에서 float로만 존재하기 때문입니다. 결측치가 있는 진짜 정수가 필요하다면 일반 `int64` 대신 pandas의 nullable `Int64` 타입(`df["col"].astype("Int64")`)을 사용하세요.

### Mistake 2 — Calling `.sum()` on numbers that are actually text
**English:** `pd.Series(["1000", "2000"]).sum()` doesn't return `3000` — it returns `"10002000"`, because the values are still text (dtype `str` on pandas 3.0+, `object` on older pandas), and `+` on strings means concatenation, exactly like JS's `"1000" + "2000"`.  
**한글:** `pd.Series(["1000", "2000"]).sum()`은 `3000`이 아니라 `"10002000"`을 반환합니다. 값이 여전히 텍스트이기 때문입니다(dtype은 pandas 3.0 이상에서 `str`, 이전 버전에서는 `object`), 문자열에 대한 `+`는 이어붙이기를 의미합니다 — JS의 `"1000" + "2000"`과 정확히 같은 원리입니다.  
**✅ Fix / 해결법:**  
Always check `.dtype` before doing math on a column you didn't create yourself. Convert with `.astype(int)` or `pd.to_numeric()` first.  
직접 만들지 않은 열에 연산을 하기 전에는 항상 `.dtype`을 먼저 확인하세요. `.astype(int)`나 `pd.to_numeric()`으로 먼저 변환해야 합니다.

### Mistake 3 — Using `.iloc[]` with a string label
**English:** After `df.index = pd.Index(["E001", "E002", "E003"])`, calling `df.iloc["E002"]` raises a `TypeError` — `.iloc` only ever accepts integer positions, no matter what the index actually looks like.  
**한글:** `df.index = pd.Index(["E001", "E002", "E003"])` 이후 `df.iloc["E002"]`를 호출하면 `TypeError`가 발생합니다 — `.iloc`은 인덱스가 실제로 어떻게 생겼든 상관없이 오직 정수 위치만 받습니다.  
**✅ Fix / 해결법:**  
If the index is made of strings or dates, always use `.loc[]`. Reserve `.iloc[]` for the default `RangeIndex` or when you truly mean "the Nth row regardless of its label."  
인덱스가 문자열이나 날짜로 되어 있다면 항상 `.loc[]`을 사용하세요. `.iloc[]`은 기본 `RangeIndex`이거나 "라벨과 상관없이 진짜 N번째 행"을 의미할 때만 사용하세요.

# 💡 Tips
Useful tips or shortcuts / 유용한 팁과 단축법

- Run `.dtypes` right after loading any new DataFrame — catching type issues early saves a lot of debugging later.  
새 DataFrame을 불러오면 바로 `.dtypes`부터 확인하세요 — 타입 문제를 미리 잡으면 나중 디버깅 시간이 크게 줄어듭니다.
- Use `set_index()` whenever a column is a natural unique key (an ID, a currency code, a date) — it turns repeated lookups into one clean `.loc[]` line.  
열에 자연스러운 고유 키(ID, 통화 코드, 날짜)가 있다면 `set_index()`를 사용하세요 — 반복되는 조회가 `.loc[]` 한 줄로 끝납니다.
- `df["col"]` (single brackets) returns a **Series**; `df[["col"]]` (double brackets) returns a **DataFrame**. This distinction becomes important starting in Section 4 (Selection & Filtering).  
`df["col"]`(대괄호 1개)은 **Series**를, `df[["col"]]`(대괄호 2개)은 **DataFrame**을 반환합니다. 이 차이는 4번 섹션(선택 & 필터링)부터 중요해집니다.
- `df.shape` returns a `(rows, cols)` tuple you can unpack directly: `rows, cols = df.shape`.  
`df.shape`는 구조분해할당이 가능한 `(행, 열)` 튜플을 반환합니다: `rows, cols = df.shape`.

# 💼 Business Example
*How would a Business Analyst use this?*

**Scenario / 시나리오**

**English:** You just pulled a raw JSON export of last week's coffee shop orders from the POS system. Before running any analysis, you need to load it into a DataFrame, confirm its shape and dtypes, and set a meaningful index so you can look up any single order instantly.

**한글:** 방금 POS 시스템에서 지난주 커피숍 주문 내역을 JSON 원본으로 받았습니다. 분석을 시작하기 전에, 이를 DataFrame으로 불러와서 크기와 dtype을 확인하고, 특정 주문을 바로 조회할 수 있도록 의미 있는 인덱스를 설정해야 합니다.

**To Do / 할 일**
- [x] Load the list of dicts into a DataFrame  
list of dicts를 DataFrame으로 불러오기
- [x] Check `shape` and `dtypes`  
`shape`와 `dtypes` 확인하기
- [x] Fix the `price` column's dtype if needed  
필요하다면 `price` 열의 dtype 수정하기
- [x] Set `order_id` as the index  
`order_id`를 인덱스로 설정하기
- [x] Look up one order directly by its ID  
ID로 특정 주문 바로 조회하기

````python
import pandas as pd

pos_export = [
    {"order_id": "CF1001", "item": "Americano", "price": "4500", "qty": 2},
    {"order_id": "CF1002", "item": "Latte", "price": "5000", "qty": 1},
    {"order_id": "CF1003", "item": "Cold Brew", "price": "5500", "qty": 3},
]

# 1) Load into a DataFrame / DataFrame으로 불러오기
orders = pd.DataFrame(pos_export)

# 2) Check shape and dtypes / shape와 dtypes 확인
print("shape:", orders.shape)
print(orders.dtypes)
print()

# 3) price came in as text -- fix it / price가 문자열로 들어왔으므로 수정
orders["price"] = orders["price"].astype(int)

# 4) Set order_id as the index / order_id를 인덱스로 설정
orders = orders.set_index("order_id")

# 5) Look up one order directly / 특정 주문 바로 조회
print("Lookup CF1002:")
print(orders.loc["CF1002"])
print()

# Bonus: total revenue for the week, now that price is numeric
# 보너스: price가 숫자가 되었으니 주간 총매출 계산
orders["revenue"] = orders["price"] * orders["qty"]
print("Total weekly revenue / 주간 총매출:", orders["revenue"].sum())
````

**English**
A pandas Series is a 1D labeled array, and a DataFrame is a 2D table made of Series that share an index. DataFrames can be built from a dict, a list of dicts, or a 2D list with column names — the second one matches JSON API responses almost exactly. Five attributes — `shape`, `columns`, `index`, `dtypes`, `values` — are the first thing worth checking on any new DataFrame. `set_index()` promotes a meaningful column into the row label so `.loc[]` can look it up directly, and `reset_index()` reverses that. **한글**
pandas Series는 1차원 라벨 배열이고, DataFrame은 같은 인덱스를 공유하는 여러 Series로 이루어진 2차원 테이블입니다.
