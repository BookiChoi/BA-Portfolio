# 🧠 Concept

## What is it?
*(Explain it in your own words.)*

**English**
EDA (Exploratory Data Analysis) is the "getting to know the data" phase that happens right after loading a DataFrame and right before you touch any real analysis. It's a fixed routine of quick checks — row/column count, dtypes, missing values, distribution, duplicates — that tells you whether the data is trustworthy and what shape it's actually in.

**한글**
EDA(탐색적 데이터 분석)는 DataFrame을 불러온 직후, 실제 분석에 손대기 직전에 거치는 "데이터와 친해지는" 단계입니다. 행/열 개수, dtype, 결측치, 분포, 중복 여부를 확인하는 고정된 루틴으로, 데이터를 믿을 수 있는지와 실제로 어떤 모양인지를 알려줍니다.

## Why do we use it?
*(When is it useful?)*

**English**
Jumping straight into analysis on unfamiliar data is how an "average conversion rate" ends up silently wrong because 15% of the rows were duplicates, or a `groupby()` crashes because a column you assumed was numeric turned out to be text. A five-minute EDA routine — `.info()`, `.describe()`, `.isna().sum()`, `.duplicated().sum()` — catches these problems before they quietly corrupt every calculation downstream.

**한글**
낯선 데이터에 곧바로 분석부터 들어가면, 행의 15%가 중복이었다는 걸 몰라서 "평균 전환율"이 조용히 틀리거나, 숫자라고 생각했던 열이 사실 텍스트였다는 걸 몰라서 `groupby()`가 터지는 일이 생깁니다. `.info()`, `.describe()`, `.isna().sum()`, `.duplicated().sum()`으로 이루어진 5분짜리 EDA 루틴은 이런 문제가 이후의 모든 계산을 조용히 망가뜨리기 전에 미리 잡아줍니다.

## When is it used in Business Analytics?
*(Real-world use case)*

**English**
Every dataset you receive — a fresh CSV export, a query result, a shared spreadsheet — gets the same few minutes of EDA before a single chart gets made. It's the difference between reporting a number you can defend in a meeting and one that falls apart under the first follow-up question ("wait, are there duplicates in here?").

**한글**
새로 받는 모든 데이터셋 — 방금 내보낸 CSV, 쿼리 결과, 공유받은 스프레드시트 — 은 차트 하나 만들기 전에 똑같이 몇 분짜리 EDA를 거칩니다. 이는 회의에서 방어할 수 있는 숫자를 보고하는 것과, 첫 번째 후속 질문("혹시 여기 중복 있어요?")에 무너지는 숫자를 보고하는 것의 차이를 만듭니다.

## Example 1 — head / tail / sample: Previewing Data
*(Covers source section 3-1)*

**English:** `.head()` and `.tail()` show the first/last N rows — the fastest way to sanity-check that a load actually worked. `.sample()` grabs random rows, which is more honest about what the *middle* of the data looks like — but only if you fix `random_state` so the "random" sample is reproducible every time you rerun the notebook.  
**한글:** `.head()`와 `.tail()`은 처음/마지막 N개 행을 보여줍니다 — 불러오기가 실제로 잘 됐는지 가장 빠르게 확인하는 방법입니다. `.sample()`은 무작위 행을 가져오는데, 데이터의 *중간* 부분이 어떻게 생겼는지 더 정직하게 보여줍니다 — 단, `random_state`를 고정해야 노트북을 다시 실행해도 같은 "무작위" 샘플이 재현됩니다.

````python
import pandas as pd
import numpy as np

campaigns = pd.DataFrame({
    "campaign_id": ["C001", "C002", "C003", "C004", "C005", "C006"],
    "channel": ["Instagram", "YouTube", "Instagram", "Google", "YouTube", "Instagram"],
    "clicks": [320, 580, 210, np.nan, 450, 320],
    "cost": [50000, 120000, 35000, 80000, 95000, 50000],
    "converted": [12, 28, 8, 15, 20, 12]
})

print("head(3):")
print(campaigns.head(3))
print()

print("tail(2):")
print(campaigns.tail(2))
print()

# random_state fixes the "randomness" so everyone gets the same sample
# random_state로 "무작위성"을 고정해서 모두가 같은 샘플을 얻게 됨
print("sample(2, random_state=1):")
print(campaigns.sample(2, random_state=1))
````

## Example 2 — info() / describe(): Structure & Distribution
*(Covers source section 3-2)*

**English:** `.info()` answers "how big is this, and what's actually inside each column" in one call — row count, dtypes, and (critically) the non-null count per column, so a missing value shows up immediately as a smaller number than the total. `.describe()` answers the numeric question — mean, spread, quartiles — and `.describe(include="all")` extends that to text columns too (most frequent value, how many unique values).  
**한글:** `.info()`는 한 번의 호출로 "이게 얼마나 크고, 각 열에 실제로 뭐가 들어있는지"를 알려줍니다 — 행 개수, dtype, 그리고 (중요하게도) 열별 결측치 아닌 값의 개수를 보여주므로, 결측치가 있으면 전체 행 수보다 작은 숫자로 바로 드러납니다. `.describe()`는 숫자에 대한 질문에 답합니다 — 평균, 분산, 사분위수 — 그리고 `.describe(include="all")`은 이를 텍스트 열까지 확장해줍니다(가장 흔한 값, 고유값 개수).

````python
import pandas as pd
import numpy as np

campaigns = pd.DataFrame({
    "campaign_id": ["C001", "C002", "C003", "C004", "C005", "C006"],
    "channel": ["Instagram", "YouTube", "Instagram", "Google", "YouTube", "Instagram"],
    "clicks": [320, 580, 210, np.nan, 450, 320],
    "cost": [50000, 120000, 35000, 80000, 95000, 50000],
    "converted": [12, 28, 8, 15, 20, 12]
})

campaigns.info()
# How to read it / 읽는 법:
# "Non-Null Count" smaller than the total row count (6) -> that column has missing values
# "Non-Null Count"가 전체 행 수(6)보다 작은 열 -> 결측치가 있다는 뜻
print()

print("describe() -- numeric columns only:")
print(campaigns.describe())
print()

print("describe(include='all') -- every column:")
print(campaigns.describe(include="all"))
````

## Example 3 — unique / nunique / value_counts: Exploring Categorical Columns
*(Covers source section 3-3)*

**English:** For a text column, `.unique()` lists every distinct value, `.nunique()` counts how many there are, and `.value_counts()` counts how often each one shows up — add `normalize=True` to get percentages instead of raw counts.  
**한글:** 텍스트 열에 대해 `.unique()`는 모든 고유값을 나열하고, `.nunique()`는 그 개수를 세며, `.value_counts()`는 각 값이 몇 번 나오는지 셉니다 — `normalize=True`를 추가하면 원래 개수 대신 비율(퍼센트)을 얻습니다.

````python
import pandas as pd
import numpy as np

campaigns = pd.DataFrame({
    "campaign_id": ["C001", "C002", "C003", "C004", "C005", "C006"],
    "channel": ["Instagram", "YouTube", "Instagram", "Google", "YouTube", "Instagram"],
    "clicks": [320, 580, 210, np.nan, 450, 320],
    "cost": [50000, 120000, 35000, 80000, 95000, 50000],
    "converted": [12, 28, 8, 15, 20, 12]
})

print("unique channels:", list(campaigns["channel"].unique()))
print("number of unique channels:", campaigns["channel"].nunique())
print()

print("value_counts():")
print(campaigns["channel"].value_counts())
print()

print("value_counts(normalize=True) -- as proportions:")
print(campaigns["channel"].value_counts(normalize=True).round(3))
````

## Example 4 — isna / duplicated: Missing Values & Duplicate Rows
*(Covers source section 3-4)*

**English:** `.isna().sum()` counts missing values per column in one line. `.duplicated()` flags rows that are exact repeats of an earlier row — but checking every column (including an ID) almost always finds zero duplicates, so `subset=[...]` (ignoring the ID) is what actually catches "this campaign got entered twice."  
**한글:** `.isna().sum()`은 한 줄로 열별 결측치 개수를 셉니다. `.duplicated()`는 이전 행과 완전히 똑같은 행에 표시를 하는데 — ID까지 포함한 모든 열을 기준으로 확인하면 거의 항상 중복이 0건으로 나오므로, (ID를 제외한) `subset=[...]`이야말로 "이 캠페인이 두 번 입력됐다"를 실제로 잡아냅니다.

````python
import pandas as pd
import numpy as np

campaigns = pd.DataFrame({
    "campaign_id": ["C001", "C002", "C003", "C004", "C005", "C006"],
    "channel": ["Instagram", "YouTube", "Instagram", "Google", "YouTube", "Instagram"],
    "clicks": [320, 580, 210, np.nan, 450, 320],
    "cost": [50000, 120000, 35000, 80000, 95000, 50000],
    "converted": [12, 28, 8, 15, 20, 12]
})

# Missing values per column / 열별 결측치 개수
print("isna().sum():")
print(campaigns.isna().sum())
print()

# Show only the rows that actually have a missing value / 결측치가 있는 행만 보기
print("rows with a missing 'clicks':")
print(campaigns[campaigns["clicks"].isna()])
print()

# Duplicate check -- across ALL columns (including the unique campaign_id) / 전체 열 기준 중복
print("duplicated() across all columns:", campaigns.duplicated().sum())

# Duplicate check -- ignoring campaign_id, so an identical campaign entered twice DOES get caught
# campaign_id를 제외하고 중복 확인 -- 그래야 두 번 입력된 동일 캠페인을 잡아냄
dup_count = campaigns.duplicated(subset=["channel", "clicks", "cost", "converted"]).sum()
print("duplicated() ignoring campaign_id:", dup_count)
print()

print("the actual duplicate rows (keep=False shows every copy):")
print(campaigns[campaigns.duplicated(subset=["channel", "clicks", "cost", "converted"], keep=False)])
````

## Example 5 — Common Combo: The Standard EDA Routine
*(Covers source section 3-5)*

**English:** Pattern 1 is the fixed 3-step routine to run on any new dataset: `.info()` for shape+dtypes, `.describe()` for distribution, and a missing-value percentage check. Pattern 2 uses `.nunique()` to instantly tell an ID column (unique count == row count) apart from a category column (few unique values) — the first is a join key, the second is a `groupby()` candidate.  
**한글:** 패턴 1은 새 데이터셋에 항상 실행하는 고정된 3단계 루틴입니다: 크기+dtype을 위한 `.info()`, 분포를 위한 `.describe()`, 그리고 결측치 비율 확인. 패턴 2는 `.nunique()`를 이용해 ID 열(고유값 개수 == 행 개수)과 카테고리 열(고유값이 적음)을 즉시 구분합니다 — 전자는 조인 키이고, 후자는 `groupby()` 후보입니다.

````python
import pandas as pd
import numpy as np

campaigns = pd.DataFrame({
    "campaign_id": ["C001", "C002", "C003", "C004", "C005", "C006"],
    "channel": ["Instagram", "YouTube", "Instagram", "Google", "YouTube", "Instagram"],
    "clicks": [320, 580, 210, np.nan, 450, 320],
    "cost": [50000, 120000, 35000, 80000, 95000, 50000],
    "converted": [12, 28, 8, 15, 20, 12]
})

# Pattern 1: the standard 3-step EDA routine / 패턴 1: 표준 EDA 3단계 루틴
print("Step 1 -- size + dtypes:")
campaigns.info()
print()

print("Step 2 -- numeric distribution:")
print(campaigns.describe())
print()

print("Step 3 -- missing value percentage:")
print((campaigns.isna().sum() / len(campaigns) * 100).round(1))
print()

# Pattern 2: tell "ID column" and "category column" apart with nunique()
# 패턴 2: nunique()로 "ID 열"과 "카테고리 열" 구분하기
for col in campaigns.select_dtypes(include=["object", "str"]).columns:
    print(f"{col}: {campaigns[col].nunique()} unique values / {len(campaigns)} rows")
````

# ⚠️ Common Mistakes

### Mistake 1 — Trusting `.duplicated()` without excluding the ID column
**English:** `campaigns.duplicated().sum()` checks *every* column at once, including `campaign_id` — and since every ID is unique by definition, the result is almost always `0`, even if the exact same campaign was accidentally entered twice under a new ID.  
**한글:** `campaigns.duplicated().sum()`은 `campaign_id`를 포함한 *모든* 열을 한꺼번에 확인합니다 — ID는 정의상 항상 고유하므로, 완전히 똑같은 캠페인이 새 ID로 실수로 두 번 입력됐어도 결과는 거의 항상 `0`이 나옵니다.

**✅ Fix / 해결법:**  
Always run `.duplicated(subset=[...])` with the ID column excluded when checking for accidental re-entry — that's the check that actually finds "this looks like the same thing twice."  
실수로 재입력된 데이터를 확인할 때는 항상 ID 열을 제외한 `.duplicated(subset=[...])`을 실행하세요 — 이것이 "이거 두 번 들어간 것 같은데"를 실제로 찾아내는 확인법입니다.

### Mistake 2 — Forgetting `random_state` on `.sample()`
**English:** `.sample(5)` without `random_state` returns a *different* 5 rows every time the cell runs — so a chart, a screenshot, or a code-review comment referencing "row 3 in my sample" becomes impossible for anyone else (including future you) to reproduce.  
**한글:** `random_state` 없이 `.sample(5)`를 실행하면 셀을 실행할 때마다 *다른* 5개 행이 나옵니다 — 그래서 "제 샘플의 3번째 행"을 언급하는 차트, 스크린샷, 코드 리뷰 코멘트는 다른 사람(미래의 나 포함)이 재현할 수 없게 됩니다.

**✅ Fix / 해결법:**  
Always pass a fixed `random_state=` (any integer, as long as it's consistent) whenever the sample needs to be reproducible for a report or a discussion.  
보고서나 논의를 위해 샘플이 재현 가능해야 할 때는 항상 고정된 `random_state=`(정수 아무거나, 일관되게만 쓰면 됨)를 전달하세요.

### Mistake 3 — Misreading `.info()`'s "Non-Null Count" as a total
**English:** Seeing `5 non-null` in `.info()`'s output doesn't mean "there are 5 rows" — it means "5 out of however many total rows actually have a value here." You have to compare it against the total row count (shown in the `RangeIndex` line) to know whether anything is missing.  
**한글:** `.info()` 출력에서 `5 non-null`을 보고 "행이 5개구나"라고 생각하면 안 됩니다 — 이는 "전체 행 수 중 5개에 실제로 값이 있다"는 뜻입니다. 무언가 결측인지 알려면 (`RangeIndex` 줄에 나오는) 전체 행 수와 비교해야 합니다.  

**✅ Fix / 해결법:**  
Read the `RangeIndex: N entries` line first, then compare every column's `Non-Null Count` against that `N` — any smaller number means missing values in that column.  
`RangeIndex: N entries` 줄을 먼저 읽고, 각 열의 `Non-Null Count`를 그 `N`과 비교하세요 — 더 작은 숫자가 있으면 그 열에 결측치가 있다는 뜻입니다.

# 💡 Tips
Useful tips or shortcuts / 유용한 팁과 단축법

- Run `.info()` immediately after loading any new DataFrame — it's the fastest way to see row count, dtypes, and missing values all in one glance.  
새 DataFrame을 불러오면 바로 `.info()`를 실행하세요 — 행 개수, dtype, 결측치를 한눈에 보는 가장 빠른 방법입니다.
- `.value_counts(normalize=True)` turns a frequency table into percentages — more useful than raw counts when comparing categories of very different total sizes.  
`.value_counts(normalize=True)`는 빈도표를 퍼센트로 바꿔줍니다 — 전체 크기가 크게 다른 범주를 비교할 땐 원래 개수보다 더 유용합니다.
- Fix `random_state` on `.sample()` any time the output needs to be reproducible for a report, a code review, or a screenshot.  
보고서, 코드 리뷰, 스크린샷을 위해 출력이 재현 가능해야 할 때는 항상 `.sample()`에 `random_state`를 고정하세요.
- `.duplicated(subset=[...])` is almost always more useful than the bare `.duplicated()` — exclude any unique ID/key column first, or the check will never find anything.  
맨 `.duplicated()`보다는 거의 항상 `.duplicated(subset=[...])`이 더 유용합니다 — 고유한 ID/키 열을 먼저 제외하지 않으면 이 확인은 아무것도 찾아내지 못합니다.

# 💼 Business Example
*How would a Business Analyst use this?*

**Scenario / 시나리오**

**English:** You've just received this month's ad campaign export. Before building a single chart for the marketing team, run a full EDA pass to make sure the data can be trusted.

**한글:** 방금 이번 달 광고 캠페인 내보내기 파일을 받았습니다. 마케팅팀을 위한 차트를 하나라도 만들기 전에, 데이터를 믿어도 되는지 확인하기 위해 전체 EDA를 실행하세요.

**To Do / 할 일**
- [x] Preview the data and confirm the shape looks right  
데이터를 미리 보고 크기가 맞는지 확인하기
- [x] Check dtypes and missing values with `info()`  
`info()`로 dtype과 결측치 확인하기
- [x] Look at the distribution of cost and clicks with `describe()`  
`describe()`로 cost와 clicks의 분포 확인하기
- [x] Check how many channels there are and how campaigns are distributed across them  
채널이 몇 개인지, 캠페인이 채널별로 어떻게 분포하는지 확인하기
- [x] Flag any duplicate campaigns before reporting a total  
총 캠페인 수를 보고하기 전에 중복 캠페인이 있는지 확인하기

````python
import pandas as pd
import numpy as np

campaigns = pd.DataFrame({
    "campaign_id": ["C001", "C002", "C003", "C004", "C005", "C006", "C007"],
    "channel": ["Instagram", "YouTube", "Instagram", "Google", "YouTube", "Instagram", "Instagram"],
    "clicks": [320, 580, 210, np.nan, 450, 320, 320],
    "cost": [50000, 120000, 35000, 80000, 95000, 50000, 50000],
    "converted": [12, 28, 8, 15, 20, 12, 12]
})

# 1) Preview / 미리보기
print(campaigns.head())
print()

# 2) Structure + missing values / 구조 + 결측치
campaigns.info()
print()

# 3) Numeric distribution / 숫자 분포
print(campaigns.describe())
print()

# 4) Channel distribution / 채널별 분포
print(campaigns["channel"].value_counts(normalize=True).round(2))
print()

# 5) Duplicate check before trusting a "total campaigns" number
# 총 캠페인 수를 신뢰하기 전에 중복 확인
dup = campaigns[campaigns.duplicated(subset=["channel", "clicks", "cost", "converted"], keep=False)]
print(f"Possible duplicate campaigns found: {len(dup)}")
print(dup)
````

# 📝 Summary
*Write today's concept in 3~5 sentences.*

**English**
EDA is the routine you run on any new DataFrame before trusting a single number from it. `.head()` / `.tail()` / `.sample(random_state=...)` give a reproducible first look; `.info()` reveals shape, dtypes, and missing values in one call; `.describe()` (and `.describe(include="all")`) summarizes numeric and categorical distributions; `.value_counts()` breaks a categorical column down into frequencies (or percentages with `normalize=True`); and `.duplicated(subset=[...])` — checked with the ID column excluded — catches rows that were accidentally entered twice. Together these checks form a fixed, few-minutes routine worth repeating on every dataset you receive.

**한글**
EDA는 어떤 숫자든 믿기 전에 새로운 DataFrame에 실행하는 루틴입니다. `.head()` / `.tail()` / `.sample(random_state=...)`는 재현 가능한 첫인상을 주고, `.info()`는 한 번의 호출로 크기, dtype, 결측치를 드러내며, `.describe()`(그리고 `.describe(include="all")`)는 숫자형과 범주형 분포를 요약합니다. `.value_counts()`는 범주형 열을 빈도로(또는 `normalize=True`로 비율로) 분해하고, ID 열을 제외하고 확인한 `.duplicated(subset=[...])`는 실수로 두 번 입력된 행을 잡아냅니다. 이 확인들을 합치면 새로운 데이터셋을 받을 때마다 반복할 가치가 있는, 몇 분이면 끝나는 고정 루틴이 됩니다.

# ❓ Review Questions

**Q1.** Why should you fix `random_state` when using `.sample()` for a report or a code review?
**Q1.** 보고서나 코드 리뷰를 위해 `.sample()`을 쓸 때 왜 `random_state`를 고정해야 하나요?

Fixing random_state makes .sample() return the same random rows every time.  
random_state를 고정하면 .sample()을 실행할 때마다 같은 랜덤 행을 선택합니다. 시드를 고정하는 것과 비슷한 개념입니다.

**Q2.** In `.info()`'s output, how do you tell that a column has missing values without running `.isna()` separately?
**Q2.** `.info()`의 출력에서, `.isna()`를 따로 실행하지 않고도 어떤 열에 결측치가 있는지 어떻게 알 수 있나요?

Compare Non-Null Count with the total number of rows. If they are different, the column has missing values.  
Non-Null Count와 전체 행 개수를 비교하면 됩니다. 둘이 다르면 해당 열에 결측치가 있는 것입니다.

**Q3.** What does `.value_counts(normalize=True)` return instead of raw counts, and when is that more useful?
**Q3.** `.value_counts(normalize=True)`는 원래 개수 대신 무엇을 반환하며, 언제 더 유용한가요?

normalize=True returns proportions instead of raw counts.  
normalize=True는 실제 개수 대신 각 값이 전체에서 차지하는 비율을 반환합니다.

**Q4.** Why does `campaigns.duplicated().sum()` often return `0` even when two rows describe the exact same real-world event, and how do you fix that check?
**Q4.** `campaigns.duplicated().sum()`이 두 행이 실제로 같은 사건을 나타내는데도 왜 자주 `0`을 반환하며, 이 확인을 어떻게 고쳐야 하나요?

duplicated() checks all columns by default, so different IDs can make otherwise identical rows look different. Use subset to check only the columns that define a real duplicate.  
duplicated()는 기본적으로 모든 열을 비교하기 때문에 ID가 다르면 실제로 같은 사건이어도 중복으로 인식하지 않습니다. subset으로 실제 사건을 구분하는 열만 지정해야 합니다.

**Q5.** How does `.nunique()` help you tell an ID column apart from a category column?
**Q5.** `.nunique()`는 ID 열과 카테고리 열을 구분하는 데 어떻게 도움이 되나요?

An ID column usually has a high nunique() value, often close to the number of rows, while a category column usually has relatively few unique values.  
ID 열은 보통 nunique() 값이 전체 행 개수에 가깝고, 카테고리 열은 고유값의 종류가 적습니다.
