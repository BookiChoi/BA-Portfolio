# 🧠 Concept

## What is it?
*(Explain it in your own words.)*

**English**
This section is about **reshaping** data — not changing any value, just rearranging rows and columns into whatever layout the next step needs. `pivot_table()` turns long data into a wide summary grid (rows × columns × aggregated value), `melt()` does the opposite (wide → long), and `stack()` / `unstack()` move labels between the columns and the index. `explode()` turns a single list-valued cell into several rows.

**한글**
이 섹션은 데이터 **재구조화**에 관한 것입니다 — 값을 바꾸는 게 아니라, 다음 단계가 필요로 하는 형태로 행과 열을 재배치하는 것입니다. `pivot_table()`은 long 데이터를 wide 요약 그리드(행 × 열 × 집계값)로 바꾸고, `melt()`는 그 반대(wide → long)를 하며, `stack()` / `unstack()`은 열과 인덱스 사이에서 라벨을 이동시킵니다. `explode()`는 리스트 값이 든 셀 하나를 여러 행으로 바꿉니다.

## Why do we use it?
*(When is it useful?)*

**English**
The same data supports two different jobs, and each job wants a different shape: a report or a chart usually wants **wide** (one row per category, one column per period), while filtering, grouping, and most calculations want **long** (one row per observation). Reshaping tools convert between the two without touching a single value.

**한글**
같은 데이터가 두 가지 다른 작업을 지원하는데, 각 작업이 원하는 형태가 다릅니다: 보고서나 차트는 보통 **wide**(카테고리당 한 행, 기간당 한 열)를 원하고, 필터링·그룹화·대부분의 계산은 **long**(관측치당 한 행)을 원합니다. 재구조화 도구는 값 하나 건드리지 않고 이 둘을 서로 변환합니다.

## When is it used in Business Analytics?
*(Real-world use case)*

**English**
Every "cross-tab" a stakeholder asks for by name — "products down the side, months across the top" — is a `pivot_table()` call. Every Excel export with "one column per month" needs `melt()` before it can be grouped or charted properly. This is the layer between raw analysis and a presentation-ready table.

**한글**
이해관계자가 이름으로 요청하는 모든 "교차표" — "제품은 세로로, 월은 가로로" — 는 `pivot_table()` 호출입니다. "월별 열 하나씩"인 모든 엑셀 내보내기는 제대로 그룹화하거나 차트로 그리기 전에 `melt()`가 필요합니다. 이는 원시 분석과 발표 가능한 테이블 사이의 계층입니다.

### Quick Comparison: Excel vs pandas / 빠른 비교

| Concept / 개념 | Excel | pandas |
|---|---|---|
| PivotTable / 피벗 테이블 | drag fields into Rows/Columns/Values / 필드를 행/열/값으로 드래그 | `pd.pivot_table(df, values=, index=, columns=, aggfunc=)` |
| Frequency cross-tab / 빈도 교차표 | PivotTable with "Count" / "Count"를 쓰는 피벗 테이블 | `pd.crosstab(a, b)` |

| Split a comma-list cell into rows / 쉼표로 구분된 셀을 행으로 분리 | Power Query "Split into Rows" | `df.explode("col")` |

# 📝 Syntax

## Basic Syntax

````python
import pandas as pd

sales = pd.DataFrame({
    "region": ["Seoul", "Seoul", "Busan", "Busan"],
    "category": ["Elec", "Cloth", "Elec", "Cloth"],
    "sales": [320000, 85000, 43000, 95000],
})

# pivot_table -- values/index/columns/aggfunc, like an Excel PivotTable
# pivot_table -- Excel PivotTable처럼 values/index/columns/aggfunc
print(pd.pivot_table(sales, values="sales", index="region", columns="category", aggfunc="sum"))
````

## Common Variations

````python
import pandas as pd

wide = pd.DataFrame({
    "product": ["Laptop", "Mouse"],
    "Jan": [320000, 85000],
    "Feb": [150000, 95000],
})

# melt -- wide (one column per month) -> long (one row per product-month) / melt -- wide -> long
long = pd.melt(wide, id_vars=["product"], var_name="month", value_name="sales")
print(long)
print()

# unstack -- move the innermost index level back into columns / unstack -- 가장 안쪽 인덱스 레벨을 다시 열로
print(long.groupby(["product", "month"])["sales"].sum().unstack())
````

# 🧪 Small Examples

## Example 1 — pivot_table: Excel-Style Pivoting
*(Covers source section 9-1)*

**English:** `pd.pivot_table(df, values=, index=, columns=, aggfunc=, fill_value=)` builds a summary grid exactly like an Excel PivotTable — `values` is what gets aggregated, `index` becomes the rows, `columns` becomes the columns, and `fill_value` replaces missing combinations (rather than leaving `NaN`). `margins=True` adds row/column subtotals, and `aggfunc` can be a list for several statistics at once.  
**한글:** `pd.pivot_table(df, values=, index=, columns=, aggfunc=, fill_value=)`는 Excel PivotTable과 똑같은 요약 그리드를 만듭니다 — `values`는 집계할 대상, `index`는 행이 되고, `columns`는 열이 되며, `fill_value`는 없는 조합을 `NaN`으로 남기는 대신 채웁니다. `margins=True`는 행/열 소계를 추가하고, `aggfunc`에 리스트를 주면 여러 통계를 한 번에 낼 수 있습니다.

````python
import pandas as pd

sales = pd.DataFrame({
    "region": ["Seoul", "Seoul", "Seoul", "Busan", "Busan", "Incheon"],
    "category": ["Elec", "Cloth", "Elec", "Cloth", "Elec", "Elec"],
    "sales": [320000, 85000, 150000, 95000, 43000, 67000],
})

# Basic pivot_table -- region as rows, category as columns / 기본 pivot_table -- 지역은 행, 카테고리는 열
print("basic pivot_table:")
print(pd.pivot_table(sales, values="sales", index="region", columns="category", aggfunc="sum", fill_value=0))
print()

# margins=True -- add row/column subtotals / margins=True -- 행/열 소계 추가
print("with margins=True:")
print(pd.pivot_table(sales, values="sales", index="region", columns="category", aggfunc="sum", fill_value=0, margins=True, margins_name="Total"))
print()

# Multiple aggfunc at once / 여러 집계 함수를 한 번에
print("multiple aggfunc:")
print(pd.pivot_table(sales, values="sales", index="region", aggfunc=["sum", "mean", "count"]))
````

## Example 2 — crosstab: Frequency Cross-Tabs
*(Covers source section 9-2)*

**English:** `pd.crosstab(a, b)` counts how often each combination of two categorical columns occurs — a lighter-weight cousin of `pivot_table()` built specifically for frequencies (no `values=` or `aggfunc=` needed). `normalize="index"` turns each row's counts into proportions that sum to 1, useful for comparing distribution shape across groups of very different sizes.  
**한글:** `pd.crosstab(a, b)`는 두 범주형 열의 각 조합이 얼마나 자주 나타나는지 셉니다 — 빈도 전용으로 만들어진 `pivot_table()`의 더 가벼운 사촌입니다(`values=`나 `aggfunc=`가 필요 없음). `normalize="index"`는 각 행의 개수를 합이 1이 되는 비율로 바꿔주는데, 크기가 크게 다른 그룹 간 분포 형태를 비교할 때 유용합니다.

````python
import pandas as pd

survey = pd.DataFrame({
    "region": ["Seoul", "Seoul", "Busan", "Seoul", "Busan", "Incheon", "Seoul", "Busan"],
    "tier": ["Gold", "Silver", "Bronze", "Gold", "Gold", "Silver", "Bronze", "Silver"],
})

print("crosstab -- raw counts:")
print(pd.crosstab(survey["region"], survey["tier"]))
print()

print("crosstab with normalize='index' -- each row sums to 1.0:")
print(pd.crosstab(survey["region"], survey["tier"], normalize="index").round(2))
````

## Example 3 — melt: Wide to Long
*(Covers source section 9-3)*

**English:** `pd.melt(df, id_vars=[...], var_name=..., value_name=...)` flattens several wide columns (like `Jan`, `Feb`, `Mar`) into two new columns: one holding the old column *names* (`var_name`), one holding the *values* (`value_name`). `id_vars` lists the columns to keep as-is. This is the standard prep step before grouping by whatever those wide columns represented.  
**한글:** `pd.melt(df, id_vars=[...], var_name=..., value_name=...)`는 (`Jan`, `Feb`, `Mar` 같은) 여러 wide 열을 두 개의 새 열로 평탄화합니다: 하나는 기존 열 *이름*을 담고(`var_name`), 하나는 그 *값*을 담습니다(`value_name`). `id_vars`는 그대로 유지할 열을 나열합니다. 이는 그 wide 열들이 나타내던 것으로 그룹화하기 전의 표준 준비 단계입니다.

````python
import pandas as pd

wide = pd.DataFrame({
    "product": ["Laptop", "Mouse", "Keyboard"],
    "Jan": [320000, 85000, 45000],
    "Feb": [150000, 95000, 38000],
    "Mar": [210000, 67000, 55000],
})
print("Wide (before):")
print(wide)
print()

long = pd.melt(
    wide,
    id_vars=["product"],
    value_vars=["Jan", "Feb", "Mar"],
    var_name="month",
    value_name="sales"
)
print("Long (after melt):")
print(long)
````

## Example 4 — pivot: Long to Wide
*(Covers source section 9-4)*

**English:** `df.pivot(index=, columns=, values=)` is `melt()`'s exact reverse. Unlike `pivot_table()`, plain `pivot()` has **no `aggfunc`** — if the same `(index, columns)` combination appears more than once, there's nothing telling it how to combine them, so it raises a `ValueError` instead of guessing.  
**한글:** `df.pivot(index=, columns=, values=)`는 `melt()`의 정확한 반대입니다. `pivot_table()`과 달리, 일반 `pivot()`에는 **`aggfunc`가 없습니다** — 같은 `(index, columns)` 조합이 두 번 이상 나타나면, 이를 어떻게 합쳐야 할지 알려주는 게 없으므로 추측하는 대신 `ValueError`를 냅니다.

````python
import pandas as pd

long = pd.DataFrame({
    "product": ["Laptop", "Mouse", "Keyboard", "Laptop", "Mouse", "Keyboard"],
    "month": ["Jan", "Jan", "Jan", "Feb", "Feb", "Feb"],
    "sales": [320000, 85000, 45000, 150000, 95000, 38000],
})

wide_again = long.pivot(index="product", columns="month", values="sales")
wide_again.columns.name = None
print("pivot() -- back to wide, no aggregation needed since each combo is unique:")
print(wide_again.reset_index())
print()

# What happens when a combination repeats / 조합이 반복되면 어떻게 되는지
duplicated_long = pd.DataFrame({
    "product": ["Laptop", "Laptop"],
    "month": ["Jan", "Jan"],   # same (product, month) combo twice / 같은 (product, month) 조합이 두 번
    "sales": [320000, 340000],
})
try:
    duplicated_long.pivot(index="product", columns="month", values="sales")
except ValueError as e:
    print("pivot() raises on duplicates:", e)
    print("-> pivot_table() would handle this fine, aggregating with its default (mean)")
    print("-> pivot_table()이라면 기본값(mean)으로 집계해서 문제없이 처리했을 것")
````

## Example 5 — stack / unstack
*(Covers source section 9-5)*

**English:** `.stack()` moves column labels **down into the index** (wide → a MultiIndex Series). `.unstack()` does the reverse, moving the innermost index level **up into columns**. The most useful real-world case: calling `.unstack()` directly on a `groupby([...])` result with a MultiIndex — an instant cross-tab, no `pivot_table()` needed.  
**한글:** `.stack()`은 열 라벨을 **인덱스 안으로** 옮깁니다(wide → MultiIndex Series). `.unstack()`은 그 반대로, 가장 안쪽 인덱스 레벨을 **열로** 옮깁니다. 실무에서 가장 유용한 경우: MultiIndex를 가진 `groupby([...])` 결과에 바로 `.unstack()`을 호출하는 것 — `pivot_table()` 없이도 즉석 크로스탭이 됩니다.

````python
import pandas as pd

sales_wide = pd.DataFrame({"Cloth": [95000, 85000], "Elec": [43000, 470000]}, index=["Busan", "Seoul"])
sales_wide.index.name = "region"
sales_wide.columns.name = "category"
print("Wide:")
print(sales_wide)
print()

stacked = sales_wide.stack()
print("stack() -- category moves into the index:")
print(stacked)
print()

print("unstack() -- back to wide:")
print(stacked.unstack())
print()

# The most useful real case -- unstack() straight after a MultiIndex groupby
# 가장 유용한 실무 사례 -- MultiIndex groupby 직후 바로 unstack()
orders = pd.DataFrame({
    "region": ["Seoul", "Seoul", "Seoul", "Busan", "Busan", "Incheon"],
    "category": ["Elec", "Cloth", "Elec", "Cloth", "Elec", "Elec"],
    "sales": [320000, 85000, 150000, 95000, 43000, 67000],
})
print("groupby(...).unstack() -- instant cross-tab, no pivot_table() needed:")
print(orders.groupby(["region", "category"])["sales"].sum().unstack(fill_value=0))
````

## Example 6 — explode: List Columns to Rows
*(Covers source section 9-6)*

**English:** When a single cell holds a list (tags, items in an order), `.explode("col")` turns each list element into its own row, duplicating every other column's value across the new rows. Useful right after loading JSON where one field is naturally an array.  
**한글:** 한 셀에 리스트(태그, 주문 내 상품 등)가 들어있을 때, `.explode("col")`은 각 리스트 항목을 자기만의 행으로 만들고, 다른 모든 열의 값을 새 행들에 복제합니다. 한 필드가 자연스럽게 배열인 JSON을 불러온 직후에 유용합니다.

````python
import pandas as pd

products = pd.DataFrame({
    "product": ["Laptop", "Mouse", "Keyboard"],
    "tags": [["Electronics", "Premium", "Work"], ["Electronics", "Budget"], ["Electronics", "Mechanical"]],
    "price": [1200000, 25000, 45000],
})
print("before explode -- one row per product, tags is a list:")
print(products)
print()

print("after explode('tags') -- one row per tag, other columns duplicated:")
print(products.explode("tags").reset_index(drop=True))
````

## Example 7 — Common Combo Patterns
*(Covers source section 9-7)*

**English:** Pattern A chains `groupby([...])` + `.unstack()` directly into a one-line cross-tab, skipping `pivot_table()` entirely when the data is already mid-pipeline. Pattern B uses `melt()` to turn a wide monthly report into an analyzable long table, then runs a normal `groupby()` + `idxmax()` on it — the exact reason wide data gets melted in the first place.  
**한글:** 패턴 A는 `groupby([...])` + `.unstack()`을 곧바로 체이닝해서 한 줄짜리 크로스탭을 만듭니다 — 데이터가 이미 파이프라인 중간에 있을 때 `pivot_table()`을 완전히 건너뜁니다. 패턴 B는 `melt()`로 wide 월간 보고서를 분석 가능한 long 테이블로 바꾼 뒤, 그 위에서 일반적인 `groupby()` + `idxmax()`를 실행합니다 — wide 데이터를 애초에 melt하는 바로 그 이유입니다.

````python
import pandas as pd

# Pattern A: groupby + unstack -- an instant cross-tab / 패턴 A: groupby + unstack -- 즉석 크로스탭
orders = pd.DataFrame({
    "region": ["Seoul", "Seoul", "Seoul", "Busan", "Busan", "Incheon"],
    "category": ["Elec", "Cloth", "Elec", "Cloth", "Elec", "Elec"],
    "sales": [320000, 85000, 150000, 95000, 43000, 67000],
})
print("Pattern A:")
print(orders.groupby(["region", "category"])["sales"].sum().unstack(fill_value=0))
print()

# Pattern B: melt + groupby -- wide report becomes analyzable / 패턴 B: melt + groupby -- wide 보고서가 분석 가능해짐
wide = pd.DataFrame({
    "product": ["Laptop", "Mouse", "Keyboard"],
    "Jan": [320000, 85000, 45000],
    "Feb": [150000, 95000, 38000],
    "Mar": [210000, 67000, 55000],
})
long = pd.melt(wide, id_vars=["product"], var_name="month", value_name="sales")
best_month = long.groupby("product")["sales"].idxmax()
print("Pattern B -- each product's best month, found only AFTER melting:")
print(long.loc[best_month][["product", "month", "sales"]])
````

# ⚠️ Common Mistakes

### Mistake 1 — Using `pivot()` when duplicate combinations are possible
**English:** `df.pivot(index=, columns=, values=)` has no way to combine two rows that share the same `(index, columns)` pair — it raises `ValueError: Index contains duplicate entries, cannot reshape` instead of guessing which one to keep.  
**한글:** `df.pivot(index=, columns=, values=)`는 같은 `(index, columns)` 쌍을 공유하는 두 행을 합칠 방법이 없습니다 — 어느 것을 유지할지 추측하는 대신 `ValueError: Index contains duplicate entries, cannot reshape`를 냅니다.

**✅ Fix / 해결법:**  
Whenever you're not 100% sure the `(index, columns)` combination is unique, use `pivot_table()` instead — it aggregates duplicates automatically (with `aggfunc`, default `"mean"`).  
`(index, columns)` 조합이 고유하다고 100% 확신할 수 없다면 `pivot()` 대신 `pivot_table()`을 사용하세요 — `aggfunc`(기본값 `"mean"`)로 중복을 자동으로 집계합니다.

### Mistake 2 — Forgetting `fill_value=0` on `pivot_table()`
**English:** A region/category combination with no data becomes `NaN` by default — which can silently break a later `.sum()` (NaN propagates) or make an empty cell look like "unknown" instead of "genuinely zero."  
**한글:** 데이터가 없는 지역/카테고리 조합은 기본적으로 `NaN`이 됩니다 — 이는 이후의 `.sum()`을 조용히 망가뜨리거나(NaN이 전파됨) 빈 셀이 "매출 0건"이 아니라 "알 수 없음"처럼 보이게 만들 수 있습니다.

**✅ Fix / 해결법:**  
Set `fill_value=0` whenever "no data" genuinely means zero (like sales or counts) — leave it as `NaN` only when "unknown" is actually the correct interpretation.  
"데이터 없음"이 실제로 0을 의미할 때는(매출이나 개수처럼) 항상 `fill_value=0`을 지정하세요 — "알 수 없음"이 실제로 맞는 해석일 때만 `NaN`으로 남겨두세요.

### Mistake 3 — Calling `unstack()` on data without a MultiIndex
**English:** `unstack()` needs at least **two index levels** to have something to move into columns. Calling it on a Series with a single, flat index either does nothing meaningful or raises an error — it's specifically the tool for MultiIndex results (like a multi-key `groupby()`).  
**한글:** `unstack()`은 열로 옮길 대상이 있으려면 최소 **두 개의 인덱스 레벨**이 필요합니다. 단일 flat 인덱스를 가진 Series에 호출하면 의미 있는 일이 일어나지 않거나 오류가 발생합니다 — 이는 특별히 (다중 키 `groupby()` 같은) MultiIndex 결과를 위한 도구입니다.

**✅ Fix / 해결법:**  
Only reach for `unstack()` right after a multi-key `groupby()` or another operation that produces a MultiIndex — check `.index` first if you're not sure.  
`unstack()`은 다중 키 `groupby()`나 MultiIndex를 만드는 다른 연산 직후에만 사용하세요 — 확신이 없다면 먼저 `.index`를 확인하세요.

# 💡 Tips
Useful tips or shortcuts / 유용한 팁과 단축법

- Default to `pivot_table()` over `pivot()` unless you're certain the `(index, columns)` combination is unique — it handles duplicates by aggregating instead of crashing.  
`(index, columns)` 조합이 고유하다고 확신하지 않는 한 `pivot()`보다 `pivot_table()`을 기본으로 사용하세요 — 오류 대신 집계로 중복을 처리합니다.
- Always set `fill_value=0` (or another sensible default) on `pivot_table()` when a missing combination should read as "zero," not "unknown."  
없는 조합이 "알 수 없음"이 아니라 "0"으로 읽혀야 한다면 `pivot_table()`에 항상 `fill_value=0`(또는 다른 적절한 기본값)을 지정하세요.
- `melt()` is the standard prep step before `groupby()` on a "month" or "period" column that started out as several separate wide columns.  
`melt()`는 원래 여러 개의 별도 wide 열이었던 "월"이나 "기간" 열로 `groupby()`하기 전의 표준 준비 단계입니다.
- `groupby([...])["col"].agg(...).unstack()` is a one-line alternative to `pivot_table()` when you're already mid-chain in a GroupBy pipeline — no need to start over with a separate `pivot_table()` call.  
GroupBy 파이프라인 중간에 있다면 `groupby([...])["col"].agg(...).unstack()`이 `pivot_table()`의 한 줄짜리 대안입니다 — 별도의 `pivot_table()` 호출로 다시 시작할 필요가 없습니다.

# 💼 Business Example
*How would a Business Analyst use this?*

**Scenario / 시나리오**

**English:** A monthly sales report arrived from Excel in wide format — one column per month. Before you can find each product's best month or build a clean cross-tab, it needs to be melted into a long, analyzable shape.

**한글:** 월간 매출 보고서가 엑셀에서 wide 형식으로 왔습니다 — 월별로 열이 하나씩입니다. 각 제품의 최고 매출 월을 찾거나 깔끔한 크로스탭을 만들기 전에, 먼저 분석 가능한 long 형태로 melt해야 합니다.

**To Do / 할 일**
- [x] Melt the wide report into a long table  
wide 보고서를 long 테이블로 melt하기
- [x] Find each product's single best month  
각 제품의 최고 매출 월 찾기
- [x] Build a pivot_table cross-tab for a quick visual scan  
빠르게 훑어볼 pivot_table 크로스탭 만들기

````python
import pandas as pd

wide_sales = pd.DataFrame({
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Jan": [3600000, 500000, 540000, 1750000],
    "Feb": [3200000, 620000, 480000, 1400000],
    "Mar": [4100000, 580000, 610000, 1900000],
})

# 1) Melt into a long, analyzable format / 분석 가능한 long 형식으로 melt
long_sales = pd.melt(wide_sales, id_vars=["product"], var_name="month", value_name="sales")

# 2) Which product had the best month? / 어느 제품이 최고의 달을 보냈나?
best_month_idx = long_sales.groupby("product")["sales"].idxmax()
print("Best month per product:")
print(long_sales.loc[best_month_idx])
print()

# 3) A pivot_table cross-tab for a quick visual scan / 빠르게 훑어볼 pivot_table 크로스탭
print("Cross-tab view:")
print(pd.pivot_table(long_sales, values="sales", index="product", columns="month", aggfunc="sum", fill_value=0))
````

# 📝 Summary
*Write today's concept in 3~5 sentences.*

**English**
`pivot_table()` is the DataFrame equivalent of an Excel PivotTable — `values` / `index` / `columns` / `aggfunc` / `fill_value` reshape and summarize in one call, with `margins=True` adding row/column subtotals. `crosstab()` is a lighter-weight cousin specifically for frequency counts, and `normalize` turns those counts into proportions. `melt()` flattens several wide columns (like one-per-month) into a long "variable, value" pair — the standard prep step before grouping by that variable — and `pivot()` reverses it back to wide, but only when there's no duplicate `(index, columns)` combination to resolve (`pivot_table()` handles that case by aggregating). `stack()` / `unstack()` move data between column labels and index levels, and `unstack()` applied directly to a MultiIndex GroupBy result is a one-line alternative to `pivot_table()`. `explode()` turns a single row with a list-valued column into one row per list item.

**한글**
`pivot_table()`은 Excel PivotTable에 해당하는 DataFrame 기능입니다 — `values` / `index` / `columns` / `aggfunc` / `fill_value`가 한 번의 호출로 재구조화와 요약을 하고, `margins=True`는 행/열 소계를 추가합니다. `crosstab()`은 빈도 집계 전용의 더 가벼운 사촌이며, `normalize`는 개수를 비율로 바꿔줍니다. `melt()`는 (월별 하나씩 같은) 여러 wide 열을 "변수, 값" 쌍의 long 형태로 평탄화합니다 — 그 변수로 그룹화하기 전의 표준 준비 단계입니다 — 그리고 `pivot()`은 이를 다시 wide로 되돌리는데, `(index, columns)` 조합에 해결할 중복이 없을 때만 가능합니다(`pivot_table()`은 집계로 이 경우를 처리함). `stack()` / `unstack()`은 열 라벨과 인덱스 레벨 사이에서 데이터를 이동시키며, MultiIndex GroupBy 결과에 직접 적용한 `unstack()`은 `pivot_table()`의 한 줄짜리 대안입니다. `explode()`는 리스트 값을 가진 열이 있는 행 하나를 리스트 항목마다 하나의 행으로 바꿉니다.
