# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** Simple linear regression fits a straight line through data to predict a numeric outcome (Y) from a single numeric predictor (X) — it's the natural next step after Section 5's correlation, which only told you HOW STRONGLY two variables move together. Regression adds a formula: Y = intercept + slope × X, letting you predict specific values and quantify exactly how much Y changes per unit of X.

**KR:** 단순선형회귀는 데이터에 직선을 맞춰 하나의 수치형 예측변수(X)로부터 수치형 결과(Y)를 예측합니다 -- 두 변수가 얼마나 강하게 함께 움직이는지만 알려주던 섹션 5의 상관관계 다음의 자연스러운 단계입니다. 회귀는 공식을 더합니다: Y = 절편 + 기울기 × X, 이를 통해 구체적인 값을 예측하고 X 한 단위당 Y가 정확히 얼마나 변하는지 정량화할 수 있습니다.

### Correlation vs. Regression (상관관계 vs 회귀)

| | Correlation (Section 5)<br>상관관계(섹션 5) | Simple Linear Regression (Section 11)<br>단순선형회귀(섹션 11) |
|---|---|---|
| Question / 질문 | How strongly do X and Y move together?<br>X와 Y가 얼마나 강하게 함께 움직이는가? | How much does Y change per unit of X?<br>X 한 단위당 Y가 얼마나 변하는가? |
| Output / 결과 | One number, r (-1 to 1)<br>숫자 하나, r(-1~1) | A formula: Y = intercept + slope × X<br>공식: Y = 절편 + 기울기 × X |
| Direction matters?<br>방향이 중요한가 | No, symmetric<br>아니오, 대칭 | Yes, X predicts Y<br>예, X가 Y를 예측 |
| Library / 라이브러리 | `stats.pearsonr()` | `smf.ols().fit()` (statsmodels) |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** Correlation tells you "ad spend and revenue move together, r=0.85" but not "if we spend $1,000 more, how much more revenue should we expect?" Regression answers exactly that second question, turning a relationship into an actionable prediction formula.

**KR:** 상관관계는 "광고비와 매출이 함께 움직인다, r=0.85"를 알려주지만 "$1,000를 더 쓰면 매출이 얼마나 더 늘 것으로 예상되는가?"는 알려주지 않습니다. 회귀는 정확히 그 두 번째 질문에 답하며, 관계를 실행 가능한 예측 공식으로 바꿔줍니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Forecasting revenue from ad spend, estimating how much a price change affects demand, predicting customer lifetime value from early engagement metrics, and as the entry point into every more advanced predictive model in this curriculum's Part 3.

**KR:** 광고비로부터 매출 예측, 가격 변경이 수요에 미치는 영향 추정, 초기 참여 지표로부터 고객 생애가치 예측, 그리고 이 커리큘럼 Part 3의 모든 더 고급 예측 모델로 들어가는 진입점.

# 📝 Syntax (문법)

## Basic Syntax (기본 문법)

````python
import numpy as np
import pandas as pd
from scipy import stats
import statsmodels.formula.api as smf

np.random.seed(42)
ad_spend = np.random.uniform(1000, 5000, 100)
revenue = ad_spend * 2.8 + np.random.normal(0, 800, 100)

df = pd.DataFrame({'ad_spend': ad_spend, 'revenue': revenue})
model = smf.ols('revenue ~ ad_spend', data=df).fit()   # the intercept is added automatically

intercept, slope = model.params
print(f"Intercept: {intercept:.2f}")
print(f"Slope: {slope:.3f}")
print(f"Equation: revenue = {intercept:.2f} + {slope:.3f} * ad_spend")
print(model.summary())
````

## Common Variations (자주 쓰는 변형)

````python
# R-squared and coefficient p-value
# R-squared와 계수 p-value
print(f"R-squared: {model.rsquared:.3f}")
print(f"Slope p-value: {model.pvalues['ad_spend']:.2e}")

# Predict a new value
# 새로운 값 예측
new_ad_spend = 3000
predicted_revenue = intercept + slope * new_ad_spend
print(f"Predicted revenue at ${new_ad_spend} ad spend: ${predicted_revenue:.2f}")
````

# 🧪 Small Examples (예제)

## Example 1 — Fitting a Regression Line: Ad Spend Predicting Revenue (광고비로 매출 예측하기)

````python
# Business question: fit a line predicting revenue from ad spend, and confirm
# it connects to Section 5's correlation coefficient.
# 비즈니스 질문: 광고비로부터 매출을 예측하는 직선을 맞추고, 이것이
# 섹션 5의 상관계수와 어떻게 연결되는지 확인.

np.random.seed(1)
ad_spend = np.random.uniform(500, 4000, 120)
revenue = ad_spend * 3.1 + 200 + np.random.normal(0, 700, 120)

df = pd.DataFrame({'ad_spend': ad_spend, 'revenue': revenue})
model = smf.ols('revenue ~ ad_spend', data=df).fit()
intercept, slope = model.params

r, _ = stats.pearsonr(ad_spend, revenue)

print(f"Regression equation: revenue = {intercept:.1f} + {slope:.3f} x ad_spend")
print(f"Pearson correlation (Section 5): r = {r:.3f}")
print(f"sqrt(R-squared) = {np.sqrt(model.rsquared):.3f}")
print()
print("For simple linear regression with one predictor, sqrt(R-squared) equals the")
print("absolute value of Pearson's r -- regression and correlation are measuring the")
print("SAME underlying linear relationship, just expressed differently.")
print("예측변수가 하나인 단순선형회귀에서는 sqrt(R-squared)가 Pearson r의 절댓값과")
print("같습니다 -- 회귀와 상관관계는 표현 방식만 다를 뿐 같은 선형 관계를")
print("측정하고 있습니다.")
````

## Example 2 — R-squared: What It Means, and Why It's Not Enough Alone (R-squared의 의미와 그것만으로 부족한 이유)

````python
# Business question: does a HIGH R-squared always mean a GOOD, trustworthy model?
# 비즈니스 질문: 높은 R-squared가 항상 좋고 신뢰할 만한 모델을 의미하는가?

np.random.seed(2)
# Model 1: a sensible relationship (ad spend -> revenue), realistic noisy data
ad_spend2 = np.random.uniform(500, 4000, 100)
revenue2 = ad_spend2 * 2.5 + np.random.normal(0, 2500, 100)
df1 = pd.DataFrame({'ad_spend2': ad_spend2, 'revenue2': revenue2})
model1 = smf.ols('revenue2 ~ ad_spend2', data=df1).fit()

# Model 2: only 4 data points -- artificially high R-squared, but nearly meaningless
tiny_x = np.array([10, 20, 30, 40])
tiny_y = np.array([15, 33, 44, 61]) + np.random.normal(0, 2, 4)
df2 = pd.DataFrame({'tiny_x': tiny_x, 'tiny_y': tiny_y})
model2 = smf.ols('tiny_y ~ tiny_x', data=df2).fit()

print(f"Model 1 (ad spend -> revenue, n=100): R-squared = {model1.rsquared:.3f}")
print(f"Model 2 (n=4 points only): R-squared = {model2.rsquared:.3f}")
print()
print("Model 2's R-squared looks more impressive, but with only 4 points, a straight")
print("line will ALWAYS fit closely almost by construction -- the high R-squared here")
print("reflects a tiny, unreliable sample, not a genuinely stronger relationship.")
print("Model 2의 R-squared가 더 인상적으로 보이지만, 점이 4개뿐이면 직선은 거의")
print("구조적으로 항상 잘 맞습니다 -- 여기서 높은 R-squared는 진짜 더 강한 관계가")
print("아니라 작고 신뢰할 수 없는 표본을 반영합니다.")
````

## Example 3 — Interpreting Coefficients in Business Terms, with Confidence Intervals (비즈니스 언어와 신뢰구간으로 계수 해석하기)

````python
# Business question: translate the regression coefficient into a plain-language
# business statement, complete with honest uncertainty.
# 비즈니스 질문: 회귀계수를 정직한 불확실성과 함께 평범한 비즈니스
# 언어로 번역하기.

np.random.seed(3)
price = np.random.uniform(20, 80, 90)
units_sold = 500 - price * 4.2 + np.random.normal(0, 30, 90)

df = pd.DataFrame({'price': price, 'units_sold': units_sold})
model = smf.ols('units_sold ~ price', data=df).fit()
intercept, slope = model.params
ci_lower, ci_upper = model.conf_int().loc['price']  # confidence interval for the slope
slope_p = model.pvalues['price']

print(f"Slope: {slope:.2f} units per $1 of price")
print(f"95% CI for slope: [{ci_lower:.2f}, {ci_upper:.2f}]")
print(f"p-value: {slope_p:.2e}")
print()
print("Business translation: 'For every $1 increase in price, we expect units sold")
print(f"to decrease by about {abs(slope):.1f}, with a 95% confidence interval of")
print(f"[{abs(ci_upper):.1f}, {abs(ci_lower):.1f}] units. This relationship is highly")
print(f"statistically significant (p={slope_p:.1e}).'")
print()
print(f"비즈니스 번역: '가격이 $1 오를 때마다 판매량은 약 {abs(slope):.1f}개 감소할")
print(f"것으로 예상되며, 95% 신뢰구간은 [{abs(ci_upper):.1f}, {abs(ci_lower):.1f}]개입니다.")
print(f"이 관계는 통계적으로 매우 유의미합니다(p={slope_p:.1e}).'")
````

## Example 4 — Residual Analysis: Checking Linearity (선형성 확인: 잔차 분석)

````python
# Business question: is a STRAIGHT line actually the right shape for this
# relationship? Check the residuals for leftover patterns.
# 비즈니스 질문: 이 관계에 직선이 정말로 맞는 형태인가? 잔차에
# 남은 패턴이 있는지 확인.

np.random.seed(4)
# Deliberately non-linear relationship: diminishing returns (square-root shaped)
experience = np.random.uniform(1, 100, 150)
salary = 30 + 8 * np.sqrt(experience) + np.random.normal(0, 2, 150)

df = pd.DataFrame({'experience': experience, 'salary': salary})
model = smf.ols('salary ~ experience', data=df).fit()
residuals = model.resid

print(f"R-squared for the (mis-specified) linear fit: {model.rsquared:.3f}")
print("(a respectably high R-squared -- but does that mean the SHAPE is right?)")
print()

# A well-behaved residual plot has NO pattern -- just random scatter around 0.
# Split X into low/mid/high thirds and compare average residuals in each:
# a genuinely curved relationship shows a systematic swing instead of noise.
df_check = pd.DataFrame({'x': experience, 'resid': residuals})
df_check['range'] = pd.qcut(df_check['x'], 3, labels=['Low X', 'Mid X', 'High X'])
range_means = df_check.groupby('range', observed=True)['resid'].mean()

print("Average residual by X range:")
print(range_means.round(2))
print()
print("If the straight line were the right shape, all three averages would hover near 0.")
print("Instead they swing negative -> positive -> negative -- a systematic pattern the")
print("straight line is missing, revealing the TRUE relationship curves (diminishing")
print("returns) even though R-squared alone looked reassuring.")
print("직선이 맞는 형태라면 세 평균 모두 0 근처여야 합니다. 대신 음수 -> 양수 ->")
print("음수로 체계적으로 바뀌는데, 이는 직선이 놓치고 있는 패턴입니다 -- R-squared만")
print("보면 안심되어 보였지만, 실제 관계는(수확체감) 곡선을 그리고 있습니다.")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Confusing Regression's Direction with Correlation's Symmetry (회귀의 방향성과 상관관계의 대칭성 혼동)
Correlation between X and Y is symmetric (corr(X,Y) = corr(Y,X)), but regression is NOT — "revenue predicted from ad spend" and "ad spend predicted from revenue" are two different lines with different slopes and different business meanings.  
X와 Y의 상관관계는 대칭입니다(corr(X,Y) = corr(Y,X)), 하지만 회귀는 그렇지 않습니다 -- "광고비로 예측한 매출"과 "매출로 예측한 광고비"는 기울기도 비즈니스적 의미도 다른 별개의 직선입니다.

✅ **Fix / 해결:** Be explicit about which variable is being predicted (Y) and which is doing the predicting (X) before fitting.

### Mistake 2 — Treating a High R-squared as Automatic Model Validity (높은 R-squared를 모델 타당성의 자동 증거로 취급)
As Example 2 showed, a tiny sample size can produce an artificially high R-squared through sheer lack of data to disagree with the line — R-squared measures fit to THIS data, not the model's real-world reliability.  
예제 2가 보여줬듯, 작은 표본 크기는 직선에 반박할 데이터가 부족해서 인위적으로 높은 R-squared를 만들 수 있습니다 -- R-squared는 이 데이터에 대한 적합도를 측정할 뿐, 모델의 실제 신뢰성을 측정하지 않습니다.

✅ **Fix / 해결:** Always check sample size and residual patterns alongside R-squared, not R-squared in isolation.

### Mistake 3 — Extrapolating Predictions Beyond the Observed Data Range (관측 범위를 벗어난 예측 외삽)
A regression line fit on ad spend between $500 and $4,000 says nothing reliable about what happens at $50,000 — the relationship could flatten, reverse, or break down entirely outside the range you actually observed.  
$500에서 $4,000 사이의 광고비로 맞춘 회귀선은 $50,000에서 무슨 일이 일어나는지에 대해 신뢰할 만한 정보를 전혀 주지 않습니다 -- 실제로 관측한 범위 밖에서는 관계가 평평해지거나, 역전되거나, 완전히 무너질 수 있습니다.

✅ **Fix / 해결:** Only use the regression line to predict within (or very close to) the range of X values actually observed in the data.

### Mistake 4 — Interpreting a Regression Coefficient as Proof of Causation (회귀계수를 인과관계의 증거로 해석)
Just like Section 5's correlation warning, a regression coefficient from observational data is still just an association — "revenue increases by $2.80 per $1 of ad spend" could partly reflect a confounder (like overall company growth) driving both, not ad spend causing revenue directly.  
섹션 5의 상관관계 경고와 마찬가지로, 관측 데이터에서 나온 회귀계수도 여전히 연관성일 뿐입니다 -- "광고비 $1당 매출이 $2.80 증가한다"는 광고비가 매출을 직접 유발하는 것이 아니라, 교란변수(전반적인 회사 성장 등)가 둘 다를 이끄는 것을 부분적으로 반영할 수 있습니다.

✅ **Fix / 해결:** Only claim causation from a properly randomized experiment (like an A/B test); treat observational regression coefficients as associations.

### Mistake 5 — Ignoring Residual Patterns and Assuming a Straight Line Always Fits (잔차 패턴을 무시하고 항상 직선이 맞다고 가정)
A model can have a respectable R-squared while still systematically over- or under-predicting across different ranges of X — exactly what Example 4's curved residual pattern revealed for a genuinely logarithmic relationship forced into a straight line.  
모델이 그럴듯한 R-squared를 가지면서도 X의 서로 다른 범위에서 체계적으로 과대·과소 예측할 수 있습니다 -- 예제 4에서 실제로는 로그형 관계를 억지로 직선에 맞췄을 때 드러난 곡선형 잔차 패턴이 정확히 그것입니다.

✅ **Fix / 해결:** Always check residuals for leftover patterns before trusting a linear model, especially when R-squared is only moderate.

### Mistake 6 — Stating a Slope Without Its Units (기울기를 단위 없이 진술)
"The slope is 2.8" is meaningless without context — is that 2.8 dollars of revenue per dollar of ad spend, or 2.8 units sold per $1 of price cut? The units carry the actual business meaning.  
"기울기가 2.8이다"는 맥락 없이는 의미가 없습니다 -- 광고비 1달러당 매출 2.8달러인가요, 아니면 가격 1달러 인하당 판매량 2.8개인가요? 단위가 실제 비즈니스적 의미를 담고 있습니다.

✅ **Fix / 해결:** Always state a slope as "[Y unit] per [X unit]," never as a bare number.

# 💡 Tips (팁)
Useful tips or shortcuts / 유용한 팁과 단축법

- `smf.ols('y ~ x', data=df)` adds the intercept automatically, so there is no `add_constant()` to forget. To force the line through the origin (intercept=0), write `'y ~ x - 1'` -- almost never what you want for business data.<br>smf.ols('y ~ x', data=df)는 절편을 자동으로 넣어주므로 잊을 `add_constant()`가 없습니다. 직선을 원점으로 강제하려면(절편=0) 'y ~ x - 1'로 쓰는데, 비즈니스 데이터에서는 거의 항상 원하지 않는 결과입니다.
- `model.params['Intercept']` is the intercept and `model.params['ad_spend']` (the predictor's column name) is the slope -- a formula model labels every coefficient by name, so you never depend on the order.<br>model.params['Intercept']는 절편이고 model.params['ad_spend'](예측변수 컬럼 이름)는 기울기입니다 -- 수식 모델은 모든 계수에 이름을 붙이므로 순서에 의존할 필요가 없습니다.
- `model.summary()` prints a full, readable regression report in one call (coefficients, R-squared, p-values, confidence intervals all at once) — worth running once on any model before pulling individual numbers out.<br>model.summary()는 한 번의 호출로 전체 회귀 보고서(계수, R-squared, p-value, 신뢰구간 전부)를 읽기 좋게 출력합니다 -- 개별 숫자를 뽑아내기 전에 어떤 모델에서든 한 번 돌려볼 가치가 있습니다.
- A slope's confidence interval crossing zero is the regression equivalent of a non-significant p-value — if [lower, upper] includes 0, you can't be confident the relationship exists in either direction.<br>기울기의 신뢰구간이 0을 포함하는 것은 유의하지 않은 p-value의 회귀 버전입니다 -- [하한, 상한]이 0을 포함하면 그 관계가 어느 방향으로든 존재한다고 확신할 수 없습니다.
- This section only covers ONE predictor — real business questions almost always have several relevant variables at once, which is exactly what Section 12's multiple regression adds.<br>이 섹션은 예측변수 하나만 다룹니다 -- 실제 비즈니스 질문은 거의 항상 여러 관련 변수를 동시에 갖는데, 이것이 정확히 섹션 12의 다중회귀가 더하는 것입니다.

# 💼 Business Example (비즈니스 예시)
*How would a Business Analyst use this?*

**Scenario / 시나리오:**

**EN:** Marketing wants to request a $2,000 increase in monthly ad spend and needs to justify the expected revenue impact to finance, including a defensible range (not just a single number) and confirmation the underlying model is trustworthy.

**KR:** 마케팅팀이 월 광고비 $2,000 증액을 요청하고 싶어하며, 재무팀에 예상 매출 영향을 방어 가능한 범위(단일 숫자가 아니라)와 함께 정당화하고, 기저 모델이 신뢰할 만한지 확인해야 합니다.

**To-do / 할 일:**
- [x] Fit the regression model and check R-squared<br>회귀모델을 맞추고 R-squared 확인하기
- [x] Get the slope's confidence interval, not just the point estimate<br>점추정치뿐 아니라 기울기의 신뢰구간 구하기
- [x] Check residuals for any leftover pattern before trusting the model<br>모델을 신뢰하기 전에 잔차에 남은 패턴이 있는지 확인하기
- [x] Translate the $2,000 increase into an expected revenue range for finance<br>$2,000 증액을 재무팀을 위한 예상 매출 범위로 번역하기

````python
np.random.seed(60)
ad_spend = np.random.uniform(2000, 8000, 150)
revenue = ad_spend * 4.1 + 1200 + np.random.normal(0, 1500, 150)

df = pd.DataFrame({'ad_spend': ad_spend, 'revenue': revenue})
model = smf.ols('revenue ~ ad_spend', data=df).fit()
intercept, slope = model.params
ci_lower, ci_upper = model.conf_int().loc['ad_spend']

# Residual check before trusting the model
residuals = model.resid
resid_vs_x_corr, _ = stats.pearsonr(ad_spend, residuals)

print(f"Model: revenue = {intercept:.0f} + {slope:.2f} x ad_spend")
print(f"R-squared: {model.rsquared:.3f}")
print(f"Slope 95% CI: [{ci_lower:.2f}, {ci_upper:.2f}]")
print(f"Residual check (X vs residuals): correlation = {resid_vs_x_corr:.3f} (near 0 is good)")
print()

increase = 2000
predicted_lift = slope * increase
lift_lower = ci_lower * increase
lift_upper = ci_upper * increase

print(f"For a ${increase:,} increase in ad spend:")
print(f"  Point estimate: +${predicted_lift:,.0f} revenue")
print(f"  95% CI: [+${lift_lower:,.0f}, +${lift_upper:,.0f}]")
print()
print("Recommendation to finance: expect a revenue increase in this range, with the")
print("usual caveat that this is an ASSOCIATION from historical data, not a guaranteed")
print("causal effect -- an A/B test on ad spend would confirm causation directly.")
print()
print(f"모델: revenue = {intercept:.0f} + {slope:.2f} x ad_spend, R-squared: {model.rsquared:.3f}")
print(f"${increase:,} 증액 시 예상 매출: +${predicted_lift:,.0f} (95% CI: [+${lift_lower:,.0f}, +${lift_upper:,.0f}])")
print("재무팀 제안: 이 범위의 매출 증가를 기대할 수 있으나, 이는 과거 데이터의")
print("연관성이지 보장된 인과효과가 아닙니다 -- 광고비에 대한 A/B 테스트가")
print("인과관계를 직접 확인해줄 것입니다.")
````

# 📝 Summary (요약)
*Write today's concept in 3~5 sentences.*

**EN:** Simple linear regression fits a straight line (Y = intercept + slope × X) to predict a numeric outcome from one predictor, extending Section 5's correlation with an actual formula — for one predictor, sqrt(R-squared) equals the absolute value of Pearson's r, confirming the two techniques measure the same underlying linear relationship. R-squared quantifies how much of Y's variation the line explains, but a high value alone doesn't validate a model — small samples can inflate it artificially, and it says nothing about whether a straight line was even the right shape to begin with (checked via residual analysis). Every coefficient should be interpreted with its units and confidence interval attached ("Y changes by B units per unit of X, 95% CI [lower, upper]"), and — echoing Section 5's warning one more time — a regression coefficient from observational data proves association, not causation, unless it comes from a genuinely randomized experiment.

**KR:** 단순선형회귀는 직선(Y = 절편 + 기울기 × X)을 맞춰 하나의 예측변수로부터 수치형 결과를 예측하며, 섹션 5의 상관관계에 실제 공식을 더한 확장입니다 -- 예측변수가 하나일 때 sqrt(R-squared)는 Pearson의 r의 절댓값과 같아서, 두 기법이 같은 근본적인 선형 관계를 측정한다는 것을 확인해줍니다. R-squared는 직선이 Y의 변동을 얼마나 설명하는지 정량화하지만, 높은 값만으로는 모델을 검증하지 못합니다 -- 작은 표본은 이를 인위적으로 부풀릴 수 있고, 애초에 직선이 맞는 형태였는지는 전혀 말해주지 않습니다(잔차 분석으로 확인). 모든 계수는 단위와 신뢰구간을 함께 붙여 해석해야 하며("X 한 단위당 Y가 B 단위 변한다, 95% CI [하한, 상한]"), 섹션 5의 경고를 다시 한번 반복하자면, 관측 데이터의 회귀계수는 진짜 무작위 실험에서 나온 것이 아니라면 인과관계가 아니라 연관성을 증명할 뿐입니다.

**Q1 (EN).** For a simple linear regression with one predictor, how does sqrt(R-squared) relate to Section 5's Pearson correlation coefficient?  
**Q1 (KR).** 예측변수가 하나인 단순선형회귀에서, sqrt(R-squared)는 섹션 5의 Pearson 상관계수와 어떤 관계인가요?

sqrt(R-squared) equals the absolute value of Pearson's r. The sign comes from the slope's direction.  
sqrt(R²)는 Pearson r의 절댓값과 같습니다. 부호는 기울기의 방향이 정합니다.

**Q2 (EN).** A model fit on only 4 data points has R-squared=0.99. Why should you be skeptical of that number?  
**Q2 (KR).** 데이터 포인트가 4개뿐인 모델의 R-squared가 0.99입니다. 왜 그 숫자를 의심해야 하나요?

With 4 points, a line can fit well by chance, so R² is unstable and overstated. The estimate carries little evidence.  
점이 4개면 우연히도 선이 잘 맞아서 R²가 불안정하고 부풀려집니다. 근거가 약한 수치입니다.

**Q3 (EN).** A regression fit on ad spend between $1,000 and $5,000 is used to predict revenue at $50,000 of ad spend. What's wrong with this?  
**Q3 (KR).** $1,000에서 $5,000 사이의 광고비로 맞춘 회귀를 $50,000 광고비의 매출을 예측하는 데 사용합니다. 이것의 문제는 무엇인가요?

This is extrapolation. Nothing in the data shows the linear trend still holds far outside $1,000–$5,000.
외삽입니다. 관측 범위($1,000–$5,000)를 크게 벗어나면 선형 관계가 유지되는지 데이터로 알 수 없습니다.

**Q4 (EN).** Why is "the slope is 2.8" an incomplete statement, and what should be added to make it a proper business interpretation?  
**Q4 (KR).** "기울기가 2.8이다"가 왜 불완전한 진술이며, 제대로 된 비즈니스 해석이 되려면 무엇을 추가해야 하나요?

It lacks units and the "per one-unit change in X" meaning, plus a confidence interval. For example: each extra $1 of ad spend is associated with $2.8 more revenue (95% CI …).  
단위와 "X가 1 늘 때"라는 기준, 그리고 신뢰구간이 빠졌습니다. 예: 광고비 $1 증가당 매출 $2.8 증가(95% 신뢰구간 …).

**Q5 (EN).** A model has a decent R-squared, but plotting residuals against X reveals a clear curved pattern. What does this suggest, and what should you consider doing?  
**Q5 (KR).** 모델의 R-squared는 괜찮지만, 잔차를 X에 대해 그려보니 명확한 곡선 패턴이 보입니다. 이것은 무엇을 시사하며, 무엇을 고려해야 하나요?

The relationship is nonlinear, so a straight line is misspecified. Consider adding a squared term, a log transform, or a nonlinear model.  
관계가 비선형이라 직선 모델이 잘못 설정된 것입니다. 제곱항 추가, 로그 변환, 비선형 모델을 고려합니다.
