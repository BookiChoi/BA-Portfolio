# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** A/B testing pitfalls are the ways a technically-correct statistical analysis (Sections 14-16) can still produce a misleading business conclusion — not through calculation errors, but through the DATA ITSELF violating an assumption the math quietly depends on. This section covers three classic traps (novelty effects, network effects, Simpson's paradox) that don't show up as a wrong formula, only as a wrong story.

**KR:** A/B 테스트 함정은 기술적으로는 올바른 통계 분석(섹션 14~16)이 계산 오류 없이도 오해를 부르는 비즈니스 결론으로 이어질 수 있는 방식들입니다 -- 계산 실수가 아니라, 데이터 자체가 수학이 조용히 의존하고 있는 가정을 위반하기 때문입니다. 이 섹션은 잘못된 공식이 아니라 잘못된 이야기로만 드러나는 세 가지 고전적인 함정(노벨티 효과, 네트워크 효과, 심슨의 역설)을 다룹니다.

### The three pitfalls at a glance (세 가지 함정 한눈에 보기)

| Pitfall / 함정 | What goes wrong / 무엇이 잘못되는가 | Detection / 탐지 방법 |
|---|---|---|
| Novelty effect<br>노벨티 효과 | Effect is temporary curiosity, not lasting improvement<br>효과가 지속적인 개선이 아니라 일시적인 호기심 | Compare early vs late window<br>초기 vs 후기 구간 비교 |
| Network effect<br>네트워크 효과 | Treatment "leaks" into control via user interaction<br>사용자 상호작용을 통해 실험군이 대조군으로 "새어나감" | Check SUTVA: does one user's assignment affect another's outcome?<br>SUTVA 확인: 한 사용자의 배정이 다른 사용자의 결과에 영향을 주는가? |
| Simpson's paradox<br>심슨의 역설 | Aggregate hides reversed segment-level trends<br>전체 결과가 뒤집힌 세그먼트별 추세를 숨김 | Break down by segment (Sections 8/15 habit)<br>세그먼트로 나눠보기(섹션 8·15의 습관) |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** A test can have a perfect pre-registered design (Section 14), adequate power (Section 15), and a clean p-value (Section 16) and STILL mislead — if the lift is really just curiosity that fades, if treatment users are secretly influencing control users, or if the overall number hides two segments moving in opposite directions. Knowing these traps by name is what separates "ran the numbers" from "understood what actually happened."

**KR:** 테스트는 완벽한 사전등록 설계(섹션 14), 충분한 검정력(섹션 15), 깨끗한 p-value(섹션 16)를 갖고도 여전히 오해를 부를 수 있습니다 -- 만약 상승폭이 그저 사라질 호기심일 뿐이라면, 실험군 사용자가 몰래 대조군 사용자에게 영향을 주고 있다면, 또는 전체 숫자가 반대 방향으로 움직이는 두 세그먼트를 숨기고 있다면. 이 함정들을 이름으로 알고 있는 것이 "숫자를 돌렸다"와 "실제로 무슨 일이 일어났는지 이해했다"를 가릅니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Deciding whether a test's early strong result should be trusted at face value or watched longer (novelty), evaluating any feature involving referrals/marketplaces/social interactions where users affect each other (network effects), and double-checking any surprising aggregate result by looking inside it (Simpson's paradox) — the final skeptical pass before a launch decision is finalized.

**KR:** 테스트의 초기 강한 결과를 액면 그대로 신뢰할지 더 지켜볼지 결정하는 것(노벨티), 추천·마켓플레이스·소셜 상호작용처럼 사용자가 서로에게 영향을 주는 기능을 평가하는 것(네트워크 효과), 그리고 놀라운 전체 결과가 있다면 안을 들여다보며 다시 확인하는 것(심슨의 역설) -- 출시 결정을 확정하기 전 마지막 회의적 검토입니다.

# 📝 Syntax (문법)

## Basic Syntax (기본 문법)

````python
import numpy as np
import pandas as pd
from scipy import stats

# Detecting a hidden pattern: compare a metric across two time windows
# 숨은 패턴 탐지: 두 시간 구간에 걸쳐 지표 비교
np.random.seed(42)
week1_lift = np.random.normal(0.03, 0.01, 200)  # early lift, strong
week3_lift = np.random.normal(0.01, 0.01, 200)  # later lift, weaker

print(f"Week 1 average lift: {week1_lift.mean():.3%}")
print(f"Week 3 average lift: {week3_lift.mean():.3%}")

t_stat, p_value = stats.ttest_ind(week1_lift, week3_lift)
print(f"Difference significant? p={p_value:.4f}")
````

## Common Variations (자주 쓰는 변형)

````python
# The same "compare two slices" pattern detects Simpson's paradox too --
# just slice by SEGMENT instead of TIME
# 같은 "두 조각 비교" 패턴이 심슨의 역설도 탐지함 -- 시간 대신 세그먼트로 나누기만 하면 됨
segment_a_lift = np.random.normal(0.02, 0.01, 150)
segment_b_lift = np.random.normal(-0.01, 0.01, 150)
print(f"Segment A lift: {segment_a_lift.mean():+.3%}")
print(f"Segment B lift: {segment_b_lift.mean():+.3%}")
print(f"Combined (naive) lift: {np.concatenate([segment_a_lift, segment_b_lift]).mean():+.3%}")
````

# 🧪 Small Examples (예제)

## Example 1 — Novelty Effects: Does the Lift Hold Up Over Time? (상승폭이 시간이 지나도 유지되는가?)

````python
# Business question: a new UI redesign shows a strong +8% lift in week 1.
# Is this a genuine improvement, or are users just reacting to novelty?
# 비즈니스 질문: 새 UI 재설계가 1주차에 강한 +8% 상승을 보입니다. 이것은
# 진짜 개선인가, 아니면 사용자가 그저 새로움에 반응하는 것인가?

np.random.seed(1)
n_per_day = 300

# Simulate a novelty effect: strong early, fading over 3 weeks
true_lasting_lift = 0.02  # the REAL, lasting effect once novelty wears off
novelty_bonus_by_week = {1: 0.06, 2: 0.03, 3: 0.005}  # extra lift from novelty, decaying

weekly_results = []
for week in [1, 2, 3]:
    novelty = novelty_bonus_by_week[week]
    control = np.random.binomial(1, 0.10, n_per_day * 7)
    treatment = np.random.binomial(1, 0.10 + true_lasting_lift + novelty, n_per_day * 7)
    lift = treatment.mean() - control.mean()
    weekly_results.append({'week': week, 'lift': lift})

df_weekly = pd.DataFrame(weekly_results)
print(df_weekly.to_string(index=False))
print()

early_lift = df_weekly.iloc[0]['lift']
late_lift = df_weekly.iloc[-1]['lift']
print(f"Week 1 lift: {early_lift:+.2%} -- Week 3 lift: {late_lift:+.2%}")
if early_lift > late_lift * 1.5:
    print("The lift shrinks substantially over time -- classic novelty effect signature.")
    print("Trusting only the week-1 number would have overstated the LASTING impact.")
    print("상승폭이 시간이 지나며 크게 줄어듭니다 -- 전형적인 노벨티 효과의 특징입니다.")
    print("1주차 숫자만 신뢰했다면 지속적인 영향을 과장했을 것입니다.")
````

## Example 2 — Network Effects: When Treatment Leaks Into Control (실험군이 대조군으로 새어나갈 때)

````python
# Business question: a new "invite a friend" feature is being tested. Treatment
# users send invites, but some invited FRIENDS are in the control group. Does
# this contaminate the comparison?
# 비즈니스 질문: 새로운 "친구 초대" 기능을 테스트하고 있습니다. 실험군
# 사용자가 초대를 보내지만, 초대받은 친구 중 일부는 대조군에 있습니다.
# 이것이 비교를 오염시키는가?

np.random.seed(2)
n_users = 2000
group = np.random.choice(['control', 'treatment'], n_users)

# Baseline engagement, with NO real treatment effect assumed for this example
baseline_engagement = np.random.normal(50, 10, n_users)

# But: treatment users send invites, and some invited friends happen to land
# in the control group, giving THEM a small engagement boost too
contamination_boost = np.where(group == 'control', np.random.poisson(0.5, n_users) * 3, 0)

observed_engagement = baseline_engagement + contamination_boost
df = pd.DataFrame({'group': group, 'engagement': observed_engagement})

naive_lift = df[df.group == 'treatment']['engagement'].mean() - df[df.group == 'control']['engagement'].mean()
print(f"Naive comparison -- treatment vs control engagement difference: {naive_lift:+.2f}")
print()
print("If control users are getting 'contaminated' engagement from invited-friend")
print("treatment users, the TRUE treatment effect is being masked -- control isn't a")
print("clean baseline anymore. This is a SUTVA violation: one unit's treatment")
print("assignment is affecting another unit's outcome, which per-user randomization")
print("alone doesn't fix. The usual remedy is randomizing at a higher level (e.g. by")
print("friend-group or geographic region) so contamination stays within an arm.")
print("대조군 사용자가 초대받은 친구인 실험군으로부터 '오염된' 참여도를 얻고 있다면,")
print("진짜 실험 효과가 가려집니다 -- 대조군이 더 이상 깨끗한 기준선이 아닙니다.")
print("이것이 SUTVA 위반입니다: 한 단위의 처치 배정이 다른 단위의 결과에 영향을")
print("주는 것이며, 사용자 단위 무작위 배정만으로는 해결되지 않습니다. 일반적인")
print("해결책은 오염이 한 그룹 안에 머물도록 더 상위 단위(친구 그룹, 지역 등)로")
print("무작위 배정하는 것입니다.")
````

## Example 3 — Simpson's Paradox: When the Aggregate Reverses the Truth (전체 결과가 진실을 뒤집을 때)

````python
# Business question: the aggregate result showed treatment converting at just
# 14% vs control's 76% -- treatment looks like a disaster. But is it?
# 비즈니스 질문: 전체 결과는 실험군이 14%, 대조군이 76% 전환율을 보여
# 실험군이 재앙처럼 보입니다. 정말 그런가?

data = {
    'mobile':  {'control_n': 50, 'control_conv': 5, 'treatment_n': 950, 'treatment_conv': 100},
    'desktop': {'control_n': 950, 'control_conv': 760, 'treatment_n': 50, 'treatment_conv': 42},
}

print("Segment-level results:")
for segment, d in data.items():
    control_rate = d['control_conv'] / d['control_n']
    treatment_rate = d['treatment_conv'] / d['treatment_n']
    verdict = 'WINS' if treatment_rate > control_rate else 'loses'
    print(f"  {segment}: control={control_rate:.1%} (n={d['control_n']}), "
          f"treatment={treatment_rate:.1%} (n={d['treatment_n']}) -> treatment {verdict}")

total_control_n = sum(d['control_n'] for d in data.values())
total_control_conv = sum(d['control_conv'] for d in data.values())
total_treatment_n = sum(d['treatment_n'] for d in data.values())
total_treatment_conv = sum(d['treatment_conv'] for d in data.values())

agg_control_rate = total_control_conv / total_control_n
agg_treatment_rate = total_treatment_conv / total_treatment_n

print()
print(f"Aggregate: control={agg_control_rate:.1%}, treatment={agg_treatment_rate:.1%}")
agg_verdict = 'WINS' if agg_treatment_rate > agg_control_rate else 'LOSES'
print(f"-> treatment {agg_verdict} in the aggregate")
print()
print("Treatment wins in EVERY segment individually, yet loses in the aggregate --")
print("this is Simpson's paradox. It happens here because 95% of control users are")
print("desktop (naturally high-converting) while 95% of treatment users are mobile")
print("(naturally low-converting) -- a segment MIX imbalance, likely from a broken")
print("randomization, not a real treatment effect. Always check segment-level results")
print("before trusting an aggregate, exactly as Sections 8 and 16 recommended.")
print("실험군은 모든 개별 세그먼트에서 이기지만 전체에서는 집니다 -- 이것이 심슨의")
print("역설입니다. 대조군의 95%가 데스크톱(원래 전환율이 높음)이고 실험군의 95%가")
print("모바일(원래 전환율이 낮음)이기 때문에 발생합니다 -- 실제 처치 효과가 아니라")
print("세그먼트 구성 불균형이며, 아마도 무작위 배정이 잘못된 결과입니다.")
````

## Example 4 — Final Pre-Launch/Post-Launch Checklist (최종 출시 전후 체크리스트)

````python
# Business question: before finalizing ANY launch decision, run one complete
# checklist synthesizing everything from Sections 14-17.
# 비즈니스 질문: 어떤 출시 결정이든 확정하기 전에, 섹션 14~17의 모든 것을
# 종합한 완전한 체크리스트를 실행.

def ab_test_checklist(design, srm_p_value, lift, p_value, mde, early_late_ratio, segment_lifts):
    """Run a complete pre-launch/post-launch A/B test checklist."""
    issues = []

    # Design checks (Section 14)
    if 'primary_metric' not in design or not design['primary_metric']:
        issues.append("No single primary metric designated (Section 14)")
    if 'randomization_unit' not in design:
        issues.append("Randomization unit not specified (Section 14)")

    # SRM check (Sections 8, 14)
    if srm_p_value < 0.01:
        issues.append(f"Sample Ratio Mismatch detected (p={srm_p_value:.4f}) -- investigate randomization")

    # Statistical vs practical significance (Sections 6, 16)
    if p_value >= 0.05:
        issues.append(f"Not statistically significant (p={p_value:.4f})")
    if lift < mde:
        issues.append(f"Lift ({lift:.2%}) below pre-registered MDE ({mde:.2%}) -- not practically significant")

    # Novelty effect check (this section)
    if early_late_ratio > 1.5:
        issues.append(f"Possible novelty effect -- early effect is {early_late_ratio:.1f}x the late effect")

    # Segment consistency (Sections 8, 16, this section's Simpson's paradox warning)
    lifts = list(segment_lifts.values())
    if any(l < 0 for l in lifts) and lift > 0:
        issues.append("Aggregate is positive but at least one segment is negative -- check for Simpson's paradox")

    return issues

# Example run: a mostly-clean test with one flagged issue
test_design = {'primary_metric': 'conversion_rate', 'randomization_unit': 'user_id'}
issues_found = ab_test_checklist(
    design=test_design, srm_p_value=0.42, lift=0.018, p_value=0.02, mde=0.01,
    early_late_ratio=1.8, segment_lifts={'mobile': 0.02, 'desktop': 0.015}
)

if issues_found:
    print("Checklist found issues to investigate before finalizing:")
    for issue in issues_found:
        print(f"  \u26a0\ufe0f  {issue}")
else:
    print("All checks passed -- test result can be trusted as-is.")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Trusting Week-1 Results Without Checking Persistence (지속성 확인 없이 1주차 결과 신뢰)
A strong early lift can be entirely novelty-driven (Example 1) — launching based on week-1 data alone risks celebrating a curiosity effect that will quietly fade to near-zero within a few weeks.   
강한 초기 상승은 완전히 노벨티가 이끄는 것일 수 있습니다(예제 1) -- 1주차 데이터만으로 출시하면 몇 주 안에 조용히 거의 0으로 사라질 호기심 효과를 축하하는 위험을 감수하는 것입니다.

✅ **Fix / 해결:** Compare early-window and late-window effect sizes (Example 1's approach) before trusting an early result as the lasting truth.

### Mistake 2 — Assuming Per-User Randomization Prevents All Contamination (사용자 단위 무작위 배정이 모든 오염을 막는다고 가정)
Even with clean, correct per-user randomization, features involving interaction between users (invites, messaging, marketplaces) can let treatment "leak" into control (Example 2) — the randomization was fine, but the SUTVA assumption behind the analysis wasn't.  
깨끗하고 올바른 사용자 단위 무작위 배정에서도, 사용자 간 상호작용이 관련된 기능(초대, 메시징, 마켓플레이스)은 실험군이 대조군으로 "새어나가게" 할 수 있습니다(예제 2) -- 무작위 배정은 괜찮았지만, 분석 뒤에 있는 SUTVA 가정이 그렇지 않았습니다.

✅ **Fix / 해결:** For any feature where users can affect each other, consider randomizing at a higher level (region, friend-group, market) instead of per-user.

### Mistake 3 — Trusting an Aggregate Without Checking Segment Consistency (세그먼트 일관성 확인 없이 전체 결과 신뢰)
As Example 3's numbers showed directly, treatment can win in EVERY individual segment and still lose in the aggregate, purely because the segment MIX differs between groups — an aggregate number alone can point in the exact opposite direction of the truth.  
예제 3의 숫자가 직접 보여줬듯, 실험군은 모든 개별 세그먼트에서 이기고도 순전히 그룹 간 세그먼트 구성이 다르다는 이유만으로 전체에서는 질 수 있습니다 -- 전체 숫자만으로는 진실과 정확히 반대 방향을 가리킬 수 있습니다.

✅ **Fix / 해결:** Always break results down by key segments (Sections 8 and 16's habit) and check that segment MIX is similar between groups before trusting an aggregate number.

### Mistake 4 — Assuming Simpson's Paradox Is a Rare, Exotic Edge Case (심슨의 역설을 드물고 특이한 극단 사례로 취급)
Simpson's paradox isn't a mathematical curiosity reserved for textbooks — it happens any time group composition differs meaningfully between Control and Treatment, which is common whenever randomization is imperfect or a feature naturally attracts different user types.  
심슨의 역설은 교과서에나 있는 수학적 진기함이 아닙니다 -- 대조군과 실험군 사이에 그룹 구성이 의미 있게 다를 때마다 일어나며, 무작위 배정이 불완전하거나 기능이 자연스럽게 다른 유형의 사용자를 끌어들일 때 흔히 발생합니다.

✅ **Fix / 해결:** Make a segment-mix comparison (are the groups similar in composition, not just size?) a routine part of every test's analysis, not a special investigation reserved for suspicious results.

### Mistake 5 — "Fixing" a Novelty Effect by Just Running the Test Longer, Silently (조용히 테스트만 더 오래 돌려 노벨티 효과를 "해결")
Simply extending a test's duration without separately reporting the early-vs-late breakdown hides the novelty pattern rather than addressing it — leadership sees one final number and never learns the effect was shrinking the whole time.  
초기 대 후기 분석을 별도로 보고하지 않고 그냥 테스트 기간만 늘리면 노벨티 패턴을 해결하는 게 아니라 숨기는 것입니다 -- 경영진은 최종 숫자 하나만 보고 효과가 내내 줄어들고 있었다는 것을 전혀 알지 못합니다.

✅ **Fix / 해결:** Always report the early-vs-late trend explicitly, even when the final aggregate number looks fine, so a fading effect is visible rather than buried in an average.

### Mistake 6 — Skipping the Final Synthesis Checklist ("We Already Checked Everything") (최종 종합 체크리스트 생략: "이미 다 확인했다")
Each individual check (design, power, SRM, significance) can pass on its own while the COMBINATION still has a problem — running one final, consolidated checklist (Example 4) catches issues that fall in the gaps between individually-passed steps.  
각 개별 확인(설계, 검정력, SRM, 유의성)이 각자 통과하더라도 조합 자체에는 여전히 문제가 있을 수 있습니다 -- 하나의 최종 통합 체크리스트(예제 4)를 실행하면 개별적으로는 통과한 단계들 사이의 틈에 있는 문제를 잡아냅니다.

✅ **Fix / 해결:** Always run one final, holistic checklist before finalizing a launch decision, even when every individual step already looked fine.

- A shrinking-but-still-positive novelty effect isn't necessarily bad news — sometimes the "true" lasting lift, even after novelty fades, is still worth shipping; the point is knowing which number you're actually looking at.<br>줄어들지만 여전히 양수인 노벨티 효과가 반드시 나쁜 소식은 아닙니다 -- 때로는 노벨티가 사라진 뒤의 "진짜" 지속 상승도 여전히 출시할 가치가 있습니다. 핵심은 지금 보고 있는 숫자가 어느 것인지 아는 것입니다.
- Marketplaces, social networks, and referral programs are the highest-risk feature TYPES for network effects — if a feature description includes words like "invite," "share," "friend," or "message," budget extra time to think through contamination risk.<br>마켓플레이스, 소셜 네트워크, 추천 프로그램은 네트워크 효과에 가장 위험한 기능 유형입니다 -- 기능 설명에 "초대", "공유", "친구", "메시지" 같은 단어가 있다면 오염 위험을 생각할 여유 시간을 더 두세요.
- Simpson's paradox is easiest to catch by comparing the SIZE of each segment between groups, not just the rate — if Segment X is 10% of Control but 80% of Treatment, that mismatch alone is worth investigating before trusting anything else.<br>심슨의 역설은 각 그룹 간 세그먼트의 비율(단순히 전환율이 아니라 크기)을 비교하는 것으로 가장 쉽게 잡아낼 수 있습니다 -- 세그먼트 X가 대조군의 10%인데 실험군의 80%라면, 다른 무엇을 신뢰하기 전에 그 불일치 자체를 조사할 가치가 있습니다.
- Keep a running, reusable checklist function (like Example 4's) across every test your team runs — the value compounds as it accumulates more of these named pitfalls over time, becoming institutional memory instead of individual vigilance.<br>팀이 실행하는 모든 테스트에 걸쳐 재사용 가능한 체크리스트 함수(예제 4처럼)를 계속 유지하세요 -- 시간이 지나며 이런 이름 붙은 함정들을 더 많이 축적할수록 가치가 누적되어, 개인의 경계심이 아니라 조직의 기억이 됩니다.

# 💼 Business Example (비즈니스 예시)
*How would a Business Analyst use this?*

**Scenario / 시나리오:**

**EN:** A marketplace app's "boost my listing" feature test just finished. Before writing the launch memo, run the complete skeptical review: check for novelty effects, consider network effects given the feature involves visibility competition between sellers, check for Simpson's paradox across seller segments, and run the final checklist.

**KR:** 마켓플레이스 앱의 "내 상품 부스트" 기능 테스트가 방금 끝났습니다. 출시 메모를 작성하기 전, 완전한 회의적 검토를 실행합니다: 노벨티 효과 확인, 판매자 간 노출 경쟁이 관련된 기능이므로 네트워크 효과 고려, 판매자 세그먼트 전반의 심슨의 역설 확인, 그리고 최종 체크리스트 실행.

**To-do / 할 일:**
- [x] Check for a novelty effect by comparing early vs late results<br>초기 vs 후기 결과 비교로 노벨티 효과 확인하기
- [x] Consider network effect risk given the feature's shared-visibility mechanism<br>공유된 노출 메커니즘을 고려해 네트워크 효과 위험 검토하기
- [x] Check segment mix for a potential Simpson's paradox<br>잠재적 심슨의 역설을 위해 세그먼트 구성 확인하기
- [x] Run the final checklist and write a recommendation<br>최종 체크리스트를 실행하고 제안 작성하기

````python
print("=== Final Review: 'Boost My Listing' Feature ===")
print()

# 1) Novelty check
np.random.seed(90)
week1 = np.random.normal(0.04, 0.01, 50)
week4 = np.random.normal(0.025, 0.01, 50)
ratio = week1.mean() / week4.mean()
print(f"1) Novelty check: week 1 lift={week1.mean():+.2%}, week 4 lift={week4.mean():+.2%}")
tag1 = "possible fading novelty" if ratio > 1.3 else "looks stable"
print(f"   Ratio: {ratio:.1f}x -- {tag1}")
print()

# 2) Network effect consideration
print("2) Network effect consideration: boosted listings compete for the SAME limited")
print("   visibility slots -- a treatment seller's boost can push a control seller's")
print("   listing down, meaning control isn't a clean, unaffected baseline. Marketplace")
print("   competition for shared inventory (search slots, ad space) is a classic SUTVA risk.")
print()

# 3) Simpson's paradox check across seller segments
segment_data = {
    'new_sellers': {'control_rate': 0.03, 'treatment_rate': 0.04, 'control_n': 800, 'treatment_n': 200},
    'established_sellers': {'control_rate': 0.15, 'treatment_rate': 0.17, 'control_n': 200, 'treatment_n': 800},
}
agg_control = (sum(d['control_rate'] * d['control_n'] for d in segment_data.values())
               / sum(d['control_n'] for d in segment_data.values()))
agg_treatment = (sum(d['treatment_rate'] * d['treatment_n'] for d in segment_data.values())
                  / sum(d['treatment_n'] for d in segment_data.values()))
print(f"3) Segment mix check: aggregate control={agg_control:.1%}, aggregate treatment={agg_treatment:.1%}")
for seg, d in segment_data.items():
    print(f"   {seg}: control n={d['control_n']}, treatment n={d['treatment_n']} (mix ratio differs)")
print()

# 4) Final checklist
issues = []
if ratio > 1.3:
    issues.append("Novelty effect suspected -- confirm with a longer holdout before full rollout")
issues.append("Network effect risk noted -- consider a market-level (not user-level) holdout test instead")
if segment_data['new_sellers']['control_n'] != segment_data['new_sellers']['treatment_n']:
    issues.append("Segment mix differs substantially between groups -- verify randomization wasn't biased by seller tenure")

print("4) Final checklist:")
for issue in issues:
    print(f"   \u26a0\ufe0f  {issue}")
print()
print("Recommendation: do not launch on the naive aggregate alone. Re-run with")
print("market-level randomization to address the network effect risk, and extend")
print("the holdout to confirm the lasting (non-novelty) effect size.")
print()
print("제안: 단순한 전체 수치만으로 출시하지 마세요. 네트워크 효과 위험을 해결하기")
print("위해 마켓 단위 무작위 배정으로 다시 실행하고, 지속적인(노벨티가 아닌)")
print("효과 크기를 확인하기 위해 홀드아웃 기간을 연장하세요.")
````

**EN:** A/B testing pitfalls are the ways a technically correct analysis can still mislead, not through calculation errors but through violated assumptions: a novelty effect makes an early lift look stronger than the lasting truth (caught by comparing early vs late windows), a network effect lets treatment "leak" into control through user-to-user interaction (a SUTVA violation that per-user randomization alone doesn't fix), and Simpson's paradox lets an aggregate result point in the exact opposite direction of every individual segment, purely because segment composition differs between groups. All three traps share the same underlying medicine: break the result down — by time, by contamination risk, by segment — before trusting a single headline number. This section closes Part 4 by assembling everything from Sections 14-17 into one final, holistic checklist, because each individual check can pass while the combination still hides a problem.
