# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** Non-parametric tests are rank-based alternatives to the t-tests and ANOVA from Sections 7 and 9 — instead of comparing means directly, they convert data to RANKS first and compare those. This sidesteps the normality assumption entirely and makes the tests far more resistant to outliers and skew, at the cost of being slightly less powerful when the data actually IS well-behaved.

**KR:** 비모수 검정은 섹션 7과 9의 t-검정·ANOVA에 대한 순위 기반 대안입니다 -- 평균을 직접 비교하는 대신, 먼저 데이터를 순위로 변환한 뒤 그 순위를 비교합니다. 이는 정규성 가정을 완전히 우회하고 검정을 이상치·왜도에 훨씬 강하게 만들지만, 데이터가 실제로 잘 행동할 때는 검정력이 약간 낮아지는 대가가 있습니다.

### Parametric vs Non-Parametric matching (모수·비모수 대응표)

| Parametric (Sections 7, 9)<br>모수 검정(섹션 7, 9) | Non-Parametric (Section 10)<br>비모수 검정(섹션 10) | Data shape / 데이터 형태 |
|---|---|---|
| Independent t-test<br>독립표본 t-검정 | Mann-Whitney U | Two separate groups<br>두 개의 별도 그룹 |
| Paired t-test<br>대응표본 t-검정 | Wilcoxon signed-rank<br>Wilcoxon 부호순위 검정 | Same subjects, before/after<br>같은 대상, 전후 |
| One-way ANOVA<br>일원배치 ANOVA | Kruskal-Wallis | 3+ separate groups<br>3개 이상의 별도 그룹 |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** Every parametric test so far has come with a nagging assumption — normality, adequate sample size for the CLT to kick in. Business data routinely fails these: response times, revenue, wait times are almost always right-skewed, and small pilot samples rarely hit the "n≥30" comfort zone. Non-parametric tests remove that worry by working directly with the ORDER of the data rather than its exact values.

**KR:** 지금까지의 모든 모수적 검정에는 성가신 가정이 따라왔습니다 -- 정규성, CLT가 작동할 만큼 충분한 표본 크기. 비즈니스 데이터는 이를 일상적으로 위반합니다: 응답 시간·매출·대기 시간은 거의 항상 오른쪽으로 치우쳐 있고, 작은 파일럿 표본은 "n≥30"의 편안한 구간에 좀처럼 도달하지 못합니다. 비모수 검정은 데이터의 정확한 값이 아니라 순서를 직접 다룸으로써 그 걱정을 없애줍니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Small pilot tests before a full rollout, any metric known to be heavily skewed (revenue, response time, session duration), data with a few extreme outliers you don't want to distort the whole test, and ordinal survey data (ratings like "1-5 stars") where the gap between categories isn't strictly meaningful as a number.

**KR:** 전면 출시 전 작은 파일럿 테스트, 심하게 치우쳐 있다고 알려진 지표(매출, 응답 시간, 세션 시간), 전체 검정을 왜곡하지 않길 원하는 소수의 극단적 이상치가 있는 데이터, 그리고 범주 간 간격이 숫자로서 엄밀하게 의미 있지 않은 순서형 설문 데이터("1-5점" 평점 등).

# 📝 Syntax (문법)

## Basic Syntax (기본 문법)

````python
import numpy as np
from scipy import stats

# Mann-Whitney U: compare two independent groups without assuming normality
# Mann-Whitney U: 정규성을 가정하지 않고 두 독립 그룹 비교
np.random.seed(42)
group_a = np.random.exponential(20, 40)  # skewed
group_b = np.random.exponential(28, 40)  # skewed

u_stat, p_value = stats.mannwhitneyu(group_a, group_b, alternative='two-sided')

print(f"U-statistic: {u_stat:.1f}")
print(f"p-value: {p_value:.4f}")
````

## Common Variations (자주 쓰는 변형)

````python
# For comparison, what would the (less appropriate) t-test say on the same data?
# 비교를 위해, 같은 데이터에 (덜 적절한) t-검정을 쓰면 어떻게 나올까?
t_stat, t_p = stats.ttest_ind(group_a, group_b)
print(f"t-test p-value (for comparison): {t_p:.4f}")
print(f"Mann-Whitney U p-value: {p_value:.4f}")
````

# 🧪 Small Examples (예제)

## Example 1 — Mann-Whitney U: Comparing Two Skewed, Outlier-Prone Groups (치우치고 이상치가 있는 두 그룹 비교)

````python
# Business question: does response time differ between the chat and email
# support channels? Response times are heavily skewed with a few extreme
# outliers -- is a t-test still trustworthy?
# 비즈니스 질문: 채팅과 이메일 지원 채널 간 응답 시간이 다른가? 응답
# 시간은 심하게 치우쳐 있고 극단적 이상치가 몇 개 있습니다 -- t-검정을
# 여전히 신뢰할 수 있는가?

np.random.seed(1)
chat_times = np.random.exponential(6, 45)
email_times = np.random.exponential(9, 45)
email_times = np.append(email_times, [180, 210])  # a couple of stuck tickets

print(f"Chat: mean={chat_times.mean():.1f}, median={np.median(chat_times):.1f}")
print(f"Email: mean={email_times.mean():.1f}, median={np.median(email_times):.1f}")
print()

t_stat, t_p = stats.ttest_ind(chat_times, email_times)
u_stat, u_p = stats.mannwhitneyu(chat_times, email_times, alternative='two-sided')

print(f"t-test:         t={t_stat:.3f}, p={t_p:.4f}")
print(f"Mann-Whitney U: U={u_stat:.1f}, p={u_p:.4f}")
print()
print("The two outliers pull the email mean up sharply, but barely affect the median")
print("or the rank-based Mann-Whitney result -- the t-test is more easily distorted by")
print("those two extreme values than the rank-based test is.")
print("두 이상치가 이메일의 평균을 크게 끌어올리지만, 중앙값이나 순위 기반")
print("Mann-Whitney 결과에는 거의 영향을 주지 않습니다 -- t-검정이 순위 기반")
print("검정보다 그 두 극단값에 더 쉽게 왜곡됩니다.")
````

## Example 2 — Wilcoxon Signed-Rank: Before/After with Non-Normal Data (전후 비교, 비정규 데이터)

````python
# Business question: did a new onboarding flow change the SAME users' time
# to first purchase? The distribution of individual changes is skewed --
# is a paired t-test still appropriate?
# 비즈니스 질문: 새 온보딩 플로우가 같은 사용자들의 첫 구매까지 걸리는
# 시간을 바꿨는가? 개별 변화의 분포가 치우쳐 있습니다 -- 대응표본
# t-검정이 여전히 적절한가?

np.random.seed(2)
n_users = 40
before = np.random.exponential(10, n_users) + 5
improvement = np.random.exponential(2, n_users)  # most improve a little, a few improve a LOT
after = np.maximum(before - improvement, 0.5)

t_stat, t_p = stats.ttest_rel(before, after)
w_stat, w_p = stats.wilcoxon(before, after)

print(f"Before mean: {before.mean():.2f}, After mean: {after.mean():.2f}")
print(f"Skewness of individual differences: {stats.skew(before - after):.2f}")
print()
print(f"Paired t-test:            t={t_stat:.3f}, p={t_p:.6f}")
print(f"Wilcoxon signed-rank:     W={w_stat:.1f}, p={w_p:.6f}")
print()
print("Both agree here that the change is significant -- Wilcoxon is the safer choice")
print("to REPORT given the skewed improvement distribution, even though the conclusion")
print("happens to match the t-test in this case.")
print("둘 다 여기서는 변화가 유의미하다는 데 동의합니다 -- 치우친 개선 분포를")
print("고려하면 Wilcoxon이 '보고'하기에 더 안전한 선택입니다. 이 경우 우연히")
print("t-검정과 결론이 일치했을 뿐입니다.")
````

## Example 3 — Kruskal-Wallis: The ANOVA Alternative, Built Out Properly (ANOVA 대안, 제대로 다루기)

````python
# Business question: compare order-processing time across 4 warehouse
# locations. Processing times are notoriously skewed with occasional
# extreme delays -- ANOVA's assumptions look shaky.
# 비즈니스 질문: 4개 물류센터의 주문 처리 시간을 비교합니다. 처리 시간은
# 악명 높게 치우쳐 있고 가끔 극단적인 지연이 있습니다 -- ANOVA의 가정이
# 불안정해 보입니다.

np.random.seed(3)
warehouse_a = np.random.exponential(4, 50)
warehouse_b = np.random.exponential(5, 50)
warehouse_c = np.random.exponential(7, 50)
warehouse_d = np.random.exponential(4.5, 50)

for name, data in [('A', warehouse_a), ('B', warehouse_b), ('C', warehouse_c), ('D', warehouse_d)]:
    print(f"Warehouse {name}: skewness={stats.skew(data):.2f}, n={len(data)}")
print()

f_stat, anova_p = stats.f_oneway(warehouse_a, warehouse_b, warehouse_c, warehouse_d)
h_stat, kw_p = stats.kruskal(warehouse_a, warehouse_b, warehouse_c, warehouse_d)

print(f"Standard ANOVA:   F={f_stat:.3f}, p={anova_p:.4f}")
print(f"Kruskal-Wallis:   H={h_stat:.3f}, p={kw_p:.4f}")
print()

if kw_p < 0.05:
    print("Kruskal-Wallis is significant -- SOME warehouse differs. Like ANOVA, it doesn't")
    print("say which -- use a rank-based post-hoc correction (e.g. Dunn's test) or pairwise")
    print("Mann-Whitney tests with a multiple-comparisons correction (Section 9's concern).")
    print("Kruskal-Wallis가 유의미합니다 -- 어떤 물류센터가 다르다는 의미입니다. ANOVA와")
    print("마찬가지로 어느 것인지는 알려주지 않습니다 -- 순위 기반 사후 보정(Dunn's test 등)")
    print("이나 다중비교 보정을 적용한 쌍별 Mann-Whitney 검정을 사용하세요(섹션 9의 관심사).")
````

## Example 4 — Choosing Parametric vs Non-Parametric: A Decision Framework (모수 vs 비모수 선택: 의사결정 틀)

````python
# Business question: build a quick decision helper that recommends parametric
# vs non-parametric based on what we've learned across Sections 7, 9, and 10.
# 비즈니스 질문: 섹션 7, 9, 10에서 배운 것을 바탕으로 모수적 vs 비모수적을
# 추천하는 빠른 의사결정 도구를 만들어보자.

def recommend_test(*groups, min_n=30, max_skew=1.0):
    """Recommend parametric or non-parametric based on sample size and skew."""
    issues = []
    for i, g in enumerate(groups):
        n = len(g)
        skew = abs(stats.skew(g))
        if n < min_n:
            issues.append(f"Group {i + 1}: small sample (n={n} < {min_n})")
        if skew > max_skew:
            issues.append(f"Group {i + 1}: notably skewed (|skew|={skew:.2f} > {max_skew})")
    if issues:
        print("Concerns found:")
        for issue in issues:
            print(f"  - {issue}")
        print("-> RECOMMENDATION: use the non-parametric alternative")
        return 'non-parametric'
    else:
        print("No major concerns with sample size or skew.")
        print("-> RECOMMENDATION: standard parametric test should be reliable")
        return 'parametric'

print("=== Scenario 1: small, skewed pilot data ===")
np.random.seed(4)
pilot_a = np.random.exponential(10, 15)
pilot_b = np.random.exponential(14, 15)
recommend_test(pilot_a, pilot_b)

print("\n=== Scenario 2: large, well-behaved data ===")
large_a = np.random.normal(50, 10, 200)
large_b = np.random.normal(53, 10, 200)
recommend_test(large_a, large_b)
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Defaulting to Parametric Tests Out of Habit (습관적으로 모수 검정을 기본으로 사용)
t-tests and ANOVA are more commonly taught and used, so it's tempting to reach for them automatically — but skipping the assumption check means you might be trusting a p-value that isn't reliable for your specific data.  
t-검정과 ANOVA가 더 흔히 가르쳐지고 사용되기 때문에 자동으로 손이 가기 쉽습니다 -- 하지만 가정 확인을 건너뛰면 특정 데이터에 신뢰할 수 없는 p-value를 믿게 될 수 있습니다.

✅ **Fix / 해결:** Make the skewness/sample-size check (Example 4's approach) a standard first step, not an afterthought.

### Mistake 2 — Assuming Non-Parametric Tests Compare Medians the Same Way t-tests Compare Means (비모수 검정이 t-검정과 똑같이 중앙값을 비교한다고 착각)
It's a common shorthand to say "Mann-Whitney tests medians," but technically it tests whether one group's values tend to be LARGER than the other's — this usually lines up with a median difference, but not always, especially when the two distributions have very different shapes.  
"Mann-Whitney는 중앙값을 검정한다"는 흔한 축약 표현이지만, 엄밀히는 한 그룹의 값이 다른 그룹보다 전반적으로 더 큰 경향이 있는지를 검정합니다 -- 보통 중앙값 차이와 일치하지만, 특히 두 분포의 모양이 매우 다를 때는 항상 그런 것은 아닙니다.

✅ **Fix / 해결:** When distribution shapes look very different between groups, interpret the result as "one tends to be larger," not strictly "the medians differ."

### Mistake 3 — Using Non-Parametric Tests as a "Safe Default" for Everything (모든 것에 비모수 검정을 "안전한 기본값"으로 사용)
Non-parametric tests give up some statistical power when the data actually IS well-behaved — reaching for them reflexively even when a parametric test's assumptions are clearly met means detecting real effects less often than you could.  
데이터가 실제로 잘 행동할 때 비모수 검정은 약간의 통계적 검정력을 포기합니다 -- 모수 검정의 가정이 명백히 충족될 때도 반사적으로 비모수 검정을 쓰면 발견할 수 있는 실제 효과를 덜 발견하게 됩니다.

✅ **Fix / 해결:** Check assumptions first; use the parametric test when they're reasonably met, and save the non-parametric alternative for when they're genuinely shaky.

### Mistake 4 — Feeding Wilcoxon Misaligned Paired Data (Wilcoxon에 정렬되지 않은 대응 데이터 입력)
Just like `ttest_rel()` (Section 7's Mistake 5), `stats.wilcoxon()` assumes row `i` in both arrays belongs to the same subject — misaligned data produces a silently meaningless result.  
ttest_rel()(섹션 7의 실수 5)과 마찬가지로, stats.wilcoxon()도 두 배열의 행 i가 같은 대상에 속한다고 가정합니다 -- 정렬되지 않은 데이터는 조용히 의미 없는 결과를 만듭니다.

✅ **Fix / 해결:** Verify both arrays are sorted/aligned by the same subject ID before running `wilcoxon()`.

### Mistake 5 — Treating a Non-Significant Non-Parametric Result as Extra Trustworthy (유의하지 않은 비모수 결과를 특별히 더 신뢰할 만하다고 착각)
"We used the more careful non-parametric test and found no significant difference" sounds extra rigorous, but a non-parametric test's lower power means a null result here is, if anything, LESS conclusive evidence of "no effect" than a well-powered parametric test would give.  
"더 신중한 비모수 검정을 사용했는데 유의미한 차이가 없었다"는 특별히 엄밀하게 들리지만, 비모수 검정의 낮은 검정력 때문에 여기서의 무효과 결과는 검정력이 충분한 모수 검정이 줬을 결과보다 오히려 "효과 없음"에 대한 증거로서 덜 결정적일 수 있습니다.

✅ **Fix / 해결:** Report the test's power (or at least sample size) alongside any non-significant result, non-parametric or not.

### Mistake 6 — Stopping After a Significant Kruskal-Wallis Without a Post-hoc Step (유의미한 Kruskal-Wallis 이후 사후 검정 없이 멈춤)
Just like ANOVA (Section 9's Mistake 2), a significant Kruskal-Wallis result only says SOME group differs, not which — and following it up with plain, uncorrected Mann-Whitney tests on every pair reintroduces the same multiple-comparisons problem Section 9 warned about.  
ANOVA(섹션 9의 실수 2)와 마찬가지로, 유의미한 Kruskal-Wallis 결과는 어떤 그룹이 다르다는 것만 말하지, 어느 것인지는 말하지 않습니다 -- 그리고 모든 쌍에 보정 없는 일반 Mann-Whitney 검정을 후속으로 돌리면 섹션 9가 경고했던 같은 다중비교 문제를 다시 끌어들입니다.

✅ **Fix / 해결:** Use a correction designed for multiple rank-based comparisons (like Dunn's test) rather than uncorrected pairwise Mann-Whitney tests.

# 💡 Tips (팁)
Useful tips or shortcuts / 유용한 팁과 단축법

- A quick mental mapping: Mann-Whitney = independent t-test, Wilcoxon = paired t-test, Kruskal-Wallis = ANOVA — memorize this table once and you'll always know which non-parametric test matches your data shape.<br>빠른 암기법: Mann-Whitney = 독립표본 t-검정, Wilcoxon = 대응표본 t-검정, Kruskal-Wallis = ANOVA -- 이 표를 한 번 외워두면 데이터 형태에 맞는 비모수 검정을 항상 알 수 있습니다.
- Non-parametric tests work by converting data to RANKS first — this is exactly the same trick Spearman correlation (Section 5) uses, and it's why both resist outliers so well.<br>비모수 검정은 데이터를 먼저 순위로 변환하는 방식으로 작동합니다 -- 이는 Spearman 상관계수(섹션 5)가 쓰는 것과 정확히 같은 트릭이며, 둘 다 이상치에 강한 이유입니다.
- When a parametric and non-parametric test on the same data strongly disagree, that disagreement itself is informative — it usually means outliers or skew are meaningfully influencing the parametric result.<br>같은 데이터에 대한 모수·비모수 검정이 크게 엇갈릴 때, 그 불일치 자체가 정보입니다 -- 보통 이상치나 왜도가 모수 검정 결과에 의미 있게 영향을 주고 있다는 뜻입니다.
- Running BOTH the parametric and non-parametric version side by side (like every example in this notebook does) costs almost nothing computationally and gives you an instant robustness check.<br>이 노트북의 모든 예제가 하듯 모수·비모수 버전을 나란히 돌리는 것은 계산 비용이 거의 들지 않으면서 즉각적인 견고성 점검을 제공합니다.
- For ordinal data (survey ratings like 1-5 stars), non-parametric tests are often the more honest choice regardless of sample size — the numbers "3" and "4" aren't guaranteed to be equally far apart in meaning the way true measurements are.<br>순서형 데이터(1-5점 평점 같은 설문)에는 표본 크기와 무관하게 비모수 검정이 더 정직한 선택인 경우가 많습니다 -- "3"과 "4"라는 숫자가 실제 측정값처럼 의미상 동일한 간격이라는 보장이 없기 때문입니다.

**Scenario / 시나리오:**

**EN:** A product team ran a small pilot (n=18 per group) testing a new pricing page against the old one, measuring time-to-purchase. Given the small sample and known skew in purchase-time data, walk through deciding which test to use and running it.

**KR:** 제품팀이 새 가격 페이지와 기존 페이지를 비교하는 작은 파일럿(그룹당 n=18)을 실행해 구매까지 걸리는 시간을 측정했습니다. 작은 표본과 구매 시간 데이터에 알려진 왜도를 고려해, 어떤 검정을 사용할지 결정하고 실행하는 과정을 살펴봅니다.

**To-do / 할 일:**
- [x] Apply the decision framework from Example 4 to this pilot's data<br>이 파일럿 데이터에 예제 4의 의사결정 틀 적용하기
- [x] Run the recommended test<br>추천된 검정 실행하기
- [x] Compare against a plain t-test for reference<br>참고용으로 일반 t-검정과 비교하기
- [x] Write a recommendation on whether to proceed to a full-scale test<br>전면 테스트로 진행할지에 대한 제안 작성하기

````python
np.random.seed(50)
old_page = np.random.exponential(12, 18) + 3
new_page = np.random.exponential(8, 18) + 3

print("Step 1: check assumptions")
recommend_test(old_page, new_page, min_n=30, max_skew=1.0)

print("\nStep 2: run the recommended test")
u_stat, u_p = stats.mannwhitneyu(new_page, old_page, alternative='two-sided')
t_stat, t_p = stats.ttest_ind(new_page, old_page)

print(f"\nMann-Whitney U (recommended): U={u_stat:.1f}, p={u_p:.4f}")
print(f"(for reference) t-test:                t={t_stat:.3f}, p={t_p:.4f}")

print(f"\nOld page: median={np.median(old_page):.1f} min, mean={old_page.mean():.1f} min")
print(f"New page: median={np.median(new_page):.1f} min, mean={new_page.mean():.1f} min")

print()
if u_p < 0.05:
    print("Recommendation: the new page shows a statistically significant improvement in")
    print("time-to-purchase, even accounting for the small, skewed pilot sample. Proceed")
    print("to a full-scale A/B test to confirm before a permanent rollout.")
    print()
    print("제안: 작고 치우친 파일럿 표본을 고려해도 새 페이지는 구매까지 걸리는 시간에서")
    print("통계적으로 유의미한 개선을 보입니다. 영구 출시 전 확인을 위해 전면 A/B")
    print("테스트로 진행하세요.")
else:
    print("Recommendation: not enough evidence yet from this small pilot -- consider a")
    print("larger follow-up before deciding.")
    print()
    print("제안: 이 작은 파일럿만으로는 아직 증거가 충분하지 않습니다 -- 결정 전에")
    print("더 큰 후속 테스트를 고려하세요.")
````

# 📝 Summary (요약)
*Write today's concept in 3~5 sentences.*

**EN:** Non-parametric tests are the rank-based counterparts to Sections 7 and 9's parametric tests: Mann-Whitney U replaces the independent t-test, Wilcoxon signed-rank replaces the paired t-test, and Kruskal-Wallis replaces ANOVA — each converting raw values to ranks first, which sidesteps the normality assumption and makes the test far less sensitive to outliers and skew. This robustness isn't free: when data genuinely is well-behaved (normal, no major outliers, adequate sample size), the parametric version usually detects a real effect more reliably, so switching to non-parametric tests by default everywhere quietly costs statistical power. A simple decision framework — check skewness and sample size first — is enough to route most business comparisons to the right family of test, closing out Part 2's central lesson: every hypothesis test rests on assumptions, and knowing how to check them (and what to do when they fail) matters as much as knowing the test itself.

**KR:** 비모수 검정은 섹션 7과 9의 모수 검정에 대응하는 순위 기반 버전입니다: Mann-Whitney U는 독립표본 t-검정을, Wilcoxon 부호순위 검정은 대응표본 t-검정을, Kruskal-Wallis는 ANOVA를 대체합니다 -- 각각 원본 값을 먼저 순위로 변환해 정규성 가정을 우회하고 이상치·왜도에 훨씬 덜 민감하게 만듭니다. 이 견고함은 공짜가 아닙니다: 데이터가 실제로 잘 행동할 때(정규분포, 큰 이상치 없음, 충분한 표본 크기), 모수 버전이 보통 실제 효과를 더 신뢰성 있게 탐지하므로, 어디서나 기본적으로 비모수 검정으로 전환하면 조용히 통계적 검정력을 잃게 됩니다. 왜도와 표본 크기를 먼저 확인하는 간단한 의사결정 틀만으로도 대부분의 비즈니스 비교를 올바른 검정 계열로 안내하기에 충분하며, 이는 Part 2의 핵심 교훈을 마무리합니다: 모든 가설검정은 가정 위에 서 있으며, 그것을 확인하는 법(그리고 실패했을 때 무엇을 할지)을 아는 것이 검정 자체를 아는 것만큼 중요합니다.

**Q1 (EN).** Which non-parametric test corresponds to each of these: (a) an independent t-test, (b) a paired t-test, (c) one-way ANOVA?

(a) Mann-Whitney U (mannwhitneyu), (b) Wilcoxon signed-rank (wilcoxon), (c) Kruskal-Wallis (kruskal)

**Q2 (EN).** Why is it slightly imprecise to say "Mann-Whitney tests whether two medians differ"? What does it more precisely test?  
**Q2 (KR).** "Mann-Whitney는 두 중앙값이 다른지 검정한다"는 말이 왜 다소 부정확한가요? 더 정확히는 무엇을 검정하나요?

It works on ranks, so it tests whether one group tends to have larger values than the other, and it equals a median test only when both groups have the same shape.    
순위를 기반으로 하므로 한 집단이 다른 집단보다 큰 값을 갖는 경향이 있는지를 검정하며, 두 집단의 분포 모양이 같을 때만 중앙값 검정과 같아집니다.

**Q3 (EN).** Why shouldn't you use a non-parametric test as a "safe default" for every comparison, even when a parametric test's assumptions are clearly met?  
**Q3 (KR).** 모수 검정의 가정이 명백히 충족될 때도 왜 모든 비교에 비모수 검정을 "안전한 기본값"으로 쓰면 안 되나요?

When the assumptions are met, a parametric test has more power, so it detects a real difference more often.  
가정이 충족되면 모수 검정이 검정력이 더 높아서 실제 차이를 더 잘 찾아냅니다.

**Q4 (EN).** A significant Kruskal-Wallis result comes back for 4 groups. What's the mistake in immediately running uncorrected Mann-Whitney tests on all 6 pairs?  
**Q4 (KR).** 4개 그룹에 대해 유의미한 Kruskal-Wallis 결과가 나왔습니다. 곧바로 6개 쌍 전부에 보정 없는 Mann-Whitney 검정을 돌리는 것의 실수는 무엇인가요?

Running 6 uncorrected tests stacks the false-positive chance, so use Dunn's test or correct the p-values with Bonferroni or Holm.  
보정 없이 6번을 돌리면 거짓 양성 확률이 쌓이므로, Dunn 검정을 쓰거나 Bonferroni, Holm으로 p-value를 보정합니다.

**Q5 (EN).** A pilot test has only 15 observations per group and the data is known to be right-skewed. Walk through the decision process for choosing a test.  
**Q5 (KR).** 파일럿 테스트가 그룹당 관측치 15개뿐이고 데이터가 오른쪽으로 치우쳐 있다고 알려져 있습니다. 검정을 선택하는 의사결정 과정을 설명해보세요.

With 15 per group and right-skewed data, normality is doubtful, so choose a rank-based test (Mann-Whitney for 2 groups, Kruskal-Wallis for 3 or more) and report medians.  
그룹당 15개에 오른쪽으로 치우친 데이터라 정규성이 의심되므로, 순위 기반 검정(2그룹은 Mann-Whitney, 3그룹 이상은 Kruskal-Wallis)을 고르고 중앙값을 보고합니다.
