# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** A/B test design is the planning phase that happens BEFORE a single line of experiment code runs: translating a product idea into a testable hypothesis, choosing which metrics will decide success, deciding how users get split into groups, and locking in the analysis plan — all the decisions this curriculum's Sections 6-12 gave you the tools to make, now applied as a checklist before you ever look at data.

**KR:** A/B 테스트 설계는 실험 코드를 단 한 줄도 돌리기 전에 이루어지는 계획 단계입니다: 제품 아이디어를 검정 가능한 가설로 번역하고, 어떤 지표가 성공을 결정할지 선택하고, 사용자를 어떻게 그룹으로 나눌지 정하고, 분석 계획을 고정하는 것 -- 이 커리큘럼의 섹션 6~12가 만들어준 도구들을 데이터를 보기도 전에 체크리스트로 적용하는 것입니다.

### The pre-launch design checklist (출시 전 설계 체크리스트)

| Design element / 설계 요소 | Question it answers / 답하는 질문 | Reuses / 재사용 |
|---|---|---|
| Hypothesis / 가설 | What exactly are we testing?<br>정확히 무엇을 검정하는가? | Section 6 |
| Primary vs guardrail metrics<br>주요·가드레일 지표 | What decides success? What must NOT get worse?<br>무엇이 성공을 결정하는가? 무엇이 악화되면 안 되는가? | Sections 6-7 |
| Randomization / 무작위 배정 | Is the split fair, clean, and verifiable?<br>분할이 공정하고 깨끗하며 검증 가능한가? | Section 8 |
| Pre-registration / 사전등록 | Is the analysis plan locked before we peek at data?<br>데이터를 엿보기 전에 분석 계획이 고정되어 있는가? | Section 6's Mistake 3 |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** Every mistake this curriculum has warned about — p-hacking alpha (Section 6), forgetting to check equal variance (Section 7), the multiple-comparisons trap (Section 9), confusing correlation with causation (Section 5) — becomes far harder to accidentally commit if the test's design is locked in writing before launch. Good design is cheap insurance against expensive statistical mistakes.

**KR:** 이 커리큘럼이 경고해온 모든 실수 -- alpha p-해킹(섹션 6), 등분산 확인 잊기(섹션 7), 다중비교 함정(섹션 9), 상관관계와 인과관계 혼동(섹션 5) -- 은 테스트 설계가 출시 전에 글로 고정되어 있으면 실수로 저지르기 훨씬 어려워집니다. 좋은 설계는 값비싼 통계적 실수에 대한 값싼 보험입니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Every single A/B test, from a simple button-color change to a full checkout redesign, needs this design phase — skipping it is the single most common reason A/B test results end up disputed, misinterpreted, or simply wrong.

**KR:** 버튼 색상 변경처럼 간단한 것부터 전체 결제 재설계까지, 모든 A/B 테스트는 이 설계 단계가 필요합니다 -- 이를 건너뛰는 것이 A/B 테스트 결과가 논쟁거리가 되거나, 잘못 해석되거나, 그냥 틀리게 되는 가장 흔한 이유입니다.

````python
import numpy as np
from scipy import stats

# A minimal but complete A/B test design, expressed as a structured dict
# 최소한이지만 완전한 A/B 테스트 설계를 구조화된 dict로 표현
test_design = {
    'hypothesis_h0': 'The new checkout button color has NO effect on conversion rate',
    'hypothesis_h1': 'The new checkout button color changes conversion rate',
    'primary_metric': 'conversion_rate',
    'guardrail_metrics': ['page_load_time_ms', 'refund_rate'],
    'randomization_unit': 'user_id',
    'alpha': 0.05,
    'one_sided': False,
}

for key, value in test_design.items():
    print(f"{key}: {value}")
````

# 🧪 Small Examples (예제)

## Example 1 — Formulating a Testable Hypothesis and ONE Primary Metric (검정 가능한 가설과 하나의 주요 지표 만들기)

````python
# Business question: the product team says "let's make search better." Turn
# this vague idea into a testable hypothesis with exactly ONE primary metric.
# 비즈니스 질문: 제품팀이 "검색을 더 좋게 만들자"고 말합니다. 이 모호한
# 아이디어를 정확히 하나의 주요 지표를 가진 검정 가능한 가설로 바꾸기.

vague_idea = "let's make search better"

# Bad: multiple "primary" metrics with no clear decision rule
bad_design = {
    'metrics_that_might_matter': ['click-through rate', 'time to first click',
                                   'zero-result rate', 'search-to-purchase rate', 'session length'],
}
print(f"Vague idea: '{vague_idea}'")
print(f"Bad design -- {len(bad_design['metrics_that_might_matter'])} candidate metrics,")
print("no single one designated as THE decision-maker. If 2 of 5 improve and 3 don't,")
print("what's the verdict? This is a recipe for post-hoc storytelling.")
print()

# Good: one clear hypothesis, one clear primary metric
good_design = {
    'h0': 'New search ranking algorithm does NOT change search-to-purchase rate',
    'h1': 'New search ranking algorithm changes search-to-purchase rate',
    'primary_metric': 'search_to_purchase_rate',
    'secondary_metrics': ['click-through rate', 'zero-result rate'],  # informative, not decisive
}
print("Good design:")
for k, v in good_design.items():
    print(f"  {k}: {v}")
print()
print("One primary metric means one clear pass/fail criterion decided BEFORE launch --")
print("everything else is context, not a second chance to declare victory.")
print("주요 지표 하나는 출시 전에 결정된 하나의 명확한 성공·실패 기준을 의미합니다 --")
print("나머지는 맥락일 뿐, 승리를 선언할 두 번째 기회가 아닙니다.")
````

## Example 2 — Primary vs Guardrail Metrics: What Must Not Get Worse (무엇이 악화되면 안 되는가)

````python
# Business question: a team wants to speed up checkout by skipping an address
# confirmation step. Conversion rate is the primary metric -- but what could
# this change accidentally break?
# 비즈니스 질문: 팀이 주소 확인 단계를 건너뛰어 결제 속도를 높이려 합니다.
# 전환율이 주요 지표이지만 -- 이 변경이 실수로 무엇을 망가뜨릴 수 있는가?

design = {
    'primary_metric': 'checkout_conversion_rate',
    'primary_direction': 'higher is better',
    'guardrail_metrics': {
        'shipping_error_rate': 'must NOT increase (wrong addresses slipping through)',
        'refund_rate': 'must NOT increase (mis-shipped orders get refunded)',
        'customer_support_contacts': 'must NOT increase (confused customers calling in)',
    }
}

print(f"Primary metric: {design['primary_metric']} ({design['primary_direction']})")
print()
print("Guardrail metrics (must not regress, even if the primary metric improves):")
for metric, rule in design['guardrail_metrics'].items():
    print(f"  - {metric}: {rule}")
print()
print("If conversion rate rises 5% but shipping errors triple, this is NOT a win --")
print("guardrails exist precisely to catch a 'successful' primary metric that was")
print("secretly bought at the expense of something else that matters.")
print("전환율이 5% 오르더라도 배송 오류가 3배가 된다면 이것은 승리가 아닙니다 --")
print("가드레일은 정확히 '성공한' 주요 지표가 실은 다른 중요한 것을 희생해")
print("몰래 얻어진 경우를 잡아내기 위해 존재합니다.")
````

## Example 3 — Randomization Unit and Integrity: Avoiding Contamination (무작위 배정 단위와 무결성)

````python
# Business question: for a feature that changes how search results are
# ranked, should randomization happen per SESSION or per USER? And once
# launched, does the actual split match the intended ratio?
# 비즈니스 질문: 검색 결과 순위 방식을 바꾸는 기능에서, 무작위 배정을
# 세션 단위로 해야 하는가, 사용자 단위로 해야 하는가? 그리고 출시 후
# 실제 분할이 의도한 비율과 일치하는가?

print("Session-level randomization risk: the SAME user could see Control in one")
print("session and Treatment in the next -- a confusing, inconsistent experience that")
print("can contaminate results (their behavior in one arm may be influenced by memory")
print("of the other).")
print()
print("User-level randomization: the same user ALWAYS sees the same variant across")
print("all their sessions -- the safer default for anything affecting the user experience")
print("in a way they might notice or remember.")
print()
print("세션 단위 무작위 배정의 위험: 같은 사용자가 한 세션에서는 대조군을, 다음")
print("세션에서는 실험군을 볼 수 있습니다 -- 혼란스럽고 일관성 없는 경험이 결과를")
print("오염시킬 수 있습니다. 사용자 단위 무작위 배정: 같은 사용자는 모든 세션에서")
print("항상 같은 변형을 봅니다 -- 사용자가 알아차리거나 기억할 만한 영향을 주는")
print("것에는 더 안전한 기본 선택입니다.")
print()

# QA check: Sample Ratio Mismatch, reusing Section 8's goodness-of-fit test
np.random.seed(1)
n_total = 8000
observed_control = 4080
observed_treatment = n_total - observed_control

observed = np.array([observed_control, observed_treatment])
expected = np.array([0.5, 0.5]) * n_total

chi2_stat, srm_p = stats.chisquare(f_obs=observed, f_exp=expected)
print(f"Intended split: 50/50. Observed: {observed_control}/{observed_treatment}")
print(f"SRM check (chi-square goodness of fit): p={srm_p:.4f}")
if srm_p < 0.01:
    print("FLAG: split deviates significantly from intended -- investigate the")
    print("randomization mechanism BEFORE trusting any other result from this test.")
else:
    print("OK: split matches the intended ratio closely enough to proceed.")
````

## Example 4 — Pre-registration: Locking the Analysis Plan Before Launch (출시 전 분석 계획 고정하기)

````python
# Business question: what actually happens, in numbers, if a team checks
# results daily and stops "whenever it looks significant" instead of
# pre-registering a fixed sample size and stopping rule?
# 비즈니스 질문: 팀이 표본 크기와 중단 규칙을 미리 정하는 대신 매일 결과를
# 확인하고 "유의미해 보이면" 멈춘다면, 숫자로는 실제로 무슨 일이 일어나는가?

np.random.seed(2)
alpha = 0.05
n_simulations = 2000
false_positives_fixed = 0
false_positives_peeking = 0

for _ in range(n_simulations):
    # Scenario: NO real effect exists (H0 is true) -- simulate daily data collection
    daily_batches = [np.random.normal(0, 1, 20) for _ in range(10)]  # 10 "days" of data
    cumulative = np.array([])

    peeked_significant = False
    for batch in daily_batches:
        cumulative = np.concatenate([cumulative, batch])
        if len(cumulative) >= 20:
            _, p = stats.ttest_1samp(cumulative, popmean=0)
            if p < alpha:
                peeked_significant = True
                break  # "stop as soon as it looks significant" -- the p-hacking move

    if peeked_significant:
        false_positives_peeking += 1

    # Fixed design: only check ONCE, at the pre-registered final sample size
    final_data = np.concatenate(daily_batches)
    _, p_fixed = stats.ttest_1samp(final_data, popmean=0)
    if p_fixed < alpha:
        false_positives_fixed += 1

print(f"Pre-registered (check ONCE at the end): false positive rate = {false_positives_fixed / n_simulations:.1%}")
print(f"'Peek daily, stop when significant': false positive rate = {false_positives_peeking / n_simulations:.1%}")
print()
print("Even with NO real effect anywhere in this simulation, peeking daily and stopping")
print(f"early inflates the false-positive rate WELL above the intended {alpha:.0%} -- this is")
print("exactly why the analysis plan (sample size, stopping rule, primary metric) must")
print("be locked BEFORE launch, not adjusted based on what the data seems to show.")
print()
print(f"'사전등록(끝에 한 번만 확인)' 오탐율: {false_positives_fixed / n_simulations:.1%}")
print(f"'매일 엿보고 유의미하면 멈춤' 오탐율: {false_positives_peeking / n_simulations:.1%}")
print("이 시뮬레이션에는 실제 효과가 전혀 없는데도, 매일 엿보고 조기 종료하면 실제")
print("오탐율이 의도한 수준보다 훨씬 높아집니다 -- 이것이 정확히 분석 계획(표본 크기,")
print("중단 규칙, 주요 지표)이 출시 전에 고정되어야 하는 이유입니다.")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Tracking Multiple "Primary" Metrics With No Decisive One (여러 개의 "주요" 지표를 추적하며 결정적인 것 없음)
When 5 metrics are all candidates for "success" and none is designated as THE decision-maker in advance, a mixed result (2 up, 3 flat) invites picking whichever story fits the conclusion you wanted — exactly Example 1's bad design.  
5개의 지표 모두가 "성공"의 후보이고 미리 결정적인 것으로 지정된 것이 없으면, 혼재된 결과(2개 상승, 3개 변화 없음)는 원하는 결론에 맞는 이야기를 아무거나 고르도록 유혹합니다 -- 정확히 예제 1의 나쁜 설계입니다.

✅ **Fix / 해결:** Designate exactly ONE primary metric before launch; every other metric is context, not a tiebreaker.

### Mistake 2 — Launching Without Any Guardrail Metrics (가드레일 지표 없이 출시)
Watching only the metric you're trying to improve means a change that quietly breaks something else (refund rate, support volume, page errors) goes completely undetected until it's a real problem.  
개선하려는 지표만 지켜보면, 다른 무언가(환불율, 지원 문의량, 페이지 오류)를 조용히 망가뜨리는 변경이 실제 문제가 될 때까지 완전히 감지되지 않습니다.

✅ **Fix / 해결:** Identify at least one or two guardrail metrics tied to the specific mechanism of the change, as Example 2 did.

### Mistake 3 — Mismatching the Randomization Unit to the Feature's Scope (기능의 범위와 무작위 배정 단위 불일치)
Randomizing per session when a change affects something a user would notice and remember (like a redesigned layout) risks the same person bouncing between Control and Treatment, contaminating both groups' behavior.  
사용자가 알아차리고 기억할 만한 것(재설계된 레이아웃 등)에 영향을 주는 변경에 세션 단위로 무작위 배정하면, 같은 사람이 대조군과 실험군 사이를 오가며 두 그룹의 행동을 오염시킬 위험이 있습니다.

✅ **Fix / 해결:** Default to user-level randomization for anything visible/memorable; session-level is only safe for changes with no lasting user-side effect.

### Mistake 4 — Skipping the Sample Ratio Mismatch Check (표본 비율 불일치 확인 생략)
Trusting a test's primary result without first confirming the traffic actually split the way it was supposed to means a broken randomization mechanism could be silently invalidating everything downstream.  
트래픽이 실제로 의도한 대로 분할되었는지 먼저 확인하지 않고 테스트의 주요 결과를 신뢰하면, 고장 난 무작위 배정 메커니즘이 이후의 모든 것을 조용히 무효화하고 있을 수 있습니다.

✅ **Fix / 해결:** Run the SRM check (Example 3, Section 8's goodness-of-fit test) as a standard first step before interpreting any other result.

### Mistake 5 — Peeking at Results Repeatedly and Stopping Early (결과를 반복적으로 확인하며 조기 종료)
As Example 4's simulation showed directly, checking results daily and stopping "as soon as it looks significant" inflates the true false-positive rate well beyond the intended alpha, even when there's genuinely no effect.  
예제 4의 시뮬레이션이 직접 보여줬듯, 매일 결과를 확인하고 "유의미해 보이면 바로" 멈추는 것은 실제로 효과가 전혀 없어도 실제 오탐율을 의도한 alpha보다 훨씬 높게 부풀립니다.

✅ **Fix / 해결:** Pre-register a fixed sample size (or a formal sequential-testing method designed for early stopping) and stick to it.

### Mistake 6 — Choosing the Primary Metric After Seeing Which One Moved (어느 지표가 움직였는지 본 후에 주요 지표 선택)
Deciding "conversion rate is our primary metric" only after noticing it happened to improve — while quietly ignoring that revenue per user didn't — is HARKing (Hypothesizing After Results are Known), and it defeats the entire purpose of pre-registration.  
"전환율이 우리의 주요 지표다"라고, 그것이 우연히 개선된 것을 본 뒤에야 결정하면서 -- 사용자당 매출은 개선되지 않았다는 것은 조용히 무시하는 것은 결과를 알고 난 뒤 가설 세우기(HARKing)이며, 사전등록의 전체 목적을 무너뜨립니다.

✅ **Fix / 해결:** Write the hypothesis and primary metric down BEFORE launch, and treat that document as unchangeable once data collection begins.

# 💡 Tips (팁)
Useful tips or shortcuts / 유용한 팁과 단축법

- Write the test design as an actual document (or the structured dict from this section's Syntax) BEFORE requesting engineering resources to build it — a design that's hard to write clearly is often a sign the hypothesis itself needs more thought.<br>엔지니어링 리소스를 요청하기 전에 테스트 설계를 실제 문서(또는 이 섹션의 구조화된 dict)로 작성하세요 -- 명확하게 쓰기 어려운 설계는 종종 가설 자체를 더 다듬어야 한다는 신호입니다.
- A good guardrail metric is usually tied to the SPECIFIC mechanism of the change, not a generic "nothing bad happens" catch-all — ask "what could THIS particular change break?" rather than monitoring everything.<br>좋은 가드레일 지표는 보통 변경의 구체적인 메커니즘과 연결되어 있지, "아무 나쁜 일도 없기를"이라는 포괄적인 것이 아닙니다 -- 모든 것을 모니터링하기보다 "이 특정 변경이 무엇을 망가뜨릴 수 있는가?"를 물으세요.
- Section 15 will build directly on the primary metric and alpha locked here to calculate exactly how big a sample (and how long to run the test) is needed — design and sample size are two halves of the same planning step.<br>섹션 15는 여기서 고정한 주요 지표와 alpha를 바탕으로 정확히 얼마나 큰 표본(그리고 테스트를 얼마나 오래 돌려야 하는지)이 필요한지 계산합니다 -- 설계와 표본 크기는 같은 계획 단계의 두 반쪽입니다.
- An SRM check isn't just a launch-day formality — run it periodically throughout the test, since a randomization bug can sometimes appear (or worsen) partway through, not just at the start.<br>표본 비율 불일치 확인은 단순히 출시일의 형식적 절차가 아닙니다 -- 무작위 배정 버그는 시작 시점뿐 아니라 테스트 도중에 나타나거나 악화되기도 하므로 테스트 내내 주기적으로 확인하세요.
- If a stakeholder asks "can we just peek at how it's going," the honest answer is a fixed monitoring cadence for guardrails (to catch a genuine emergency) combined with a firm "no early stopping on the primary metric" — the two goals aren't in conflict if the design separates them clearly.<br>이해관계자가 "그냥 어떻게 되고 있는지 살짝 봐도 될까요"라고 물으면, 정직한 답은 가드레일에 대한 고정된 모니터링 주기(진짜 비상사태를 잡기 위해)와 "주요 지표는 조기 종료 없음"이라는 단호한 원칙을 결합하는 것입니다 -- 설계가 이 둘을 명확히 분리하면 두 목표는 충돌하지 않습니다.

# ❓ Review Questions (복습 질문)

**Q1 (EN).** A team tracks 5 metrics for a test with no single one designated as primary. Why is this risky, even if the test's statistics are computed correctly?  
**Q1 (KR).** 팀이 테스트를 위해 5개 지표를 추적하는데 어느 하나도 주요 지표로 지정되지 않았습니다. 테스트의 통계가 올바르게 계산되어도 왜 이것이 위험한가요?

With no primary metric, the team can pick whichever metric happened to improve after seeing the results. Checking 5 metrics also raises the chance that one looks significant by luck, so the conclusion is unreliable even if each calculation is correct.  
주요 지표가 없으면 결과를 본 뒤 우연히 좋아진 지표를 골라 따라갈 수 있습니다. 지표 5개를 보면 그중 하나가 우연히 유의하게 나올 확률도 커져서, 계산이 맞아도 결론을 믿기 어렵습니다.

**Q2 (EN).** A checkout speedup test only tracks conversion rate. What kind of problem could this design miss entirely, and how would you fix the design?  
**Q2 (KR).** 결제 속도 개선 테스트가 전환율만 추적합니다. 이 설계는 어떤 종류의 문제를 완전히 놓칠 수 있으며, 설계를 어떻게 고쳐야 하나요?

It can miss side effects, such as higher refund rate, fewer completed orders, or lower revenue per user. Fix it by adding guardrail metrics (for example payment errors, revenue per user) and locking the primary metric before launch.  
환불률 증가, 결제 완료 감소, 사용자당 매출 하락 같은 부작용을 놓칠 수 있습니다. 가드레일 지표(예: 결제 오류, 사용자당 매출)를 추가하고 주요 지표는 시작 전에 정해서 고쳐야 합니다.

**Q3 (EN).** Why is user-level randomization usually safer than session-level randomization for a change to a page's visible design?  
**Q3 (KR).** 페이지의 눈에 보이는 디자인 변경에는 왜 세션 단위보다 사용자 단위 무작위 배정이 보통 더 안전한가요?

With session-level randomization, the same person can see both versions on different visits. That confuses the user, mixes the two experiences, and breaks the independence between groups.  
세션 단위로 배정하면 같은 사람이 방문할 때마다 다른 화면을 볼 수 있습니다. 사용자가 혼란스럽고, 두 경험이 섞여 집단 간 독립성이 깨집니다.

**Q4 (EN).** Example 4's simulation showed peeking daily and stopping early inflates the false-positive rate. Why does checking the SAME data multiple times cause this, even with no real effect present?  
**Q4 (KR).** 예제 4의 시뮬레이션은 매일 결과를 엿보고 조기 종료하면 오탐율이 부풀려짐을 보여줬습니다. 실제 효과가 전혀 없어도 같은 데이터를 여러 번 확인하는 것이 왜 이런 결과를 낳나요?

Even with no real effect, the p-value moves up and down by chance as data accumulates. Each check is another chance to fall below 0.05, and stopping at the first dip counts all those chances as one, so the false-positive rate goes above 5%.  
실제 효과가 없어도 데이터가 쌓이는 동안 p-value는 우연히 오르내립니다. 확인할 때마다 0.05 아래로 내려갈 기회가 생기고, 처음 내려간 순간 멈추면 그 기회들을 한 번으로 취급해서 오탐율이 5%보다 커집니다.

**Q5 (EN).** A team picks "conversion rate" as their primary metric only after noticing it improved, quietly ignoring that revenue per user dropped. What's this practice called, and why does it undermine the test's conclusions?  
**Q5 (KR).** 팀이 전환율이 개선된 것을 본 뒤에야 이를 주요 지표로 선택하면서, 사용자당 매출이 떨어진 것은 조용히 무시합니다. 이런 관행을 무엇이라고 부르며, 왜 테스트의 결론을 무너뜨리나요?

It is called p-hacking (metric cherry-picking). It undermines the conclusion because the winning metric was chosen after the result, the drop in revenue per user was ignored, and the true false-positive rate is higher than reported.  
p-hacking(지표 골라잡기)이라고 부릅니다. 결과를 본 뒤 이긴 지표를 골랐고 사용자당 매출 하락을 무시했으며, 실제 오탐율이 보고된 값보다 높아서 결론이 무너집니다.
