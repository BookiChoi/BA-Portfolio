# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** Probability quantifies uncertainty as a number between 0 and 1 — how likely an event is to happen. The rules covered here (addition, multiplication, complement, conditional probability, Bayes' theorem) are the small set of building blocks that every statistical test and every A/B testing p-value in this curriculum is constructed from.

**KR:** 확률은 불확실성을 0과 1 사이의 숫자로 정량화합니다 -- 어떤 사건이 일어날 가능성이 얼마나 되는가. 여기서 다루는 규칙들(덧셈, 곱셈, 여사건, 조건부확률, 베이즈 정리)은 이 커리큘럼의 모든 통계 검정과 모든 A/B 테스트의 p-value가 만들어지는 작은 building block 집합입니다.

### The core rules (핵심 규칙)

| Rule / 규칙 | Formula / 공식 | Business framing / 비즈니스 예시 |
|---|---|---|
| Complement<br>여사건 | P(not A) = 1 - P(A) | "% who did NOT convert"<br>"전환하지 않은 비율" |
| Addition (mutually exclusive)<br>덧셈(배반사건) | P(A or B) = P(A) + P(B) | Channel A OR Channel B, if a visitor only ever uses one<br>한 방문자가 한 채널만 쓴다면, 채널 A 또는 채널 B |
| Addition (general)<br>덧셈(일반) | P(A or B) = P(A) + P(B) - P(A and B) | Email OR push open rate, when some users open both<br>일부가 둘 다 열 때, 이메일 또는 푸시 오픈율 |
| Multiplication (independent)<br>곱셈(독립사건) | P(A and B) = P(A) × P(B) | Two unrelated events both happening<br>서로 관련 없는 두 사건이 모두 일어남 |
| Conditional<br>조건부확률 | P(A\|B) = P(A and B) / P(B) | "% who convert GIVEN they clicked"<br>"클릭했을 때 전환하는 비율" |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** Business questions are usually really probability questions in disguise: "what fraction of visitors who see the ad actually convert?", "given that a transaction looks suspicious, how likely is it to actually be fraud?", "if two variants perform the same in a test, how likely was I to see THIS big a difference just by chance?" Without the rules in this section, none of those questions have a rigorous answer.

**KR:** 비즈니스 질문은 대개 확률 질문이 변장한 것입니다: "광고를 본 방문자 중 실제로 전환하는 비율은?", "거래가 의심스러워 보일 때 실제로 사기일 확률은?", "두 변형이 테스트에서 같은 성과를 낸다면, 우연히 이만큼 큰 차이를 볼 확률은 얼마나 될까?" 이 섹션의 규칙 없이는 이 질문 중 어느 것도 엄밀한 답을 가질 수 없습니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Funnel analysis (probability of moving from one stage to the next), fraud/anomaly detection (Bayes' theorem), targeting decisions (conditional probability of conversion given a segment), and literally every p-value you'll compute starting in Section 6.

**KR:** 퍼널 분석(한 단계에서 다음 단계로 넘어갈 확률), 사기·이상 탐지(베이즈 정리), 타겟팅 결정(세그먼트가 주어졌을 때 전환의 조건부확률), 그리고 섹션 6부터 계산하게 될 모든 p-value에 문자 그대로 사용됩니다.

# 📝 Syntax (문법)

## Basic Syntax (기본 문법)

````python
import numpy as np
import pandas as pd

# Business scenario: a marketing funnel
# 비즈니스 시나리오: 마케팅 퍼널
visitors = 10000
clicked_ad = 2500
converted = 400

p_click = clicked_ad / visitors
p_convert_given_click = converted / clicked_ad   # conditional probability
p_convert_overall = converted / visitors

print(f"P(click) = {p_click:.3f} / P(클릭) = {p_click:.3f}")
print(f"P(convert | click) = {p_convert_given_click:.3f} / P(전환 | 클릭) = {p_convert_given_click:.3f}")
print(f"P(convert) overall = {p_convert_overall:.3f} / 전체 P(전환) = {p_convert_overall:.3f}")
````

## Common Variations (자주 쓰는 변형)

````python
# Multiplication rule check: P(click) * P(convert|click) should equal P(convert overall)
# 곱셈 법칙 확인: P(클릭) * P(전환|클릭) = 전체 P(전환)과 같아야 함
p_convert_via_chain = p_click * p_convert_given_click
print(f"P(click) x P(convert|click) = {p_convert_via_chain:.3f}")
print(f"Matches direct calculation: {np.isclose(p_convert_via_chain, p_convert_overall)}")

# Complement rule
# 여사건 법칙
p_not_click = 1 - p_click
print(f"P(did NOT click) = {p_not_click:.3f}")
````

# 🧪 Small Examples (예제)

## Example 1 — Probability Rules via a Conversion Funnel (전환 퍼널로 보는 확률 법칙)

````python
# Business question: of all visitors, what fraction either clicked the ad
# OR used organic search to find the product page? (general addition rule)
# 비즈니스 질문: 전체 방문자 중 광고를 클릭했거나 오가닉 검색으로 제품
# 페이지를 찾은 비율은? (일반 덧셈 법칙)

visitors = 10000
clicked_ad = 2500
organic_search = 3200
both_ad_and_organic = 400  # visitors who used BOTH paths

p_ad = clicked_ad / visitors
p_organic = organic_search / visitors
p_both = both_ad_and_organic / visitors

# General addition rule: P(A or B) = P(A) + P(B) - P(A and B)
p_ad_or_organic = p_ad + p_organic - p_both

print(f"P(clicked ad) = {p_ad:.3f}")
print(f"P(organic search) = {p_organic:.3f}")
print(f"P(both) = {p_both:.3f}")
print(f"P(ad OR organic) = {p_ad_or_organic:.3f}")
print(f"({p_ad_or_organic * visitors:.0f} of {visitors} visitors used at least one of these paths)")
print(f"({p_ad_or_organic * visitors:.0f}명 / {visitors}명이 두 경로 중 최소 하나를 사용)")
````

## Example 2 — Conditional Probability: Conversion Rate by Segment (세그먼트별 전환율)

````python
# Business question: does the overall conversion rate hide a big difference
# between mobile and desktop users?
# 비즈니스 질문: 전체 전환율이 모바일과 데스크톱 사용자 간의 큰 차이를
# 숨기고 있는가?

np.random.seed(1)
n = 5000
device = np.random.choice(['mobile', 'desktop'], size=n, p=[0.65, 0.35])
converted = np.array([np.random.random() < (0.03 if d == 'mobile' else 0.09) for d in device])
df = pd.DataFrame({'device': device, 'converted': converted})

p_convert_overall = df['converted'].mean()
p_convert_given_mobile = df[df['device'] == 'mobile']['converted'].mean()
p_convert_given_desktop = df[df['device'] == 'desktop']['converted'].mean()

print(f"P(convert) overall = {p_convert_overall:.3f}")
print(f"P(convert | mobile) = {p_convert_given_mobile:.3f}")
print(f"P(convert | desktop) = {p_convert_given_desktop:.3f}")
print()
print("The overall rate looks unremarkable, but conditioning on device reveals desktop")
print("converts at roughly 3x the mobile rate -- a targeting opportunity.")
print("전체 비율은 평범해 보이지만, 기기로 조건을 걸면 데스크톱이 모바일보다")
print("약 3배 높은 전환율을 보입니다 -- 타겟팅 기회입니다.")
````

## Example 3 — Bayes' Theorem: Fraud Detection (베이즈 정리: 사기 탐지)

````python
# Business question: your fraud detection system flags a transaction as
# suspicious. Given the flag, what's the ACTUAL probability it's real fraud?
# 비즈니스 질문: 사기 탐지 시스템이 거래를 의심스럽다고 표시했습니다.
# 그 표시가 주어졌을 때, 실제로 사기일 확률은 얼마인가?

p_fraud = 0.001                    # base rate: 0.1% of all transactions are fraud
p_flagged_given_fraud = 0.95       # system catches 95% of real fraud (sensitivity)
p_flagged_given_not_fraud = 0.02   # system false-flags 2% of legitimate transactions

p_not_fraud = 1 - p_fraud
p_flagged = (p_flagged_given_fraud * p_fraud) + (p_flagged_given_not_fraud * p_not_fraud)

# Bayes' theorem: P(fraud | flagged) = P(flagged | fraud) * P(fraud) / P(flagged)
p_fraud_given_flagged = (p_flagged_given_fraud * p_fraud) / p_flagged

print(f"P(flagged) overall = {p_flagged:.4f}")
print(f"P(fraud | flagged) = {p_fraud_given_flagged:.4f} ({p_fraud_given_flagged * 100:.1f}%)")
print()
print(f"Even with a 95% detection rate, only about {p_fraud_given_flagged * 100:.0f}% of flagged")
print("transactions are actually fraud -- because fraud itself is so rare (0.1% base rate),")
print("the much larger pool of legitimate transactions produces more false alarms in absolute")
print("terms than true fraud cases. This is the 'base rate fallacy.'")
print(f"탐지율이 95%임에도, 표시된 거래 중 실제 사기는 약 {p_fraud_given_flagged * 100:.0f}%에")
print("불과합니다 -- 사기 자체가 워낙 드물기 때문에(기본 발생률 0.1%), 훨씬 많은 정상")
print("거래 풀에서 나오는 오탐이 실제 사기보다 절대 건수로 더 많아집니다.")
print("이것이 '기저율 오류(base rate fallacy)'입니다.")
````

## Example 4 — Classic Probability Brain-Teasers: Monty Hall & Birthday Paradox (고전 확률 문제)

````python
# Classic interview brain-teaser: should you switch doors in the Monty Hall
# problem? Simulate to find out.
# 고전 면접 확률 문제: 몬티홀 문제에서 문을 바꿔야 하는가? 시뮬레이션으로 확인.

np.random.seed(2)
n_trials = 10000
stay_wins = 0
switch_wins = 0

for _ in range(n_trials):
    car_door = np.random.randint(0, 3)
    player_pick = np.random.randint(0, 3)
    remaining_doors = [d for d in range(3) if d != player_pick and d != car_door]
    host_opens = np.random.choice(remaining_doors)
    switch_door = [d for d in range(3) if d != player_pick and d != host_opens][0]

    if player_pick == car_door:
        stay_wins += 1
    if switch_door == car_door:
        switch_wins += 1

print(f"Stay strategy win rate:   {stay_wins / n_trials:.3f}  (theory: 1/3 = 0.333)")
print(f"Switch strategy win rate: {switch_wins / n_trials:.3f}  (theory: 2/3 = 0.667)")
print()
print("Switching doubles your odds -- the host's action carries information because")
print("they always avoid opening the door with the car.")
print("문을 바꾸면 승률이 두 배가 됩니다 -- 진행자는 항상 자동차가 있는 문을 피해서")
print("열기 때문에 그 행동 자체가 정보를 담고 있습니다.")
print()

# Classic interview brain-teaser: how many people need to be in a room before
# there's a >50% chance two of them share a birthday?
# 고전 면접 확률 문제: 방에 몇 명이 있어야 두 명의 생일이 같을 확률이 50%를 넘는가?

def prob_shared_birthday(n_people):
    p_all_different = 1.0
    for i in range(n_people):
        p_all_different *= (365 - i) / 365
    return 1 - p_all_different

for n in [10, 20, 23, 30, 50]:
    p = prob_shared_birthday(n)
    print(f"{n} people: P(shared birthday) = {p:.3f}")

print()
print("The answer crosses 50% at just 23 people -- far fewer than most people guess,")
print("because the number of PAIRS grows much faster than the number of people")
print("(23 people = 253 possible pairs to compare).")
print("정답은 겨우 23명에서 50%를 넘습니다 -- 대부분의 예상보다 훨씬 적은데, 사람")
print("수가 아니라 '쌍'의 개수가 훨씬 빠르게 늘어나기 때문입니다(23명 = 비교")
print("가능한 쌍 253개).")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Confusing P(A|B) with P(B|A) (P(A|B)와 P(B|A) 혼동)
"95% of fraud gets flagged" (P(flagged|fraud)) is a completely different statement from "95% of flagged transactions are fraud" (P(fraud|flagged)) — mixing these up is called the "prosecutor's fallacy" and it's the single most common mistake with conditional probability.  
"사기의 95%가 표시된다"(P(표시\|사기))는 "표시된 거래의 95%가 사기다"(P(사기\|표시))와 완전히 다른 문장입니다 -- 이를 혼동하는 것을 "검사의 오류"라고 부르며, 조건부확률에서 가장 흔한 실수입니다.

✅ **Fix / 해결:** Always write out exactly which event is being conditioned on, and use Bayes' theorem to convert between the two directions.

### Mistake 2 — Adding Overlapping Probabilities Without Subtracting the Overlap (겹치는 확률을 겹침 제거 없이 더하기)
P(A) + P(B) overcounts anyone in both groups — if 40% opened email and 30% opened push, "70% opened at least one" is only correct if NO ONE opened both.  
P(A) + P(B)는 두 그룹 모두에 속한 사람을 이중으로 셉니다 -- 이메일 40%, 푸시 30%가 열었다면, "최소 하나를 연 사람이 70%"라는 말은 아무도 둘 다 열지 않았을 때만 맞습니다.

✅ **Fix / 해결:** Use the general addition rule, P(A or B) = P(A) + P(B) - P(A and B), unless you've confirmed the events are mutually exclusive.

### Mistake 3 — Assuming Independence Without Checking (확인 없이 독립성 가정)
The multiplication rule P(A and B) = P(A) x P(B) only works if A and B are independent — device type and purchase amount, for example, are usually NOT independent, so multiplying their individual probabilities gives a wrong answer.  
곱셈 법칙 P(A and B) = P(A) x P(B)는 A와 B가 독립일 때만 성립합니다 -- 예를 들어 기기 종류와 구매 금액은 보통 독립이 아니므로, 각각의 확률을 곱하면 틀린 답이 나옵니다.

✅ **Fix / 해결:** Check independence first, or use the conditional-probability form that always works: P(A and B) = P(A) x P(B\|A).

### Mistake 4 — Base Rate Neglect (기저율 무시)
A test that's "95% accurate" sounds trustworthy, but if the thing being tested for is rare (like 0.1% fraud), most positive results will still be false positives — the base rate matters as much as the test's accuracy.  
"95% 정확도"인 검사는 신뢰할 만해 보이지만, 검사 대상이 드문 경우(0.1% 사기처럼) 대부분의 양성 결과가 여전히 오탐일 수 있습니다 -- 기저율은 검사의 정확도만큼 중요합니다.

✅ **Fix / 해결:** Always run the full Bayes' theorem calculation before trusting a "% accurate" claim on a rare event.

### Mistake 5 — In Monty Hall, Assuming It's 50/50 After a Door Opens (몬티홀에서 문이 열린 뒤 50/50이라고 착각)
It feels like 2 remaining doors should mean 50/50 odds, but the host's choice isn't random — they always avoid the car, so their action carries information that shifts the odds to 1/3 (stay) vs 2/3 (switch).  
남은 문이 2개니까 50/50일 것 같지만, 진행자의 선택은 무작위가 아닙니다 -- 항상 자동차를 피하기 때문에 그 행동 자체가 정보를 담고 있어 확률이 1/3(유지) 대 2/3(바꾸기)로 바뀝니다.

✅ **Fix / 해결:** Remember that the host's non-random behavior is the whole point of the puzzle — simulate it if the intuition doesn't click (see Example 4).

### Mistake 6 — Misjudging the Birthday Paradox's Scale (생일 역설의 규모 오판)
Intuition says you'd need close to 183 people (half of 365) for a coin-flip chance of a shared birthday, but the real answer is 23 — because the number of PAIRS to compare grows roughly with the square of the number of people, not linearly.  
직관적으로는 365의 절반인 183명 정도가 필요할 것 같지만, 실제 정답은 23명입니다 -- 비교해야 할 '쌍'의 개수가 사람 수에 비례하는 게 아니라 대략 제곱에 비례해서 늘어나기 때문입니다.

✅ **Fix / 해결:** Whenever a problem involves comparing every pair in a group, expect the answer to be much smaller than intuition suggests.

# 💼 Business Example (비즈니스 예시)
*How would a Business Analyst use this?*

**Scenario / 시나리오:**

**EN:** Your ad platform can target users by a "high purchase intent" signal, but the signal is imperfect. Leadership wants to know: if we ONLY show ads to users flagged as high-intent, how much of that ad spend is actually reaching people who were going to buy anyway versus people who wouldn't convert regardless?

**KR:** 광고 플랫폼은 "높은 구매 의도" 신호로 사용자를 타겟팅할 수 있지만, 이 신호는 완벽하지 않습니다. 경영진은 알고 싶어합니다: 높은 의도로 표시된 사용자에게만 광고를 보여준다면, 그 광고비 중 실제로 전환할 사람에게 도달하는 비율은 얼마나 되는가?

**To-do / 할 일:**
- [x] Compute the overall conversion rate and the conversion rate conditional on being flagged high-intent<br>전체 전환율과 높은 의도로 표시된 조건부 전환율 계산하기
- [x] Use Bayes' theorem to compute P(convert | flagged) and compare it to the overall rate<br>베이즈 정리로 P(전환 \| 표시됨)을 계산하고 전체 비율과 비교하기
- [x] Quantify the targeting lift and the remaining wasted spend<br>타겟팅 개선 배율과 남은 낭비 광고비 정량화하기
- [x] Write a recommendation for the ad team<br>광고팀을 위한 제안 작성하기

````python
# 100,000 users, 4% overall conversion rate. The "high-intent" signal isn't
# perfect: it correctly identifies 70% of actual converters, but also
# flags 15% of non-converters.
n_users = 100000
p_convert = 0.04
p_flagged_given_convert = 0.70
p_flagged_given_not_convert = 0.15

p_not_convert = 1 - p_convert
p_flagged = (p_flagged_given_convert * p_convert) + (p_flagged_given_not_convert * p_not_convert)
p_convert_given_flagged = (p_flagged_given_convert * p_convert) / p_flagged

n_flagged = n_users * p_flagged
n_flagged_who_convert = n_users * p_flagged_given_convert * p_convert
lift = p_convert_given_flagged / p_convert

print(f"Overall conversion rate: {p_convert:.1%}")
print(f"Conversion rate among flagged users: {p_convert_given_flagged:.1%}")
print(f"Lift from targeting: {lift:.1f}x")
print()
print(f"Users flagged as high-intent: {n_flagged:.0f} ({p_flagged:.1%} of all users)")
print(f"Of those, expected converters: {n_flagged_who_convert:.0f}")
print(f"So {(1 - p_convert_given_flagged) * 100:.0f}% of ad spend on flagged users still goes to")
print(f"people who won't convert -- but that's still a {lift:.1f}x improvement over targeting everyone.")
print()
print(f"전체 전환율: {p_convert:.1%}")
print(f"표시된 사용자 중 전환율: {p_convert_given_flagged:.1%}")
print(f"타겟팅으로 인한 개선: {lift:.1f}배")
print(f"높은 의도로 표시된 사용자: {n_flagged:.0f}명 (전체의 {p_flagged:.1%})")
print(f"그중 예상 전환자: {n_flagged_who_convert:.0f}명")
print(f"즉 표시된 사용자 광고비의 {(1 - p_convert_given_flagged) * 100:.0f}%는 여전히 전환하지")
print(f"않을 사람에게 가지만, 그래도 전체 타겟팅 대비 {lift:.1f}배 개선입니다.")
````

# 📝 Summary (요약)
*Write today's concept in 3~5 sentences.*

**EN:** Probability quantifies uncertainty using a small set of rules: the addition rule (with a subtraction for overlap) combines "or" events, the multiplication rule (valid only for independent events) combines "and" events, and conditional probability, P(A\|B), narrows the sample space to only the cases where B happened. Bayes' theorem flips a conditional probability around — from P(evidence\|cause) to P(cause\|evidence) — and reveals the base-rate fallacy: even a highly accurate test on a rare event produces mostly false positives, because the sheer number of non-events dwarfs the number of true events. Classic brain-teasers like Monty Hall (switching doubles your odds because the host's choice isn't random) and the birthday paradox (23 people is enough because you're comparing pairs, not individuals) show up in interviews specifically to test whether you reach for a structured approach — like simulation — instead of trusting gut intuition.

**KR:** 확률은 작은 규칙 집합으로 불확실성을 정량화합니다: 덧셈 법칙(겹침은 빼줌)은 "또는" 사건을 결합하고, 곱셈 법칙(독립일 때만 성립)은 "그리고" 사건을 결합하며, 조건부확률 P(A\|B)는 표본공간을 B가 일어난 경우로만 좁힙니다. 베이즈 정리는 조건부확률의 방향을 뒤집어 P(증거\|원인)에서 P(원인\|증거)로 바꾸며, 기저율 오류를 드러냅니다: 드문 사건에 대해서는 아무리 정확한 검사라도 대부분 오탐이 나오는데, 정상 사례의 절대 수가 실제 사건보다 압도적으로 많기 때문입니다. 몬티홀(진행자의 선택이 무작위가 아니므로 바꾸면 확률이 두 배)과 생일 역설(개인이 아니라 쌍을 비교하므로 23명이면 충분함) 같은 고전 문제는 면접에서 암기가 아니라 직관 대신 구조적 접근(시뮬레이션 등)을 하는지 보기 위해 등장합니다.

**Q1 (EN).** What's the difference between the addition rule for mutually exclusive events and the general addition rule, and when do you need the general version?  
**Q1 (KR).** 배반사건의 덧셈 법칙과 일반 덧셈 법칙의 차이는 무엇이며, 언제 일반 버전이 필요한가요?

Mutually exclusive events use P(A or B) = P(A) + P(B) because both cannot happen. The general rule subtracts the overlap, P(A) + P(B) - P(A and B), whenever both can happen.  
배반사건은 동시에 일어날 수 없어서 P(A 또는 B) = P(A) + P(B)입니다. 둘 다 일어날 수 있으면 겹친 부분을 빼는 P(A) + P(B) - P(A 그리고 B)를 씁니다.

The denominators are different groups: fraud transactions versus flagged transactions. Fraud is so rare that most flags are false alarms.  
분모가 다릅니다. 한쪽은 사기 거래이고, 다른 쪽은 표시된 거래입니다. 사기가 매우 드물어 표시된 거래의 대부분이 오탐입니다.

The condition is extremely rare, so a 99% accurate test can still produce mostly false positives. Bayes' theorem is how you calculate that.  
그 상태가 매우 드물어, 99% 정확한 검사도 양성 대부분이 오탐일 수 있습니다. 베이즈 정리는 그 확률을 계산하는 방법입니다.
