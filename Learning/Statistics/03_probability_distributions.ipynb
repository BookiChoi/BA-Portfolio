# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** A probability distribution describes all the possible values a random variable can take, and how likely each one is. Some business quantities follow well-known, mathematically precise shapes — the Normal distribution for continuous measurements clustered around an average, the Binomial for counting successes out of a fixed number of trials, and the Poisson for counting rare events over a period of time. Knowing which shape your data follows tells you exactly which formulas and tests apply to it.

**KR:** 확률분포는 확률변수가 취할 수 있는 모든 가능한 값과 각각의 가능성을 설명합니다. 일부 비즈니스 지표는 잘 알려진, 수학적으로 정확한 모양을 따릅니다 -- 평균 주변에 몰린 연속형 측정값에는 정규분포, 고정된 시행 횟수 중 성공 횟수를 셀 때는 이항분포, 일정 기간 동안 드문 사건의 횟수를 셀 때는 포아송분포. 데이터가 어떤 모양을 따르는지 알면 정확히 어떤 공식과 검정이 적용되는지 알 수 있습니다.

### The three core distributions (핵심 분포 세 가지)

| Distribution / 분포 | Models / 모델링 대상 | Parameters / 파라미터 | Example / 예시 |
|---|---|---|---|
| Normal (정규분포) | Continuous values clustered around a mean<br>평균 주변에 몰린 연속형 값 | mean (μ), std dev (σ) | Order amounts, session lengths<br>주문금액, 세션 시간 |
| Binomial (이항분포) | Number of successes in n independent trials<br>n번의 독립 시행 중 성공 횟수 | n (trials/시행 수), p (success probability/성공확률) | Conversions out of visitors<br>방문자 중 전환 |
| Poisson (포아송분포) | Number of rare events in a fixed interval<br>고정된 구간 동안 드문 사건의 횟수 | λ (average rate/평균 발생률) | Support tickets per hour<br>시간당 지원 티켓 |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** Once you know a variable follows a known distribution, you can answer precise probability questions without collecting more data — "what's the probability a random customer spends over $100?", "what's the probability we get more than 5 support tickets in the next hour?", "is 15 conversions out of 100 trials unusually high, or just normal variation?" All of these have exact mathematical answers if you know the right distribution.

**KR:** 어떤 변수가 알려진 분포를 따른다는 것을 알면 데이터를 더 모으지 않고도 정밀한 확률 질문에 답할 수 있습니다 -- "무작위 고객이 $100 이상 쓸 확률은?", "다음 한 시간 안에 지원 티켓이 5건 넘게 들어올 확률은?", "100번 시도 중 15번 전환은 특이하게 높은가 그냥 정상적인 변동인가?" 올바른 분포를 알면 이 모든 질문에 정확한 수학적 답이 있습니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Modeling customer metrics that cluster around an average (Normal), A/B test conversion counts (Binomial), call-center/support-ticket volume forecasting (Poisson), and — through the Central Limit Theorem — justifying why nearly every hypothesis test and confidence interval in this curriculum is allowed to assume normality even when the raw data isn't normal.

**KR:** 평균 주변에 몰린 고객 지표 모델링(정규분포), A/B 테스트 전환 횟수(이항분포), 콜센터·지원 티켓 물량 예측(포아송분포), 그리고 중심극한정리를 통해 이 커리큘럼의 거의 모든 가설검정과 신뢰구간이 원본 데이터가 정규분포가 아니어도 정규성을 가정할 수 있는 이유를 정당화합니다.

# 📝 Syntax (문법)

## Basic Syntax (기본 문법)

````python
import numpy as np
from scipy import stats

# Normal distribution: order amounts averaging $50, std dev $12
# 정규분포: 평균 $50, 표준편차 $12인 주문 금액
mean, std = 50, 12
order_dist = stats.norm(loc=mean, scale=std)

# probability a random order is under $40
p_under_40 = order_dist.cdf(40)
print(f"P(order < $40) = {p_under_40:.3f} / P(주문 < $40) = {p_under_40:.3f}")

# z-score: how many std devs is $65 from the mean?
z = (65 - mean) / std
print(f"z-score for $65: {z:.2f}")
````

## Common Variations (자주 쓰는 변형)

````python
# Empirical rule (68-95-99.7): what fraction of orders fall within 1/2/3 std devs?
# 경험적 규칙(68-95-99.7): 주문의 몇 %가 표준편차 1/2/3배 안에 있는가?
for k in [1, 2, 3]:
    p_within = order_dist.cdf(mean + k * std) - order_dist.cdf(mean - k * std)
    print(f"P(within {k} std dev) = {p_within:.4f}")

# .ppf() is the inverse of .cdf(): given a probability, find the value
# .ppf()는 .cdf()의 역함수: 확률이 주어지면 그에 해당하는 값을 찾음
value_95th = order_dist.ppf(0.95)
print(f"\n95th percentile order amount: ${value_95th:.2f}")
````

# 🧪 Small Examples (예제)

## Example 1 — Normal Distribution: z-scores & the Empirical Rule (z-점수와 경험적 규칙)

````python
# Business question: a customer places an $86 order. Given that typical orders
# average $50 with a std dev of $12, is this unusually high?
# 비즈니스 질문: 고객이 $86짜리 주문을 했습니다. 평균 주문금액이 $50,
# 표준편차가 $12일 때, 이는 특이하게 높은가?

mean, std = 50, 12
order_dist = stats.norm(loc=mean, scale=std)

z_score = (86 - mean) / std
percentile = order_dist.cdf(86)

print(f"z-score: {z_score:.2f}")
print(f"This order is at the {percentile * 100:.2f}th percentile")
print(f"이 주문은 백분위 {percentile * 100:.2f}에 해당합니다 (상위 {(1 - percentile) * 100:.2f}%)")
print()
print("z=3.0 means 3 standard deviations above the mean -- under the empirical rule,")
print("about 99.7% of orders fall within 3 std devs, so this order is genuinely rare.")
print("z=3.0은 평균보다 표준편차 3배 위를 의미합니다 -- 경험적 규칙에 따르면 약")
print("99.7%의 주문이 표준편차 3배 이내에 있으므로, 이 주문은 정말로 드뭅니다.")
````

## Example 2 — Binomial Distribution: A/B Test Conversion Counts (A/B 테스트 전환 횟수)

````python
# Business question: if the true conversion rate is 10%, how likely is it to
# see 15 OR MORE conversions out of 100 visitors just by random chance?
# 비즈니스 질문: 실제 전환율이 10%라면, 100명 중 15명 이상이 우연히
# 전환할 확률은 얼마나 되는가?

n_trials = 100
p_success = 0.10
binom_dist = stats.binom(n=n_trials, p=p_success)

p_exactly_15 = binom_dist.pmf(15)
p_15_or_more = 1 - binom_dist.cdf(14)  # P(X >= 15) = 1 - P(X <= 14)
expected_conversions = binom_dist.mean()
std_conversions = binom_dist.std()

print(f"Expected conversions: {expected_conversions:.1f} (std dev: {std_conversions:.2f})")
print(f"P(exactly 15 conversions) = {p_exactly_15:.4f}")
print(f"P(15 or more conversions) = {p_15_or_more:.4f}")
print()
print(f"Seeing 15+ conversions happens by chance about {p_15_or_more * 100:.1f}% of the time")
print("even with NO real improvement -- this is exactly the logic behind A/B test p-values.")
print("실제 개선이 전혀 없어도 우연히 15명 이상 전환할 확률이 약")
print(f"{p_15_or_more * 100:.1f}%입니다 -- 이것이 바로 A/B 테스트 p-value의 논리입니다.")
````

## Example 3 — Poisson Distribution: Support Ticket Volume (지원 티켓 물량)

````python
# Business question: the support team averages 4 tickets per hour. What's the
# probability of a "slammed" hour with 8 or more tickets? How many tickets
# should staffing plan for to cover 95% of hours?
# 비즈니스 질문: 지원팀은 시간당 평균 4건의 티켓을 받습니다. 8건 이상
# 몰리는 "바쁜" 시간대의 확률은? 95%의 시간대를 커버하려면 몇 건을
# 기준으로 인력을 계획해야 하는가?

avg_rate = 4
ticket_dist = stats.poisson(mu=avg_rate)

p_8_or_more = 1 - ticket_dist.cdf(7)  # P(X >= 8) = 1 - P(X <= 7)
staffing_threshold = ticket_dist.ppf(0.95)

print(f"P(8 or more tickets in an hour) = {p_8_or_more:.4f}")
print(f"To cover 95% of hours, staff for {staffing_threshold:.0f} tickets/hour")
print()
print(f"약 {p_8_or_more * 100:.1f}%의 시간대는 8건 이상의 '바쁜' 시간대가 됩니다.")
print(f"95%의 시간대를 커버하려면 시간당 {staffing_threshold:.0f}건 기준으로")
print("인력을 계획해야 합니다.")
````

## Example 4 — Central Limit Theorem: Why Sample Means Become Normal (표본평균이 정규분포가 되는 이유)

````python
# Business question: individual order amounts are heavily right-skewed (NOT
# normal at all) -- does that break every statistical test that assumes
# normality? Simulate to find out.
# 비즈니스 질문: 개별 주문 금액은 심하게 오른쪽으로 치우쳐 있습니다(전혀
# 정규분포가 아님) -- 이것이 정규성을 가정하는 모든 통계 검정을 무너뜨리는가?
# 시뮬레이션으로 확인.

np.random.seed(42)
population = np.random.exponential(scale=50, size=100000)  # heavily skewed

sample_means_5 = [np.random.choice(population, size=5, replace=False).mean() for _ in range(2000)]
sample_means_30 = [np.random.choice(population, size=30, replace=False).mean() for _ in range(2000)]

print(f"Population skewness: {stats.skew(population):.2f} (very skewed)")
print(f"Skewness of means from samples of size 5:  {stats.skew(sample_means_5):.2f}")
print(f"Skewness of means from samples of size 30: {stats.skew(sample_means_30):.2f}")
print()
print("Even though individual orders are heavily skewed, the AVERAGE of just 30 orders")
print("is far closer to symmetric (normal) -- this is the Central Limit Theorem in action,")
print("and it's why t-tests and confidence intervals still work on skewed business data.")
print("개별 주문은 심하게 치우쳐 있지만, 주문 30개의 평균은 훨씬 더 대칭(정규분포)에")
print("가깝습니다 -- 이것이 중심극한정리의 실제 작동이며, t-검정과 신뢰구간이")
print("치우친 비즈니스 데이터에서도 여전히 작동하는 이유입니다.")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Assuming Raw Data Must Be Normal Before Any Test (검정 전에 원본 데이터가 정규분포여야 한다고 착각)
Many t-tests and confidence intervals only require the SAMPLE MEAN to be approximately normal, not the raw individual data points — thanks to the CLT, this is usually true even when the raw data is heavily skewed (see Example 4).  
많은 t-검정과 신뢰구간은 원본 개별 데이터가 아니라 표본평균이 대략 정규분포이기만 하면 됩니다 -- CLT 덕분에 원본 데이터가 심하게 치우쳐 있어도 이는 보통 성립합니다(예제 4 참고).

✅ **Fix / 해결:** Check the distribution of the SAMPLE MEAN (or trust the CLT for n≥30ish), not the raw data's shape, before deciding a test is invalid.

### Mistake 2 — Mixing Up Binomial's n and p (이항분포의 n과 p 혼동)
`n` is the number of trials (e.g., 100 visitors), `p` is the probability of success on ONE trial (e.g., 10% conversion rate) — swapping them into `stats.binom(n=0.10, p=100)` produces nonsense or an error.  
n은 시행 횟수(예: 방문자 100명), p는 한 번의 시행에서 성공할 확률(예: 전환율 10%)입니다 -- 이를 뒤바꿔 stats.binom(n=0.10, p=100)처럼 쓰면 말이 안 되거나 에러가 납니다.

✅ **Fix / 해결:** n is always a whole-number count, p is always between 0 and 1 — check both before running the calculation.

### Mistake 3 — Using Binomial When Trials Aren't Independent (시행이 독립적이지 않을 때 이항분포 사용)
The binomial distribution assumes each trial's outcome doesn't affect the others — but if one customer's purchase influences their friends' purchases (a referral or viral effect), the independence assumption breaks and the binomial's variance estimate becomes too small.  
이항분포는 각 시행의 결과가 다른 시행에 영향을 주지 않는다고 가정합니다 -- 하지만 한 고객의 구매가 친구들의 구매에 영향을 준다면(추천이나 바이럴 효과), 독립성 가정이 깨지고 이항분포의 분산 추정치가 너무 작아집니다.

✅ **Fix / 해결:** Before using the binomial, confirm trials are genuinely independent — social/network effects are a common violation in consumer data.

### Mistake 4 — Assuming a Constant Rate for Poisson (포아송분포에 일정한 발생률 가정)
Poisson assumes events occur at a constant average rate — but support tickets spike during outages, and website traffic spikes during sales, so a single λ estimated from "normal" hours will badly underestimate the probability of a spike.  
포아송분포는 사건이 일정한 평균 발생률로 일어난다고 가정합니다 -- 하지만 지원 티켓은 장애 중에, 웹사이트 트래픽은 세일 중에 급증하므로, "평범한" 시간대에서 추정한 하나의 λ는 급증 확률을 심하게 과소추정합니다.

✅ **Fix / 해결:** Check whether the rate truly stays constant across the period you're modeling, or split the data into separate periods (normal vs. incident) with their own λ.

### Mistake 5 — Reading a z-score Directly as a Percentage (z-점수를 그대로 퍼센트로 읽기)
A z-score of 2.0 does NOT mean "98%" — it means "2 standard deviations above the mean," which corresponds to roughly the 97.7th percentile only after converting through the CDF.  
z-점수 2.0은 "98%"를 의미하지 않습니다 -- "평균보다 표준편차 2배 위"를 의미하며, CDF를 거쳐야 대략 97.7번째 백분위수에 해당한다는 것을 알 수 있습니다.

✅ **Fix / 해결:** Always convert a z-score to a percentile with `.cdf()` before reporting it as a probability or percentage.

### Mistake 6 — Treating n=30 as a Magic, Universal CLT Cutoff (n=30을 만능 CLT 기준으로 취급)
"n≥30 is enough for the CLT" is a common rule of thumb, not a law — a population that's extremely skewed or has extreme outliers may need a much larger sample before its mean looks normal.  
"n≥30이면 CLT에 충분하다"는 흔한 경험칙이지 법칙이 아닙니다 -- 극도로 치우쳐 있거나 극단적인 이상치가 있는 모집단은 평균이 정규분포처럼 보이려면 훨씬 큰 표본이 필요할 수 있습니다.

✅ **Fix / 해결:** When the underlying data is severely skewed or has extreme outliers, check the sample-mean distribution directly (like Example 4 does) instead of assuming n=30 is automatically enough.

# 💡 Tips (팁)
Useful tips or shortcuts / 유용한 팁과 단축법

- `scipy.stats` distributions all share the same four methods: `.pdf()`/`.pmf()` (exact probability/density), `.cdf()` (probability of X or less), `.ppf()` (inverse of cdf — find the value at a percentile), and `.rvs()` (generate random samples) — learn these once, reuse them for every distribution.<br>scipy.stats 분포들은 모두 같은 네 가지 메서드를 공유합니다: .pdf()/.pmf()(정확한 확률·밀도), .cdf()(X 이하일 확률), .ppf()(cdf의 역함수 -- 백분위수에 해당하는 값 찾기), .rvs()(무작위 표본 생성) -- 한 번 배우면 모든 분포에 재사용할 수 있습니다.
- Binomial with a large n and small p starts to look like Poisson — if you ever see both give nearly the same answer, that's not a coincidence, it's a known mathematical relationship.<br>n이 크고 p가 작은 이항분포는 포아송분포처럼 보이기 시작합니다 -- 둘이 거의 같은 답을 준다면 우연이 아니라 잘 알려진 수학적 관계입니다.
- The empirical rule (68-95-99.7) is a fast mental-math shortcut for the Normal distribution specifically — don't apply it to skewed, binomial, or Poisson data.<br>경험적 규칙(68-95-99.7)은 정규분포에 특화된 빠른 암산 지름길입니다 -- 치우친 데이터나 이항·포아송 데이터에는 적용하지 마세요.
- When in doubt about which distribution fits your data, plot a histogram first (Visualization Section 5) — the shape usually makes the right distribution obvious before you do any math.<br>어떤 분포가 데이터에 맞는지 헷갈릴 때는 먼저 히스토그램을 그려보세요(시각화 커리큘럼 섹션 5) -- 계산을 하기도 전에 모양만으로 맞는 분포가 대체로 명확해집니다.
- The Central Limit Theorem is the single most important idea in this section for everything that follows — Section 4's confidence intervals and Section 7's t-tests both lean on it directly.<br>중심극한정리는 이 섹션에서 이후 모든 내용에 가장 중요한 아이디어입니다 -- 섹션 4의 신뢰구간과 섹션 7의 t-검정 모두 이를 직접 활용합니다.

# 💼 Business Example (비즈니스 예시)
*How would a Business Analyst use this?*

**Scenario / 시나리오:**

**EN:** Your customer support team needs to decide staffing for a new live-chat feature. Historical data suggests chat volume follows a Poisson pattern, and each chat's response time is roughly Normal. Leadership wants a staffing plan that handles 95% of hours without a backlog, and wants to know how often SLA misses will happen even when properly staffed.

**KR:** 고객지원팀이 새로운 실시간 채팅 기능의 인력을 결정해야 합니다. 과거 데이터로는 채팅량이 포아송 패턴을 따르고, 채팅별 응답 시간은 대략 정규분포를 따릅니다. 경영진은 95%의 시간대를 백로그 없이 처리할 수 있는 인력 계획을 원하며, 인력을 제대로 배치해도 SLA를 놓치는 경우가 얼마나 자주 발생하는지 알고 싶어합니다.

**To-do / 할 일:**
- [x] Model chat volume with a Poisson distribution using the historical average rate<br>과거 평균 발생률로 포아송분포를 사용해 채팅량 모델링하기
- [x] Find the staffing level that covers 95% of hours<br>95%의 시간대를 커버하는 인력 수준 찾기
- [x] Model response time with a Normal distribution and estimate the SLA-miss rate<br>정규분포로 응답 시간을 모델링하고 SLA 초과율 추정하기
- [x] Combine both to estimate expected SLA misses in a well-staffed hour<br>둘을 결합해 인력이 충분한 시간대에도 예상되는 SLA 초과 건수 추정하기

````python
avg_chats_per_hour = 12
chat_dist = stats.poisson(mu=avg_chats_per_hour)

staffing_level = chat_dist.ppf(0.95)
p_crisis = 1 - chat_dist.cdf(19)  # P(20+ chats in an hour, a "crisis" level)

print(f"Average chat volume: {avg_chats_per_hour} chats/hour")
print(f"Staffing level to cover 95% of hours: {staffing_level:.0f} chats/hour")
print(f"P(20+ chats in an hour, a 'crisis' level) = {p_crisis:.4f} ({p_crisis * 100:.2f}%)")
print()
print(f"평균 채팅량: 시간당 {avg_chats_per_hour}건")
print(f"95%의 시간대를 커버하는 인력 수준: 시간당 {staffing_level:.0f}건")
print(f"P(시간당 20건 이상, '위기' 수준) = {p_crisis:.4f} ({p_crisis * 100:.2f}%)")

# Response time: average 4 minutes, std dev 1.5 minutes; SLA target is 3 minutes
resp_dist = stats.norm(loc=4, scale=1.5)
p_over_sla = 1 - resp_dist.cdf(3)
expected_missed = staffing_level * p_over_sla

print()
print(f"P(a single chat response exceeds the 3-min SLA target) = {p_over_sla:.3f}")
print(f"At {staffing_level:.0f} chats/hour, expect about {expected_missed:.1f} SLA misses")
print("even during a well-staffed hour -- worth flagging to leadership as an expected")
print("baseline, not a sign something is broken.")
print()
print(f"단일 채팅 응답이 3분 SLA 목표를 초과할 확률 = {p_over_sla:.3f}")
print(f"시간당 {staffing_level:.0f}건 기준, 인력이 충분한 시간대에도 약")
print(f"{expected_missed:.1f}건의 SLA 초과가 예상됩니다 -- 문제가 있다는 신호가 아니라")
print("예상되는 기준선으로 경영진에게 알릴 가치가 있습니다.")
````

# 📝 Summary (요약)
*Write today's concept in 3~5 sentences.*

**EN:** Three distributions cover most business scenarios: the Normal distribution models continuous values clustered around a mean (with z-scores measuring distance from that mean in standard-deviation units, and the empirical rule giving quick 68-95-99.7 benchmarks), the Binomial distribution counts successes across a fixed number of independent trials with a constant success probability (exactly the shape of an A/B test's conversion counts), and the Poisson distribution counts rare events over a fixed interval at a constant average rate (ideal for forecasting support-ticket or server-error volume). The Central Limit Theorem ties all of this together: no matter how skewed the underlying population is, the distribution of a SAMPLE MEAN approaches Normal as sample size grows — the mathematical reason confidence intervals and t-tests, covered starting next section, are allowed to assume normality even on messy real-world business data.

**KR:** 세 가지 분포가 대부분의 비즈니스 시나리오를 다룹니다: 정규분포는 평균 주변에 몰린 연속형 값을 모델링하고(z-점수는 표준편차 단위로 평균으로부터의 거리를 측정하며, 경험적 규칙은 68-95-99.7 빠른 기준을 제공), 이항분포는 고정된 횟수의 독립 시행에서 일정한 성공 확률로 성공 횟수를 세며(A/B 테스트의 전환 횟수와 정확히 같은 모양), 포아송분포는 일정한 평균 발생률로 고정된 구간 동안 드문 사건의 횟수를 셉니다(지원 티켓·서버 오류 물량 예측에 이상적). 중심극한정리는 이 모든 것을 하나로 묶습니다: 기저 모집단이 아무리 치우쳐 있어도, 표본평균의 분포는 표본 크기가 커질수록 정규분포에 가까워집니다 -- 이것이 다음 섹션부터 다룰 신뢰구간과 t-검정이 지저분한 실제 비즈니스 데이터에서도 정규성을 가정할 수 있는 수학적 이유입니다.

Today's topic in ONE sentence. / 오늘의 주제를 한 문장으로.

> **EN:** Normal, Binomial, and Poisson distributions each model a different kind of business randomness (continuous averages, success counts, rare-event counts), and the Central Limit Theorem guarantees that sample MEANS look Normal regardless of the shape of the underlying data — the single fact that makes every confidence interval and hypothesis test in the rest of this curriculum possible.

> **KR:** 정규·이항·포아송 분포는 각각 다른 종류의 비즈니스 무작위성(연속형 평균, 성공 횟수, 드문 사건 횟수)을 모델링하며, 중심극한정리는 기저 데이터의 모양과 상관없이 표본평균이 정규분포처럼 보인다는 것을 보장합니다 -- 이 하나의 사실이 이 커리큘럼 나머지의 모든 신뢰구간과 가설검정을 가능하게 합니다.

# ❓ Review Questions (복습 질문)

**Q1 (EN).** A customer's order has a z-score of 2.5. What does that number mean, and what additional step do you need to turn it into a percentile?  
**Q1 (KR).** 어떤 고객의 주문 z-점수가 2.5입니다. 이 숫자는 무엇을 의미하며, 백분위수로 바꾸려면 어떤 추가 단계가 필요한가요?

A z-score of 2.5 means the order is 2.5 standard deviations above the mean. To turn that into a percentile, run it through the cumulative probability (the CDF).  
z-점수 2.5는 그 주문이 평균보다 표준편차 2.5걸음 위라는 뜻입니다. 백분위수로 바꾸려면 그 걸음 수를 누적 확률(CDF)에 통과시켜야 합니다.

**Q2 (EN).** What two conditions does the Binomial distribution require that the Poisson distribution doesn't?  
**Q2 (KR).** 이항분포는 포아송분포가 요구하지 않는 어떤 두 가지 조건을 요구하나요?

The binomial needs a fixed number of trials, and a success probability on each trial.  
이항분포는 미리 정해진 시행 횟수와, 각 시행의 성공 확률이 필요합니다.

**Q3 (EN).** Why might using a single Poisson rate (λ) to model support tickets across an entire day, including both quiet and busy periods, give misleading results?  
**Q3 (KR).** 조용한 시간대와 바쁜 시간대를 모두 포함하는 하루 전체의 지원 티켓을 하나의 포아송 발생률(λ)로 모델링하면 왜 오해를 부르는 결과가 나올 수 있나요?

One rate averaged over the whole day makes quiet hours look busier than they are and busy hours look quieter, so the chance of a spike is understated.  
하루 전체를 평균 낸 하나의 발생률은 조용한 시간대를 실제보다 바쁘게, 바쁜 시간대를 실제보다 한가하게 보이게 해서, 몰리는 시간의 확률을 작게 봅니다.

**Q4 (EN).** Why does the Central Limit Theorem let you use a t-test's normality-based formula even when your raw sales data is heavily right-skewed?  
**Q4 (KR).** 원본 매출 데이터가 심하게 오른쪽으로 치우쳐 있어도 중심극한정리가 정규성 기반의 t-검정 공식을 사용할 수 있게 해주는 이유는 무엇인가요?

The t-test applies the normal curve to the sample mean, and the Central Limit Theorem says that mean becomes approximately normal as more sales are averaged, even when each individual sale is right-skewed.  
t-검정은 종 모양을 표본평균에 적용합니다. 중심극한정리는 매출 한 건이 오른쪽으로 치우쳐 있어도, 더 많이 평균 낼수록 그 평균이 종 모양에 가까워진다고 말합니다.

**Q5 (EN).** Is n≥30 a strict guarantee that a sample mean will look normal? What kind of population might need a larger sample?  
**Q5 (KR).** n≥30이면 표본평균이 정규분포처럼 보인다는 것이 절대적으로 보장되나요? 어떤 종류의 모집단이 더 큰 표본을 필요로 할 수 있나요?

No. n≥30 is only a rule of thumb. A population that is extremely skewed, or that has extreme outliers, can need a larger sample before its mean looks normal.
절대적으로 보장되지는 않습니다. n≥30은 경험적인 기준일 뿐입니다. 극도로 치우쳐 있거나 극단적인 이상치가 있는 모집단은 평균이 종 모양으로 보이려면 더 큰 표본이 필요할 수 있습니다.
