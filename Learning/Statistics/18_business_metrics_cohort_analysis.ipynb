# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** Business metrics translate the statistical toolkit from Sections 1-17 into the specific numbers a subscription or e-commerce business lives and dies by: how many users stick around (retention/cohort analysis), how much each user is worth over their lifetime (LTV), and whether it costs less to acquire them than they're worth (LTV:CAC).

**KR:** 비즈니스 지표는 섹션 1~17의 통계 도구를 구독형·이커머스 비즈니스가 사활을 거는 구체적인 숫자로 번역합니다: 얼마나 많은 사용자가 남아있는지(리텐션·코호트 분석), 각 사용자가 생애 동안 얼마의 가치를 갖는지(LTV), 그리고 그들을 확보하는 비용이 그 가치보다 적은지(LTV:CAC).

### Key definitions (핵심 정의)

| Metric / 지표 | Formula / 공식 | Question it answers / 답하는 질문 |
|---|---|---|
| N-day retention<br>N일 리텐션 | users active on day N / cohort size<br>N일차 활성 사용자 / 코호트 크기 | Are users still here after N days?<br>N일 후에도 사용자가 남아있는가? |
| Churn rate / 이탈률 | 1 - retention rate<br>1 - 리텐션율 | What fraction left?<br>얼마나 이탈했는가? |
| LTV | ARPU × avg. lifespan (or ARPU / churn rate)<br>ARPU × 평균 생존기간(또는 ARPU / 이탈률) | What's a customer worth, total?<br>고객의 총 가치는 얼마인가? |
| LTV:CAC | LTV / CAC | Is this customer worth what we paid?<br>지불한 비용만큼 가치가 있는가? |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** A raw user count going up says almost nothing on its own — a company could be growing while quietly leaking most new users within a month, or spending more to acquire a customer than that customer will ever generate. Cohort analysis, retention, LTV, and CAC together are what separate "we have more users" from "we have a sustainable business."

**KR:** 늘어나는 사용자 수 자체는 거의 아무것도 말해주지 않습니다 -- 회사는 성장하는 동시에 조용히 대부분의 신규 사용자를 한 달 안에 잃고 있을 수도 있고, 고객이 평생 만들어낼 가치보다 더 많은 비용을 들여 확보하고 있을 수도 있습니다. 코호트 분석, 리텐션, LTV, CAC를 함께 보는 것이 "사용자가 늘었다"와 "지속 가능한 비즈니스다"를 가릅니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Nearly every recurring analytics review at a subscription or marketplace company: monthly retention dashboards, LTV:CAC health checks before a fundraising round, and cohort tables that reveal whether a product change (perhaps validated by an A/B test in Part 4) actually improved how long users stick around.

**KR:** 구독형·마켓플레이스 회사의 거의 모든 정기 분석 리뷰: 월간 리텐션 대시보드, 자금 조달 라운드 전 LTV:CAC 건전성 점검, 그리고 제품 변경(Part 4의 A/B 테스트로 검증됐을 수 있는)이 실제로 사용자가 머무는 기간을 개선했는지 드러내는 코호트 표.

## Basic Syntax (기본 문법)

````python
import numpy as np
import pandas as pd

# A cohort's retention rate on a given day: active users / cohort size
# 특정 날짜의 코호트 리텐션율: 활성 사용자 / 코호트 크기
cohort_size = 1000
active_on_day_7 = 620

day7_retention = active_on_day_7 / cohort_size
churn_rate = 1 - day7_retention

print(f"Day 7 retention: {day7_retention:.1%}")
print(f"Day 7 churn rate: {churn_rate:.1%}")
````

# 🧪 Small Examples (예제)

## Example 1 — Cohort Analysis: Building a Retention Curve (리텐션 곡선 만들기)

````python
# Business question: build a cohort retention table showing what % of each
# monthly signup cohort is still active in subsequent months.
# 비즈니스 질문: 매달 가입 코호트의 몇 %가 이후 달에도 계속 활성 상태인지
# 보여주는 코호트 리텐션 표를 만들기.

np.random.seed(1)
cohort_months = ['2026-01', '2026-02', '2026-03', '2026-04']
cohort_sizes = {'2026-01': 1000, '2026-02': 1200, '2026-03': 900, '2026-04': 1100}
monthly_retention_decay = 0.75  # each month retains ~75% of the PREVIOUS month's active users

records = []
for cohort in cohort_months:
    size = cohort_sizes[cohort]
    cohort_idx = cohort_months.index(cohort)
    for months_since in range(len(cohort_months) - cohort_idx):
        if months_since == 0:
            active_this_month = size
        else:
            active_this_month = int(size * (monthly_retention_decay ** months_since))
        records.append({'cohort': cohort, 'months_since_signup': months_since,
                         'active_users': active_this_month})

df_cohort = pd.DataFrame(records)
retention_table = df_cohort.pivot_table(index='cohort', columns='months_since_signup', values='active_users')

# Convert to retention PERCENTAGE (relative to month 0)
retention_pct = retention_table.div(retention_table[0], axis=0) * 100
print("Cohort retention table (% of cohort still active):")
print(retention_pct.round(1))
print()
print("The triangular shape (more NaN toward the bottom-right) is standard -- newer")
print("cohorts simply haven't had time to reach later months yet. Reading DOWN a")
print("column compares cohorts at the SAME age; reading ACROSS a row shows one")
print("cohort's decay over time.")
print("삼각형 모양(오른쪽 아래로 갈수록 NaN 증가)은 표준적입니다 -- 더 최근 코호트는")
print("아직 나중 달에 도달할 시간이 없었을 뿐입니다. 열을 아래로 읽으면 같은 나이의")
print("코호트를 비교하고, 행을 가로로 읽으면 한 코호트의 시간에 따른 감소를 봅니다.")
````

## Example 2 — Retention & Churn Rate: Precise Definitions (정확한 정의)

````python
# Business question: from raw "days a user was last active" data, compute
# Day 1, Day 7, and Day 30 retention and churn rates.
# 비즈니스 질문: "사용자가 마지막으로 활성이었던 날"이라는 원본 데이터로부터
# 1일, 7일, 30일 리텐션과 이탈률을 계산.

np.random.seed(2)
n_users = 2000
# time-to-churn is often exponentially distributed (Section 3's callback)
days_active = np.random.exponential(scale=25, size=n_users)

for day_n in [1, 7, 30]:
    still_active = (days_active >= day_n).mean()
    churn = 1 - still_active
    print(f"Day {day_n}: retention={still_active:.1%}, churn={churn:.1%}")

print()
print("Time-to-churn following an exponential distribution (Section 3) means a")
print("constant HAZARD rate -- users are equally likely to churn on any given day")
print("they're still active, regardless of how long they've already stuck around.")
print("이탈까지 걸리는 시간이 지수분포를 따른다는 것(섹션 3)은 일정한 위험률을")
print("의미합니다 -- 사용자는 이미 얼마나 오래 남아있었든 상관없이, 활성 상태인")
print("어느 날이든 이탈할 확률이 동일합니다.")
````

## Example 3 — LTV: Calculating from Retention/Churn and ARPU (리텐션·이탈률과 ARPU로 LTV 계산)

````python
# Business question: calculate a customer's lifetime value from their monthly
# churn rate and average revenue per user.
# 비즈니스 질문: 월간 이탈률과 사용자당 평균 매출로부터 고객 생애가치 계산.

monthly_churn = 0.05  # 5% of customers churn each month
arpu_monthly = 30      # average revenue per user per month

# Method 1: sum revenue over an assumed lifespan, month by month
retained_fraction = 1.0
ltv_summed = 0
for month in range(1, 61):  # sum over 5 years -- effectively "forever" given the decay
    retained_fraction *= (1 - monthly_churn)
    ltv_summed += arpu_monthly * retained_fraction

# Method 2: the shortcut formula -- LTV = ARPU / churn_rate
ltv_formula = arpu_monthly / monthly_churn

print(f"LTV (summed month-by-month over 5 years): ${ltv_summed:.2f}")
print(f"LTV (shortcut formula: ARPU / churn):      ${ltv_formula:.2f}")
print()
print("Both methods agree closely -- LTV is fundamentally a bet that the churn rate")
print("observed so far will hold steady into the future. If churn actually rises over")
print("time (common as easy-to-retain early adopters give way to a broader audience),")
print("this LTV estimate is too optimistic.")
print("두 방법이 거의 일치합니다 -- LTV는 근본적으로 지금까지 관측된 이탈률이")
print("미래에도 그대로 유지된다는 데 거는 내기입니다. 실제로 이탈률이 시간이")
print("지나며 오른다면(초기에 남기 쉬운 얼리어답터들이 더 폭넓은 사용자층으로")
print("바뀌면서 흔히 발생), 이 LTV 추정치는 너무 낙관적인 것입니다.")
````

## Example 4 — CAC and the LTV:CAC Ratio (CAC와 LTV:CAC 비율)

````python
# Business question: is this customer segment profitable to acquire? Compute
# the LTV:CAC ratio and payback period.
# 비즈니스 질문: 이 고객 세그먼트를 확보하는 것이 수익성이 있는가?
# LTV:CAC 비율과 회수 기간 계산.

ltv = 600
cac = 150
arpu_monthly = 30

ltv_cac_ratio = ltv / cac
payback_months = cac / arpu_monthly

print(f"LTV: ${ltv}, CAC: ${cac}")
print(f"LTV:CAC ratio: {ltv_cac_ratio:.1f}:1")
print(f"Payback period: {payback_months:.1f} months")
print()
if ltv_cac_ratio >= 3:
    print("A ratio of 3:1 or higher is a commonly cited healthy benchmark -- this")
    print("customer segment generates meaningfully more value than it costs to acquire.")
    print("3:1 이상의 비율은 흔히 인용되는 건전한 기준입니다 -- 이 고객 세그먼트는")
    print("확보 비용보다 의미 있게 더 많은 가치를 만들어냅니다.")
elif ltv_cac_ratio >= 1:
    print("Ratio is positive but below the common 3:1 benchmark -- profitable, but")
    print("with less margin for error (rising costs, retention dips) than ideal.")
else:
    print("Ratio below 1:1 -- this segment costs more to acquire than it's worth.")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Comparing Cohorts at Different Ages (다른 나이의 코호트를 직접 비교)
Comparing January's month-1 retention to March's month-6 retention isn't a fair comparison — always compare cohorts at the SAME age (reading down a column in the cohort table, not diagonally or across rows at different points).  
1월 코호트의 1개월차 리텐션을 3월 코호트의 6개월차 리텐션과 비교하는 것은 공정한 비교가 아닙니다 -- 항상 같은 나이의 코호트를 비교하세요(코호트 표에서 대각선이나 다른 지점의 행이 아니라 같은 열을 아래로 읽기).

✅ **Fix / 해결:** Only compare retention numbers within the same column (same months-since-signup) of the cohort table.

### Mistake 2 — Averaging Retention Across Cohorts Without Checking the Trend (추세 확인 없이 코호트 전체 평균 리텐션 사용)
A single "average retention" number can mask whether retention is improving or declining cohort-over-cohort — a flat average could hide newer cohorts retaining much better (or worse) than older ones.  
단일 "평균 리텐션" 숫자는 리텐션이 코호트마다 개선되는지 악화되는지를 숨길 수 있습니다 -- 평평한 평균이 더 최근 코호트가 예전보다 훨씬 더 잘(또는 더 나쁘게) 남아있다는 것을 가릴 수 있습니다.

✅ **Fix / 해결:** Always look at the cohort table itself (Example 1), not just a single blended average, to catch a trend across cohort vintages.

### Mistake 3 — Assuming Churn Rate Stays Constant Forever for LTV (LTV 계산에서 이탈률이 영원히 일정하다고 가정)
The LTV shortcut formula (ARPU / churn) assumes a constant hazard rate (Example 2's exponential assumption) — but churn is often HIGHER in a customer's early months and lower later, so a churn rate measured only from long-tenured survivors underestimates true LTV risk.  
LTV 단축 공식(ARPU / 이탈률)은 일정한 위험률(예제 2의 지수분포 가정)을 가정합니다 -- 하지만 이탈률은 종종 고객의 초기 몇 달 동안 더 높고 이후엔 낮아지므로, 오래 남은 생존자로부터만 측정한 이탈률은 실제 LTV 위험을 과소평가합니다.

✅ **Fix / 해결:** Check whether churn rate itself changes with customer tenure before trusting the constant-churn LTV shortcut; use the month-by-month cohort curve (Example 3's Method 1) when it doesn't.

### Mistake 4 — Computing CAC Without Separating Paid from Organic Acquisition (유료·오가닉 획득을 분리하지 않고 CAC 계산)
Dividing total marketing spend by ALL new users (including free organic signups that cost nothing to acquire) understates the true cost of PAID acquisition — the number that actually matters for an LTV:CAC decision about ad spend.  
전체 마케팅 지출을 모든 신규 사용자(획득 비용이 들지 않는 무료 오가닉 가입자 포함)로 나누면 유료 획득의 실제 비용을 과소평가합니다 -- 광고비에 대한 LTV:CAC 결정에서 실제로 중요한 숫자입니다.

✅ **Fix / 해결:** Calculate CAC separately for paid vs organic channels, and use the paid-channel CAC when evaluating whether ad spend is profitable.

### Mistake 5 — Treating LTV:CAC as a Fixed, One-Time Calculation (LTV:CAC를 고정된 일회성 계산으로 취급)
LTV:CAC changes whenever retention, pricing, or acquisition costs change — a healthy 3:1 ratio calculated a year ago can quietly become unhealthy if churn crept up or ad costs rose, without anyone recalculating.  
LTV:CAC는 리텐션, 가격, 획득 비용이 바뀔 때마다 변합니다 -- 1년 전에 계산한 건강한 3:1 비율은 아무도 다시 계산하지 않는 사이 이탈률이 슬금슬금 오르거나 광고비가 오르면 조용히 나빠질 수 있습니다.

✅ **Fix / 해결:** Recalculate LTV:CAC on a regular cadence (e.g. quarterly), not as a one-time historical fact.

### Mistake 6 — Confusing "N-Day Retention" with "N-Day Rolling Retention" (N일 리텐션과 N일 롤링 리텐션 혼동)
"Day 30 retention" (still active exactly on day 30) and "Day 30 rolling retention" (active at ANY point during the first 30 days) are different definitions that produce very different numbers — reporting one while a stakeholder assumes the other causes confusion.  
"30일 리텐션"(정확히 30일째에 활성)과 "30일 롤링 리텐션"(첫 30일 중 어느 시점에든 활성)은 다른 정의이며 매우 다른 숫자를 만듭니다 -- 이해관계자가 다른 쪽을 가정하는 동안 한쪽을 보고하면 혼란이 생깁니다.

✅ **Fix / 해결:** Always state explicitly which retention definition is being used, especially in a document shared across teams.

# 💡 Tips (팁)
Useful tips or shortcuts / 유용한 팁과 단축법

- `pivot_table`'s `aggfunc='nunique'` (not `'count'`) is essential when raw data has multiple event rows per user per month — `'count'` would double-count a user who logged multiple events.<br>원본 데이터에 사용자당 월별 여러 이벤트 행이 있을 때는 pivot_table의 aggfunc='nunique'(단순 'count'가 아니라)가 필수입니다 -- 'count'는 여러 이벤트를 기록한 사용자를 중복 계산합니다.
- The LTV:CAC ratio and payback period tell complementary stories — a good ratio with a very long payback period still means the business needs a lot of cash on hand to fund growth, even if it's ultimately profitable per customer.<br>LTV:CAC 비율과 회수 기간은 서로 보완적인 이야기를 합니다 -- 좋은 비율이라도 회수 기간이 매우 길다면, 고객당 궁극적으로 수익성이 있더라도 성장 자금을 대기 위해 많은 현금이 필요하다는 뜻입니다.
- Retention curves almost always flatten out eventually rather than decaying to zero — that flattened "floor" represents your most loyal, hard-core users, and is often a more useful number for LTV than an early, steep decay rate.<br>리텐션 곡선은 거의 항상 결국 0으로 수렴하기보다 평평해집니다 -- 그 평평해진 "바닥"이 가장 충성도 높은 핵심 사용자를 나타내며, 초기의 가파른 감소율보다 LTV에 더 유용한 숫자인 경우가 많습니다.
- Section 11's regression toolkit works directly on a cohort table too — regressing retention rate on months-since-signup (and cohort vintage) can quantify a retention trend far more precisely than eyeballing the pivot table.<br>섹션 11의 회귀 도구는 코호트 표에도 직접 적용됩니다 -- 리텐션율을 가입 후 경과 개월(과 코호트 시기)에 대해 회귀하면 피벗 테이블을 눈으로 훑어보는 것보다 훨씬 정밀하게 리텐션 추세를 정량화할 수 있습니다.
- When presenting LTV or CAC to non-technical stakeholders, always pair the number with its key assumption ("assumes current churn holds steady") — these are projections built on today's data, not guaranteed facts about the future.<br>비기술직 이해관계자에게 LTV나 CAC를 제시할 때는 항상 그 숫자에 핵심 가정("현재 이탈률이 유지된다고 가정")을 함께 붙이세요 -- 이는 오늘의 데이터로 만든 예측이지, 미래에 대해 보장된 사실이 아닙니다.

# 💼 Business Example (비즈니스 예시)
*How would a Business Analyst use this?*

**Scenario / 시나리오:**

**EN:** Leadership is preparing for a fundraising conversation and needs a complete health check: build a cohort retention table, compute LTV from the observed churn rate, and calculate the LTV:CAC ratio to include in the pitch deck.

**KR:** 경영진이 자금 조달 대화를 준비하고 있으며 완전한 건전성 점검이 필요합니다: 코호트 리텐션 표를 만들고, 관측된 이탈률로 LTV를 계산하고, 피치덱에 포함할 LTV:CAC 비율을 계산.

**To-do / 할 일:**
- [x] Build the cohort retention table<br>코호트 리텐션 표 만들기
- [x] Compute the observed monthly churn rate and LTV<br>관측된 월간 이탈률과 LTV 계산하기
- [x] Calculate the LTV:CAC ratio and payback period<br>LTV:CAC 비율과 회수 기간 계산하기
- [x] Write a one-paragraph summary for the pitch deck<br>피치덱을 위한 한 문단 요약 작성하기

````python
np.random.seed(100)

# 1) Cohort retention table
cohort_months = ['2025-10', '2025-11', '2025-12', '2026-01']
cohort_sizes = {'2025-10': 2000, '2025-11': 2200, '2025-12': 1900, '2026-01': 2400}
decay = 0.95  # 5% monthly churn

records = []
for cohort in cohort_months:
    size = cohort_sizes[cohort]
    idx = cohort_months.index(cohort)
    for m in range(len(cohort_months) - idx):
        active = size if m == 0 else int(size * (decay ** m))
        records.append({'cohort': cohort, 'month': m, 'active': active})

df = pd.DataFrame(records)
table = df.pivot_table(index='cohort', columns='month', values='active')
pct_table = table.div(table[0], axis=0) * 100

print("=== Fundraising Health Check ===")
print("\n1) Cohort retention table (%):")
print(pct_table.round(1))

# 2) Observed monthly churn rate (matches the decay rate used above)
observed_monthly_churn = 1 - decay
print(f"\n2) Observed monthly churn rate: {observed_monthly_churn:.1%}")

# 3) LTV
arpu = 22
ltv = arpu / observed_monthly_churn
print(f"3) LTV (ARPU=${arpu} / churn): ${ltv:.2f}")

# 4) LTV:CAC
cac = 90
ratio = ltv / cac
payback = cac / arpu
print(f"\n4) CAC: ${cac}, LTV:CAC ratio: {ratio:.1f}:1, Payback: {payback:.1f} months")

oldest_cohort_latest_pct = pct_table.loc['2025-10'].dropna().iloc[-1]
verdict = "a healthy 3:1 or better" if ratio >= 3 else "below the common 3:1 benchmark"
verdict_kr = "건전한 3:1 이상" if ratio >= 3 else "3:1 기준 미만"
print()
print(f"Summary for pitch deck: the oldest cohort retains {oldest_cohort_latest_pct:.0f}% of users")
print(f"after 3 months, LTV:CAC is {verdict} at {ratio:.1f}:1, with a {payback:.1f}-month payback.")
print()
print(f"피치덱 요약: 가장 오래된 코호트는 3개월 후 사용자의 {oldest_cohort_latest_pct:.0f}%를 유지하며,")
print(f"LTV:CAC는 {ratio:.1f}:1로 {verdict_kr}이고, 회수 기간은 {payback:.1f}개월입니다.")
````

# ❓ Review Questions (복습 질문)

**Q1 (EN).** Why is comparing January's month-1 retention to March's month-6 retention an unfair comparison, and how should you compare cohorts instead?
**Q1 (KR).** 1월 코호트의 1개월차 리텐션을 3월 코호트의 6개월차 리텐션과 비교하는 것이 왜 불공정한 비교이며, 대신 코호트를 어떻게 비교해야 하나요?

Cohorts are at different ages, so compare them at the same month since signup (month-1 vs month-1, month-6 vs month-6).  
코호트의 나이가 달라서 불공정합니다. 가입 후 같은 시점끼리(1개월차는 1개월차끼리, 6개월차는 6개월차끼리) 비교해야 합니다.

**Q2 (EN).** What does it mean for time-to-churn to follow an exponential distribution, and why does that assumption matter for the LTV shortcut formula?
**Q2 (KR).** 이탈까지 걸리는 시간이 지수분포를 따른다는 것은 무엇을 의미하며, 그 가정이 LTV 단축 공식에 왜 중요한가요?

It means a constant churn rate each period regardless of tenure, so expected lifetime is 1/churn and LTV ≈ margin per period / churn; if churn varies with tenure, the shortcut is off.  
가입 기간과 상관없이 기간마다 이탈률이 일정하다는 뜻입니다. 그래서 기대 수명이 1/이탈률이고 LTV ≈ 기간당 마진 / 이탈률이 되며, 이탈률이 기간에 따라 변하면 이 단축 공식은 틀립니다.

**Q3 (EN).** A company reports CAC by dividing total marketing spend by all new signups, including free organic ones. Why might this understate the true cost of paid acquisition?
**Q3 (KR).** 어떤 회사가 전체 마케팅 지출을 무료 오가닉 가입자를 포함한 모든 신규 가입자로 나눠 CAC를 보고합니다. 이것이 왜 유료 획득의 실제 비용을 과소평가할 수 있나요?

Organic signups cost nothing, so they shrink the average and hide how expensive each paid customer really is.  
무료 오가닉 가입자가 분모를 키워 평균을 낮추기 때문에, 유료로 얻은 고객 한 명의 실제 비용이 숨겨집니다.

**Q4 (EN).** An LTV:CAC ratio of 4:1 was calculated a year ago and hasn't been revisited since. Why is this risky, even if the ratio was healthy at the time?
**Q4 (KR).** 1년 전에 계산된 LTV:CAC 비율 4:1이 그 이후로 재검토되지 않았습니다. 당시엔 건전한 비율이었어도 왜 이것이 위험한가요?

Costs, churn, and margins change over a year, so the ratio may no longer be 4:1 and decisions rest on old numbers.  
1년 사이 비용, 이탈률, 마진이 바뀌어서 지금은 4:1이 아닐 수 있고, 오래된 숫자로 의사결정을 하게 됩니다.

**Q5 (EN).** What's the difference between "Day 30 retention" and "Day 30 rolling retention," and why does the distinction matter when reporting to stakeholders?
**Q5 (KR).** "30일 리텐션"과 "30일 롤링 리텐션"의 차이는 무엇이며, 이해관계자에게 보고할 때 이 구분이 왜 중요한가요?

Day 30 retention counts users active exactly on day 30, while rolling retention counts anyone active on day 30 or later, so rolling is always higher or equal; stakeholders must know which one is reported.  
30일 리텐션은 정확히 30일째에 활동한 사용자만 세고, 롤링 리텐션은 30일째 또는 그 이후에 활동한 사용자까지 세서 항상 같거나 더 높습니다. 어느 쪽인지 알아야 숫자를 올바르게 읽습니다.
