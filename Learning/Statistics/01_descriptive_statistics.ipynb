# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** Descriptive statistics summarize a dataset's center, spread, and shape using a handful of numbers, turning thousands of individual rows into a few interpretable figures. Every hypothesis test and A/B test in this curriculum builds directly on these summaries — you can't tell whether a difference is "significant" without first knowing what "typical" and "spread out" mean for your data.

**KR:** 기술통계(Descriptive Statistics)는 데이터셋의 중심, 퍼짐 정도, 모양을 몇 개의 숫자로 요약해, 수천 개의 개별 행을 몇 개의 해석 가능한 수치로 압축합니다. 이 커리큘럼의 모든 가설검정과 A/B 테스트는 이 요약 통계 위에 직접 세워집니다 -- 데이터에서 "전형적"과 "퍼져있음"이 무엇을 의미하는지 먼저 알지 못하면 차이가 "유의미"한지 말할 수 없습니다.

### Measures of center (중심 측정값)

| Measure / 측정값 | Best when / 언제 적합한가 | Sensitive to outliers? / 이상치에 민감한가 | Code |
|---|---|---|---|
| Mean (평균) | Data is roughly symmetric<br>데이터가 대체로 대칭적일 때 | Yes, very<br>매우 민감함 | `.mean()` |
| Median (중앙값) | Data is skewed or has outliers<br>데이터가 치우쳐 있거나 이상치가 있을 때 | No<br>민감하지 않음 | `.median()` |
| Mode (최빈값) | Data is categorical, or you want the most common value<br>범주형 데이터이거나 가장 흔한 값이 필요할 때 | No<br>민감하지 않음 | `.mode()` |

### Measures of spread (산포도 측정값)

| Measure / 측정값 | What it tells you / 무엇을 말해주는가 | Same units as data? / 원 데이터와 단위가 같은가 |
|---|---|---|
| Range (범위) | max - min, simplest but very sensitive to outliers<br>최댓값-최솟값, 가장 단순하지만 이상치에 매우 민감 | Yes |
| IQR (사분위범위) | Spread of the middle 50%<br>중간 50%의 퍼짐 정도 | Yes |
| Variance (분산) | Average squared deviation from the mean<br>평균으로부터의 편차 제곱의 평균 | No (squared units) |
| Std Dev (표준편차) | Square root of variance — a "typical" deviation<br>분산의 제곱근 -- "전형적인" 편차 | Yes |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** A raw column of 1,000 order amounts tells you nothing at a glance. "Average order value is $58, but half of orders are under $40, and a small group of high-spending outliers is pulling that average up" tells a business decision-maker something they can act on immediately.

**KR:** 주문 금액 1,000개짜리 원본 컬럼은 한눈에 아무것도 말해주지 않습니다. "평균 주문금액은 $58이지만 절반은 $40 미만이고, 소수의 고액 이상치가 평균을 끌어올리고 있다"는 설명은 비즈니스 의사결정자가 바로 행동으로 옮길 수 있는 정보를 줍니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Every KPI report (average order value, median session duration), every "here's what our users look like" slide, and the very first step of any deeper statistical analysis — you cannot run a meaningful hypothesis test without first understanding what "typical" and "spread out" mean for your data.

**KR:** 모든 KPI 보고서(평균 주문금액, 중앙값 세션 시간), "우리 사용자는 이렇습니다" 슬라이드, 그리고 더 깊은 통계 분석의 첫 단계 어디에나 사용됩니다 -- 데이터에서 "전형적"과 "퍼져있음"이 무엇을 의미하는지 먼저 이해하지 않고는 의미 있는 가설검정을 할 수 없습니다.

# 📝 Syntax (문법)

## Basic Syntax (기본 문법)

````python
import numpy as np
import pandas as pd
from scipy import stats

np.random.seed(42)
order_amounts = np.random.gamma(shape=2, scale=25, size=1000)  # right-skewed, like real order data
df = pd.DataFrame({'order_amount': order_amounts})

mean_val = df['order_amount'].mean()
median_val = df['order_amount'].median()
std_val = df['order_amount'].std()

print(f"Mean: ${mean_val:.2f} / 평균: ${mean_val:.2f}")
print(f"Median: ${median_val:.2f} / 중앙값: ${median_val:.2f}")
print(f"Std Dev: ${std_val:.2f} / 표준편차: ${std_val:.2f}")
````

## Common Variations (자주 쓰는 변형)

````python
# describe() gives you most of this in one call
# describe()는 이 대부분을 한 번에 알려줌
print(df['order_amount'].describe())
print()

# skewness: positive = right-skewed (mean > median), negative = left-skewed
# 왜도: 양수면 오른쪽 치우침(평균>중앙값), 음수면 왼쪽 치우침
skewness = stats.skew(df['order_amount'])
print(f"Skewness: {skewness:.2f}")
if skewness > 0.5:
    print("Right-skewed: a few large values are pulling the mean above the median.")
    print("오른쪽으로 치우침: 소수의 큰 값들이 평균을 중앙값보다 위로 끌어올리고 있음.")
````

# 🧪 Small Examples (예제)

## Example 1 — Mean vs Median vs Mode: Why "Average" Can Mislead (평균이 오해를 부르는 이유)

````python
# Business question: does "average order value" tell the whole story when a
# small group of whale customers spends much more than everyone else?
# 비즈니스 질문: 소수의 큰손 고객이 훨씬 더 많이 쓸 때 "평균 주문금액"이
# 전체 이야기를 말해주는가?

np.random.seed(1)
typical = np.random.normal(35, 8, 950)     # 950 typical customers
whales = np.random.normal(400, 60, 50)     # 50 high-spending customers
order_amounts = np.concatenate([typical, whales])
df = pd.DataFrame({'order_amount': order_amounts})

mean_val = df['order_amount'].mean()
median_val = df['order_amount'].median()
mode_val = df['order_amount'].round(-1).mode()[0]  # round to nearest 10 for a meaningful mode

print(f"Mean order value: ${mean_val:.2f}")
print(f"평균 주문금액: ${mean_val:.2f}")
print(f"Median order value: ${median_val:.2f}")
print(f"중앙값 주문금액: ${median_val:.2f}")
print(f"Most common order value (rounded to nearest $10): ${mode_val:.0f}")
print(f"가장 흔한 주문금액(10달러 단위 반올림): ${mode_val:.0f}")
print(f"\nGap between mean and median: ${mean_val - median_val:.2f}")
print(f"평균과 중앙값의 차이: ${mean_val - median_val:.2f}")
print("The 5% of whale customers pulled the mean well above what a typical customer spends.")
print("5%의 큰손 고객이 평균을 전형적인 고객의 지출보다 훨씬 위로 끌어올렸습니다.")
````

## Example 2 — Measuring Spread: Comparing Two Channels' Consistency (두 채널의 안정성 비교)

````python
# Business question: two marketing channels have the same average conversion
# rate -- but is one more RELIABLE than the other?
# 비즈니스 질문: 두 마케팅 채널의 평균 전환율이 같다면 -- 어느 쪽이 더 안정적인가?

np.random.seed(2)
channel_a = np.random.normal(50, 5, 200)   # consistent
channel_b = np.random.normal(50, 20, 200)  # volatile
df_ab = pd.DataFrame({'channel_a': channel_a, 'channel_b': channel_b})

print(f"Channel A: mean={df_ab['channel_a'].mean():.1f}, std={df_ab['channel_a'].std():.1f}")
print(f"Channel B: mean={df_ab['channel_b'].mean():.1f}, std={df_ab['channel_b'].std():.1f}")
print()
print("Both channels average the same conversion rate, but Channel A's daily results")
print("cluster tightly around 50, while Channel B swings wildly -- Channel A is the")
print("more predictable bet for planning purposes.")
print("두 채널의 평균 전환율은 같지만, 채널 A의 일별 결과는 50 근처에 촘촘히")
print("모여 있고, 채널 B는 크게 요동칩니다 -- 계획 수립 관점에서는 채널 A가")
print("더 예측 가능한 선택입니다.")
````

## Example 3 — Percentiles & the Five-Number Summary (백분위수와 5수 요약)

````python
# Business question: what does "typical" session length look like across the
# FULL range, not just the average? (connects to the Box Plot from Section 5
# of the Visualization curriculum)
# 비즈니스 질문: 평균뿐 아니라 전체 범위에서 "전형적인" 세션 길이는 어떤
# 모습인가? (시각화 커리큘럼 섹션 5의 Box Plot과 연결됨)

np.random.seed(3)
session_minutes = np.random.exponential(8, 500)
df_sess = pd.DataFrame({'session_minutes': session_minutes})

five_num = df_sess['session_minutes'].describe()[['min', '25%', '50%', '75%', 'max']]
print(five_num.round(1))

q1, q3 = five_num['25%'], five_num['75%']
iqr = q3 - q1
print(f"\nIQR (middle 50% spread): {iqr:.1f} minutes")
print(f"IQR (중간 50%의 퍼짐): {iqr:.1f}분")
print("This five-number summary is exactly what a Box Plot draws: the box spans")
print("Q1-Q3, the line inside is the median, and the whiskers reach toward min/max.")
print("이 5수 요약이 바로 Box Plot이 그리는 것입니다: 박스는 Q1~Q3, 안의 선은")
print("중앙값, 수염은 최솟값·최댓값 방향으로 뻗어나갑니다.")
````

## Example 4 — Detecting Outliers with the IQR Rule (IQR 규칙으로 이상치 탐지하기)

````python
# Business question: which orders look like data-entry errors or genuine
# outlier events that deserve a closer look?
# 비즈니스 질문: 어떤 주문이 데이터 입력 오류이거나 자세히 살펴볼 가치가
# 있는 진짜 이상치 이벤트로 보이는가?

np.random.seed(4)
normal_orders = np.random.normal(50, 12, 300)
flagged_orders = [350, 420, -10]  # a couple of whale orders + a likely data-entry error
all_orders = np.concatenate([normal_orders, flagged_orders])
df_out = pd.DataFrame({'order_amount': all_orders})

q1, q3 = df_out['order_amount'].quantile([0.25, 0.75])
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df_out[(df_out['order_amount'] < lower_bound) | (df_out['order_amount'] > upper_bound)]

print(f"IQR bounds: [{lower_bound:.1f}, {upper_bound:.1f}]")
print(f"IQR 경계: [{lower_bound:.1f}, {upper_bound:.1f}]")
print(f"Outliers found: {len(outliers)}")
print(f"발견된 이상치: {len(outliers)}개")
print(outliers.sort_values('order_amount'))
print()
print("The negative value is almost certainly a data-entry error worth fixing. The two")
print("high values could be legitimate whale orders worth investigating, not deleting.")
print("음수 값은 거의 확실히 수정이 필요한 데이터 입력 오류입니다. 두 개의 높은")
print("값은 삭제가 아니라 조사해볼 가치가 있는 정당한 큰손 주문일 수 있습니다.")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Reporting Only the Mean (평균만 보고하기)
Reporting "average order value: $58" without also reporting spread hides whether that $58 is a reliable number everyone is near, or an average of wildly different customers.  
"평균 주문금액: $58"만 보고하고 산포도를 함께 보고하지 않으면, 그 $58이 모두가 근접한 신뢰할 만한 숫자인지, 아니면 서로 크게 다른 고객들의 평균인지 알 수 없습니다.

✅ **Fix / 해결:** Always pair a center measure with a spread measure (e.g., "$58 ± $23").

### Mistake 2 — Using Mean on Heavily Skewed Data (심하게 치우친 데이터에 평균 사용)
A handful of whale customers or one huge outlier order can drag the mean far from what a "typical" customer actually looks like.  
소수의 큰손 고객이나 하나의 거대한 이상치 주문이 평균을 실제 "전형적인" 고객의 모습에서 크게 벗어나게 할 수 있습니다.

✅ **Fix / 해결:** When skewness is noticeable, lead with the median and mention the mean as a secondary figure.

### Mistake 3 — Confusing Variance and Standard Deviation (분산과 표준편차 혼동)
Variance is in SQUARED units (e.g., dollars²), which isn't directly interpretable — a variance of "144" doesn't mean anything on its own until you take its square root.  
분산은 제곱 단위(예: 달러²)이므로 직접 해석할 수 없습니다 -- 분산 "144"는 제곱근을 취하기 전까지는 그 자체로 아무 의미가 없습니다.

✅ **Fix / 해결:** Report standard deviation (not variance) when communicating with a business audience; use variance mainly as an intermediate calculation.

### Mistake 4 — numpy and pandas Disagree on the Default Standard Deviation (numpy·pandas의 표준편차 기본값 불일치)
`numpy.std()` defaults to the POPULATION standard deviation (`ddof=0`), while `pandas.Series.std()` defaults to the SAMPLE standard deviation (`ddof=1`) — running both on the same data gives two different numbers. Verified directly: for `[2,4,4,4,5,5,7,9]`, `np.std()` gives `2.0` but `pd.Series(...).std()` gives `2.14`.  
numpy.std()는 기본적으로 모집단 표준편차(ddof=0)를, pandas.Series.std()는 기본적으로 표본표준편차(ddof=1)를 계산합니다 -- 같은 데이터에 둘 다 돌리면 서로 다른 숫자가 나옵니다. 직접 확인: [2,4,4,4,5,5,7,9]에서 np.std()는 2.0, pd.Series(...).std()는 2.14.

✅ **Fix / 해결:** Know which library you're using, and set `ddof=` explicitly if precision matters — almost always use `ddof=1` (sample) for real business data.

### Mistake 5 — Deleting Outliers Automatically (이상치를 자동으로 삭제)
Not every outlier is an error — a $400 order might be a real VIP purchase worth celebrating, not a mistake to remove.  
모든 이상치가 오류는 아닙니다 -- $400짜리 주문은 삭제할 실수가 아니라 축하할 가치가 있는 진짜 VIP 구매일 수 있습니다.

✅ **Fix / 해결:** Investigate each flagged outlier before deciding whether to keep, fix, or remove it.

### Mistake 6 — Treating 1.5×IQR as a Universal Law (1.5×IQR을 절대 법칙으로 취급)
The 1.5x multiplier is a widely used convention (from statistician John Tukey), not a law of nature — different fields and different distributions may call for a different threshold, or a transformation instead of filtering.  
1.5배 승수는 널리 쓰이는 관행(통계학자 John Tukey가 제안)이지 자연의 법칙이 아닙니다 -- 분야와 분포에 따라 다른 임계값이나, 필터링 대신 변환이 필요할 수 있습니다.

✅ **Fix / 해결:** Treat the IQR rule as a starting heuristic for flagging candidates to investigate, not an automatic decision rule.

# 💡 Tips (팁)
Useful tips or shortcuts / 유용한 팁과 단축법

- `.describe()` gives you count, mean, std, min, 25%/50%/75%, and max in one call — often the fastest way to get oriented on a new column.<br>.describe()는 count, mean, std, min, 25%/50%/75%, max를 한 번에 알려줍니다 -- 새 컬럼을 빠르게 파악하는 가장 빠른 방법인 경우가 많습니다.
- When mean and median are far apart, that gap itself is a useful signal — it tells you roughly how skewed your data is, before you even compute skewness.<br>평균과 중앙값이 크게 다르다면 그 차이 자체가 유용한 신호입니다 -- 왜도를 계산하기도 전에 데이터가 대략 얼마나 치우쳐 있는지 알려줍니다.
- Standard deviation is easiest to communicate in the SAME units as the original data ("orders vary by about $23") — resist the urge to lead with variance.<br>표준편차는 원 데이터와 같은 단위로 전달하기 가장 쉽습니다("주문은 약 $23 정도 차이 남") -- 분산으로 먼저 설명하고 싶은 유혹을 참으세요.
- The five-number summary (min, Q1, median, Q3, max) is exactly what feeds a Box Plot — if you've already built one in the Visualization curriculum, you already know how to read this table visually.<br>5수 요약(최솟값, Q1, 중앙값, Q3, 최댓값)은 Box Plot을 그리는 데 그대로 쓰입니다 -- 시각화 커리큘럼에서 이미 만들어봤다면 이 표를 시각적으로 읽는 법을 이미 알고 있는 것입니다.
- Before you ever run a hypothesis test (Section 6+), always look at descriptive stats first — a huge outlier or unexpected skew can completely change which test is appropriate.<br>가설검정(섹션 6 이후)을 돌리기 전에는 항상 기술통계부터 보세요 -- 큰 이상치나 예상치 못한 왜도가 어떤 검정이 적절한지를 완전히 바꿀 수 있습니다.

# 💼 Business Example (비즈니스 예시)
*How would a Business Analyst use this?*

**Scenario / 시나리오:**

**EN:** Leadership asks: "How's our new checkout flow performing compared to the old one, in terms of order value?" You have 3 months of order data and need to give a defensible, complete answer — not just one number.

**KR:** 경영진이 묻습니다: "새 결제 플로우가 주문 금액 측면에서 예전과 비교해 어떻게 되고 있나요?" 3개월치 주문 데이터가 있고, 숫자 하나가 아니라 방어 가능하고 완전한 답을 줘야 합니다.

**To-do / 할 일:**
- [x] Compute mean AND median to check whether outliers are distorting the average<br>평균과 중앙값을 모두 계산해 이상치가 평균을 왜곡하고 있는지 확인하기
- [x] Compute standard deviation to report how consistent order values are, not just their center<br>표준편차를 계산해 중심뿐 아니라 주문금액이 얼마나 일관적인지 보고하기
- [x] Flag any outlier orders with the IQR rule and decide whether they need investigation<br>IQR 규칙으로 이상치 주문을 표시하고 조사가 필요한지 판단하기
- [x] Write a two-sentence summary leadership can act on<br>경영진이 실행할 수 있는 두 문장짜리 요약 작성하기

````python
np.random.seed(11)
old_flow = np.random.gamma(shape=2.2, scale=22, size=600)
new_flow = np.random.gamma(shape=2.4, scale=23, size=600)
df_checkout = pd.DataFrame({
    'order_amount': np.concatenate([old_flow, new_flow]),
    'flow': ['Old'] * 600 + ['New'] * 600,
})

summary = df_checkout.groupby('flow')['order_amount'].agg(['mean', 'median', 'std']).round(2)
print(summary)
print()

for flow in ['Old', 'New']:
    sub = df_checkout[df_checkout['flow'] == flow]['order_amount']
    q1, q3 = sub.quantile([0.25, 0.75])
    iqr = q3 - q1
    upper = q3 + 1.5 * iqr
    n_outliers = (sub > upper).sum()
    print(f"{flow} flow: {n_outliers} orders flagged as outliers above ${upper:.0f}")

mean_lift = (summary.loc['New', 'mean'] - summary.loc['Old', 'mean']) / summary.loc['Old', 'mean'] * 100
median_lift = (summary.loc['New', 'median'] - summary.loc['Old', 'median']) / summary.loc['Old', 'median'] * 100

print()
print(f"Summary: The new flow's average order value is {mean_lift:+.1f}% vs the old flow, and the")
print(f"median tells a consistent story ({median_lift:+.1f}%), so this isn't just an outlier effect.")
print(f"Standard deviation is similar between the two, so the new flow isn't introducing more risk.")
print(f"요약: 새 플로우의 평균 주문금액은 예전 대비 {mean_lift:+.1f}%이며, 중앙값도 일관된 결과를")
print(f"보여({median_lift:+.1f}%) 이상치 효과가 아님을 알 수 있습니다. 표준편차는 두 플로우가")
print(f"비슷해 새 플로우가 추가 리스크를 만들지는 않습니다.")
````

# 📝 Summary (요약)
*Write today's concept in 3~5 sentences.*

**EN:** Descriptive statistics compress a dataset into three types of summary numbers: center (mean, median, mode — each best in different situations, with the median resisting outliers that distort the mean), spread (range, IQR, variance, and standard deviation, with standard deviation being the most business-interpretable since it shares units with the original data), and shape (skewness, which reveals when mean and median diverge, and outliers, which the IQR rule can flag for investigation). These aren't just abstract math — `.mean()`, `.median()`, `.std()`, and `.describe()` are pandas methods already used in earlier curriculum work, and every more advanced technique in this curriculum, from confidence intervals to A/B testing, is built directly on top of these same numbers.

**KR:** 기술통계는 데이터셋을 세 종류의 요약 숫자로 압축합니다: 중심(평균·중앙값·최빈값 -- 각각 다른 상황에서 적합하며, 중앙값은 평균을 왜곡하는 이상치에 저항성이 있음), 산포도(범위·IQR·분산·표준편차 -- 표준편차는 원 데이터와 단위가 같아 비즈니스적으로 가장 해석하기 쉬움), 모양(왜도는 평균과 중앙값이 언제 갈라지는지 보여주고, 이상치는 IQR 규칙으로 조사 대상으로 표시할 수 있음). 이것들은 추상적인 수학이 아닙니다 -- .mean(), .median(), .std(), .describe()는 이전 커리큘럼에서 이미 사용했던 pandas 메서드이며, 신뢰구간부터 A/B 테스트까지 이 커리큘럼의 더 고급 기법들은 모두 이 같은 숫자들 위에 직접 세워집니다.

# ❓ Review Questions (복습 질문)

**Q1 (EN).** A dataset has a mean of $85 and a median of $52. What does this gap tell you about the data's shape, and which measure should you lead with in a report?  
**Q1 (KR).** 데이터셋의 평균이 $85이고 중앙값이 $52입니다. 이 차이는 데이터의 모양에 대해 무엇을 말해주며, 보고서에서 어떤 측정값을 먼저 제시해야 하나요?

The mean ($85) sits well above the median ($52), so the data is right-skewed: a few large values pull the mean up. Lead with the median, and give the mean second.  
평균($85)이 중앙값($52)보다 훨씬 위에 있으므로 데이터가 오른쪽으로 치우쳐 있습니다. 큰 값 몇 개가 평균을 끌어올립니다. 보고서에서는 중앙값을 먼저 쓰고, 평균은 그다음에 제시합니다.

**Q2 (EN).** Why is standard deviation usually reported instead of variance when communicating with a business audience?  
**Q2 (KR).** 비즈니스 청중에게 전달할 때 왜 보통 분산 대신 표준편차를 보고하나요?

Variance is in squared units, so it is hard to read. Standard deviation is its square root and uses the same units as the original data.  
분산은 단위가 제곱되어 있어 읽기 어렵습니다. 표준편차는 그 제곱근이라 원래 데이터와 같은 단위를 씁니다.

**Q3 (EN).** You run `np.std(data)` and `pd.Series(data).std()` on the exact same data and get two different numbers. Why, and how do you fix it?  
**Q3 (KR).** 같은 데이터에 `np.std(data)`와 `pd.Series(data).std()`를 각각 돌렸더니 서로 다른 숫자가 나왔습니다. 왜 그런가요, 어떻게 고치나요?

The data is the same. np.std() divides by n (ddof=0), while pd.Series.std() divides by n - 1 (ddof=1). Set ddof= to the same value.  
데이터는 같습니다. np.std()는 n으로 나누고(ddof=0), pd.Series.std()는 n - 1로 나눕니다(ddof=1). ddof=를 같게 맞추면 됩니다.

**Q4 (EN).** You flag an order as a statistical outlier using the 1.5xIQR rule. What should you do next, and what should you NOT automatically do?  
**Q4 (KR).** 1.5×IQR 규칙으로 어떤 주문을 통계적 이상치로 표시했습니다. 다음으로 무엇을 해야 하며, 자동으로 하면 안 되는 것은 무엇인가요?

Check whether that order is a data error or a real large purchase. The 1.5×IQR rule only flags a candidate, so do not delete it automatically.  
그 주문이 입력 실수인지, 실제 큰 구매인지 확인합니다. 1.5×IQR 규칙은 조사 후보를 표시할 뿐이므로 자동으로 삭제하지 않습니다.

**Q5 (EN).** What does the five-number summary (min, Q1, median, Q3, max) directly correspond to in a chart you learned in the Visualization curriculum?  
**Q5 (KR).** 5수 요약(최솟값, Q1, 중앙값, Q3, 최댓값)은 시각화 커리큘럼에서 배운 어떤 차트와 직접 대응되나요?

The five-number summary is what a box plot draws: minimum, Q1, median, Q3, and maximum.  
5수 요약은 박스플롯이 그리는 값입니다. 최솟값, Q1, 중앙값, Q3, 최댓값입니다.
