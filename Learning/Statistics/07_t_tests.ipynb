# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** A t-test is the specific hypothesis test (from Section 6's general framework) used to compare means. There are three flavors depending on your data's shape: a one-sample t-test checks a single group's mean against a known value, an independent samples t-test compares the means of two separate groups, and a paired samples t-test compares two measurements taken on the SAME subjects (like before/after).

**KR:** t-검정은 평균을 비교하는 데 사용되는(섹션 6의 일반적인 틀에서 나온) 구체적인 가설검정입니다. 데이터의 형태에 따라 세 가지 유형이 있습니다: 단일표본 t-검정은 하나의 그룹 평균을 알려진 값과 비교하고, 독립표본 t-검정은 두 개의 서로 다른 그룹의 평균을 비교하며, 대응표본 t-검정은 같은 대상에서 측정한 두 값(전후 비교 등)을 비교합니다.

### The three t-test flavors (세 가지 t-검정 유형)

| Type / 유형 | Compares / 비교 대상 | Business example / 비즈니스 예시 | scipy function |
|---|---|---|---|
| One-sample<br>단일표본 | Sample mean vs a known/claimed value<br>표본평균 vs 알려진 값 | Is average deal size = $50K target?<br>평균 거래 규모 = $50K 목표인가? | `stats.ttest_1samp()` |
| Independent<br>독립표본 | Mean of Group A vs Mean of Group B (different subjects)<br>그룹 A 평균 vs 그룹 B 평균(다른 대상) | Variant A vs Variant B conversion<br>변형 A vs 변형 B 전환 | `stats.ttest_ind()` |
| Paired<br>대응표본 | Mean difference within the SAME subjects (before vs after)<br>같은 대상 내 평균 차이(전 vs 후) | Same users before/after redesign<br>같은 사용자의 재설계 전후 | `stats.ttest_rel()` |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** The three t-test flavors map directly onto the three most common business comparison shapes: "did this month's number hit the target?" (one-sample), "did Variant B outperform Variant A?" (independent), "did the SAME customers behave differently after a change?" (paired). Picking the wrong flavor for your data's structure gives a technically-runs-but-wrong answer.

**KR:** 세 가지 t-검정 유형은 가장 흔한 세 가지 비즈니스 비교 형태에 정확히 대응합니다: "이번 달 숫자가 목표를 달성했는가?"(단일표본), "변형 B가 변형 A보다 나았는가?"(독립표본), "같은 고객이 변경 후 다르게 행동했는가?"(대응표본). 데이터 구조에 맞지 않는 유형을 고르면 기술적으로는 실행되지만 틀린 답이 나옵니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** One-sample: checking if this quarter's average deal size hit the $50K target. Independent: the bread-and-butter test behind most A/B tests comparing two separate user groups. Paired: measuring the effect of a change on the same set of users' behavior before and after, which removes person-to-person variability and often needs a much smaller sample to detect the same effect.

**KR:** 단일표본: 이번 분기 평균 거래 규모가 $50K 목표를 달성했는지 확인. 독립표본: 두 개의 별도 사용자 그룹을 비교하는 대부분의 A/B 테스트 뒤에 있는 가장 기본적인 검정. 대응표본: 같은 사용자 집단의 행동을 변경 전후로 측정해 개인차를 제거하며, 같은 효과를 탐지하는 데 훨씬 적은 표본으로도 충분한 경우가 많음.

# 📝 Syntax (문법)

## Basic Syntax (기본 문법)

````python
import numpy as np
from scipy import stats

# One-sample t-test: is this quarter's average deal size different from the $50K target?
# 단일표본 t-검정: 이번 분기 평균 거래 규모가 $50K 목표와 다른가?
np.random.seed(42)
deal_sizes = np.random.normal(loc=54000, scale=12000, size=40)

t_stat, p_value = stats.ttest_1samp(deal_sizes, popmean=50000)

print(f"Sample mean: ${deal_sizes.mean():,.0f}")
print(f"t-statistic: {t_stat:.3f}")
print(f"p-value: {p_value:.4f}")
````

## Common Variations (자주 쓰는 변형)

````python
# One-sided test: is the mean GREATER than the target, specifically? (not just "different")
# 단측검정: 평균이 특히 목표보다 '큰가'? ("다른가"가 아니라)
t_stat, p_value_two_sided = stats.ttest_1samp(deal_sizes, popmean=50000)
p_value_one_sided = p_value_two_sided / 2 if t_stat > 0 else 1 - p_value_two_sided / 2

print(f"Two-sided p-value: {p_value_two_sided:.4f}")
print(f"One-sided p-value (testing mean > $50K specifically): {p_value_one_sided:.4f}")
````

# 🧪 Small Examples (예제)

## Example 1 — One-Sample t-test: Did This Quarter Hit the Target? (이번 분기 목표를 달성했는가?)

````python
# Business question: did this quarter's average deal size meaningfully differ
# from the $50K target sales leadership set?
# 비즈니스 질문: 이번 분기 평균 거래 규모가 영업 경영진이 설정한 $50K
# 목표와 의미 있게 달랐는가?

np.random.seed(1)
deal_sizes = np.random.normal(loc=54200, scale=13000, size=45)
target = 50000

t_stat, p_value = stats.ttest_1samp(deal_sizes, popmean=target)

print(f"H0: true average deal size = ${target:,}")
print(f"H1: true average deal size != ${target:,}")
print(f"Sample mean: ${deal_sizes.mean():,.0f} (n={len(deal_sizes)})")
print(f"t-statistic: {t_stat:.3f}, p-value: {p_value:.4f}")
print()
alpha = 0.05
if p_value < alpha:
    print(f"p={p_value:.4f} < {alpha} -> REJECT H0: this quarter's average deal size is")
    print(f"statistically significantly different from the ${target:,} target.")
    print(f"p={p_value:.4f} < {alpha} -> H0 기각: 이번 분기 평균 거래 규모는 ${target:,}")
    print("목표와 통계적으로 유의미하게 다릅니다.")
else:
    print(f"p={p_value:.4f} >= {alpha} -> FAIL TO REJECT H0: not enough evidence that this")
    print(f"quarter's average differs from the ${target:,} target.")
````

## Example 2 — Independent Samples t-test: A/B Test Comparison (A/B 테스트 비교)

````python
# Business question: does Variant B (new pricing display) lead to a higher
# average order value than Variant A (control)?
# 비즈니스 질문: 변형 B(새 가격 표시)가 변형 A(대조군)보다 평균
# 주문금액이 더 높은가?

np.random.seed(2)
variant_a = np.random.normal(48, 14, 200)
variant_b = np.random.normal(52, 14, 200)

t_stat, p_value = stats.ttest_ind(variant_b, variant_a)

print("H0: mean(Variant B) = mean(Variant A)")
print("H1: mean(Variant B) != mean(Variant A)")
print(f"Variant A mean: ${variant_a.mean():.2f} (n={len(variant_a)})")
print(f"Variant B mean: ${variant_b.mean():.2f} (n={len(variant_b)})")
print(f"t-statistic: {t_stat:.3f}, p-value: {p_value:.4f}")
print()
lift = (variant_b.mean() - variant_a.mean()) / variant_a.mean() * 100
print(f"Observed lift: {lift:+.1f}%")
if p_value < 0.05:
    print(f"p={p_value:.4f} < 0.05 -> statistically significant difference; Variant B")
    print("appears to genuinely outperform Variant A.")
    print(f"p={p_value:.4f} < 0.05 -> 통계적으로 유의미한 차이; 변형 B가 실제로")
    print("변형 A보다 나은 것으로 보입니다.")
````

## Example 3 — Paired Samples t-test: Before/After the Same Users (같은 사용자의 전후 비교)

````python
# Business question: did a UI redesign change the SAME users' time to
# complete a core task (measured on the same 50 users, before and after)?
# 비즈니스 질문: UI 재설계가 같은 사용자들의 핵심 작업 완료 시간을
# 바꿨는가? (같은 50명을 재설계 전후로 측정)

np.random.seed(3)
n_users = 50
before = np.random.normal(85, 15, n_users)  # seconds to complete task
individual_improvement = np.random.normal(8, 6, n_users)  # each user improves by a bit
after = before - individual_improvement

t_stat, p_value = stats.ttest_rel(before, after)

print("H0: mean(before) = mean(after), for the SAME users")
print("H1: mean(before) != mean(after)")
print(f"Before mean: {before.mean():.1f}s, After mean: {after.mean():.1f}s")
print(f"Mean individual improvement: {(before - after).mean():.1f}s")
print(f"t-statistic: {t_stat:.3f}, p-value: {p_value:.6f}")
print()
print("A paired test uses each user as their OWN baseline, removing person-to-person")
print("variability -- this is usually far more sensitive than treating before/after as")
print("two independent groups would be, for the exact same underlying data.")

# For comparison: what if we'd (incorrectly) treated this as two independent groups?
t_stat_wrong, p_value_wrong = stats.ttest_ind(before, after)
print("\nFor comparison, treating before/after as independent groups (WRONG for this")
print(f"design): t={t_stat_wrong:.3f}, p={p_value_wrong:.6f} -- notice how much less")
print("significant this looks, purely from throwing away the pairing information.")
print("비교를 위해 전후를 (이 설계에는 잘못된 방식으로) 독립 그룹으로 취급하면:")
print(f"t={t_stat_wrong:.3f}, p={p_value_wrong:.6f} -- 대응 정보를 버리는 것만으로")
print("얼마나 덜 유의미해 보이는지 주목하세요.")
````

## Example 4 — Checking t-test Assumptions (t-검정의 가정 확인하기)

````python
# Business question: before trusting a t-test comparing two support channels'
# resolution times, are its assumptions actually met?
# 비즈니스 질문: 두 지원 채널의 해결 시간을 비교하는 t-검정을 신뢰하기
# 전에, 그 가정이 실제로 충족되는가?

np.random.seed(4)
channel_chat = np.random.exponential(scale=12, size=80)   # skewed, common for wait times
channel_email = np.random.exponential(scale=18, size=80)  # also skewed, different scale

# 1) Normality check (skewness as a quick proxy; Section 3 covered this)
# 1) 정규성 확인(왜도를 빠른 대용치로 사용; 섹션 3에서 다룸)
print(f"Chat skewness: {stats.skew(channel_chat):.2f}")
print(f"Email skewness: {stats.skew(channel_email):.2f}")
print("Both are heavily right-skewed -- individual times are NOT normal.")
print()

# 2) Equal variance check (Levene's test)
# 2) 등분산 확인(Levene 검정)
levene_stat, levene_p = stats.levene(channel_chat, channel_email)
print(f"Levene's test for equal variance: statistic={levene_stat:.3f}, p={levene_p:.4f}")
if levene_p < 0.05:
    print("p < 0.05 -> variances are significantly different; use Welch's t-test (equal_var=False)")
else:
    print("p >= 0.05 -> no evidence variances differ; standard t-test is fine")

# 3) Run the appropriate t-test
t_stat, p_value = stats.ttest_ind(channel_chat, channel_email, equal_var=(levene_p >= 0.05))
print(f"\nt-test result: t={t_stat:.3f}, p={p_value:.4f}")
print()
print("Note: even though the RAW data is skewed (violating normality), the sample sizes")
print("(n=80 each) are large enough that the CLT (Section 3) makes the SAMPLE MEANS")
print("approximately normal -- the t-test is likely still reasonably valid here.")
print("참고: 원본 데이터는 치우쳐 있지만(정규성 위반), 표본 크기(각 n=80)가 충분히")
print("커서 CLT(섹션 3)가 표본평균을 대략 정규분포로 만들어줍니다 -- 여기서는")
print("t-검정이 여전히 상당히 타당할 가능성이 높습니다.")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Using an Independent t-test on Paired Data (대응표본 데이터에 독립표본 검정 사용)
When the same subjects are measured twice (before/after), using `ttest_ind()` instead of `ttest_rel()` throws away the pairing information and treats person-to-person variability as noise — this typically makes a real effect look weaker (or non-significant) than it actually is, exactly as Example 3 demonstrated.  
같은 대상을 두 번 측정했을 때(전후), ttest_rel() 대신 ttest_ind()를 쓰면 대응 정보를 버리고 개인 간 변동을 노이즈로 취급합니다 -- 이는 예제 3에서 보여줬듯, 실제 효과를 실제보다 약하게(또는 유의미하지 않게) 보이게 만드는 경우가 많습니다.

✅ **Fix / 해결:** If the SAME subjects contribute both measurements, always use `ttest_rel()`.

### Mistake 2 — Defaulting to `equal_var=True` Without Checking (확인 없이 equal_var=True를 기본값으로 사용)
`stats.ttest_ind()`'s default assumes both groups have equal variance — if that's false (common with business data of different scales or volatility), the p-value can be meaningfully off.  
stats.ttest_ind()의 기본값은 두 그룹의 분산이 같다고 가정합니다 -- 이것이 거짓이면(규모나 변동성이 다른 비즈니스 데이터에서 흔함), p-value가 의미 있게 틀릴 수 있습니다.

✅ **Fix / 해결:** Run Levene's test first (Example 4), or simply default to `equal_var=False` (Welch's t-test), which is nearly as powerful when variances ARE equal and much safer when they aren't.

### Mistake 3 — Trusting Normality on Skewed Data with a Small Sample (작은 표본의 치우친 데이터에서 정규성을 신뢰)
The Central Limit Theorem (Section 3) rescues heavily skewed data once the sample is reasonably large, but with a small sample (rough rule of thumb: under ~30) and severe skew, a t-test's p-value can be unreliable.  
중심극한정리(섹션 3)는 표본이 충분히 크면 심하게 치우친 데이터도 구제해주지만, 표본이 작고(대략 30 미만) 왜도가 심하면 t-검정의 p-value가 신뢰할 수 없을 수 있습니다.

✅ **Fix / 해결:** For small, skewed samples, check skewness first (Example 4's approach) and consider a non-parametric alternative (covered later in this curriculum) if in doubt.

### Mistake 4 — Mismatching a Two-Sided Test with a One-Sided Business Question (양측검정과 단측 비즈니스 질문의 불일치)
`scipy`'s t-test functions run a two-sided test by default ("is the mean different, in either direction?"), but many business questions are really one-sided ("is Variant B BETTER, specifically?") — reporting a two-sided p-value for a one-sided claim overstates the evidence needed.  
scipy의 t-검정 함수는 기본적으로 양측검정을 실행합니다("평균이 어느 방향으로든 다른가?") -- 하지만 많은 비즈니스 질문은 사실 단측입니다("특히 변형 B가 더 나은가?") -- 단측 주장에 양측 p-value를 보고하면 필요한 증거 기준이 과장됩니다.

✅ **Fix / 해결:** Decide (before running the test) whether your question is one-sided or two-sided, and adjust the p-value accordingly (usually halving it for a one-sided claim in the predicted direction).

### Mistake 5 — Accidentally Treating Paired Data as Independent (대응 데이터를 실수로 독립으로 취급)
It's easy to lose track of which rows belong to the same subject after some data wrangling — feeding `ttest_rel()` two arrays that AREN'T actually aligned row-by-row to the same subject silently produces a meaningless result.  
데이터를 다루다 보면 어떤 행이 같은 대상에 속하는지 놓치기 쉽습니다 -- 실제로 같은 대상에 행별로 정렬되지 않은 두 배열을 ttest_rel()에 넣으면 조용히 의미 없는 결과가 나옵니다.

✅ **Fix / 해결:** Before running a paired test, explicitly verify row `i` in both arrays really does belong to the same subject (e.g., by sorting both by a shared ID column first).

### Mistake 6 — Treating a Non-Significant t-test as Proof of Equivalence ("유의미하지 않음"을 동등성의 증거로 취급)
"We ran a t-test comparing the new and old checkout flows and found no significant difference" does NOT mean they perform equally well — it might just mean the test lacked the power to detect a real, smaller difference (Section 6's Mistake 4, specifically relevant here).  
"새 결제 플로우와 기존 플로우를 t-검정으로 비교했는데 유의미한 차이가 없었다"는 둘의 성과가 동등하다는 뜻이 아닙니다 -- 그저 검정이 실재하지만 더 작은 차이를 탐지할 검정력이 부족했을 수 있습니다(섹션 6의 실수 4가 여기서 특히 관련됨).

✅ **Fix / 해결:** If you need to formally claim two things are equivalent (not just "not proven different"), use a dedicated equivalence test, not a standard t-test's non-significant result.

# 💡 Tips (팁)
Useful tips or shortcuts / 유용한 팁과 단축법

- A quick way to remember which function to use: `ttest_1samp` (one group vs a number), `ttest_ind` (two INDEPENDENT groups), `ttest_rel` (two RELATED/paired measurements) — the function names literally describe the data shape.<br>어떤 함수를 쓸지 빠르게 기억하는 법: ttest_1samp(그룹 하나 vs 숫자), ttest_ind(두 개의 독립 그룹), ttest_rel(두 개의 관련된/대응 측정값) -- 함수 이름이 데이터 형태를 그대로 설명합니다.
- A paired design is almost always more statistically powerful than an independent one for the same number of observations — if you CAN measure the same subjects twice, it's usually worth the extra effort to set up.<br>대응 설계는 같은 관측치 수에서 거의 항상 독립 설계보다 통계적으로 더 강력합니다 -- 같은 대상을 두 번 측정할 수 있다면, 그렇게 설계하는 추가 노력이 보통 가치가 있습니다.
- `equal_var=False` (Welch's t-test) is a very safe default — it loses almost nothing when variances ARE equal, and protects you when they aren't. Many statisticians recommend using it by default rather than testing for equal variance first.<br>equal_var=False(Welch's t-test)는 매우 안전한 기본값입니다 -- 분산이 실제로 같을 때도 거의 손해가 없고, 다를 때는 보호해줍니다. 많은 통계학자들이 먼저 분산 검정을 하기보다 이를 기본값으로 쓰는 것을 권장합니다.
- Always report the effect size (the actual difference in means, or lift %) right alongside the t-test's p-value — the p-value alone answers "is it real?" but not "does it matter?"<br>t-검정의 p-value 옆에는 항상 효과 크기(실제 평균 차이나 상승률 %)를 함께 보고하세요 -- p-value만으로는 "실제인가?"에는 답하지만 "중요한가?"에는 답하지 않습니다.
- The p-values you're computing here are the exact same p-values that power every A/B test result you'll see later in this curriculum — a t-test comparing two conversion-related metrics IS an A/B test's statistical engine.<br>여기서 계산하는 p-value는 이 커리큘럼 후반에서 보게 될 모든 A/B 테스트 결과를 구동하는 것과 정확히 같은 p-value입니다 -- 전환 관련 지표 두 개를 비교하는 t-검정이 바로 A/B 테스트의 통계 엔진입니다.

# 💼 Business Example (비즈니스 예시)
*How would a Business Analyst use this?*

**Scenario / 시나리오:**

**EN:** Your support team just finished a new training program. Leadership wants a complete before/after picture: (1) is the trained team's average handle time now under the 6-minute target, (2) how does the trained team compare to a separate team that didn't receive training, and (3) did the SAME trained agents individually get faster?

**KR:** 지원팀이 새 교육 프로그램을 방금 마쳤습니다. 경영진은 완전한 전후 그림을 원합니다: (1) 교육받은 팀의 평균 처리 시간이 이제 6분 목표 아래인가, (2) 교육받은 팀이 교육받지 않은 별도 팀과 비교해 어떤가, (3) 같은 교육받은 상담원들이 개별적으로 더 빨라졌는가?

**To-do / 할 일:**
- [x] One-sample t-test: is the trained team's average handle time under the 6-minute target?<br>단일표본 t-검정: 교육받은 팀의 평균 처리 시간이 6분 목표 아래인가?
- [x] Independent samples t-test: trained team vs a separate untrained team<br>독립표본 t-검정: 교육받은 팀 vs 별도의 교육받지 않은 팀
- [x] Paired samples t-test: the same trained agents, before vs after training<br>대응표본 t-검정: 같은 교육받은 상담원, 교육 전 vs 후
- [x] Summarize which comparison gives the strongest evidence<br>어느 비교가 가장 강한 증거를 주는지 요약하기

````python
np.random.seed(20)

# (1) One-sample: trained team's handle time vs 6-min target
trained_team_now = np.random.normal(5.6, 1.3, 35)
t1, p1 = stats.ttest_1samp(trained_team_now, popmean=6.0)

# (2) Independent: trained team vs separate untrained team
untrained_team = np.random.normal(6.3, 1.4, 30)
t2, p2 = stats.ttest_ind(trained_team_now, untrained_team, equal_var=False)

# (3) Paired: SAME trained agents, before vs after
n_agents = 25
before_training = np.random.normal(6.4, 1.1, n_agents)
after_training = before_training - np.random.normal(0.7, 0.5, n_agents)
t3, p3 = stats.ttest_rel(before_training, after_training)

print("(1) One-sample: trained team vs 6-min target")
print(f"    mean={trained_team_now.mean():.2f} min, t={t1:.3f}, p={p1:.4f}")
print()
print("(2) Independent: trained vs untrained team")
print(f"    trained={trained_team_now.mean():.2f} min, untrained={untrained_team.mean():.2f} min")
print(f"    t={t2:.3f}, p={p2:.4f}")
print()
print("(3) Paired: same agents, before vs after")
print(f"    before={before_training.mean():.2f} min, after={after_training.mean():.2f} min")
print(f"    t={t3:.3f}, p={p3:.6f}")
print()

results = {'One-sample': p1, 'Independent': p2, 'Paired': p3}
strongest = min(results, key=results.get)
print(f"Strongest evidence: {strongest} test (p={results[strongest]:.6f})")
print("The paired test isolates each agent's own improvement and removes team-to-team")
print("variability, which is usually why it shows the clearest signal.")

print()
print(f"(1) 단일표본: 교육팀 vs 6분 목표 -- 평균={trained_team_now.mean():.2f}분, p={p1:.4f}")
print(f"(2) 독립표본: 교육팀 vs 비교육팀 -- t={t2:.3f}, p={p2:.4f}")
print(f"(3) 대응표본: 같은 상담원 전후 -- t={t3:.3f}, p={p3:.6f}")
print(f"가장 강한 증거: {strongest} 검정")
print("대응 검정은 각 상담원 자신의 개선을 분리하고 팀 간 변동을 제거하므로")
print("보통 가장 명확한 신호를 보여줍니다.")
````

# 📝 Summary (요약)
*Write today's concept in 3~5 sentences.*

**EN:** A t-test is Section 6's hypothesis-testing framework applied specifically to means, and comes in three flavors matched to three data shapes: a one-sample test compares a group's mean to a known value, an independent-samples test compares two separate groups, and a paired-samples test compares two measurements on the same subjects — with the paired version typically far more statistically powerful for the same sample size, since it removes person-to-person variability entirely. Before trusting any t-test result, its assumptions are worth a quick check: is variance roughly equal between groups (Levene's test decides whether to use the standard test or Welch's correction), and is the sample large enough for the Central Limit Theorem to rescue non-normal raw data. None of the specific tests are new concepts — they're Section 6's H0/p-value/alpha vocabulary wearing three different, concrete formulas.

**KR:** t-검정은 섹션 6의 가설검정 틀을 평균에 구체적으로 적용한 것이며, 세 가지 데이터 형태에 맞는 세 가지 유형으로 나뉩니다: 단일표본 검정은 그룹의 평균을 알려진 값과 비교하고, 독립표본 검정은 두 개의 별도 그룹을 비교하며, 대응표본 검정은 같은 대상에서 측정한 두 값을 비교합니다 -- 대응표본 버전은 개인 간 변동을 완전히 제거하기 때문에 같은 표본 크기에서 보통 통계적으로 훨씬 더 강력합니다. 어떤 t-검정 결과든 신뢰하기 전에 그 가정을 빠르게 확인할 가치가 있습니다: 그룹 간 분산이 대략 같은지(Levene 검정이 표준 검정과 Welch 보정 중 무엇을 쓸지 결정), 그리고 표본이 중심극한정리가 비정규 원본 데이터를 구제할 만큼 충분히 큰지. 구체적인 검정들은 새로운 개념이 아닙니다 -- 섹션 6의 H0/p-value/alpha 용어가 세 가지 다른 구체적인 공식을 입은 것뿐입니다.

ttest_rel() removes the differences between people, so the standard error and the p-value get smaller.  
ttest_rel()은 사람 간 차이를 없애서 표준오차와 p-value가 작아집니다.

It tests equal variances; if the p-value is small, run Welch's t-test.  
등분산을 검정하며, p-value가 작으면 Welch t-검정을 돌립니다.

**Q4 (EN).** Why is `equal_var=False` (Welch's t-test) often recommended as a safe default, even when you haven't checked variances first?
**Q4 (KR).** 먼저 분산을 확인하지 않았어도 `equal_var=False`(Welch's t-test)가 안전한 기본값으로 자주 권장되는 이유는 무엇인가요?

Welch's test is about the same as the standard test when variances are equal, and safer when they differ.  
Welch 검정은 분산이 같으면 일반 검정과 비슷하고, 다르면 더 안전합니다.

**Q5 (EN).** A t-test comparing a new feature's engagement to the old version comes back with p=0.35. Can you conclude the two versions perform equally well? Why or why not?
**Q5 (KR).** 새 기능의 참여도를 기존 버전과 비교한 t-검정 결과 p=0.35가 나왔습니다. 두 버전이 동등하게 잘 작동한다고 결론 내릴 수 있나요? 왜 그런가요, 왜 아닌가요?

No. p=0.35 only means there is not enough evidence to reject H0, not that they are equal.  
아니요. p=0.35는 H0을 기각할 증거가 부족하다는 뜻이지, 같다는 뜻이 아닙니다.
