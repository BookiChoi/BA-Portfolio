# 🧠 Concept (개념)

## What is it? (이것은 무엇인가?)
*(Explain it in your own words.)*

**EN:** Hypothesis testing is a formal procedure for deciding whether a pattern you observe in data (like a difference between two groups) is likely a real effect, or just the kind of random variation you'd expect even if nothing changed. It starts by stating two competing claims — the null hypothesis (H0, "nothing's going on") and the alternative hypothesis (H1, "something's going on") — then uses the data to decide which one the evidence favors.

**KR:** 가설검정은 데이터에서 관찰한 패턴(두 그룹 간 차이 등)이 실제 효과일 가능성이 높은지, 아니면 아무것도 바뀌지 않았어도 예상되는 종류의 무작위 변동일 뿐인지를 판단하는 공식적인 절차입니다. 두 개의 경쟁하는 주장 — 귀무가설(H0, "아무 일도 없다")과 대립가설(H1, "무언가 일어나고 있다") — 을 세우는 것으로 시작해, 데이터가 어느 쪽을 지지하는지 판단합니다.

### Framing a hypothesis (가설 프레이밍)

| | Null (H0) / 귀무가설 | Alternative (H1) / 대립가설 |
|---|---|---|
| Meaning / 의미 | No effect / no difference<br>효과 없음 / 차이 없음 | There IS an effect / difference<br>효과가 있다 / 차이가 있다 |
| A/B test example<br>A/B 테스트 예시 | New flow has the SAME conversion rate as old<br>새 플로우가 기존과 같은 전환율 | New flow has a DIFFERENT conversion rate<br>새 플로우가 다른 전환율 |
| Default assumption<br>기본 가정 | Assumed true until evidence says otherwise<br>증거가 없으면 참으로 가정 | What you're trying to find evidence FOR<br>증거를 찾으려는 대상 |

### The four possible outcomes (네 가지 가능한 결과)

| | H0 is actually TRUE / 실제로 H0 참 | H0 is actually FALSE / 실제로 H0 거짓 |
|---|---|---|
| Reject H0<br>H0 기각 | Type I Error (false positive), rate = α<br>1종 오류(오탐), 비율 = α | Correct decision (true positive)<br>올바른 결정(진양성) |
| Fail to reject H0<br>H0 기각 실패 | Correct decision (true negative)<br>올바른 결정(진음성) | Type II Error (false negative), rate = β<br>2종 오류(미탐), 비율 = β |

## Why do we use it? (왜 사용하는가?)
*(When is it useful?)*

**EN:** Without a formal test, "Variant B converted at 5.2% vs Variant A's 4.8%" could mean B is genuinely better, or it could mean you got a slightly lucky sample — humans are notoriously bad at telling real signal from noise by eyeballing numbers. Hypothesis testing gives you a disciplined, repeatable way to draw that line, using the exact same standard error and distribution machinery built in Sections 3-5.

**KR:** 공식적인 검정 없이는 "변형 B가 5.2%, 변형 A가 4.8% 전환"이 B가 정말 더 나은지, 아니면 살짝 운이 좋은 표본을 얻은 것인지 알 수 없습니다 -- 사람은 숫자만 봐서 실제 신호와 노이즈를 구분하는 데 악명 높게 서투릅니다. 가설검정은 섹션 3~5에서 만든 표준오차·분포 도구를 정확히 그대로 사용해, 그 경계를 규율 있고 반복 가능한 방식으로 그어줍니다.

## When is it used in Business Analytics? (비즈니스 분석에서 언제 사용되는가?)
*(Real-world use case)*

**EN:** Every A/B test result, every "did this change actually help" question, and any claim that starts with "the data shows a difference" — the formal vocabulary (null/alternative, p-value, significance, power) covered here is the exact language used starting with Section 7's t-tests all the way through the A/B testing sections later in this curriculum.

**KR:** 모든 A/B 테스트 결과, "이 변경이 실제로 도움이 됐는가"라는 모든 질문, 그리고 "데이터가 차이를 보여준다"로 시작하는 모든 주장. 여기서 다루는 공식 용어(귀무/대립가설, p-value, 유의성, 검정력)는 섹션 7의 t-검정부터 이 커리큘럼 후반의 A/B 테스트 섹션까지 정확히 그대로 사용되는 언어입니다.

# 📝 Syntax (문법)

## Basic Syntax (기본 문법)

````python
import numpy as np
from scipy import stats

# Business scenario: is the new checkout flow's average order value different
# from the historical average of $50?
# H0: true mean = $50 (no change) / H1: true mean != $50 (something changed)
# 비즈니스 시나리오: 새 결제 플로우의 평균 주문금액이 과거 평균 $50과 다른가?
# H0: 실제 평균 = $50(변화 없음) / H1: 실제 평균 != $50(변화 있음)

np.random.seed(42)
new_flow_orders = np.random.normal(loc=54, scale=15, size=100)
historical_mean = 50

t_statistic, p_value = stats.ttest_1samp(new_flow_orders, popmean=historical_mean)

print(f"Sample mean: ${new_flow_orders.mean():.2f}")
print(f"t-statistic: {t_statistic:.3f}")
print(f"p-value: {p_value:.4f}")
````

## Common Variations (자주 쓰는 변형)

````python
# The decision rule: compare the p-value to a significance level (alpha)
# 결정 규칙: p-value를 유의수준(alpha)과 비교
alpha = 0.05
if p_value < alpha:
    print(f"p={p_value:.4f} < alpha={alpha} -> REJECT H0 (statistically significant)")
    print(f"p={p_value:.4f} < alpha={alpha} -> H0 기각 (통계적으로 유의미함)")
else:
    print(f"p={p_value:.4f} >= alpha={alpha} -> FAIL TO REJECT H0 (not statistically significant)")
    print(f"p={p_value:.4f} >= alpha={alpha} -> H0 기각 실패 (통계적으로 유의미하지 않음)")
````

# 🧪 Small Examples (예제)

## Example 1 — Framing a Business Question as H0 vs H1 (비즈니스 질문을 H0 vs H1로 프레이밍하기)

````python
# Business question: does a new email subject line change the open rate,
# compared to the historical baseline of 22%?
# H0: true open rate = 22% (no change) / H1: true open rate != 22% (changed)
# 비즈니스 질문: 새 이메일 제목이 과거 기준 오픈율 22%와 비교해 오픈율을
# 바꾸는가? H0: 실제 오픈율 = 22%(변화 없음) / H1: 실제 오픈율 != 22%(변화 있음)

np.random.seed(1)
n_sent = 2000
historical_open_rate = 0.22
opened = np.random.random(n_sent) < 0.245  # true new rate is actually 24.5%

observed_rate = opened.mean()
print(f"H0: true open rate = {historical_open_rate:.1%}")
print(f"H1: true open rate != {historical_open_rate:.1%}")
print(f"Observed open rate: {observed_rate:.1%} (n={n_sent})")
print()
print("Framing it this way turns a vague question ('did the new subject line work?')")
print("into something we can formally test with the machinery from Sections 3-5")
print("(distributions, standard error, confidence intervals).")
print("이렇게 프레이밍하면 모호한 질문('새 제목이 효과가 있었나?')이 섹션 3~5의")
print("도구(분포, 표준오차, 신뢰구간)로 공식적으로 검정할 수 있는 것으로 바뀝니다.")
````

## Example 2 — What a p-value Actually Means (p-value가 실제로 의미하는 것)

````python
# Business question: what does a p-value ACTUALLY mean? Let's find out by
# simulating many "no real effect" experiments and seeing how often we get a
# result as extreme as our actual observation, purely by chance.
# 비즈니스 질문: p-value는 실제로 무엇을 의미하는가? 실제 효과가 없는 실험을
# 많이 시뮬레이션해서, 우리가 관측한 것만큼 극단적인 결과가 순전히 우연으로
# 얼마나 자주 나오는지 확인해보자.

np.random.seed(2)
n = 100
observed_mean = 54    # what we actually saw
true_null_mean = 50   # H0 claim
true_null_std = 15

# Simulate 10,000 experiments where H0 is TRUE (no real effect)
n_simulations = 10000
simulated_means = [np.random.normal(true_null_mean, true_null_std, n).mean()
                    for _ in range(n_simulations)]

# How many "no effect" simulations produced a mean AS OR MORE extreme than our observation?
as_extreme = sum(1 for m in simulated_means
                  if abs(m - true_null_mean) >= abs(observed_mean - true_null_mean))
simulated_p = as_extreme / n_simulations

# Compare to the actual t-test p-value on a real-ish sample
sample_data = np.random.normal(observed_mean, true_null_std, n)
t_stat, actual_p = stats.ttest_1samp(sample_data, popmean=true_null_mean)

print(f"Simulated p-value (from {n_simulations} 'no effect' experiments): {simulated_p:.4f}")
print(f"Actual t-test p-value: {actual_p:.4f}")
print()
print("A p-value is the probability of seeing data this extreme (or more extreme), IF H0")
print("were actually true. It is NOT the probability that H0 is true, and it is NOT the")
print("probability your result is due to chance -- both are common, incorrect readings.")
print("p-value는 H0가 실제로 참이라면 이 정도로 극단적인(또는 더 극단적인) 데이터를 볼")
print("확률입니다. H0가 참일 확률이 아니며, 결과가 우연 때문일 확률도 아닙니다 -- 둘 다")
print("흔하지만 잘못된 해석입니다.")
````

## Example 3 — Significance Level (alpha) and the Decision Rule (유의수준과 결정 규칙)

````python
# Business question: why is 0.05 the "standard" significance level, and what
# happens to your decision if you use a stricter or looser threshold?
# 비즈니스 질문: 왜 0.05가 "표준" 유의수준이며, 더 엄격하거나 느슨한
# 기준을 쓰면 결정이 어떻게 달라지는가?

np.random.seed(3)
sample = np.random.normal(53, 15, 80)
t_stat, p_value = stats.ttest_1samp(sample, popmean=50)

print(f"p-value: {p_value:.4f}")
print()
for a in [0.01, 0.05, 0.10]:
    decision = "REJECT H0" if p_value < a else "FAIL TO REJECT H0"
    print(f"alpha={a}: {decision}")

print()
print("The SAME data can lead to different decisions depending on the threshold you")
print("chose BEFORE looking at the data. alpha=0.05 (a 5% false-positive tolerance) is")
print("a widely used convention (from statistician Ronald Fisher), not a law of nature --")
print("fields with higher stakes (e.g. medicine) often use alpha=0.01.")
print("같은 데이터라도 데이터를 보기 '전에' 선택한 임계값에 따라 다른 결정으로 이어질")
print("수 있습니다. alpha=0.05(5% 오탐 허용)는 널리 쓰이는 관행(통계학자 Ronald")
print("Fisher가 제안)이지 자연의 법칙이 아닙니다 -- 위험 부담이 더 큰 분야(의학 등)는")
print("보통 alpha=0.01을 사용합니다.")
````

## Example 4 — Type I/II Errors and Statistical Power (1종·2종 오류와 검정력)

````python
# Business question: if we ship a change with NO real effect, how often will
# we falsely conclude it worked (Type I error)? If it DOES have a real, modest
# effect, how often will we correctly detect it (power)?
# 비즈니스 질문: 실제 효과가 없는 변경을 배포했을 때, 효과가 있다고 잘못
# 결론 내릴 확률(1종 오류)은? 실제로 적당한 효과가 있다면, 그것을
# 올바르게 탐지할 확률(검정력)은?

np.random.seed(4)
alpha = 0.05
n = 60
true_std = 15
n_sims = 2000

# Scenario A: NO real effect (H0 is actually true) -- how often do we wrongly reject H0?
false_positives = 0
for _ in range(n_sims):
    sample = np.random.normal(50, true_std, n)  # true mean IS 50, matching H0
    _, p = stats.ttest_1samp(sample, popmean=50)
    if p < alpha:
        false_positives += 1

# Scenario B: a REAL, modest effect exists (true mean is 55) -- how often do we detect it?
true_positives = 0
for _ in range(n_sims):
    sample = np.random.normal(55, true_std, n)  # true mean IS 55, H0 (50) is false
    _, p = stats.ttest_1samp(sample, popmean=50)
    if p < alpha:
        true_positives += 1

type_1_rate = false_positives / n_sims
power = true_positives / n_sims

print(f"Type I error rate (false positive, when H0 is TRUE): {type_1_rate:.3f} (target: ~{alpha})")
print(f"Statistical power (correct detection, when H0 is FALSE): {power:.3f}")
print(f"Type II error rate (missed real effect): {1 - power:.3f}")
print()
print(f"Type I error rate lands close to alpha={alpha} by design -- that's exactly what alpha")
print(f"controls. Power of {power:.1%} means we'd MISS a real effect this size about")
print(f"{(1 - power) * 100:.0f}% of the time -- a business argument for a bigger sample size.")
print(f"1종 오류율이 alpha={alpha}에 근접하는 것은 설계상 당연합니다 -- 바로 alpha가")
print(f"통제하는 것입니다. 검정력 {power:.1%}는 이만한 크기의 실제 효과를 약")
print(f"{(1 - power) * 100:.0f}%의 확률로 놓친다는 뜻입니다 -- 더 큰 표본 크기가 필요하다는")
print("비즈니스적 근거가 됩니다.")
````

# ⚠️ Common Mistakes (자주 하는 실수)

### Mistake 1 — Interpreting a p-value as "the Probability H0 Is True" (p-value를 "H0가 참일 확률"로 오해)
A p-value is calculated ASSUMING H0 is true — it can never turn around and tell you the probability that assumption itself is correct. "p=0.03" means "if nothing had changed, data this extreme would show up 3% of the time," not "there's a 3% chance nothing changed."  
p-value는 H0가 참이라고 가정한 상태에서 계산됩니다 -- 그 가정 자체가 맞을 확률을 거꾸로 말해줄 수 없습니다. "p=0.03"은 "아무것도 바뀌지 않았다면, 이 정도로 극단적인 데이터가 3%의 확률로 나타난다"는 뜻이지, "아무것도 바뀌지 않았을 확률이 3%다"라는 뜻이 아닙니다.

✅ **Fix / 해결:** Always phrase it as "assuming no effect, this result (or more extreme) would occur X% of the time" — never as "there's an X% chance the null is true."

### Mistake 2 — Treating Statistical Significance as Proof of a BIG Effect (통계적 유의성을 큰 효과의 증거로 취급)
With a large enough sample, even a tiny, practically meaningless difference (like a 0.01% conversion lift) can produce a very small p-value. Statistical significance answers "is this probably real?", not "is this big enough to matter?"  
충분히 큰 표본이면 실질적으로 무의미하게 작은 차이(0.01% 전환율 상승 등)도 매우 작은 p-value를 만들 수 있습니다. 통계적 유의성은 "이것이 아마 실제일까?"에 답할 뿐, "이것이 신경 쓸 만큼 큰가?"에는 답하지 않습니다.

✅ **Fix / 해결:** Always look at the effect size (and its confidence interval) alongside the p-value, not the p-value alone.

### Mistake 3 — Choosing Alpha After Seeing the p-value (p-value를 본 뒤 alpha를 정함)
If a test gives p=0.06 and you're tempted to say "well, let's use alpha=0.10 instead of 0.05," you've turned an objective test into a foregone conclusion — the threshold has to be locked in BEFORE looking at results to mean anything.  
검정 결과 p=0.06이 나왔는데 "그럼 0.05 대신 alpha=0.10을 쓰자"라고 하고 싶어진다면, 객관적인 검정을 이미 정해진 결론으로 바꿔버린 것입니다 -- 임계값은 결과를 보기 전에 고정되어야만 의미가 있습니다.

✅ **Fix / 해결:** Set alpha as part of your study design, before collecting or looking at any data, and don't move it afterward.

### Mistake 4 — Treating "Fail to Reject H0" as "H0 Is Proven True" ("H0 기각 실패"를 "H0가 증명됨"으로 취급)
Failing to find enough evidence for an effect isn't the same as proving there's no effect — it might just mean your sample was too small (low power) to detect a real but modest difference.  
효과에 대한 충분한 증거를 찾지 못한 것은 효과가 없다는 것을 증명한 것과 다릅니다 -- 그저 표본이 너무 작아서(낮은 검정력) 실재하지만 적당한 크기의 차이를 탐지하지 못한 것일 수 있습니다.

✅ **Fix / 해결:** Say "we didn't find sufficient evidence of an effect" rather than "we proved there's no effect," and check the test's power before treating a null result as final.

### Mistake 5 — Ignoring Statistical Power Until After a "No Effect" Result (검정력을 "효과 없음" 결과 이후에야 확인)
A study with too few observations can have low power even for a real, meaningful effect — running the test anyway and reporting "no significant difference" without checking power first can mask a real business win.  
관측치가 너무 적은 연구는 실재하고 의미 있는 효과에 대해서도 낮은 검정력을 가질 수 있습니다 -- 검정력을 먼저 확인하지 않고 검정을 돌려 "유의미한 차이 없음"이라고 보고하면 실제 비즈니스 성과를 가릴 수 있습니다.

✅ **Fix / 해결:** Estimate the required sample size for a target power (commonly 80%) BEFORE running the test — a later section covers exactly this for A/B tests.

### Mistake 6 — Assuming Type I and Type II Errors Cost the Business the Same (1종·2종 오류가 비즈니스에 같은 비용이라고 가정)
Falsely flagging a good customer as fraud (Type I) and missing an actual fraudster (Type II) have very different costs to a business — the "right" balance between alpha and power should reflect which error is more expensive to make, not just default statistical conventions.  
정상 고객을 사기로 잘못 표시하는 것(1종)과 실제 사기꾼을 놓치는 것(2종)은 비즈니스에 매우 다른 비용을 안깁니다 -- alpha와 검정력 사이의 "올바른" 균형은 단순한 기본 통계 관행이 아니라 어느 오류가 더 비싼지를 반영해야 합니다.

✅ **Fix / 해결:** Before setting alpha, explicitly ask which error type is more costly in this specific business context, and lean the threshold accordingly.

# 💡 Tips (팁)
Useful tips or shortcuts / 유용한 팁과 단축법

- Frame your hypotheses BEFORE looking at the data, in writing — this single habit prevents most of the mistakes in this section, from p-hacking alpha to over-interpreting a significant result.<br>데이터를 보기 전에 가설을 글로 미리 세워두세요 -- 이 하나의 습관이 이 섹션의 실수 대부분(alpha p-해킹부터 유의미한 결과의 과잉 해석까지)을 막아줍니다.
- "Statistically significant" and "practically significant" are different claims — always ask "how big is the effect, and does that size matter for the business?" right after checking the p-value.<br>"통계적으로 유의미하다"와 "실질적으로 유의미하다"는 다른 주장입니다 -- p-value를 확인한 직후 항상 "효과가 얼마나 크며, 그 크기가 비즈니스적으로 중요한가?"를 물으세요.
- `scipy.stats` test functions return `(statistic, p_value)` as a tuple in that order — unpacking them the wrong way around is a common, silent bug.<br>scipy.stats 검정 함수는 (statistic, p_value)를 그 순서의 튜플로 반환합니다 -- 반대로 언패킹하면 흔하지만 조용한 버그가 생깁니다.
- The four-quadrant confusion matrix (Type I/II errors) from this section is the exact same shape as any classification model's confusion matrix — the vocabulary transfers directly if you later work with ML classifiers.<br>이 섹션의 4분면 혼동 행렬(1종·2종 오류)은 어떤 분류 모델의 혼동 행렬과도 정확히 같은 형태입니다 -- 나중에 ML 분류기를 다루게 되면 이 용어가 그대로 통합니다.
- When someone says "the result wasn't significant, so nothing happened," ask what the test's statistical power was — a low-power test proves much less than it sounds like it does.<br>누군가 "결과가 유의미하지 않아서 아무 일도 없었다"고 말하면, 그 검정의 검정력이 얼마였는지 물어보세요 -- 낮은 검정력의 검정은 들리는 것보다 훨씬 적은 것을 증명합니다.

# 💼 Business Example (비즈니스 예시)
*How would a Business Analyst use this?*

**Scenario / 시나리오:**

**EN:** Your company is deciding whether to launch a new pricing page. Leadership wants you to formally set up the test BEFORE it runs: what's the hypothesis, what significance level makes sense given the cost of a false positive (an expensive redesign for no real gain) versus a false negative (missing real revenue growth), and how would you interpret the result once it's in?

**KR:** 회사가 새로운 가격 페이지를 출시할지 결정하려 합니다. 경영진은 테스트를 돌리기 전에 공식적으로 설계하기를 원합니다: 가설은 무엇인지, 오탐(실제 이득 없이 비용이 큰 재설계)과 미탐(실제 매출 성장을 놓침)의 비용을 고려했을 때 어떤 유의수준이 타당한지, 그리고 결과가 나오면 어떻게 해석할 것인지.

**To-do / 할 일:**
- [x] State H0 and H1 clearly for the pricing page test<br>가격 페이지 테스트의 H0와 H1을 명확히 진술하기
- [x] Decide on alpha based on the relative cost of a false positive vs false negative<br>오탐 대비 미탐의 상대적 비용에 근거해 alpha 결정하기
- [x] Run the test on collected data and make the reject/fail-to-reject decision<br>수집된 데이터로 검정을 돌리고 기각/기각실패 결정 내리기
- [x] Write the correct interpretation of the p-value for the executive summary<br>경영진 요약을 위한 p-value의 올바른 해석 작성하기

````python
# Business setup: re-designing the pricing page costs ~$50K in dev/design time
# (expensive if we do it for no real gain -- Type I concern). Missing a real
# revenue lift, though, costs far more over a year -- so we lean toward a
# LESS strict alpha to avoid missing a real effect (Type II concern).
print("H0: new pricing page has the SAME average order value as the current page")
print("H1: new pricing page has a DIFFERENT average order value")
print()
print("Cost of Type I error (false positive): ~$50K wasted redesign")
print("Cost of Type II error (false negative): missed recurring revenue lift, likely bigger")
print("-> given Type II is costlier here, we choose a slightly LOOSER alpha = 0.10")
print("   instead of the conventional 0.05, to reduce the chance of missing a real effect")

np.random.seed(10)
alpha = 0.10
current_page = np.random.normal(48, 14, 90)
new_page = np.random.normal(52, 14, 90)

t_stat, p_value = stats.ttest_ind(new_page, current_page)

print()
print(f"Current page mean: ${current_page.mean():.2f}")
print(f"New page mean: ${new_page.mean():.2f}")
print(f"p-value: {p_value:.4f}")

decision = "REJECT H0 -- launch the new page" if p_value < alpha else "FAIL TO REJECT H0 -- keep the current page"
print(f"Decision at alpha={alpha}: {decision}")

print()
print("Correct interpretation for the exec summary: 'If the new page truly had no effect,")
print(f"we would see a difference this large (or larger) only {p_value * 100:.1f}% of the time")
print("by chance. Given our pre-agreed 10% threshold, we recommend launching.'")

print()
print("H0: 새 가격 페이지의 평균 주문금액이 현재 페이지와 같다")
print("H1: 새 가격 페이지의 평균 주문금액이 현재 페이지와 다르다")
print("1종 오류(오탐) 비용: 재설계 비용 약 $50K 낭비")
print("2종 오류(미탐) 비용: 놓친 매출 성장, 아마 더 큼")
print("-> 2종 오류가 더 비싸므로 관습적인 0.05 대신 조금 더 느슨한 alpha=0.10 선택")
print(f"현재 페이지 평균: ${current_page.mean():.2f}, 새 페이지 평균: ${new_page.mean():.2f}")
print(f"p-value: {p_value:.4f}, alpha={alpha}에서의 결정: {decision}")
````

# 📝 Summary (요약)
*Write today's concept in 3~5 sentences.*

**EN:** Hypothesis testing formalizes the question "is this difference real, or just noise?" by setting up two competing claims (H0: no effect, H1: there is an effect) and using a p-value — the probability of seeing data this extreme IF H0 were true — to decide between them. That decision compares the p-value to a pre-chosen significance level (alpha, conventionally 0.05, though not a law of nature), and every such decision carries risk in two directions: a Type I error (false positive, rejecting a true H0) at a rate you control directly via alpha, and a Type II error (false negative, missing a real effect) at a rate controlled by statistical power. None of these tools are new math — a p-value reuses Section 3's distributions and Section 4's standard error, applied to formally answer the question those sections could only describe informally.

**KR:** 가설검정은 "이 차이가 실제인가, 그냥 노이즈인가?"라는 질문을 두 개의 경쟁하는 주장(H0: 효과 없음, H1: 효과 있음)을 세우고, p-value(H0가 참이라면 이 정도로 극단적인 데이터를 볼 확률)를 사용해 둘 중 하나를 선택하는 방식으로 공식화합니다. 그 결정은 p-value를 사전에 정한 유의수준(alpha, 관습적으로 0.05이지만 자연의 법칙은 아님)과 비교하며, 모든 그런 결정은 두 방향의 위험을 안고 있습니다: alpha로 직접 통제하는 비율의 1종 오류(오탐, 참인 H0를 기각)와 검정력으로 통제하는 비율의 2종 오류(미탐, 실제 효과를 놓침). 이 도구들은 새로운 수학이 아닙니다 -- p-value는 섹션 3의 분포와 섹션 4의 표준오차를 재사용해, 그 섹션들이 비공식적으로만 설명할 수 있었던 질문에 공식적으로 답합니다.

"효과가 없다"가 증명된 것이 아닙니다. 표본이 작으면 실제 효과도 못 찾을 수 있습니다. "충분한 증거를 찾지 못했다"고 말합니다.
